## Results 1 – Genome-wide Z-DNA density (Fig. 1b–f, Suppl. Fig. 1a–c, Table 1)

### Analysis (not rerun)

#### 1 · HPRC haplotype metadata

```python
"""1 - HPRC haplotype metadata -> 1_hprc_metadata.tsv (464 haplotypes, 232 individuals; superpopulation per haplotype).
"""
from pathlib import Path

import pandas as pd

BASE = Path("/path/to/pangenomes_zdna")
META_FILE = BASE / "0_hprc_release2_sample_metadata.csv"
CSV_DIR = BASE / "hprc_zseeker"
OUT_TSV = BASE / "1_hprc_metadata.tsv"

SKIP_FILES = {
    "CHM13_v2.0_maskedY_rCRS.fa.PanSN.zseeker.csv",
    "GRCh38_GCA_000001405.15_no_alt_analysis_set.PanSN.zseeker.csv",
}
N_EXPECTED = 464

SUPERPOPULATION_MAP = {
    "ACB": "AFR", "ASL": "AFR", "ASW": "AFR", "ESN": "AFR", "GWD": "AFR",
    "LWK": "AFR", "MKK": "AFR", "MSL": "AFR", "YRI": "AFR",
    "CLM": "AMR", "MXL": "AMR", "PEL": "AMR", "PUR": "AMR",
    "CDX": "EAS", "CHB": "EAS", "CHS": "EAS", "JPT": "EAS", "KHV": "EAS",
    "FIN": "EUR", "GBR": "EUR", "IBS": "EUR", "TSI": "EUR",
    "BEB": "SAS", "GIH": "SAS", "ITU": "SAS", "PJL": "SAS", "STU": "SAS",
    "N/A": None,
}
SAMPLE_SUPERPOPULATION = {   # samples without a population code (NIST reference materials)
    "HG002": "EUR",          # RM 8391, GM24385: Eastern European Ashkenazi Jewish ancestry
    "HG005": "EAS",          # RM 8393, GM24631: Chinese ancestry
}
SUPERPOPULATIONS = {"AFR", "AMR", "EAS", "EUR", "SAS"}


def main():
    meta = pd.read_csv(META_FILE, dtype=str, keep_default_na=False)
    if meta["sample_id"].duplicated().any():
        raise ValueError("Duplicate sample_id in sample metadata")

    records = []
    for f in sorted(CSV_DIR.glob("*.zseeker.csv")):
        if f.name in SKIP_FILES:
            continue
        low = f.name.lower()
        records.append({
            "filename": f.name,
            "sample_id": f.name.split("_")[0],
            "pat_or_mat": "pat" if "pat" in low else "mat" if "mat" in low else "0",
            "hap1_or_hap2": "hap1" if "hap1" in low else "hap2" if "hap2" in low else "0",
        })
    out = pd.DataFrame(records).merge(meta, on="sample_id", how="left", validate="many_to_one", indicator=True)
    missing = out.loc[out["_merge"].ne("both"), "filename"].tolist()
    if missing:
        raise ValueError(f"Haplotypes without sample metadata: {missing}")
    out = out.drop(columns="_merge")

    unknown = sorted(set(out["population_abbreviation"]) - set(SUPERPOPULATION_MAP))
    if unknown:
        raise ValueError(f"Population codes missing from SUPERPOPULATION_MAP: {unknown}")
    out["superpopulation"] = out["population_abbreviation"].map(SUPERPOPULATION_MAP)
    out["superpopulation"] = out["sample_id"].map(SAMPLE_SUPERPOPULATION).fillna(out["superpopulation"])

    if len(out) != N_EXPECTED or not out["filename"].is_unique:
        raise ValueError(f"Expected {N_EXPECTED} unique haplotypes, found {len(out)} ({out['filename'].nunique()} unique)")
    bad = out.loc[~out["superpopulation"].isin(SUPERPOPULATIONS), "filename"].tolist()
    if bad:
        raise ValueError(f"Haplotypes without a superpopulation: {bad}")
    if out.duplicated(["sample_id", "pat_or_mat", "hap1_or_hap2"]).any():
        raise ValueError("More than one haplotype per sample and haplotype label")

    out.to_csv(OUT_TSV, sep="\t", index=False)
    label = out["pat_or_mat"].where(out["pat_or_mat"].ne("0"), out["hap1_or_hap2"])
    print(f"Haplotypes: {len(out)} from {out['sample_id'].nunique()} individuals")
    print(f"Superpopulation: {out['superpopulation'].value_counts().reindex(sorted(SUPERPOPULATIONS), fill_value=0).to_dict()}")
    print(f"Haplotype label: {label.value_counts().to_dict()}")
    print(f"Sex (individuals): {out.drop_duplicates('sample_id')['sex'].value_counts().to_dict()}")
    print(f"Assigned without population code: "
          f"{out[out['sample_id'].isin(SAMPLE_SUPERPOPULATION)][['filename', 'superpopulation']].to_dict('records')}")
    print(f"Output: {OUT_TSV.name}")


if __name__ == "__main__":
    main()
```

#### 2 · CHM13 and GRCh38 densities → Fig. 1b, 1f, Table 1

```python
"""2 - Genome-wide and per-chromosome Z-DNA density of CHM13 and GRCh38 primary chromosomes (Fig. 1b, 1f, Table 1 inputs).
"""
from __future__ import annotations

import gzip
from pathlib import Path

import numpy as np
import pandas as pd
from Bio.SeqIO.FastaIO import SimpleFastaParser


ASSEMBLIES = {
    "GRCh38": {
        "fasta": Path(
            "/path/to/pangenomes_zdna/tmp/grch38/grch38_refseq/"
            "GCF_000001405.40_GRCh38.p14_primary_noMT_ucsc.fna"
        ),
        "zseeker_csv": Path(
            "/path/to/pangenomes_zdna/tmp/grch38/grch38_refseq/"
            "GCF_000001405.40_GRCh38.p14_primary_noMT_ucscnames_zseeker.csv"
        ),
    },
    "CHM13": {
        "fasta": Path(
            "/path/to/pangenomes_zdna/tmp/chm13/chm13_refseq/"
            "GCF_009914755.1_T2T-CHM13v2.0_genomic_ucsc.fna"
        ),
        "zseeker_csv": Path(
            "/path/to/pangenomes_zdna/tmp/chm13/chm13_zdna_files/"
            "GCF_009914755.1_T2T-CHM13v2.0_genomic_zdna_score.csv"
        ),
    },
}

OUT_DIR = Path("/path/to/pangenomes_zdna")
MIN_SEGMENT_TIERS = (1, 9, 50)


def open_fasta(path: Path):
    if path.suffix == ".gz":
        return gzip.open(path, "rt")
    return open(path, "rt", encoding="utf-8")


def segment_lengths(seq: str) -> np.ndarray:
    arr = np.frombuffer(
        seq.upper().encode("ascii", errors="replace"),
        dtype=np.uint8,
    )

    mask = (
        (arr == ord("A"))
        | (arr == ord("C"))
        | (arr == ord("G"))
        | (arr == ord("T"))
    )

    if not mask.any():
        return np.empty(0, dtype=np.int64)

    padded = np.concatenate(([False], mask, [False]))
    diffs = np.diff(padded.astype(np.int8))

    starts = np.flatnonzero(diffs == 1)
    ends = np.flatnonzero(diffs == -1)

    return (ends - starts).astype(np.int64)


def compute_effective_lengths(fasta_path: Path) -> pd.DataFrame:
    rows = []

    with open_fasta(fasta_path) as fh:
        for header, seq in SimpleFastaParser(fh):
            chrom = header.split()[0]
            raw_len = len(seq)
            seg_lens = segment_lengths(seq)

            row = {
                "ucsc_chr": chrom,
                "raw_length_bp": raw_len,
            }

            for tier in MIN_SEGMENT_TIERS:
                row[f"eff_len_min{tier}_bp"] = int(
                    seg_lens[seg_lens >= tier].sum()
                )

            rows.append(row)

    return pd.DataFrame(rows)


def chr_sort_key(chrom: str):
    chrom = str(chrom)

    if chrom.startswith("chr"):
        rest = chrom[3:]

        if rest.isdigit():
            return (0, int(rest))
        if rest == "X":
            return (1, 23)
        if rest == "Y":
            return (1, 24)
        if rest in ("M", "MT"):
            return (2, 25)

        return (3, rest)

    return (4, chrom)


def compute_density(
    zseeker_csv: Path,
    eff_df: pd.DataFrame,
) -> tuple[pd.DataFrame, pd.DataFrame]:

    z = pd.read_csv(zseeker_csv)

    required = {"Chromosome", "Start", "End"}
    missing = required - set(z.columns)

    if missing:
        raise KeyError(
            f"Missing columns in {zseeker_csv}: {sorted(missing)}"
        )

    z["ucsc_chr"] = z["Chromosome"].astype(str).str.strip()
    z["Start"] = pd.to_numeric(z["Start"], errors="coerce")
    z["End"] = pd.to_numeric(z["End"], errors="coerce")

    bad = z[z[["ucsc_chr", "Start", "End"]].isna().any(axis=1)]

    if len(bad):
        raise ValueError(
            f"NA in ucsc_chr/Start/End:\n{bad.head(10)}"
        )

    z["zdna_bp"] = (z["End"] - z["Start"] + 1).astype("int64")

    if (z["zdna_bp"] <= 0).any():
        raise ValueError("Non-positive zdna_bp detected.")

    per_chr_zdna = (
        z.groupby("ucsc_chr", as_index=False)["zdna_bp"]
        .sum()
        .rename(columns={"zdna_bp": "total_zdna_bp"})
    )

    per_chr = eff_df.merge(
        per_chr_zdna,
        on="ucsc_chr",
        how="left",
    )

    per_chr["total_zdna_bp"] = (
        per_chr["total_zdna_bp"]
        .fillna(0)
        .astype("int64")
    )

    for tier in MIN_SEGMENT_TIERS:
        eff_col = f"eff_len_min{tier}_bp"
        denom = per_chr[eff_col].astype("float64")

        per_chr[f"density_fraction_min{tier}"] = np.where(
            denom > 0,
            per_chr["total_zdna_bp"] / denom,
            np.nan,
        )

        per_chr[f"density_bp_per_Mb_min{tier}"] = np.where(
            denom > 0,
            per_chr["total_zdna_bp"] / (denom / 1e6),
            np.nan,
        )

    per_chr = (
        per_chr
        .assign(__sort=per_chr["ucsc_chr"].map(chr_sort_key))
        .sort_values("__sort")
        .drop(columns="__sort")
        .reset_index(drop=True)
    )

    total_zdna = int(per_chr["total_zdna_bp"].sum())
    total_raw = int(per_chr["raw_length_bp"].sum())

    summary_rows = []

    for tier in MIN_SEGMENT_TIERS:
        total_eff = int(
            per_chr[f"eff_len_min{tier}_bp"].sum()
        )

        summary_rows.append({
            "min_segment": tier,
            "total_zdna_bp": total_zdna,
            "total_raw_length_bp": total_raw,
            "total_effective_length_bp": total_eff,
            "fraction_of_raw_excluded":
                1.0 - total_eff / total_raw
                if total_raw > 0 else np.nan,
            "density_fraction":
                total_zdna / total_eff
                if total_eff > 0 else np.nan,
            "density_bp_per_Mb":
                total_zdna / (total_eff / 1e6)
                if total_eff > 0 else np.nan,
        })

    return per_chr, pd.DataFrame(summary_rows)


def make_comparison(
    chm13: pd.DataFrame,
    grch38: pd.DataFrame,
) -> pd.DataFrame:

    a = chm13[
        ["ucsc_chr", "density_bp_per_Mb_min1"]
    ].rename(
        columns={
            "density_bp_per_Mb_min1":
                "CHM13_density_bp_per_Mb"
        }
    )

    b = grch38[
        ["ucsc_chr", "density_bp_per_Mb_min1"]
    ].rename(
        columns={
            "density_bp_per_Mb_min1":
                "GRCh38_density_bp_per_Mb"
        }
    )

    comp = a.merge(
        b,
        on="ucsc_chr",
        how="inner",
    )

    comp["difference_CHM13_minus_GRCh38_bp_per_Mb"] = (
        comp["CHM13_density_bp_per_Mb"]
        - comp["GRCh38_density_bp_per_Mb"]
    )

    comp["absolute_difference_bp_per_Mb"] = (
        comp[
            "difference_CHM13_minus_GRCh38_bp_per_Mb"
        ].abs()
    )

    comp["percent_difference_vs_GRCh38"] = (
        comp[
            "difference_CHM13_minus_GRCh38_bp_per_Mb"
        ]
        / comp["GRCh38_density_bp_per_Mb"]
        * 100
    )

    comp["absolute_percent_difference"] = (
        comp["percent_difference_vs_GRCh38"].abs()
    )

    comp = (
        comp
        .assign(__sort=comp["ucsc_chr"].map(chr_sort_key))
        .sort_values("__sort")
        .drop(columns="__sort")
        .reset_index(drop=True)
    )

    return comp


def main():
    OUT_DIR.mkdir(parents=True, exist_ok=True)

    results = {}

    for assembly_name, paths in ASSEMBLIES.items():

        fasta_path = paths["fasta"]
        zseeker_csv = paths["zseeker_csv"]

        if not fasta_path.is_file():
            raise FileNotFoundError(fasta_path)

        if not zseeker_csv.is_file():
            raise FileNotFoundError(zseeker_csv)

        print(f"\nProcessing {assembly_name}")
        print(f"FASTA:   {fasta_path}")
        print(f"ZSeeker: {zseeker_csv}")

        eff_df = compute_effective_lengths(fasta_path)

        per_chr, summary = compute_density(
            zseeker_csv,
            eff_df,
        )

        eff_file = (
            OUT_DIR
            / f"2_{assembly_name}_effective_chrom_sizes.tsv"
        )

        per_chr_file = (
            OUT_DIR
            / f"2_{assembly_name}_zdna_density_per_chr.tsv"
        )

        summary_file = (
            OUT_DIR
            / f"2_{assembly_name}_zdna_density_genome_summary.tsv"
        )

        eff_df.to_csv(
            eff_file,
            sep="\t",
            index=False,
        )

        per_chr.to_csv(
            per_chr_file,
            sep="\t",
            index=False,
        )

        summary.to_csv(
            summary_file,
            sep="\t",
            index=False,
        )

        results[assembly_name] = per_chr

        print(f"Wrote {eff_file}")
        print(f"Wrote {per_chr_file}")
        print(f"Wrote {summary_file}")

    comparison = make_comparison(
        results["CHM13"],
        results["GRCh38"],
    )

    comparison_file = (
        OUT_DIR
        / "2_CHM13_vs_GRCh38_zdna_density_per_chr.tsv"
    )

    comparison.to_csv(
        comparison_file,
        sep="\t",
        index=False,
    )

    print(f"\nWrote {comparison_file}")

    print("\nGenome-wide summaries:")

    for assembly_name in ("GRCh38", "CHM13"):
        summary_file = (
            OUT_DIR
            / f"2_{assembly_name}_zdna_density_genome_summary.tsv"
        )
        print(f"\n{assembly_name}")
        print(
            pd.read_csv(
                summary_file,
                sep="\t",
            ).to_string(index=False)
        )

    print("\nCHM13 vs GRCh38:")
    print(comparison.to_string(index=False))


if __name__ == "__main__":
    main()
```

#### 2 · GRCh38 with unlocalized/unplaced contigs (text)

```python
"""2 - GRCh38 Z-DNA density over primary, unlocalized and unplaced contigs (text value).
"""
from __future__ import annotations

from pathlib import Path

import numpy as np
import pandas as pd
from Bio.SeqIO.FastaIO import SimpleFastaParser


FASTA = Path(
    "/path/to/pangenomes_zdna/tmp/grch38/grch38_refseq/"
    "grch38.p14.primary_and_unlocalized.fna"
)

ZSEEKER_CSV = Path(
    "/path/to/pangenomes_zdna/tmp/grch38/grch38_refseq/"
    "grch38.p14.primary_and_unlocalized_zseeker.csv"
)

OUT_TSV = Path(
    "/path/to/pangenomes_zdna/"
    "2_GRCh38_primary_and_unlocalized_zdna_density_summary.tsv"
)

MIN_SEGMENT_TIERS = (1, 9, 50)


def segment_lengths(seq: str) -> np.ndarray:
    arr = np.frombuffer(
        seq.upper().encode("ascii", errors="replace"),
        dtype=np.uint8,
    )

    mask = (
        (arr == ord("A"))
        | (arr == ord("C"))
        | (arr == ord("G"))
        | (arr == ord("T"))
    )

    if not mask.any():
        return np.empty(0, dtype=np.int64)

    padded = np.concatenate(([False], mask, [False]))
    diffs = np.diff(padded.astype(np.int8))

    starts = np.flatnonzero(diffs == 1)
    ends = np.flatnonzero(diffs == -1)

    return (ends - starts).astype(np.int64)


def compute_effective_genome_lengths(fasta_path: Path):
    total_raw = 0
    total_eff = {tier: 0 for tier in MIN_SEGMENT_TIERS}
    n_sequences = 0

    with open(fasta_path, "rt", encoding="utf-8") as fh:
        for header, seq in SimpleFastaParser(fh):
            n_sequences += 1
            total_raw += len(seq)

            seg_lens = segment_lengths(seq)

            for tier in MIN_SEGMENT_TIERS:
                total_eff[tier] += int(
                    seg_lens[seg_lens >= tier].sum()
                )

    return n_sequences, total_raw, total_eff


def compute_total_zdna(zseeker_csv: Path):
    z = pd.read_csv(zseeker_csv)

    required = {"Chromosome", "Start", "End"}
    missing = required - set(z.columns)

    if missing:
        raise KeyError(
            f"Missing columns in {zseeker_csv}: {sorted(missing)}"
        )

    if z["Chromosome"].isna().any():
        bad = z[z["Chromosome"].isna()]
        raise ValueError(
            f"Rows with missing Chromosome:\n{bad.head(10)}"
        )

    z["Chromosome"] = z["Chromosome"].astype(str).str.strip()
    z["Start"] = pd.to_numeric(z["Start"], errors="coerce")
    z["End"] = pd.to_numeric(z["End"], errors="coerce")

    no_prediction = z["Start"].isna() & z["End"].isna()

    malformed = z["Start"].isna() ^ z["End"].isna()

    if malformed.any():
        raise ValueError(
            "Rows with only one of Start/End missing:\n"
            f"{z.loc[malformed].head(10)}"
        )

    predictions = z.loc[~no_prediction].copy()

    predictions["zdna_bp"] = (
        predictions["End"] - predictions["Start"] + 1
    ).astype("int64")

    if (predictions["zdna_bp"] <= 0).any():
        raise ValueError(
            "Non-positive Z-DNA interval length detected."
        )

    total_zdna_bp = int(predictions["zdna_bp"].sum())
    n_predictions = len(predictions)

    n_sequences_in_zseeker = z["Chromosome"].nunique()

    n_sequences_with_predictions = (
        predictions["Chromosome"].nunique()
    )

    n_sequences_without_predictions = (
        n_sequences_in_zseeker
        - n_sequences_with_predictions
    )

    return (
        total_zdna_bp,
        n_predictions,
        n_sequences_in_zseeker,
        n_sequences_with_predictions,
        n_sequences_without_predictions,
    )

def main():
    if not FASTA.is_file():
        raise FileNotFoundError(FASTA)

    if not ZSEEKER_CSV.is_file():
        raise FileNotFoundError(ZSEEKER_CSV)

    print(f"FASTA:   {FASTA}")
    print(f"ZSeeker: {ZSEEKER_CSV}")

    n_sequences, total_raw, total_eff = (
        compute_effective_genome_lengths(FASTA)
    )

    (
        total_zdna_bp,
        n_predictions,
        n_sequences_in_zseeker,
        n_sequences_with_predictions,
        n_sequences_without_predictions,
    ) = compute_total_zdna(ZSEEKER_CSV)
    rows = []

    for tier in MIN_SEGMENT_TIERS:
        eff = total_eff[tier]

        rows.append({
            "min_segment": tier,
            "n_fasta_sequences": n_sequences,
            "n_sequences_in_zseeker": n_sequences_in_zseeker,
            "n_sequences_with_zdna": n_sequences_with_predictions,
            "n_sequences_without_zdna": n_sequences_without_predictions,
            "n_zdna_predictions": n_predictions,
            "total_zdna_bp": total_zdna_bp,
            "total_raw_length_bp": total_raw,
            "total_effective_length_bp": eff,
            "fraction_of_raw_excluded":
                1.0 - eff / total_raw
                if total_raw > 0 else np.nan,
            "density_fraction":
                total_zdna_bp / eff
                if eff > 0 else np.nan,
            "density_bp_per_Mb":
                total_zdna_bp / (eff / 1e6)
                if eff > 0 else np.nan,
        })

    summary = pd.DataFrame(rows)

    OUT_TSV.parent.mkdir(parents=True, exist_ok=True)

    summary.to_csv(
        OUT_TSV,
        sep="\t",
        index=False,
    )

    print(f"\nWrote: {OUT_TSV}")
    print("\nSummary:")
    print(summary.to_string(index=False))


if __name__ == "__main__":
    main()
```

#### 3 · HPRC per-contig densities

```python
"""3 - Per-contig Z-DNA and ACGT bp for the 464 HPRC haplotypes, contig classification, chromosome and haplotype tables
(Fig. 1b, 1c, 1f, Suppl. Fig. 1b, 1c inputs). Haplotypes in parallel.
"""
from __future__ import annotations

import os
import gzip
from pathlib import Path
from multiprocessing import get_context

import numpy as np
import pandas as pd


# NPROC = int(os.environ.get("SLURM_CPUS_ON_NODE", os.cpu_count() or 8))

NPROC = 128

WORK_DIR = Path(
    "/path/to/pangenomes_zdna"
)

META_PATH = (
    WORK_DIR / "1_hprc_metadata.tsv"
)

ZDNA_DIR = Path(
    "/path/to/pangenomes_zdna/hprc_zseeker"
)

FASTA_DIR = Path(
    "/path/to/scratch/human_pangenome/assemblies"
)

ALIAS_DIR = Path(
    "/path/to/scratch/human_pangenome/chrom_aliases"
)

HAPLOTYPE_OUT = (
    WORK_DIR / "3_hprc_haplotype_tables.final.tsv"
)

CONTIG_OUT = (
    WORK_DIR / "3_hprc_chromosome_tables.final.tsv"
)

MIN_SEGMENT_TIERS = (1, 9, 50)

PRIMARY_CHRS = (
    {f"chr{i}" for i in range(1, 23)}
    | {"chrX", "chrY"}
)

CATEGORIES = (
    "primary_chr",
    "chr_random",
    "mitochondrial",
    "chr_un",
    "no_alias",
)


HG002_FASTAS = {
    "HG002_v1.1.mat_MT.PanSN.zseeker.csv":
        "hg002v1.1.mat_MT.PanSN.fa",

    "HG002_v1.1.pat.PanSN.zseeker.csv":
        "hg002v1.1.pat.PanSN.fa",
}


def chr_class(ucsc) -> str:

    if ucsc is None:
        return "no_alias"

    if isinstance(ucsc, float) and np.isnan(ucsc):
        return "no_alias"

    ucsc = str(ucsc)

    if ucsc in PRIMARY_CHRS:
        return "primary_chr"

    if ucsc.startswith("chrM"):
        return "mitochondrial"

    if ucsc.startswith("chrUn"):
        return "chr_un"

    if (
        ucsc.startswith("chr")
        and "_random" in ucsc
    ):
        return "chr_random"

    if ucsc.startswith("chr"):
        return "chr_random"

    return "no_alias"


def pansn_canonical(name: str) -> str:
    """
    HG002 PanSN example:

        HG002#1#chr1 -> chr1
    """

    return (
        name.split("#")[-1]
        if "#" in name
        else name
    )


def open_fasta(path: Path):
    """
    Open either:

        *.fa.gz
        *.fa

    transparently in text mode.
    """

    if path.suffix == ".gz":
        return gzip.open(
            path,
            "rt",
        )

    return open(
        path,
        "rt",
        encoding="utf-8",
    )


def fasta_stem(path: Path) -> str:
    """
    Remove either:

        .fa.gz
        .fa

    from a FASTA filename.

    Examples:

        HG00096_pat.fa.gz -> HG00096_pat
        HG00096_pat.fa    -> HG00096_pat
    """

    name = path.name

    if name.endswith(".fa.gz"):
        return name[:-6]

    if name.endswith(".fa"):
        return name[:-3]

    raise ValueError(
        f"Unexpected FASTA extension: {path}"
    )


def find_fasta_from_zseeker(
    filename: str,
) -> Path:
    """
    Find the FASTA corresponding to an ordinary HPRC
    ZSeeker CSV.

    Search order:

        1. *.fa.gz
        2. *.fa

    Example:

        HG00096_pat.zseeker.csv

    searches for:

        HG00096_pat.fa.gz
        HG00096_pat.fa
    """

    if not filename.endswith(
        ".zseeker.csv"
    ):
        raise ValueError(
            "Expected ZSeeker filename ending "
            f"in '.zseeker.csv': {filename}"
        )

    base = filename[
        :-len(".zseeker.csv")
    ]

    candidates = [
        FASTA_DIR / f"{base}.fa.gz",
        FASTA_DIR / f"{base}.fa",
    ]

    existing = [
        p for p in candidates
        if p.is_file()
    ]

    if not existing:
        raise FileNotFoundError(
            f"No FASTA found for {filename}.\n"
            "Tried:\n"
            + "\n".join(
                str(p)
                for p in candidates
            )
        )

    return existing[0]


def infer_alias_key(name: str) -> str:

    n = name.lower()

    if (
        "_pat" in n
        or "#1#" in n
        or "_hap1" in n
    ):
        return "pat"

    if (
        "_mat" in n
        or "#2#" in n
        or "_hap2" in n
    ):
        return "mat"

    return ""


def find_alias_file(
    stem: str,
    prefix: str,
    key: str,
):

    exact = sorted(
        ALIAS_DIR.glob(
            f"{stem}.chromAlias.txt"
        )
    )

    if exact:
        return exact[0]

    if key:

        cand = sorted(
            ALIAS_DIR.glob(
                f"{prefix}*{key}*.chromAlias.txt"
            )
        )

        if cand:
            return cand[0]

    cand = sorted(
        ALIAS_DIR.glob(
            f"{prefix}*.chromAlias.txt"
        )
    )

    return (
        cand[0]
        if cand
        else None
    )


def load_alias_map(
    alias_file: Path,
) -> dict:

    mapping = {}

    with open(alias_file) as fh:

        for line in fh:

            line = line.strip()

            if (
                not line
                or line.startswith("#")
            ):
                continue

            parts = line.split()

            if len(parts) < 2:
                continue

            assembly = (
                parts[0].strip()
            )

            ucsc = (
                parts[1].strip()
            )

            mapping[assembly] = ucsc

    return mapping


def segment_lengths(
    seq: str,
) -> np.ndarray:
    """
    Return lengths of every contiguous
    A/C/G/T-only run.

    Lowercase a/c/g/t are accepted.

    N, IUPAC ambiguity characters, gaps,
    etc. terminate a run.
    """

    if not seq:
        return np.empty(
            0,
            dtype=np.int64,
        )

    arr = np.frombuffer(
        seq.upper().encode(
            "ascii",
            errors="replace",
        ),
        dtype=np.uint8,
    )

    mask = (
        (arr == ord("A"))
        | (arr == ord("C"))
        | (arr == ord("G"))
        | (arr == ord("T"))
    )

    if not mask.any():
        return np.empty(
            0,
            dtype=np.int64,
        )

    padded = np.concatenate(
        (
            [False],
            mask,
            [False],
        )
    )

    diffs = np.diff(
        padded.astype(
            np.int8
        )
    )

    starts = np.flatnonzero(
        diffs == 1
    )

    ends = np.flatnonzero(
        diffs == -1
    )

    return (
        ends - starts
    ).astype(
        np.int64
    )


def genome_by_contig(
    fasta_path: Path,
    alias_map: dict | None,
    hg002: bool,
) -> pd.DataFrame:
    """
    Stream either a .fa.gz or .fa FASTA.

    For every FASTA contig calculate:

        raw genome bp
        effective bp min1
        effective bp min9
        effective bp min50

    The FASTA, not the ZSeeker CSV, defines
    which contigs belong in the denominator.
    """

    rows = []

    contig = None
    seq_chunks = []

    def flush(
        name,
        chunks,
    ):

        if name is None:
            return

        seq = "".join(
            chunks
        )

        if hg002:

            assembly_contig = name

            ucsc = pansn_canonical(
                name
            )

        else:

            assembly_contig = name

            ucsc = alias_map.get(
                name
            )

        category = chr_class(
            ucsc
        )

        seg_lens = segment_lengths(
            seq
        )

        row = {
            "assembly_contig":
                assembly_contig,

            "ucsc_name":
                ucsc
                if ucsc is not None
                else "NA",

            "category":
                category,

            "genome_bp_raw":
                len(seq),
        }

        for tier in MIN_SEGMENT_TIERS:

            row[
                f"genome_bp_effective_min{tier}"
            ] = int(
                seg_lens[
                    seg_lens >= tier
                ].sum()
            )

        rows.append(
            row
        )

    with open_fasta(
        fasta_path
    ) as fh:

        for line in fh:

            if line.startswith(">"):

                flush(
                    contig,
                    seq_chunks,
                )

                contig = (
                    line[1:]
                    .strip()
                    .split()[0]
                )

                seq_chunks = []

            else:

                seq_chunks.append(
                    line.strip()
                )

        flush(
            contig,
            seq_chunks,
        )

    df = pd.DataFrame(
        rows
    )

    if df.empty:
        raise ValueError(
            f"No FASTA records found in "
            f"{fasta_path}"
        )

    if (
        df[
            "assembly_contig"
        ]
        .duplicated()
        .any()
    ):

        dup = df.loc[
            df[
                "assembly_contig"
            ].duplicated(
                keep=False
            ),
            "assembly_contig",
        ]

        raise ValueError(
            "Duplicate FASTA contig names "
            f"in {fasta_path}:\n"
            + "\n".join(
                dup
                .astype(str)
                .head(20)
            )
        )

    return df


def zdna_by_contig(
    csv_path: Path,
):
    """
    Read ZSeeker predictions.

    A row such as:

        Chromosome = contig_name
        Start      = NA
        End        = NA

    is interpreted as a valid zero-prediction
    contig.

    It contributes:

        0 Z-DNA bp
        0 intervals

    but the contig remains in the FASTA-derived
    denominator.

    A row where only ONE of Start/End is missing
    is considered malformed.

    ZSeeker coordinates are treated as
    0-based inclusive:

        length = End - Start + 1
    """

    df = pd.read_csv(
        csv_path,
        usecols=[
            "Chromosome",
            "Start",
            "End",
        ],
        dtype=str,
    )

    if (
        df[
            "Chromosome"
        ]
        .isna()
        .any()
    ):

        bad = df[
            df[
                "Chromosome"
            ].isna()
        ]

        raise ValueError(
            f"Missing Chromosome in "
            f"{csv_path}:\n"
            f"{bad.head(10)}"
        )

    df["Chromosome"] = (
        df[
            "Chromosome"
        ]
        .astype(str)
        .str.strip()
    )

    if (
        df[
            "Chromosome"
        ]
        .eq("")
        .any()
    ):

        raise ValueError(
            f"Empty Chromosome value "
            f"in {csv_path}"
        )

    df["Start"] = pd.to_numeric(
        df["Start"],
        errors="coerce",
    )

    df["End"] = pd.to_numeric(
        df["End"],
        errors="coerce",
    )

    no_prediction = (
        df["Start"].isna()
        & df["End"].isna()
    )

    malformed = (
        df["Start"].isna()
        ^ df["End"].isna()
    )

    if malformed.any():

        raise ValueError(
            f"Malformed Start/End rows "
            f"in {csv_path}:\n"
            f"{df.loc[malformed].head(10)}"
        )

    all_zseeker_contigs = set(
        df[
            "Chromosome"
        ].unique()
    )

    pred = (
        df.loc[
            ~no_prediction
        ]
        .copy()
    )

    if pred.empty:

        return (
            pd.DataFrame(
                columns=[
                    "assembly_contig",
                    "zdna_bp",
                    "n_intervals",
                ]
            ),
            all_zseeker_contigs,
            0,
        )

    pred["len"] = (
        pred["End"]
        - pred["Start"]
        + 1
    ).astype(
        np.int64
    )

    if (
        pred["len"] <= 0
    ).any():

        bad = pred[
            pred["len"] <= 0
        ]

        raise ValueError(
            f"Non-positive Z-DNA intervals "
            f"in {csv_path}:\n"
            f"{bad.head(10)}"
        )

    pred[
        "assembly_contig"
    ] = pred[
        "Chromosome"
    ]

    agg = (
        pred
        .groupby(
            "assembly_contig",
            as_index=False,
        )
        .agg(
            zdna_bp=(
                "len",
                "sum",
            ),
            n_intervals=(
                "len",
                "size",
            ),
        )
    )

    agg["zdna_bp"] = (
        agg[
            "zdna_bp"
        ]
        .astype(
            np.int64
        )
    )

    agg["n_intervals"] = (
        agg[
            "n_intervals"
        ]
        .astype(
            np.int64
        )
    )

    return (
        agg,
        all_zseeker_contigs,
        len(pred),
    )


def process_one(item):
    """
    Process one HPRC/HG002 haplotype.

    Returns:
        haplotype summary dictionary
        contig-level DataFrame
        status/error information
    """

    kind, target_csv = item

    try:
        zpath = ZDNA_DIR / target_csv

        if target_csv in HG002_FASTAS:

            fasta = (
                FASTA_DIR
                / HG002_FASTAS[target_csv]
            )

            alias_map = None
            alias_file_name = "INFERRED_FROM_PANSN_HEADER"
            hg002 = True

        else:

            fasta = find_fasta_from_zseeker(
                target_csv
            )

            stem = fasta_stem(
                fasta
            )

            prefix = (
                fasta.name
                .split("_")[0]
            )

            key = infer_alias_key(
                fasta.name
            )

            alias_file = find_alias_file(
                stem,
                prefix,
                key,
            )

            if alias_file is None:
                raise FileNotFoundError(
                    f"Alias file not found "
                    f"for {target_csv}"
                )

            alias_map = load_alias_map(
                alias_file
            )

            alias_file_name = (
                alias_file.name
            )

            hg002 = False

        if not zpath.is_file():
            raise FileNotFoundError(
                f"ZSeeker CSV not found: "
                f"{zpath}"
            )

        if not fasta.is_file():
            raise FileNotFoundError(
                f"FASTA not found: "
                f"{fasta}"
            )

        genome_df = genome_by_contig(
            fasta,
            alias_map,
            hg002,
        )

        (
            zdna_df,
            zseeker_contigs,
            n_predictions,
        ) = zdna_by_contig(
            zpath,
        )

        fasta_contigs = set(
            genome_df[
                "assembly_contig"
            ].astype(str)
        )

        missing_from_zseeker = (
            fasta_contigs
            - zseeker_contigs
        )

        extra_in_zseeker = (
            zseeker_contigs
            - fasta_contigs
        )

        if missing_from_zseeker:

            examples = sorted(
                missing_from_zseeker
            )[:10]

            raise ValueError(
                f"{len(missing_from_zseeker)} "
                f"FASTA contigs are completely absent "
                f"from ZSeeker output for "
                f"{target_csv}. "
                f"Examples: {examples}"
            )

        if extra_in_zseeker:

            examples = sorted(
                extra_in_zseeker
            )[:10]

            raise ValueError(
                f"{len(extra_in_zseeker)} "
                f"ZSeeker contigs are absent "
                f"from FASTA for "
                f"{target_csv}. "
                f"Examples: {examples}"
            )

        contig_df = genome_df.merge(
            zdna_df,
            on="assembly_contig",
            how="left",
            validate="one_to_one",
        )

        contig_df[
            "zdna_bp"
        ] = (
            contig_df[
                "zdna_bp"
            ]
            .fillna(0)
            .astype(
                np.int64
            )
        )

        contig_df[
            "n_intervals"
        ] = (
            contig_df[
                "n_intervals"
            ]
            .fillna(0)
            .astype(
                np.int64
            )
        )

        for tier in MIN_SEGMENT_TIERS:

            denom_col = (
                f"genome_bp_effective_min{tier}"
            )

            density_col = (
                f"zdna_bp_per_mb_min{tier}"
            )

            denom = (
                contig_df[
                    denom_col
                ]
                .astype(float)
            )

            contig_df[
                density_col
            ] = np.where(
                denom > 0,
                contig_df[
                    "zdna_bp"
                ]
                / (
                    denom
                    / 1e6
                ),
                np.nan,
            )

        contig_df.insert(
            0,
            "filename",
            target_csv,
        )

        contig_df.insert(
            1,
            "fasta_file",
            fasta.name,
        )

        contig_df.insert(
            2,
            "alias_file",
            alias_file_name,
        )

        summary = {
            "filename":
                target_csv,

            "fasta_file":
                fasta.name,

            "alias_file":
                alias_file_name,

            "kind":
                kind,

            "n_fasta_contigs":
                len(contig_df),

            "n_zseeker_contigs":
                len(
                    zseeker_contigs
                ),

            "n_contigs_with_zdna":
                int(
                    (
                        contig_df[
                            "n_intervals"
                        ] > 0
                    ).sum()
                ),

            "n_contigs_without_zdna":
                int(
                    (
                        contig_df[
                            "n_intervals"
                        ] == 0
                    ).sum()
                ),

            "n_intervals_total":
                int(
                    n_predictions
                ),
        }

        for category in CATEGORIES:

            sub = contig_df[
                contig_df[
                    "category"
                ] == category
            ]

            summary[
                f"n_contigs_{category}"
            ] = int(
                len(sub)
            )

            summary[
                f"zdna_bp_{category}"
            ] = int(
                sub[
                    "zdna_bp"
                ].sum()
            )

            summary[
                f"genome_bp_raw_{category}"
            ] = int(
                sub[
                    "genome_bp_raw"
                ].sum()
            )

            for tier in MIN_SEGMENT_TIERS:

                eff_col = (
                    f"genome_bp_effective_min{tier}"
                )

                eff = int(
                    sub[
                        eff_col
                    ].sum()
                )

                summary[
                    f"genome_bp_{category}_min{tier}"
                ] = eff

                if eff > 0:

                    summary[
                        f"zdna_bp_per_mb_{category}_min{tier}"
                    ] = (
                        summary[
                            f"zdna_bp_{category}"
                        ]
                        / (
                            eff
                            / 1e6
                        )
                    )

                else:

                    summary[
                        f"zdna_bp_per_mb_{category}_min{tier}"
                    ] = np.nan

        summary[
            "zdna_bp_all"
        ] = int(
            contig_df[
                "zdna_bp"
            ].sum()
        )

        summary[
            "genome_bp_raw_all"
        ] = int(
            contig_df[
                "genome_bp_raw"
            ].sum()
        )

        for tier in MIN_SEGMENT_TIERS:

            eff = int(
                contig_df[
                    f"genome_bp_effective_min{tier}"
                ].sum()
            )

            summary[
                f"genome_bp_all_min{tier}"
            ] = eff

            summary[
                f"fraction_raw_excluded_min{tier}"
            ] = (
                1.0
                - (
                    eff
                    / summary[
                        "genome_bp_raw_all"
                    ]
                )
                if summary[
                    "genome_bp_raw_all"
                ] > 0
                else np.nan
            )

            summary[
                f"zdna_bp_per_mb_all_min{tier}"
            ] = (
                summary[
                    "zdna_bp_all"
                ]
                / (
                    eff
                    / 1e6
                )
                if eff > 0
                else np.nan
            )

        return {
            "status":
                "ok",

            "target_csv":
                target_csv,

            "summary":
                summary,

            "contigs":
                contig_df,

            "error":
                "",
        }

    except Exception:

        import traceback

        return {
            "status":
                "fail",

            "target_csv":
                target_csv,

            "summary":
                None,

            "contigs":
                None,

            "error":
                traceback.format_exc(),
        }


meta = pd.read_csv(
    META_PATH,
    sep="\t",
    dtype=str,
).fillna("")

required_meta = {
    "filename",
    "sample_id",
}

missing = (
    required_meta
    - set(
        meta.columns
    )
)

if missing:

    raise KeyError(
        f"Metadata missing columns: "
        f"{sorted(missing)}"
    )


hprc_mask = (
    meta["sample_id"]
    .str.startswith(("H", "N"))
)

hprc_csvs = (
    meta.loc[hprc_mask, "filename"]
    .loc[
        lambda s: ~s.isin(HG002_FASTAS)
    ]
    .drop_duplicates()
    .tolist()
)

tasks = [
    ("hprc", filename)
    for filename in hprc_csvs
]

tasks.extend(
    ("hg002", filename)
    for filename in HG002_FASTAS
)

task_filenames = [
    filename
    for _, filename in tasks
]

if len(task_filenames) != len(set(task_filenames)):
    raise ValueError(
        "Duplicate ZSeeker filenames detected "
        "in task list."
    )

from collections import Counter

counts = Counter(
    filename
    for _, filename in tasks
)

duplicates = {
    filename: count
    for filename, count in counts.items()
    if count > 1
}

assert not duplicates, (
    f"Duplicate tasks: {duplicates}"
)


def task_size(
    item,
):

    kind, target_csv = item

    zpath = (
        ZDNA_DIR
        / target_csv
    )

    try:

        return (
            zpath
            .stat()
            .st_size
        )

    except FileNotFoundError:

        return 0


tasks.sort(
    key=task_size,
    reverse=True,
)


print(
    f"Tasks: "
    f"{len(tasks):,}"
)

print(
    f"Workers: "
    f"{NPROC}"
)

print(
    f"HPRC: "
    f"{sum(k == 'hprc' for k, _ in tasks):,}"
)

print(
    f"HG002: "
    f"{sum(k == 'hg002' for k, _ in tasks):,}"
)


with get_context(
    "fork"
).Pool(
    NPROC
) as pool:

    results = list(
        pool.imap_unordered(
            process_one,
            tasks,
            chunksize=1,
        )
    )


failed = [
    r
    for r in results
    if r[
        "status"
    ] != "ok"
]

if failed:

    print(
        f"\nFAILED: "
        f"{len(failed):,} / "
        f"{len(results):,}"
    )

    for r in failed:

        print(
            "\n"
            + "=" * 80
        )

        print(
            r[
                "target_csv"
            ]
        )

        print(
            r[
                "error"
            ]
        )

    raise RuntimeError(
        f"{len(failed)} haplotypes "
        f"failed. "
        f"No final output written."
    )


haplotype_df = pd.DataFrame(
    [
        r[
            "summary"
        ]
        for r in results
    ]
)

haplotype_df = (
    haplotype_df
    .sort_values(
        "filename"
    )
    .reset_index(
        drop=True
    )
)


contig_df = pd.concat(
    [
        r[
            "contigs"
        ]
        for r in results
    ],
    ignore_index=True,
)

contig_df = (
    contig_df
    .sort_values(
        [
            "filename",
            "category",
            "ucsc_name",
            "assembly_contig",
        ]
    )
    .reset_index(
        drop=True
    )
)


haplotype_df.to_csv(
    HAPLOTYPE_OUT,
    sep="\t",
    index=False,
)

contig_df.to_csv(
    CONTIG_OUT,
    sep="\t",
    index=False,
)


print(
    f"\nHaplotypes: "
    f"{len(haplotype_df):,}"
)

print(
    f"Contigs: "
    f"{len(contig_df):,}"
)

print(
    f"\nHaplotype output:\n"
    f"{HAPLOTYPE_OUT}"
)

print(
    f"\nContig output:\n"
    f"{CONTIG_OUT}"
)


print(
    "\nContigs without any "
    "Z-DNA prediction:"
)

print(
    haplotype_df[
        [
            "filename",
            "fasta_file",
            "n_fasta_contigs",
            "n_zseeker_contigs",
            "n_contigs_with_zdna",
            "n_contigs_without_zdna",
        ]
    ]
    .head()
    .to_string(
        index=False
    )
)


print(
    "\nPrimary chromosome densities:"
)

print(
    haplotype_df[
        [
            "filename",
            "zdna_bp_per_mb_primary_chr_min1",
            "zdna_bp_per_mb_primary_chr_min9",
            "zdna_bp_per_mb_primary_chr_min50",
        ]
    ]
    .head()
    .to_string(
        index=False
    )
)
```

#### 3 · HPRC haplotype densities → Fig. 1b, 1c

```python
"""3 - Genome-wide Z-DNA density per HPRC haplotype (primary_chr + chr_random + chr_un, min1 ACGT bp).
"""
from pathlib import Path
import pandas as pd

base = Path("/path/to/pangenomes_zdna")

df = pd.read_csv(base / "3_hprc_haplotype_tables.final.tsv", sep="\t")

assert len(df) == 464, f"Expected 464 haplotypes, found {len(df)}"
assert df["filename"].is_unique, "Duplicate haplotypes detected"

out = df[["filename", "kind"]].copy()

for cutoff in (1, 9, 50):
    out[f"density_primary_min{cutoff}"] = (
        df["zdna_bp_primary_chr"]
        / df[f"genome_bp_primary_chr_min{cutoff}"].replace(0, float("nan"))
        * 1e6
    )

    categories = ("primary_chr", "chr_random", "chr_un")

    zdna_bp = sum(df[f"zdna_bp_{c}"] for c in categories)
    genome_bp = sum(df[f"genome_bp_{c}_min{cutoff}"] for c in categories)

    out[f"density_nuclear_min{cutoff}"] = (
        zdna_bp / genome_bp.replace(0, float("nan")) * 1e6
    )

out.to_csv(
    base / "3_hprc_haplotype_densities.final.tsv",
    sep="\t",
    index=False,
)

print(out.shape)
print(out.head().to_string(index=False))
```

#### 4 · Length and score samples → Fig. 1e, Suppl. Fig. 1a

```python
"""4 - Reservoir samples (2,000 per assembly and metric) of Z-DNA length and ZSeeker score for 466 assemblies
(Fig. 1e, Suppl. Fig. 1a inputs). Files in parallel.
"""
from __future__ import annotations
from pathlib import Path
from multiprocessing import get_context
import hashlib
import numpy as np
import pandas as pd

BASE = Path('/path/to/pangenomes_zdna')
META = BASE / '1_hprc_metadata.tsv'
CHROM_TABLE = BASE / '3_hprc_chromosome_tables.final.tsv'
HPRC_DIR = BASE / 'hprc_zseeker'
SPECIAL = {
    'CHM13': BASE / 'tmp/chm13/chm13_zdna_files/GCF_009914755.1_T2T-CHM13v2.0_genomic_zdna_score.csv',
    'GRCh38': BASE / 'tmp/grch38/grch38_refseq/GCF_000001405.40_GRCh38.p14_primary_noMT_ucscnames_zseeker.csv',
}
SP_ORDER = ['AFR', 'AMR', 'EAS', 'EUR', 'SAS', 'CHM13', 'GRCh38']
KEEP_CATEGORIES = {'primary_chr', 'chr_random', 'chr_un'}

SEED = 42
NPROC = 32
CHUNK_SIZE = 250_000
CANDIDATES_PER_FILE = 2000  # Each input gets an unbiased reservoir of this size, per metric.


def stable_seed(name: str, metric: str) -> int:
    payload = f'{SEED}|{name}|{metric}'.encode()
    return int.from_bytes(hashlib.sha256(payload).digest()[:8], 'little')


def reservoir_update(reservoir, seen, values, rng, capacity):
    """Exact reservoir sampling across a streamed file (Algorithm R)."""
    for v in values:
        seen += 1
        if len(reservoir) < capacity:
            reservoir.append(float(v))
        else:
            j = rng.integers(0, seen)
            if j < capacity:
                reservoir[int(j)] = float(v)
    return seen


def worker(task):
    name, path_str, group, allowed = task
    path = Path(path_str)
    if not path.is_file():
        raise FileNotFoundError(path)
    rngs = {m: np.random.default_rng(stable_seed(name, m)) for m in ('score', 'length')}
    reservoirs = {'score': [], 'length': []}
    counts = {'score': 0, 'length': 0}
    for chunk in pd.read_csv(path, usecols=['Chromosome', 'Start', 'End', 'Z-DNA Score'], chunksize=CHUNK_SIZE):
        if allowed is not None:
            chunk = chunk.loc[chunk['Chromosome'].isin(allowed)]
        else:
            ch = chunk['Chromosome'].astype('string')
            chunk = chunk.loc[ch.str.startswith('chr', na=False) & ~ch.str.startswith('chrM', na=False)]
        if chunk.empty:
            continue
        for c in ('Start', 'End', 'Z-DNA Score'):
            chunk[c] = pd.to_numeric(chunk[c], errors='coerce')
        chunk = chunk.dropna(subset=['Start', 'End', 'Z-DNA Score'])
        chunk = chunk.loc[chunk['End'] >= chunk['Start']]
        if chunk.empty:
            continue
        metrics = {
            'score': chunk['Z-DNA Score'].to_numpy(dtype='float64'),
            'length': (chunk['End'] - chunk['Start'] + 1).to_numpy(dtype='float64'),
        }
        for metric, arr in metrics.items():
            arr = arr[np.isfinite(arr) & (arr > 0)]
            counts[metric] = reservoir_update(
                reservoirs[metric], counts[metric], arr,
                rngs[metric], CANDIDATES_PER_FILE,
            )
    return {'filename': name, 'group': group, 'counts': counts, 'samples': reservoirs}


def main():
    meta = pd.read_csv(META, sep='\t', dtype=str)[['filename', 'superpopulation']]
    assert meta['filename'].is_unique
    lookup = meta.set_index('filename')['superpopulation'].to_dict()
    chrom = pd.read_csv(CHROM_TABLE, sep='\t', dtype=str)
    allowed = (chrom.loc[chrom['category'].isin(KEEP_CATEGORIES)]
               .groupby('filename')['assembly_contig'].agg(set).to_dict())
    tasks = []
    for path in sorted(HPRC_DIR.glob('*.zseeker.csv')):
        group = lookup.get(path.name)
        if group not in SP_ORDER[:5]:
            continue
        if path.name not in allowed:
            raise ValueError(f'No chromosome table entries for {path.name}')
        tasks.append((path.name, str(path), group, allowed[path.name]))
    for group, path in SPECIAL.items():
        tasks.append((group, str(path), group, None))
    if not tasks:
        raise RuntimeError('No input datasets')
    print(f'Processing {len(tasks)} files using {NPROC} workers', flush=True)
    results = []
    with get_context('fork').Pool(processes=NPROC) as pool:
        for result in pool.imap_unordered(worker, tasks, chunksize=1):
            results.append(result)
    results.sort(key=lambda r: r['filename'])
    counts_df = pd.DataFrame([
        {'filename': r['filename'], 'superpopulation': r['group'],
         'n_score_intervals': r['counts']['score'], 'n_length_intervals': r['counts']['length']}
        for r in results
    ])
    counts_df.to_csv(BASE / '4_score_length_source_counts.tsv', sep='\t', index=False)

    for metric in ('score', 'length'):
        rows = [
            {
                'source_filename': r['filename'],
                metric: value,
            }
            for r in results
            for value in r['samples'][metric]
        ]

        out = BASE / f'4_{metric}_per_haplotype_distribution.tsv'

        pd.DataFrame(rows).to_csv(
            out,
            sep='\t',
            index=False,
            float_format='%.10g',
        )

        print(f'Wrote {out}: {len(rows):,} observations')


if __name__ == '__main__':
    main()
```

#### 4 · Assembly-specific sequences → Fig. 1d

```python
"""4 - Assembly-specific Z-DNA sequences (exact sequences present in one assembly only) per assembly (Fig. 1d input).
"""
from pathlib import Path
from collections import Counter
import pandas as pd

BASE = Path("/path/to/pangenomes_zdna")

ZDNA_DIR = BASE / "hprc_zseeker"
CHROM_TABLE = BASE / "3_hprc_chromosome_tables.final.tsv"

OUTFILE = BASE / "4_hprc_private_zdna_sequences.tsv"

KEEP_CATEGORIES = {
    "primary_chr",
    "chr_random",
    "chr_un",
}

SPECIAL = {
    "GRCh38_primary": (
        BASE / "tmp/grch38/grch38_refseq/"
        "GCF_000001405.40_GRCh38.p14_primary_noMT_ucscnames_zseeker.csv"
    ),
    "CHM13": (
        BASE / "tmp/chm13/chm13_zdna_files/"
        "GCF_009914755.1_T2T-CHM13v2.0_genomic_zdna_score.csv"
    ),
}


chrom = pd.read_csv(
    CHROM_TABLE,
    sep="\t",
    dtype=str,
)

chrom = chrom[
    chrom["category"].isin(KEEP_CATEGORIES)
]

allowed_contigs = (
    chrom.groupby("filename")["assembly_contig"]
    .apply(set)
    .to_dict()
)


input_files = {
    filename: ZDNA_DIR / filename
    for filename in allowed_contigs
}

input_files.update(SPECIAL)

print(f"Input datasets: {len(input_files):,}")


dataset_sets = {}
global_counter = Counter()

for dataset, path in input_files.items():

    df = pd.read_csv(
        path,
        usecols=["Chromosome", "Sequence"],
        dtype=str,
    )

    df = df.dropna(
        subset=["Chromosome", "Sequence"]
    )

    if dataset in allowed_contigs:

        df = df[
            df["Chromosome"].isin(
                allowed_contigs[dataset]
            )
        ]

    else:
        chroms = df["Chromosome"]

        df = df[
            chroms.str.startswith("chr")
            & ~chroms.str.startswith("chrM")
        ]

    seq_set = set(
        df["Sequence"].str.upper()
    )

    dataset_sets[dataset] = seq_set
    global_counter.update(seq_set)

    print(
        f"{dataset}: "
        f"{len(seq_set):,} unique sequences"
    )


rows = []

for dataset, seq_set in dataset_sets.items():

    n_unique = len(seq_set)

    n_private = sum(
        global_counter[seq] == 1
        for seq in seq_set
    )

    rows.append({
        "dataset": dataset,
        "n_unique_zdna_sequences": n_unique,
        "n_private_zdna_sequences": n_private,
        "fraction_private": (
            n_private / n_unique
            if n_unique
            else 0.0
        ),
    })


out = (
    pd.DataFrame(rows)
    .sort_values(
        "n_private_zdna_sequences",
        ascending=False,
    )
    .reset_index(drop=True)
)

out.to_csv(
    OUTFILE,
    sep="\t",
    index=False,
)

print(f"\nDatasets: {len(out):,}")
print(f"Wrote: {OUTFILE}")
print(out.head().to_string(index=False))
```

#### 5 · GC content → Suppl. Fig. 1b

```python
"""5 - GC content of HPRC haplotypes (per selected contig and genome-wide) and reference chromosomes (Suppl. Fig. 1b
input). Each FASTA streamed once, in parallel.
"""
from __future__ import annotations

import gzip
import os
import re
import traceback
from multiprocessing import get_context
from pathlib import Path

import numpy as np
import pandas as pd


BASE = Path('/path/to/pangenomes_zdna')
FASTA_DIR = Path('/path/to/scratch/human_pangenome/assemblies')
CHROM_TABLE = BASE / '3_hprc_chromosome_tables.final.tsv'

CONTIG_OUT = BASE / '5_hprc_chromosome_gc_min1.tsv'
HAPLOTYPE_OUT = BASE / '5_hprc_haplotype_gc_content.tsv'
REFERENCE_OUT = BASE / '5_reference_chromosome_gc.tsv'

MIN_SEGMENT_TIERS = (1, 9, 50)

KEEP_CATEGORIES = {'primary_chr', 'chr_random', 'chr_un'}
CHR_CATEGORIES = {'primary_chr', 'chr_random'}

CHROMS = [f'chr{i}' for i in range(1, 23)] + ['chrX', 'chrY']
CHR_RE = re.compile(r'^(chr(?:[1-9]|1[0-9]|2[0-2]|X|Y))(?=_|$)')

NPROC = 64 # min(32, int(os.environ.get('SLURM_CPUS_PER_TASK', '32')))

REFERENCES = {
    'GRCh38': Path(
        '/path/to/pangenomes_zdna/tmp/grch38/grch38_refseq/'
        'GCF_000001405.40_GRCh38.p14_primary_noMT_ucsc.fna'
    ),
    'CHM13': Path(
        '/path/to/pangenomes_zdna/tmp/chm13/chm13_refseq/'
        'GCF_009914755.1_T2T-CHM13v2.0_genomic_ucsc.fna'
    ),
}

ACGT_RUN = re.compile(rb'[ACGT]+')


def canonical_chrom(value: str) -> str | None:
    """chr13 or chr13_..._random -> chr13; rejects chrUn / chrM."""
    m = CHR_RE.match(str(value))
    return m.group(1) if m else None


def find_haplotype_fasta(zseeker_filename: str) -> Path:
    special = {
        'HG002_v1.1.mat_MT.PanSN.zseeker.csv': 'hg002v1.1.mat_MT.PanSN.fa',
        'HG002_v1.1.pat.PanSN.zseeker.csv': 'hg002v1.1.pat.PanSN.fa',
    }
    if zseeker_filename in special:
        candidates = [FASTA_DIR / special[zseeker_filename]]
    else:
        if not zseeker_filename.endswith('.zseeker.csv'):
            raise ValueError(f'Unexpected filename: {zseeker_filename}')
        prefix = zseeker_filename[:-len('.zseeker.csv')]
        candidates = [FASTA_DIR / f'{prefix}.fa', FASTA_DIR / f'{prefix}.fa.gz']
    for path in candidates:
        if path.is_file():
            return path
    raise FileNotFoundError(
        f'No FASTA for {zseeker_filename}:\n' + '\n'.join(map(str, candidates))
    )


def tier_counts(seq: bytes):
    """Effective ACGT bp and GC bp per tier.

    A tier counts a contiguous A/C/G/T-only segment only if its length >= tier.
    `seq` must be upper-case bytes.
    """
    eff = dict.fromkeys(MIN_SEGMENT_TIERS, 0)
    gc = dict.fromkeys(MIN_SEGMENT_TIERS, 0)
    for m in ACGT_RUN.finditer(seq):
        s, e = m.span()
        n = e - s
        g = seq.count(b'G', s, e) + seq.count(b'C', s, e)
        for t in MIN_SEGMENT_TIERS:
            if n >= t:
                eff[t] += n
                gc[t] += g
    return eff, gc


def scan_fasta(fasta: Path, wanted) -> dict:
    """Stream a FASTA once; return {contig: {raw, eff{tier}, gc{tier}}}."""
    records: dict = {}
    contig = None
    keep = False
    chunks: list = []

    def flush():
        if contig is None or not keep:
            return
        if contig in records:
            raise RuntimeError(f'{fasta.name}: duplicate FASTA record {contig}')
        seq = b''.join(chunks).upper()
        eff, gc = tier_counts(seq)
        records[contig] = {'raw': len(seq), 'eff': eff, 'gc': gc}

    opener = gzip.open if fasta.suffix == '.gz' else open
    with opener(fasta, 'rb') as fh:
        for line in fh:
            if line.startswith(b'>'):
                flush()
                contig = line[1:].split(maxsplit=1)[0].decode('utf-8')
                keep = wanted(contig)
                chunks = []
            elif keep:
                chunks.append(line.strip())
        flush()
    return records


def worker(task):
    kind, key, payload = task
    try:
        if kind == 'hap':
            allowed = payload
            fasta = find_haplotype_fasta(key)
            rec = scan_fasta(fasta, lambda c: c in allowed)
            missing = allowed - set(rec)
            if missing:
                raise RuntimeError(
                    f'{key}: {len(missing)} contigs from analysis 3 absent '
                    f'in FASTA; e.g. {sorted(missing)[:5]}'
                )
        else:
            fasta = payload
            rec = scan_fasta(
                fasta,
                lambda c: c.startswith('chr') and not c.startswith('chrM'),
            )
        return {'status': 'ok', 'kind': kind, 'key': key,
                'fasta': Path(fasta).name, 'records': rec, 'error': ''}
    except Exception:
        return {'status': 'fail', 'kind': kind, 'key': key,
                'fasta': '', 'records': None, 'error': traceback.format_exc()}


def load_chrom_table() -> pd.DataFrame:
    ct = pd.read_csv(CHROM_TABLE, sep='\t', dtype=str)
    required = {'filename', 'assembly_contig', 'ucsc_name', 'category',
                'genome_bp_raw', 'genome_bp_effective_min1',
                'genome_bp_effective_min9', 'genome_bp_effective_min50'}
    if required - set(ct):
        raise ValueError(f'Analysis-3 columns missing: {sorted(required - set(ct))}')
    ct = ct.loc[ct['category'].isin(KEEP_CATEGORIES)].copy()

    num_cols = ['genome_bp_raw'] + [f'genome_bp_effective_min{t}'
                                    for t in MIN_SEGMENT_TIERS]
    for c in num_cols:
        ct[c] = pd.to_numeric(ct[c], errors='raise').astype('int64')

    key = ['filename', 'assembly_contig']
    conflicts = ct.groupby(key)[num_cols].nunique()
    if (conflicts > 1).any(axis=None):
        print(conflicts[(conflicts > 1).any(axis=1)].head(30).to_string())
        raise RuntimeError('Duplicate contig rows contain different values.')
    before = len(ct)
    ct = ct.drop_duplicates(key).copy()
    print(f'Removed {before - len(ct):,} identical duplicate filename/contig rows.')
    return ct


def select_chromosome_contigs(ct: pd.DataFrame) -> pd.DataFrame:
    """primary_chr (longest per haplotype+chromosome) + all chr_random."""
    sel = ct.loc[ct['category'].isin(CHR_CATEGORIES)].copy()
    sel['chr'] = sel['ucsc_name'].map(canonical_chrom)
    sel = sel.loc[sel['chr'].isin(CHROMS)].copy()
    primary = (sel.loc[sel['category'].eq('primary_chr')]
               .sort_values('genome_bp_raw', ascending=False, kind='mergesort')
               .drop_duplicates(['filename', 'chr'], keep='first'))
    random = sel.loc[sel['category'].eq('chr_random')]
    sel = pd.concat([primary, random], ignore_index=True)
    if sel.duplicated(['filename', 'assembly_contig']).any():
        raise RuntimeError('Contigs duplicated after primary selection')
    return sel[['filename', 'assembly_contig']]


def main():
    ct = load_chrom_table()
    allowed = ct.groupby('filename')['assembly_contig'].apply(set).to_dict()
    selected = select_chromosome_contigs(ct)

    tasks = [('ref', name, path) for name, path in REFERENCES.items()]
    tasks += [('hap', fn, contigs) for fn, contigs in allowed.items()]
    print(f'Haplotype datasets: {len(allowed):,}')
    print(f'Reference genomes:  {len(REFERENCES):,}')
    print(f'Workers:            {NPROC:,}', flush=True)

    results = []
    with get_context('fork').Pool(NPROC) as pool:
        for i, r in enumerate(pool.imap_unordered(worker, tasks, chunksize=1), 1):
            results.append(r)
            if i % 20 == 0 or i == len(tasks):
                print(f'  {i}/{len(tasks)} FASTAs', flush=True)

    failed = [r for r in results if r['status'] != 'ok']
    if failed:
        for r in failed:
            print('\n' + '=' * 80 + f"\n{r['kind']} {r['key']}\n{r['error']}")
        raise RuntimeError(f'{len(failed)} FASTA tasks failed.')

    flat_rows, fasta_names = [], {}
    for r in results:
        if r['kind'] != 'hap':
            continue
        fasta_names[r['key']] = r['fasta']
        for contig, rec in r['records'].items():
            row = {'filename': r['key'], 'assembly_contig': contig,
                   'fasta_raw_bp': rec['raw']}
            for t in MIN_SEGMENT_TIERS:
                row[f'acgt_bp_min{t}'] = rec['eff'][t]
                row[f'gc_bp_min{t}'] = rec['gc'][t]
            flat_rows.append(row)
    flat = pd.DataFrame(flat_rows)

    qc = ct.merge(flat, on=['filename', 'assembly_contig'], how='outer',
                  validate='one_to_one', indicator=True)
    if not qc['_merge'].eq('both').all():
        raise RuntimeError(
            f'Contig set mismatch between FASTAs and analysis 3: '
            f'{qc["_merge"].value_counts().to_dict()}'
        )
    checks = [('raw length', 'genome_bp_raw', 'fasta_raw_bp')]
    checks += [(f'min{t} effective length', f'genome_bp_effective_min{t}',
                f'acgt_bp_min{t}') for t in MIN_SEGMENT_TIERS]
    for label, exp_col, obs_col in checks:
        bad = qc.loc[qc[exp_col].astype('int64') != qc[obs_col].astype('int64'),
                     ['filename', 'assembly_contig', exp_col, obs_col]]
        if len(bad):
            print(bad.head(30).to_string())
            raise RuntimeError(f'{label} QC failed for {len(bad)} contigs.')
        print(f'{label} QC: PASS')

    contig_cols = ['filename', 'assembly_contig', 'fasta_raw_bp',
                   'acgt_bp_min1', 'gc_bp_min1']
    contig_df = (flat.merge(selected, on=['filename', 'assembly_contig'],
                            how='inner', validate='one_to_one')[contig_cols]
                 .sort_values(['filename', 'assembly_contig'])
                 .reset_index(drop=True))
    if len(contig_df) != len(selected):
        raise RuntimeError('Selected contigs missing from FASTA results.')

    sums = flat.groupby('filename').sum(numeric_only=True)
    hap_rows = []
    for fn, s in sums.iterrows():
        row = {'filename': fn, 'fasta_file': fasta_names[fn],
               'raw_length_bp': int(s['fasta_raw_bp'])}
        for t in MIN_SEGMENT_TIERS:
            eff, gc = int(s[f'acgt_bp_min{t}']), int(s[f'gc_bp_min{t}'])
            row[f'acgt_length_bp_min{t}'] = eff
            row[f'gc_count_min{t}'] = gc
            row[f'gc_fraction_min{t}'] = gc / eff if eff > 0 else np.nan
        hap_rows.append(row)
    haplotype_df = (pd.DataFrame(hap_rows).sort_values('filename')
                    .reset_index(drop=True))

    ref_rows = []
    for r in results:
        if r['kind'] != 'ref':
            continue
        for contig, rec in r['records'].items():
            row = {'assembly': r['key'], 'ucsc_chr': contig,
                   'raw_length_bp': rec['raw']}
            for t in MIN_SEGMENT_TIERS:
                eff, gc = rec['eff'][t], rec['gc'][t]
                row[f'acgt_length_bp_min{t}'] = eff
                row[f'gc_count_min{t}'] = gc
                row[f'gc_fraction_min{t}'] = gc / eff if eff > 0 else np.nan
            ref_rows.append(row)
    reference_df = (pd.DataFrame(ref_rows)
                    .sort_values(['assembly', 'ucsc_chr']).reset_index(drop=True))

    contig_df.to_csv(CONTIG_OUT, sep='\t', index=False)
    haplotype_df.to_csv(HAPLOTYPE_OUT, sep='\t', index=False)
    reference_df.to_csv(REFERENCE_OUT, sep='\t', index=False)

    print(f'\nContig rows:               {len(contig_df):,}')
    print(f'Haplotypes:                {len(haplotype_df):,}')
    print(f'Reference chromosome rows: {len(reference_df):,}')
    print(f'\nWrote:\n  {CONTIG_OUT}\n  {HAPLOTYPE_OUT}\n  {REFERENCE_OUT}')


if __name__ == '__main__':
    main()
```

```python
"""Fig. 1b - genome-wide Z-DNA density of 464 HPRC haplotypes (histogram) with GRCh38 and CHM13 (dashed lines).
"""
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from matplotlib.ticker import MaxNLocator

BASE = Path("/path/to/pangenomes_zdna")
PLOT_DIR, TEST_DIR = BASE / "plots", BASE / "tests"
STEM = "fig1b_zdna_density_distribution"
READABLE = {
    "dataset": "Assembly", "filename": "Assembly", "source_filename": "Assembly", "assembly": "Reference genome",
    "dataset_type": "Assembly type", "sample_id": "Sample", "superpopulation": "Superpopulation",
    "pat_or_mat": "Parental origin", "chr": "Chromosome", "ucsc_chr": "Chromosome", "n_contigs": "Contigs (n)",
    "n_chr_random": "chr_random contigs (n)", "zdna_bp": "Z-DNA (bp)", "acgt_bp_min1": "ACGT (bp)", "genome_bp": "ACGT (bp)",
    "zdna_density_bp_per_Mb": "Z-DNA density (bp/Mb)", "density_bp_per_Mb": "Z-DNA density (bp/Mb)",
    "density_bp_per_Mb_min1": "Z-DNA density (bp/Mb)", "gc_pct": "GC (%)", "rank": "Rank",
    "n_unique_zdna_sequences": "Distinct Z-DNA sequences (n)",
    "n_private_zdna_sequences": "Assembly-specific Z-DNA sequences (n)", "fraction_private": "Fraction assembly-specific",
    "length": "Z-DNA length (bp)", "score": "ZSeeker score", "metric_value": "Value", "transformed_coordinate": "Plot coordinate",
    "kde_percent_per_visual_bin": "KDE (% per visual bin)", "n_sampled_predictions": "Sampled predictions (n)",
    "n_in_display_range": "Predictions in display range (n)", "percent_in_display_range": "Predictions in display range (%)",
    "scott_bandwidth_multiplier": "Scott bandwidth multiplier",
    "visual_bin_width_custom_axis_units": "Visual bin width (axis units)", "n_autosomes": "Autosomes (n)",
    "paternal_bp_per_Mb": "Paternal density (bp/Mb)", "maternal_bp_per_Mb": "Maternal density (bp/Mb)",
    "delta_pat_minus_mat": "Paternal - maternal (bp/Mb)", "absolute_delta": "Absolute difference (bp/Mb)",
    "n_chromosomes": "Chromosomes (n)", "spearman_rho": "Spearman rho", "spearman_p": "Spearman P value",
    "ols_slope_bp_per_Mb_per_pct_GC": "OLS slope (bp/Mb per % GC)", "ols_intercept_bp_per_Mb": "OLS intercept (bp/Mb)",
    "ols_r_squared": "OLS R squared", "ols_p": "OLS P value", "test": "Test", "n_pairs": "Pairs (n)",
    "density_denominator": "Density denominator", "contig_rule": "Contig rule", "n_pat_only": "Paternal only (n)",
    "n_mat_only": "Maternal only (n)", "median_autosomes_per_pair": "Median autosomes per pair",
    "min_autosomes_per_pair": "Minimum autosomes per pair",
    "median_delta_pat_minus_mat_bp_per_Mb": "Median paternal - maternal (bp/Mb)",
    "mean_delta_bp_per_Mb": "Mean paternal - maternal (bp/Mb)", "sd_delta_bp_per_Mb": "SD paternal - maternal (bp/Mb)",
    "mean_absolute_delta_bp_per_Mb": "Mean absolute difference (bp/Mb)",
    "mean_absolute_delta_pct_grand_median": "Mean absolute difference (% of grand median)",
    "wilcoxon_W": "Wilcoxon W", "p_value": "P value",
}
HAP_FILE = BASE / "3_hprc_haplotype_densities.final.tsv"
HAP_TABLE = BASE / "3_hprc_haplotype_tables.final.tsv"
META = BASE / "1_hprc_metadata.tsv"
OUT_FILE = PLOT_DIR / f"{STEM}.png"
PLOT_DATA = TEST_DIR / f"{STEM}.tsv"
NUCLEAR = ("primary_chr", "chr_random", "chr_un")                         # as density_nuclear_min1
PLOT_DIR.mkdir(parents=True, exist_ok=True)
TEST_DIR.mkdir(parents=True, exist_ok=True)

CHM13_COLOR, GRCH38_COLOR, HIST_COLOR = "#536078", "#547853", "#BDBDBD"
FIG_WIDTH, FIG_HEIGHT = 4.0, 2.5
FONT_BASE, FONT_LABEL, FONT_TICK, FONT_ANNOTATION = 10, 11.5, 10, 10
AXIS_WIDTH, TICK_WIDTH, TICK_LENGTH = 1, 0.8, 3

df = pd.read_csv(HAP_FILE, sep="\t")
assert len(df) == 464, f"Expected 464 haplotypes, found {len(df)}"
assert df["filename"].is_unique, "Duplicate haplotypes detected"
density = pd.to_numeric(df["density_nuclear_min1"], errors="raise").dropna()
assert len(density) == 464


def reference_row(filename):
    ref = pd.read_csv(BASE / filename, sep="\t")
    row = ref.loc[ref["min_segment"] == 1]
    assert len(row) == 1, f"Expected one min_segment=1 row in {filename}"
    return row.iloc[0]


ref_rows = {"CHM13": reference_row("2_CHM13_zdna_density_genome_summary.tsv"),
            "GRCh38": reference_row("2_GRCh38_zdna_density_genome_summary.tsv")}
chm13 = float(ref_rows["CHM13"]["density_bp_per_Mb"])
grch38 = float(ref_rows["GRCh38"]["density_bp_per_Mb"])

tab = pd.read_csv(HAP_TABLE, sep="\t")
tab["zdna_bp_nuclear"] = sum(tab[f"zdna_bp_{c}"] for c in NUCLEAR)
tab["acgt_bp_nuclear_min1"] = sum(tab[f"genome_bp_{c}_min1"] for c in NUCLEAR)
meta = pd.read_csv(META, sep="\t", dtype=str)[["filename", "sample_id", "superpopulation"]]
hprc = (df[["filename", "density_nuclear_min1"]]
        .merge(tab[["filename", "zdna_bp_nuclear", "acgt_bp_nuclear_min1"]], on="filename", validate="one_to_one")
        .merge(meta, on="filename", how="left", validate="one_to_one"))
recomputed = hprc["zdna_bp_nuclear"] / hprc["acgt_bp_nuclear_min1"] * 1e6
assert np.allclose(recomputed, hprc["density_nuclear_min1"], rtol=1e-9), "Density does not match numerator/denominator"
plot_data = pd.concat([
    pd.DataFrame({"dataset": hprc["filename"], "dataset_type": "HPRC haplotype", "sample_id": hprc["sample_id"],
                  "superpopulation": hprc["superpopulation"], "zdna_bp": hprc["zdna_bp_nuclear"],
                  "acgt_bp_min1": hprc["acgt_bp_nuclear_min1"], "zdna_density_bp_per_Mb": hprc["density_nuclear_min1"]}),
    pd.DataFrame([{"dataset": name, "dataset_type": "reference genome", "sample_id": name, "superpopulation": "",
                   "zdna_bp": int(r["total_zdna_bp"]), "acgt_bp_min1": int(r["total_effective_length_bp"]),
                   "zdna_density_bp_per_Mb": float(r["density_bp_per_Mb"])} for name, r in ref_rows.items()]),
], ignore_index=True)
plot_data.rename(columns=READABLE).to_csv(PLOT_DATA, sep="\t", index=False, float_format="%.10g")

plt.rcParams.update({
    "font.family": "Nimbus Sans", "font.size": FONT_BASE, "axes.labelsize": FONT_LABEL,
    "xtick.labelsize": FONT_TICK, "ytick.labelsize": FONT_TICK, "axes.linewidth": AXIS_WIDTH,
    "axes.spines.top": False, "axes.spines.right": False, "pdf.fonttype": 42, "ps.fonttype": 42,
})
fig, ax = plt.subplots(figsize=(FIG_WIDTH, FIG_HEIGHT))
ax.hist(density, bins=20, color=HIST_COLOR, edgecolor="black", linewidth=0.5, zorder=2)
ax.axvline(grch38, color=GRCH38_COLOR, linestyle="--", linewidth=1.2, zorder=3)
ax.axvline(chm13, color=CHM13_COLOR, linestyle="--", linewidth=1.2, zorder=3)
ymax = ax.get_ylim()[1]
ax.annotate(f"GRCh38\n({grch38:.1f})", xy=(grch38, ymax * 0.95), xytext=(40, 0), textcoords="offset points",
            ha="right", va="top", fontsize=FONT_ANNOTATION, fontweight="bold", color=GRCH38_COLOR,
            linespacing=1.1, zorder=5)
ax.annotate(f"CHM13\n({chm13:.1f})", xy=(chm13, ymax * 0.95), xytext=(4, 0), textcoords="offset points",
            ha="left", va="top", fontsize=FONT_ANNOTATION, fontweight="bold", color=CHM13_COLOR,
            linespacing=1.1, zorder=5)
ax.set_xlabel("Z-DNA density (bp/Mb)", fontsize=FONT_LABEL, labelpad=5)
ax.set_ylabel("Number of haplotypes", fontsize=FONT_LABEL, labelpad=5)
ax.xaxis.set_major_locator(MaxNLocator(nbins=6))
ax.yaxis.set_major_locator(MaxNLocator(nbins=8, integer=True))
ax.tick_params(axis="both", which="major", labelsize=FONT_TICK, width=TICK_WIDTH, length=TICK_LENGTH,
               direction="out", pad=3)
plt.setp(ax.get_xticklabels(), rotation=45, ha="right", rotation_mode="anchor")
ax.set_ylim(0, ymax)
ax.grid(True, axis="both", which="major", linestyle="--", linewidth=0.4, color="#B0B0B0", alpha=0.5, zorder=0)
ax.set_axisbelow(True)
fig.subplots_adjust(left=0.17, right=0.97, bottom=0.25, top=0.96)
fig.savefig(OUT_FILE, dpi=600, bbox_inches=None)
fig.savefig(OUT_FILE.with_suffix(".pdf"), bbox_inches=None)
plt.show()

print(f"Haplotypes: {len(density)}; median HPRC density {density.median():.2f} bp/Mb "
      f"(range {density.min():.2f}-{density.max():.2f}); histogram bins: 20")
print(f"GRCh38: {grch38:.2f} bp/Mb; CHM13: {chm13:.2f} bp/Mb")
```

#### Fig. 1c

```python
"""Fig. 1c - genome-wide Z-DNA density by superpopulation (boxes + points) + legend; Kruskal-Wallis and Dunn's test
(Bonferroni).
"""
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import scikit_posthocs as sp
from matplotlib.patches import Patch
from matplotlib.ticker import MaxNLocator
from scipy import stats

BASE = Path("/path/to/pangenomes_zdna")
PLOT_DIR, TEST_DIR = BASE / "plots", BASE / "tests"
PLOT_DIR.mkdir(parents=True, exist_ok=True)
TEST_DIR.mkdir(parents=True, exist_ok=True)
STEM = "fig1c_zdna_density_by_superpopulation"
NUCLEAR = ("primary_chr", "chr_random", "chr_un")
READABLE = {
    "dataset": "Assembly", "filename": "Assembly", "source_filename": "Assembly", "assembly": "Reference genome",
    "dataset_type": "Assembly type", "sample_id": "Sample", "superpopulation": "Superpopulation",
    "pat_or_mat": "Parental origin", "chr": "Chromosome", "ucsc_chr": "Chromosome", "n_contigs": "Contigs (n)",
    "n_chr_random": "chr_random contigs (n)", "zdna_bp": "Z-DNA (bp)", "acgt_bp_min1": "ACGT (bp)", "genome_bp": "ACGT (bp)",
    "zdna_density_bp_per_Mb": "Z-DNA density (bp/Mb)", "density_bp_per_Mb": "Z-DNA density (bp/Mb)",
    "density_bp_per_Mb_min1": "Z-DNA density (bp/Mb)", "gc_pct": "GC (%)", "rank": "Rank",
    "n_unique_zdna_sequences": "Distinct Z-DNA sequences (n)",
    "n_private_zdna_sequences": "Assembly-specific Z-DNA sequences (n)", "fraction_private": "Fraction assembly-specific",
    "length": "Z-DNA length (bp)", "score": "ZSeeker score", "metric_value": "Value", "transformed_coordinate": "Plot coordinate",
    "kde_percent_per_visual_bin": "KDE (% per visual bin)", "n_sampled_predictions": "Sampled predictions (n)",
    "n_in_display_range": "Predictions in display range (n)", "percent_in_display_range": "Predictions in display range (%)",
    "scott_bandwidth_multiplier": "Scott bandwidth multiplier",
    "visual_bin_width_custom_axis_units": "Visual bin width (axis units)", "n_autosomes": "Autosomes (n)",
    "paternal_bp_per_Mb": "Paternal density (bp/Mb)", "maternal_bp_per_Mb": "Maternal density (bp/Mb)",
    "delta_pat_minus_mat": "Paternal - maternal (bp/Mb)", "absolute_delta": "Absolute difference (bp/Mb)",
    "n_chromosomes": "Chromosomes (n)", "spearman_rho": "Spearman rho", "spearman_p": "Spearman P value",
    "ols_slope_bp_per_Mb_per_pct_GC": "OLS slope (bp/Mb per % GC)", "ols_intercept_bp_per_Mb": "OLS intercept (bp/Mb)",
    "ols_r_squared": "OLS R squared", "ols_p": "OLS P value", "test": "Test", "n_pairs": "Pairs (n)",
    "density_denominator": "Density denominator", "contig_rule": "Contig rule", "n_pat_only": "Paternal only (n)",
    "n_mat_only": "Maternal only (n)", "median_autosomes_per_pair": "Median autosomes per pair",
    "min_autosomes_per_pair": "Minimum autosomes per pair",
    "median_delta_pat_minus_mat_bp_per_Mb": "Median paternal - maternal (bp/Mb)",
    "mean_delta_bp_per_Mb": "Mean paternal - maternal (bp/Mb)", "sd_delta_bp_per_Mb": "SD paternal - maternal (bp/Mb)",
    "mean_absolute_delta_bp_per_Mb": "Mean absolute difference (bp/Mb)",
    "mean_absolute_delta_pct_grand_median": "Mean absolute difference (% of grand median)",
    "wilcoxon_W": "Wilcoxon W", "p_value": "P value",
}
OUT_FILE = PLOT_DIR / f"{STEM}.png"
LEGEND_FILE = PLOT_DIR / f"{STEM}_legend.png"

SP_ORDER = ["AFR", "AMR", "EAS", "EUR", "SAS"]
SP_COLORS = {"AFR": "#A559AA", "AMR": "#59A89C", "EAS": "#F0C571", "EUR": "#E02B35", "SAS": "#082A54"}
DOT_COLOR = "#000000"
FIG_WIDTH, FIG_HEIGHT = 4.0, 2.5
FONT_BASE, FONT_LABEL, FONT_TICK, FONT_ANNOTATION = 10, 11.5, 10, 10
AXIS_WIDTH, TICK_WIDTH, TICK_LENGTH = 1.0, 0.8, 3
BOX_WIDTH, BOX_LINEWIDTH, MEDIAN_LINEWIDTH, WHISKER_LINEWIDTH = 0.55, 0.8, 1.0, 0.8
DOT_SIZE, DOT_ALPHA, JITTER_SPREAD = 8, 0.5, 0.05

df = pd.read_csv(BASE / "3_hprc_haplotype_densities.final.tsv", sep="\t")
meta = pd.read_csv(BASE / "1_hprc_metadata.tsv", sep="\t", dtype=str)
assert len(df) == 464 and df["filename"].is_unique and meta["filename"].is_unique
df = df.merge(meta[["filename", "sample_id", "superpopulation"]], on="filename", how="left", validate="one_to_one")
df["density"] = pd.to_numeric(df["density_nuclear_min1"], errors="raise")
tab = pd.read_csv(BASE / "3_hprc_haplotype_tables.final.tsv", sep="\t")
tab["zdna_bp"] = sum(tab[f"zdna_bp_{c}"] for c in NUCLEAR)
tab["acgt_bp_min1"] = sum(tab[f"genome_bp_{c}_min1"] for c in NUCLEAR)
df = df.merge(tab[["filename", "zdna_bp", "acgt_bp_min1"]], on="filename", how="left", validate="one_to_one")
sp_df = df[df["superpopulation"].isin(SP_ORDER)].copy()
assert len(sp_df) == 464 and not sp_df["density"].isna().any()
groups = [sp_df.loc[sp_df["superpopulation"] == pop, "density"].to_numpy() for pop in SP_ORDER]

sp_df[["filename", "sample_id", "superpopulation", "zdna_bp", "acgt_bp_min1", "density"]].rename(
    columns={"density": "zdna_density_bp_per_Mb"}).rename(columns=READABLE).to_csv(
    TEST_DIR / f"{STEM}.tsv", sep="\t", index=False, float_format="%.10g")
kw_stat, kw_p = stats.kruskal(*groups)
dunn = sp.posthoc_dunn(sp_df, val_col="density", group_col="superpopulation",
                       p_adjust="bonferroni").reindex(index=SP_ORDER, columns=SP_ORDER)
stats_rows = [{"Test": "Kruskal-Wallis", "Group 1": "all", "Group 2": "", "Haplotypes (n)": len(sp_df),
               "H statistic": kw_stat, "Degrees of freedom": len(SP_ORDER) - 1, "P value": kw_p}]
stats_rows += [{"Test": "Dunn (Bonferroni-adjusted)", "Group 1": a, "Group 2": b, "P value": dunn.loc[a, b]}
               for i_, a in enumerate(SP_ORDER) for b in SP_ORDER[i_ + 1:]]
pd.DataFrame(stats_rows).to_csv(TEST_DIR / f"{STEM}_statistics.tsv", sep="\t", index=False, float_format="%.10g")

plt.rcParams.update({
    "font.family": "Nimbus Sans", "font.size": FONT_BASE, "axes.labelsize": FONT_LABEL,
    "xtick.labelsize": FONT_TICK, "ytick.labelsize": FONT_TICK, "axes.linewidth": AXIS_WIDTH,
    "axes.spines.top": False, "axes.spines.right": False, "pdf.fonttype": 42, "ps.fonttype": 42,
})
fig, ax = plt.subplots(figsize=(FIG_WIDTH, FIG_HEIGHT))
bp = ax.boxplot(groups, patch_artist=True, widths=BOX_WIDTH, showfliers=False, zorder=2)
for box, pop in zip(bp["boxes"], SP_ORDER):
    box.set_facecolor(SP_COLORS[pop]); box.set_edgecolor("black")
    box.set_linewidth(BOX_LINEWIDTH); box.set_alpha(0.85)
for artist in bp["medians"]:
    artist.set_color("black"); artist.set_linewidth(MEDIAN_LINEWIDTH)
for name in ("whiskers", "caps"):
    for artist in bp[name]:
        artist.set_color("black"); artist.set_linewidth(WHISKER_LINEWIDTH)
rng = np.random.default_rng(1)
for i, values in enumerate(groups, start=1):
    ax.scatter(i + rng.normal(0, JITTER_SPREAD, len(values)), values, s=DOT_SIZE, color=DOT_COLOR,
               alpha=DOT_ALPHA, edgecolors="none", zorder=3)
ax.set_xlabel("Superpopulation", fontsize=FONT_LABEL, labelpad=5)
ax.set_ylabel("Z-DNA density (bp/Mb)", fontsize=FONT_LABEL, labelpad=5)
ax.set_xticks(range(1, len(SP_ORDER) + 1))
ax.set_xticklabels(SP_ORDER, rotation=0, ha="center")
ax.yaxis.set_major_locator(MaxNLocator(nbins=8))
ax.tick_params(axis="both", which="major", labelsize=FONT_TICK, width=TICK_WIDTH, length=TICK_LENGTH,
               direction="out", pad=3)
ymin, ymax = ax.get_ylim()
padding = (ymax - ymin) * 0.05
ax.set_ylim(ymin, ymax + padding * 2)
for i, values in enumerate(groups, start=1):
    ax.text(i, ymax + padding * 0.5, f"n={len(values)}", ha="center", va="bottom", fontsize=FONT_ANNOTATION,
            color="black", zorder=5)
ax.grid(True, axis="both", which="major", linestyle="--", linewidth=0.4, color="#B0B0B0", alpha=0.5, zorder=0)
ax.set_axisbelow(True)
fig.subplots_adjust(left=0.17, right=0.97, bottom=0.25, top=0.96)
fig.savefig(OUT_FILE, dpi=600, bbox_inches=None)
fig.savefig(OUT_FILE.with_suffix(".pdf"), bbox_inches=None)
plt.show()

legend_handles = [Patch(facecolor=SP_COLORS[p], edgecolor="black", linewidth=0.8, alpha=0.85, label=p)
                  for p in SP_ORDER]
legend_fig, legend_ax = plt.subplots(figsize=(1.8, 1.6))
legend_ax.axis("off")
legend_ax.legend(handles=legend_handles, title="Superpopulation", loc="center", fontsize=FONT_BASE,
                 title_fontsize=FONT_LABEL, frameon=False, ncol=1, labelspacing=0.4, handlelength=1.3,
                 handleheight=0.8, handletextpad=0.5)
legend_fig.savefig(LEGEND_FILE, dpi=600, bbox_inches="tight", pad_inches=0.05, transparent=True)
legend_fig.savefig(LEGEND_FILE.with_suffix(".pdf"), bbox_inches="tight", pad_inches=0.05, transparent=True)
plt.close(legend_fig)

print(f"Kruskal-Wallis: H = {kw_stat:.4f}, p = {kw_p:.3g} (n = {len(sp_df)})")
print("Dunn's test, Bonferroni-adjusted p:\n" + dunn.round(4).to_string())
```

#### Fig. 1d

```python
"""Fig. 1d - ranked number of assembly-specific Z-DNA sequences across 466 assemblies, bars by superpopulation.
"""
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from matplotlib.ticker import (
    FixedLocator,
    FuncFormatter,
    NullLocator,
)


BASE = Path("/path/to/pangenomes_zdna")

PRIVATE = BASE / "4_hprc_private_zdna_sequences.tsv"
META = BASE / "1_hprc_metadata.tsv"

PLOT_DIR = BASE / "plots"
TEST_DIR = BASE / "tests"

PLOT_DIR.mkdir(parents=True, exist_ok=True)
TEST_DIR.mkdir(parents=True, exist_ok=True)

STEM = "fig1d_assembly_specific_zdna_sequences"
READABLE = {
    "dataset": "Assembly", "filename": "Assembly", "source_filename": "Assembly", "assembly": "Reference genome",
    "dataset_type": "Assembly type", "sample_id": "Sample", "superpopulation": "Superpopulation",
    "pat_or_mat": "Parental origin", "chr": "Chromosome", "ucsc_chr": "Chromosome", "n_contigs": "Contigs (n)",
    "n_chr_random": "chr_random contigs (n)", "zdna_bp": "Z-DNA (bp)", "acgt_bp_min1": "ACGT (bp)", "genome_bp": "ACGT (bp)",
    "zdna_density_bp_per_Mb": "Z-DNA density (bp/Mb)", "density_bp_per_Mb": "Z-DNA density (bp/Mb)",
    "density_bp_per_Mb_min1": "Z-DNA density (bp/Mb)", "gc_pct": "GC (%)", "rank": "Rank",
    "n_unique_zdna_sequences": "Distinct Z-DNA sequences (n)",
    "n_private_zdna_sequences": "Assembly-specific Z-DNA sequences (n)", "fraction_private": "Fraction assembly-specific",
    "length": "Z-DNA length (bp)", "score": "ZSeeker score", "metric_value": "Value", "transformed_coordinate": "Plot coordinate",
    "kde_percent_per_visual_bin": "KDE (% per visual bin)", "n_sampled_predictions": "Sampled predictions (n)",
    "n_in_display_range": "Predictions in display range (n)", "percent_in_display_range": "Predictions in display range (%)",
    "scott_bandwidth_multiplier": "Scott bandwidth multiplier",
    "visual_bin_width_custom_axis_units": "Visual bin width (axis units)", "n_autosomes": "Autosomes (n)",
    "paternal_bp_per_Mb": "Paternal density (bp/Mb)", "maternal_bp_per_Mb": "Maternal density (bp/Mb)",
    "delta_pat_minus_mat": "Paternal - maternal (bp/Mb)", "absolute_delta": "Absolute difference (bp/Mb)",
    "n_chromosomes": "Chromosomes (n)", "spearman_rho": "Spearman rho", "spearman_p": "Spearman P value",
    "ols_slope_bp_per_Mb_per_pct_GC": "OLS slope (bp/Mb per % GC)", "ols_intercept_bp_per_Mb": "OLS intercept (bp/Mb)",
    "ols_r_squared": "OLS R squared", "ols_p": "OLS P value", "test": "Test", "n_pairs": "Pairs (n)",
    "density_denominator": "Density denominator", "contig_rule": "Contig rule", "n_pat_only": "Paternal only (n)",
    "n_mat_only": "Maternal only (n)", "median_autosomes_per_pair": "Median autosomes per pair",
    "min_autosomes_per_pair": "Minimum autosomes per pair",
    "median_delta_pat_minus_mat_bp_per_Mb": "Median paternal - maternal (bp/Mb)",
    "mean_delta_bp_per_Mb": "Mean paternal - maternal (bp/Mb)", "sd_delta_bp_per_Mb": "SD paternal - maternal (bp/Mb)",
    "mean_absolute_delta_bp_per_Mb": "Mean absolute difference (bp/Mb)",
    "mean_absolute_delta_pct_grand_median": "Mean absolute difference (% of grand median)",
    "wilcoxon_W": "Wilcoxon W", "p_value": "P value",
}


FIG_WIDTH = 4.0
FIG_HEIGHT = 2.5

FONT_BASE = 10
FONT_LABEL = 11.5
FONT_TICK = 10

AXIS_WIDTH = 1.0
TICK_WIDTH = 0.8
TICK_LENGTH = 3


YMAX = 2500

Y_TICKS = [
    0,
    500,
    1000,
    1500,
    2000,
    2500,
]


BAR_WIDTH = 1.04

BAR_ANTIALIASED = False

BAR_RASTERIZED = True


SP_ORDER = ["AFR", "AMR", "EAS", "EUR", "SAS"]

SP_COLORS = {
    "AFR": "#A559AA",
    "AMR": "#59A89C",
    "EAS": "#F0C571",
    "EUR": "#E02B35",
    "SAS": "#082A54",
}


REF_COLORS = {
    "GRCh38_primary": "#547853",
    "CHM13": "#536078",
}

OTHER_COLOR = "#BDBDBD"


REF_LINEWIDTH = 1.2
REF_MARKER_SIZE = 8


plt.rcParams.update({
    "font.family": "Nimbus Sans",
    "font.size": FONT_BASE,
    "axes.labelsize": FONT_LABEL,
    "xtick.labelsize": FONT_TICK,
    "ytick.labelsize": FONT_TICK,
    "axes.linewidth": AXIS_WIDTH,
    "axes.spines.top": False,
    "axes.spines.right": False,
    "pdf.fonttype": 42,
    "ps.fonttype": 42,
    "savefig.facecolor": "white",
})


def main():

    priv = pd.read_csv(
        PRIVATE,
        sep="\t",
    )

    required = {
        "dataset",
        "n_unique_zdna_sequences",
        "n_private_zdna_sequences",
        "fraction_private",
    }

    missing = required.difference(priv.columns)

    if missing:
        raise ValueError(
            f"Missing columns in {PRIVATE}: "
            f"{sorted(missing)}"
        )

    if (
        priv["dataset"].isna().any()
        or priv["dataset"].duplicated().any()
    ):
        raise ValueError(
            "Private-sequence table must contain "
            "unique, non-null datasets"
        )

    meta = pd.read_csv(
        META,
        sep="\t",
        dtype=str,
        usecols=[
            "filename",
            "superpopulation",
        ],
    )

    if meta["filename"].duplicated().any():
        raise ValueError(
            "Duplicate filenames in metadata"
        )

    df = priv.merge(
        meta,
        left_on="dataset",
        right_on="filename",
        how="left",
        validate="one_to_one",
    )

    numeric_cols = [
        "n_unique_zdna_sequences",
        "n_private_zdna_sequences",
        "fraction_private",
    ]

    for col in numeric_cols:
        df[col] = pd.to_numeric(
            df[col],
            errors="raise",
        )

    if df["n_private_zdna_sequences"].isna().any():
        raise ValueError(
            "Null private-sequence counts"
        )

    if (df["n_private_zdna_sequences"] < 0).any():
        raise ValueError(
            "Negative private-sequence counts"
        )

    if (
        df["n_private_zdna_sequences"]
        > df["n_unique_zdna_sequences"]
    ).any():
        raise ValueError(
            "Private counts exceed total "
            "distinct sequence counts"
        )

    for ref in REF_COLORS:

        if (df["dataset"] == ref).sum() != 1:
            raise ValueError(
                f"Expected exactly one row for {ref}"
            )

    df["superpopulation"] = (
        df["superpopulation"]
        .fillna("")
        .str.strip()
    )

    df.loc[
        ~df["superpopulation"].isin(SP_COLORS),
        "superpopulation"
    ] = "Unassigned"

    df.loc[
        df["dataset"].isin(REF_COLORS),
        "superpopulation"
    ] = "Reference"

    df = df.sort_values(
        [
            "n_private_zdna_sequences",
            "dataset",
        ],
        ascending=[False, True],
        kind="mergesort",
    ).reset_index(drop=True)

    df["rank"] = np.arange(
        1,
        len(df) + 1,
    )

    plot_data_file = (
        TEST_DIR / f"{STEM}.tsv"
    )

    df.drop(
        columns=["filename"]
    ).rename(columns=READABLE).to_csv(
        plot_data_file,
        sep="\t",
        index=False,
        float_format="%.10g",
    )

    fig, ax = plt.subplots(
        figsize=(FIG_WIDTH, FIG_HEIGHT)
    )

    x = np.arange(len(df))

    y = df[
        "n_private_zdna_sequences"
    ].to_numpy(dtype=float)

    is_ref = df[
        "dataset"
    ].isin(REF_COLORS).to_numpy()

    ordinary = df.loc[~is_ref]

    ord_x = x[~is_ref]

    ordinary_colors = [
        SP_COLORS.get(
            sp,
            OTHER_COLOR,
        )
        for sp in ordinary["superpopulation"]
    ]

    ax.bar(
        ord_x,
        ordinary[
            "n_private_zdna_sequences"
        ].to_numpy(dtype=float),
        width=BAR_WIDTH,
        align="center",
        color=ordinary_colors,
        edgecolor="none",
        linewidth=0,
        antialiased=BAR_ANTIALIASED,
        rasterized=BAR_RASTERIZED,
        zorder=2,
    )

    ymax_observed = max(
        float(y.max()),
        1.0,
    )

    for ref, color in REF_COLORS.items():

        idx = int(
            df.index[
                df["dataset"] == ref
            ][0]
        )

        value = float(y[idx])

        ax.vlines(
            idx,
            0,
            value,
            colors=color,
            linewidth=REF_LINEWIDTH,
            zorder=5,
        )

        ax.scatter(
            [idx],
            [value],
            s=REF_MARKER_SIZE,
            color=color,
            edgecolors="none",
            zorder=6,
        )

    ax.set_xlim(
        -0.52,
        len(df) - 0.48,
    )

    ax.set_xticks([])

    ax.set_xlabel(
        "Haplotype / reference genome (ranked)",
        fontsize=FONT_LABEL,
        labelpad=5,
    )

    ax.set_ylim(
        0,
        YMAX,
    )

    ax.yaxis.set_major_locator(
        FixedLocator(Y_TICKS)
    )

    ax.yaxis.set_major_formatter(
        FuncFormatter(
            lambda value, pos: f"{value:,.0f}"
        )
    )

    ax.yaxis.set_minor_locator(
        NullLocator()
    )

    ax.yaxis.offsetText.set_visible(False)

    ax.set_ylabel(
        "Assembly-specific Z-DNA sequences",
        fontsize=FONT_LABEL,
        labelpad=5,
    )

    if ymax_observed > YMAX:

        n_clipped = int(
            (y > YMAX).sum()
        )

        print(
            f"WARNING: {n_clipped} counts exceed "
            f"{YMAX:,} and are clipped in the figure. "
            "Original values remain in the TSV."
        )

    ax.tick_params(
        axis="both",
        which="major",
        labelsize=FONT_TICK,
        width=TICK_WIDTH,
        length=TICK_LENGTH,
        direction="out",
        pad=3,
    )

    ax.grid(
        True,
        axis="y",
        which="major",
        linestyle="--",
        linewidth=0.4,
        color="#B0B0B0",
        alpha=0.5,
        zorder=0,
    )

    ax.set_axisbelow(True)

    fig.subplots_adjust(
        left=0.17,
        right=0.97,
        bottom=0.25,
        top=0.96,
    )

    png = PLOT_DIR / f"{STEM}.png"
    pdf = PLOT_DIR / f"{STEM}.pdf"

    fig.savefig(
        png,
        dpi=600,
        bbox_inches=None,
    )

    fig.savefig(
        pdf,
        dpi=600,
        bbox_inches=None,
    )

    plt.show()
    plt.close(fig)

    print(
        f"Datasets plotted: {len(df)}"
    )

    print(
        f"Private-count range: "
        f"{int(y.min()):,}–{int(y.max()):,}"
    )
    hap = df[~is_ref]
    n_afr = int((hap.superpopulation == "AFR").sum())
    print(f"AFR among the {n_afr} highest-ranked haplotypes: {int((hap.head(n_afr).superpopulation == 'AFR').sum())}")
    print("Median private sequences per superpopulation: "
            + hap.groupby("superpopulation").n_private_zdna_sequences.median().round(0).to_dict().__str__())
    print("\nReferences:")

    print(
        df.loc[
            is_ref,
            [
                "dataset",
                "rank",
                "n_private_zdna_sequences",
                "fraction_private",
            ],
        ].to_string(index=False)
    )

    print(f"\nSaved PNG: {png}")
    print(f"Saved PDF: {pdf}")
    print(f"Plot data: {plot_data_file}")

    print(
        f"Figure dimensions: "
        f"{FIG_WIDTH} x {FIG_HEIGHT} inches"
    )


if __name__ == "__main__":
    main()
```

#### Fig. 1e, Suppl. Fig. 1a

```python
"""Fig. 1e and Suppl. Fig. 1a - KDE of Z-DNA length (Fig. 1e) and ZSeeker score (Suppl. Fig. 1a), one curve per
assembly (2,000 sampled predictions each; grey HPRC, green GRCh38, blue CHM13) + legend.
"""
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from matplotlib.lines import Line2D
from matplotlib.ticker import MaxNLocator
from scipy.stats import gaussian_kde


BASE = Path("/path/to/pangenomes_zdna")

READABLE = {
    "dataset": "Assembly", "filename": "Assembly", "source_filename": "Assembly", "assembly": "Reference genome",
    "dataset_type": "Assembly type", "sample_id": "Sample", "superpopulation": "Superpopulation",
    "pat_or_mat": "Parental origin", "chr": "Chromosome", "ucsc_chr": "Chromosome", "n_contigs": "Contigs (n)",
    "n_chr_random": "chr_random contigs (n)", "zdna_bp": "Z-DNA (bp)", "acgt_bp_min1": "ACGT (bp)", "genome_bp": "ACGT (bp)",
    "zdna_density_bp_per_Mb": "Z-DNA density (bp/Mb)", "density_bp_per_Mb": "Z-DNA density (bp/Mb)",
    "density_bp_per_Mb_min1": "Z-DNA density (bp/Mb)", "gc_pct": "GC (%)", "rank": "Rank",
    "n_unique_zdna_sequences": "Distinct Z-DNA sequences (n)",
    "n_private_zdna_sequences": "Assembly-specific Z-DNA sequences (n)", "fraction_private": "Fraction assembly-specific",
    "length": "Z-DNA length (bp)", "score": "ZSeeker score", "metric_value": "Value", "transformed_coordinate": "Plot coordinate",
    "kde_percent_per_visual_bin": "KDE (% per visual bin)", "n_sampled_predictions": "Sampled predictions (n)",
    "n_in_display_range": "Predictions in display range (n)", "percent_in_display_range": "Predictions in display range (%)",
    "scott_bandwidth_multiplier": "Scott bandwidth multiplier",
    "visual_bin_width_custom_axis_units": "Visual bin width (axis units)", "n_autosomes": "Autosomes (n)",
    "paternal_bp_per_Mb": "Paternal density (bp/Mb)", "maternal_bp_per_Mb": "Maternal density (bp/Mb)",
    "delta_pat_minus_mat": "Paternal - maternal (bp/Mb)", "absolute_delta": "Absolute difference (bp/Mb)",
    "n_chromosomes": "Chromosomes (n)", "spearman_rho": "Spearman rho", "spearman_p": "Spearman P value",
    "ols_slope_bp_per_Mb_per_pct_GC": "OLS slope (bp/Mb per % GC)", "ols_intercept_bp_per_Mb": "OLS intercept (bp/Mb)",
    "ols_r_squared": "OLS R squared", "ols_p": "OLS P value", "test": "Test", "n_pairs": "Pairs (n)",
    "density_denominator": "Density denominator", "contig_rule": "Contig rule", "n_pat_only": "Paternal only (n)",
    "n_mat_only": "Maternal only (n)", "median_autosomes_per_pair": "Median autosomes per pair",
    "min_autosomes_per_pair": "Minimum autosomes per pair",
    "median_delta_pat_minus_mat_bp_per_Mb": "Median paternal - maternal (bp/Mb)",
    "mean_delta_bp_per_Mb": "Mean paternal - maternal (bp/Mb)", "sd_delta_bp_per_Mb": "SD paternal - maternal (bp/Mb)",
    "mean_absolute_delta_bp_per_Mb": "Mean absolute difference (bp/Mb)",
    "mean_absolute_delta_pct_grand_median": "Mean absolute difference (% of grand median)",
    "wilcoxon_W": "Wilcoxon W", "p_value": "P value",
}
PLOT_DIR = BASE / "plots"
TEST_DIR = BASE / "tests"

PLOT_DIR.mkdir(parents=True, exist_ok=True)
TEST_DIR.mkdir(parents=True, exist_ok=True)


FIG_WIDTH = 4.0
FIG_HEIGHT = 2.1

FONT_BASE = 10
FONT_LABEL = 11.5
FONT_TICK = 10

AXIS_WIDTH = 1.0
TICK_WIDTH = 0.8
TICK_LENGTH = 3


CHM13_COLOR = "#536078"
GRCH38_COLOR = "#547853"
HIST_COLOR = "#BDBDBD"

HPRC_COLOR = HIST_COLOR


HPRC_LINEWIDTH = 0.45
HPRC_ALPHA = 0.30

REF_LINEWIDTH = 1.2
REF_ALPHA = 1.0


HPRC_RASTERIZED = True


LENGTH_X_TICKS = np.array(
    [9, 50, 100, 500, 5000],
    dtype=float,
)

SCORE_X_TICKS = np.array(
    [50, 75, 100, 500, 5000],
    dtype=float,
)


N_GRID = 300


BANDWIDTH_MULTIPLIER = 1.35


VISUAL_BIN_WIDTH = 4.0 / 90.0


MIN_POINTS = 30


plt.rcParams.update({
    "font.family": "Nimbus Sans",
    "font.size": FONT_BASE,
    "axes.labelsize": FONT_LABEL,
    "xtick.labelsize": FONT_TICK,
    "ytick.labelsize": FONT_TICK,
    "axes.linewidth": AXIS_WIDTH,
    "axes.spines.top": False,
    "axes.spines.right": False,
    "pdf.fonttype": 42,
    "ps.fonttype": 42,
    "savefig.facecolor": "white",
})


def forward_custom_axis(x, ticks):

    x = np.asarray(x, dtype=float)

    z = np.full_like(x, np.nan)

    for i, (lo, hi) in enumerate(
        zip(ticks[:-1], ticks[1:])
    ):

        mask = (x >= lo) & (x <= hi)

        z[mask] = (
            i
            + np.log(x[mask] / lo)
            / np.log(hi / lo)
        )

    mask = (x > 0) & (x < ticks[0])

    z[mask] = (
        np.log(x[mask] / ticks[0])
        / np.log(ticks[1] / ticks[0])
    )

    mask = x > ticks[-1]

    z[mask] = (
        len(ticks) - 1
        + np.log(x[mask] / ticks[-1])
        / np.log(ticks[-1] / ticks[-2])
    )

    return z


def inverse_custom_axis(z, ticks):

    z = np.asarray(z, dtype=float)

    x = np.full_like(z, np.nan)

    for i, (lo, hi) in enumerate(
        zip(ticks[:-1], ticks[1:])
    ):

        mask = (z >= i) & (z <= i + 1)

        x[mask] = (
            lo * (hi / lo) ** (z[mask] - i)
        )

    mask = z < 0

    x[mask] = (
        ticks[0]
        * (ticks[1] / ticks[0]) ** z[mask]
    )

    mask = z > len(ticks) - 1

    x[mask] = (
        ticks[-1]
        * (ticks[-1] / ticks[-2]) **
        (z[mask] - len(ticks) + 1)
    )

    return x


def input_path(metric):
      path = BASE / f"4_{metric}_per_haplotype_distribution.tsv"
      if not path.is_file():
          raise FileNotFoundError(path)
      return path


def make_panel(metric, xlabel):

    filename = input_path(metric)

    df = pd.read_csv(
        filename,
        sep="\t",
        usecols=[
            "source_filename",
            metric,
        ],
    )

    df[metric] = pd.to_numeric(
        df[metric],
        errors="raise",
    )

    df = df.loc[
        np.isfinite(df[metric])
        & (df[metric] > 0)
    ].copy()

    if df.empty:
        raise ValueError(
            f"No valid values in {filename}"
        )

    ticks = (
        SCORE_X_TICKS
        if metric == "score"
        else LENGTH_X_TICKS
    )

    z_grid = np.linspace(
        0,
        len(ticks) - 1,
        N_GRID,
    )

    x_grid = inverse_custom_axis(
        z_grid,
        ticks,
    )

    curves = []
    summaries = []

    for source, group in df.groupby(
        "source_filename",
        sort=True,
    ):

        values = group[
            metric
        ].to_numpy(dtype=float)

        z_values = forward_custom_axis(
            values,
            ticks,
        )

        z_values = z_values[
            np.isfinite(z_values)
        ]

        n_display = int(
            np.count_nonzero(
                (z_values >= 0)
                & (z_values <= 4)
            )
        )

        kind = (
            source
            if source in ("CHM13", "GRCh38")
            else "HPRC"
        )

        if (
            len(z_values) < MIN_POINTS
            or np.std(z_values) < 1e-12
        ):

            print(
                f"WARNING: skipping {source}: "
                "insufficient or constant values "
                f"({len(z_values)})"
            )

            continue

        kde = gaussian_kde(
            z_values,
            bw_method=lambda k: (
                k.scotts_factor()
                * BANDWIDTH_MULTIPLIER
            ),
        )

        percent = (
            kde(z_grid)
            * 100.0
            * VISUAL_BIN_WIDTH
        )

        curves.append((
            source,
            kind,
            percent,
            len(z_values),
            n_display,
        ))

        summaries.append({
            "source_filename": source,
            "dataset_type": kind,
            "n_sampled_predictions": len(z_values),
            "n_in_display_range": n_display,
            "percent_in_display_range": (
                n_display / len(z_values) * 100.0
            ),
            "scott_bandwidth_multiplier": (
                BANDWIDTH_MULTIPLIER
            ),
            "visual_bin_width_custom_axis_units": (
                VISUAL_BIN_WIDTH
            ),
        })

    if not any(
        c[1] == "HPRC"
        for c in curves
    ):
        raise ValueError(
            "No HPRC haplotypes found"
        )

    for ref in ("GRCh38", "CHM13"):

        n_ref = sum(
            c[0] == ref
            for c in curves
        )

        if n_ref != 1:
            raise ValueError(
                f"Expected exactly one {ref} "
                f"reference, found {n_ref}"
            )

    stem = {
        "length": "fig1e_zdna_length_kde",
        "score": "suppl_fig1a_zseeker_score_kde",
    }[metric]

    df.assign(
        dataset_type=np.where(df["source_filename"].isin(["CHM13", "GRCh38"]),
                              df["source_filename"], "HPRC haplotype")
    )[["source_filename", "dataset_type", metric]].rename(columns=READABLE).to_csv(
        TEST_DIR / f"{stem}_sampled_values.tsv", sep="\t", index=False, float_format="%.10g",
    )

    pd.DataFrame(
        summaries
    ).rename(columns=READABLE).to_csv(
        TEST_DIR / f"{stem}_sampling_summary.tsv",
        sep="\t",
        index=False,
    )

    pd.concat(
        [
            pd.DataFrame({
                "source_filename": source,
                "dataset_type": kind,
                "metric_value": x_grid,
                "transformed_coordinate": z_grid,
                "kde_percent_per_visual_bin": pct,
            })
            for source, kind, pct, _, _ in curves
        ],
        ignore_index=True,
    ).rename(columns=READABLE).to_csv(
        TEST_DIR / f"{stem}_curves.tsv",
        sep="\t",
        index=False,
        float_format="%.10g",
    )

    fig, ax = plt.subplots(
        figsize=(FIG_WIDTH, FIG_HEIGHT),
        layout="constrained",
    )

    for source, kind, percent, n, _ in curves:

        if kind == "HPRC":

            ax.plot(
                x_grid,
                percent,
                color=HPRC_COLOR,
                linewidth=HPRC_LINEWIDTH,
                alpha=HPRC_ALPHA,
                zorder=2,
                rasterized=HPRC_RASTERIZED,
            )

    for ref, color, layer in [
        ("GRCh38", GRCH38_COLOR, 5),
        ("CHM13", CHM13_COLOR, 6),
    ]:

        this = [
            c for c in curves
            if c[0] == ref
        ][0]

        ax.plot(
            x_grid,
            this[2],
            color=color,
            linewidth=REF_LINEWIDTH,
            alpha=REF_ALPHA,
            zorder=layer,
        )

    ax.set_xscale(
        "function",
        functions=(
            lambda x: forward_custom_axis(
                x, ticks
            ),
            lambda z: inverse_custom_axis(
                z, ticks
            ),
        ),
    )

    ax.set_xlim(
        ticks[0],
        ticks[-1],
    )

    ax.set_xticks(ticks)

    ax.set_xticklabels(
        [str(int(x)) for x in ticks],
        rotation=0,
        ha="center",
    )

    ax.set_ylim(bottom=0)

    ax.yaxis.set_major_locator(
        MaxNLocator(
            nbins=8,
        )
    )

    ax.set_xlabel(
        xlabel,
        fontsize=FONT_LABEL,
        labelpad=5,
    )

    ax.set_ylabel(
        "Z-DNA predictions per bin (%)",
        fontsize=FONT_LABEL,
        labelpad=5,
    )

    ax.tick_params(
        axis="both",
        which="major",
        labelsize=FONT_TICK,
        width=TICK_WIDTH,
        length=TICK_LENGTH,
        direction="out",
        pad=3,
    )

    ax.minorticks_off()

    ax.grid(
        True,
        axis="both",
        which="major",
        linestyle="--",
        linewidth=0.4,
        color="#B0B0B0",
        alpha=0.5,
        zorder=0,
    )

    ax.set_axisbelow(True)

    png = PLOT_DIR / f"{stem}.png"
    pdf = PLOT_DIR / f"{stem}.pdf"

    fig.savefig(
        png,
        dpi=600,
        bbox_inches=None,
    )

    fig.get_layout_engine().set(
        w_pad=0.03,
        h_pad=0.03,
        wspace=0.02,
        hspace=0.02,
    )

    fig.savefig(
        pdf,
        dpi=600,
        bbox_inches=None,
    )

    plt.show()
    plt.close(fig)

    print(
        f"{metric}: "
        f"{sum(c[1] == 'HPRC' for c in curves)} "
        "HPRC haplotypes"
    )

    print(f"Saved PNG: {png}")
    print(f"Saved PDF: {pdf}")


def make_legend():

    handles = [
        Line2D(
            [0], [0],
            color=HPRC_COLOR,
            linewidth=1.2,
            label="HPRC haplotypes",
        ),
        Line2D(
            [0], [0],
            color=GRCH38_COLOR,
            linewidth=REF_LINEWIDTH,
            label="GRCh38",
        ),
        Line2D(
            [0], [0],
            color=CHM13_COLOR,
            linewidth=REF_LINEWIDTH,
            label="CHM13",
        ),
    ]

    legend_fig, legend_ax = plt.subplots(
        figsize=(2.0, 1.1)
    )

    legend_ax.axis("off")

    legend_ax.legend(
        handles=handles,
        loc="center",
        fontsize=FONT_BASE,
        frameon=False,
        ncol=1,
        labelspacing=0.45,
        handlelength=1.6,
        handletextpad=0.6,
    )

    legend_png = PLOT_DIR / "fig1e_kde_legend.png"
    legend_pdf = PLOT_DIR / "fig1e_kde_legend.pdf"

    legend_fig.savefig(
        legend_png,
        dpi=600,
        bbox_inches="tight",
        pad_inches=0.05,
        transparent=True,
    )

    legend_fig.savefig(
        legend_pdf,
        bbox_inches="tight",
        pad_inches=0.05,
        transparent=True,
    )

    plt.close(legend_fig)

    print(f"Saved legend PNG: {legend_png}")
    print(f"Saved legend PDF: {legend_pdf}")


if __name__ == "__main__":

    make_panel(
        "score",
        "Z-DNA score",
    )

    make_panel(
        "length",
        "Z-DNA length (bp)",
    )

    make_legend()


for metric in ("length", "score"):
    df = pd.read_csv(
        BASE / f"4_{metric}_per_haplotype_distribution.tsv",
        sep="\t"
    )

    medians = df.groupby("source_filename")[metric].median()
    print(
        f"{metric}: median range = "
        f"{medians.min():.1f}–{medians.max():.1f} "
        f"{'bp' if metric == 'length' else ''}"
        f"(n={len(medians)} assemblies)"
    )
```

#### Fig. 1f

```python
"""Fig. 1f - chromosome-level Z-DNA density across HPRC haplotypes (violins; longest primary contig + assigned
chr_random), GRCh38 (triangles) and CHM13 (circles) + legend.
"""
import re
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from matplotlib.lines import Line2D
from matplotlib.patches import Patch
from matplotlib.ticker import MaxNLocator

BASE = Path("/path/to/pangenomes_zdna")
HAP_FILE = BASE / "3_hprc_chromosome_tables.final.tsv"
REF_FILE = BASE / "2_CHM13_vs_GRCh38_zdna_density_per_chr.tsv"
PLOT_DIR, TEST_DIR = BASE / "plots", BASE / "tests"
PLOT_DIR.mkdir(parents=True, exist_ok=True)
TEST_DIR.mkdir(parents=True, exist_ok=True)
STEM = "fig1f_zdna_density_per_chromosome"
READABLE = {
    "dataset": "Assembly", "filename": "Assembly", "source_filename": "Assembly", "assembly": "Reference genome",
    "dataset_type": "Assembly type", "sample_id": "Sample", "superpopulation": "Superpopulation",
    "pat_or_mat": "Parental origin", "chr": "Chromosome", "ucsc_chr": "Chromosome", "n_contigs": "Contigs (n)",
    "n_chr_random": "chr_random contigs (n)", "zdna_bp": "Z-DNA (bp)", "acgt_bp_min1": "ACGT (bp)", "genome_bp": "ACGT (bp)",
    "zdna_density_bp_per_Mb": "Z-DNA density (bp/Mb)", "density_bp_per_Mb": "Z-DNA density (bp/Mb)",
    "density_bp_per_Mb_min1": "Z-DNA density (bp/Mb)", "gc_pct": "GC (%)", "rank": "Rank",
    "n_unique_zdna_sequences": "Distinct Z-DNA sequences (n)",
    "n_private_zdna_sequences": "Assembly-specific Z-DNA sequences (n)", "fraction_private": "Fraction assembly-specific",
    "length": "Z-DNA length (bp)", "score": "ZSeeker score", "metric_value": "Value", "transformed_coordinate": "Plot coordinate",
    "kde_percent_per_visual_bin": "KDE (% per visual bin)", "n_sampled_predictions": "Sampled predictions (n)",
    "n_in_display_range": "Predictions in display range (n)", "percent_in_display_range": "Predictions in display range (%)",
    "scott_bandwidth_multiplier": "Scott bandwidth multiplier",
    "visual_bin_width_custom_axis_units": "Visual bin width (axis units)", "n_autosomes": "Autosomes (n)",
    "paternal_bp_per_Mb": "Paternal density (bp/Mb)", "maternal_bp_per_Mb": "Maternal density (bp/Mb)",
    "delta_pat_minus_mat": "Paternal - maternal (bp/Mb)", "absolute_delta": "Absolute difference (bp/Mb)",
    "n_chromosomes": "Chromosomes (n)", "spearman_rho": "Spearman rho", "spearman_p": "Spearman P value",
    "ols_slope_bp_per_Mb_per_pct_GC": "OLS slope (bp/Mb per % GC)", "ols_intercept_bp_per_Mb": "OLS intercept (bp/Mb)",
    "ols_r_squared": "OLS R squared", "ols_p": "OLS P value", "test": "Test", "n_pairs": "Pairs (n)",
    "density_denominator": "Density denominator", "contig_rule": "Contig rule", "n_pat_only": "Paternal only (n)",
    "n_mat_only": "Maternal only (n)", "median_autosomes_per_pair": "Median autosomes per pair",
    "min_autosomes_per_pair": "Minimum autosomes per pair",
    "median_delta_pat_minus_mat_bp_per_Mb": "Median paternal - maternal (bp/Mb)",
    "mean_delta_bp_per_Mb": "Mean paternal - maternal (bp/Mb)", "sd_delta_bp_per_Mb": "SD paternal - maternal (bp/Mb)",
    "mean_absolute_delta_bp_per_Mb": "Mean absolute difference (bp/Mb)",
    "mean_absolute_delta_pct_grand_median": "Mean absolute difference (% of grand median)",
    "wilcoxon_W": "Wilcoxon W", "p_value": "P value",
}
OUT_FILE = PLOT_DIR / f"{STEM}.png"
LEGEND_FILE = PLOT_DIR / f"{STEM}_legend.png"
PLOT_DATA = TEST_DIR / f"{STEM}.tsv"

CHROMS = [f"chr{i}" for i in range(1, 23)] + ["chrX", "chrY"]
CHR_RE = re.compile(r"^(chr(?:[1-9]|1[0-9]|2[0-2]|X|Y))(?=_|$)")
POSITIONS = np.arange(1, len(CHROMS) + 1)

FIG_WIDTH, FIG_HEIGHT = 8.0, 3.0
FONT_BASE, FONT_LABEL, FONT_TICK, FONT_ANNOTATION = 10, 11.5, 10, 8
AXIS_WIDTH, TICK_WIDTH, TICK_LENGTH = 1.0, 0.8, 3
CHM13_COLOR, GRCH38_COLOR, HPRC_COLOR = "#536078", "#547853", "#BDBDBD"
VIOLIN_WIDTH, VIOLIN_EDGE_WIDTH, VIOLIN_MEDIAN_WIDTH, VIOLIN_ALPHA = 0.78, 0.9, 1.2, 1.0
REF_OFFSET, GRCH38_MARKER_SIZE, CHM13_MARKER_SIZE, REF_EDGE_WIDTH = 0.2, 68, 66, 1

plt.rcParams.update({
    "font.family": "Nimbus Sans", "font.size": FONT_BASE, "axes.labelsize": FONT_LABEL,
    "xtick.labelsize": FONT_TICK, "ytick.labelsize": FONT_TICK, "axes.linewidth": AXIS_WIDTH,
    "axes.spines.top": False, "axes.spines.right": False, "pdf.fonttype": 42, "ps.fonttype": 42,
    "savefig.facecolor": "white",
})

ct = pd.read_csv(HAP_FILE, sep="\t")
ct = ct[ct["category"].isin(["primary_chr", "chr_random"])].copy()
ct["chr"] = ct["ucsc_name"].astype(str).str.extract(CHR_RE.pattern, expand=False)
ct = ct[ct["chr"].isin(CHROMS)].copy()
for col in ("genome_bp_raw", "genome_bp_effective_min1", "zdna_bp"):
    ct[col] = pd.to_numeric(ct[col], errors="raise").astype("int64")
ct = ct.drop_duplicates(["filename", "assembly_contig"])
primary = (ct[ct["category"].eq("primary_chr")]
           .sort_values("genome_bp_raw", ascending=False, kind="mergesort")
           .drop_duplicates(["filename", "chr"], keep="first"))
random = ct[ct["category"].eq("chr_random")].merge(primary[["filename", "chr"]], on=["filename", "chr"])
sel = pd.concat([primary, random], ignore_index=True)
df = (sel.groupby(["filename", "chr"], as_index=False)
      .agg(n_contigs=("assembly_contig", "size"), n_chr_random=("category", lambda s: int((s == "chr_random").sum())),
           zdna_bp=("zdna_bp", "sum"), acgt_bp_min1=("genome_bp_effective_min1", "sum")))
df["density"] = df["zdna_bp"] / df["acgt_bp_min1"] * 1e6
assert df["filename"].nunique() == 464, f"Expected 464 HPRC haplotypes, found {df['filename'].nunique()}"
assert np.isfinite(df["density"]).all(), "Non-finite HPRC chromosome densities"

ref = pd.read_csv(REF_FILE, sep="\t")
assert ref["ucsc_chr"].is_unique
ref = ref.set_index("ucsc_chr")
assert not set(CHROMS) - set(ref.index), "Missing reference chromosomes"
ref = ref.loc[CHROMS]
grch38 = pd.to_numeric(ref["GRCh38_density_bp_per_Mb"], errors="raise").to_numpy()
chm13 = pd.to_numeric(ref["CHM13_density_bp_per_Mb"], errors="raise").to_numpy()
assert not np.isinf(grch38).any() and not np.isinf(chm13).any()

groups = [df.loc[df["chr"] == c, "density"].to_numpy() for c in CHROMS]
assert all(len(g) for g in groups), "A chromosome has no HPRC observations"

pd.concat([
    df.assign(dataset_type="HPRC haplotype").rename(columns={"filename": "dataset", "density": "zdna_density_bp_per_Mb"}),
    pd.DataFrame({"dataset": "GRCh38", "chr": CHROMS, "dataset_type": "reference genome",
                  "zdna_density_bp_per_Mb": grch38}),
    pd.DataFrame({"dataset": "CHM13", "chr": CHROMS, "dataset_type": "reference genome",
                  "zdna_density_bp_per_Mb": chm13}),
], ignore_index=True)[["dataset", "dataset_type", "chr", "n_contigs", "n_chr_random", "zdna_bp", "acgt_bp_min1",
                       "zdna_density_bp_per_Mb"]].assign(chr_order=lambda d: d["chr"].map(CHROMS.index)).sort_values(
    ["dataset_type", "dataset", "chr_order"]).drop(columns="chr_order").rename(columns=READABLE).to_csv(
    PLOT_DATA, sep="\t", index=False, float_format="%.10g")
summary = (df.groupby("chr")["density"].agg(n="size", mean="mean", median="median", std="std", min="min", max="max")
           .reindex(CHROMS))
summary["GRCh38"], summary["CHM13"] = grch38, chm13
summary["n_with_chr_random"] = df.groupby("chr")["n_chr_random"].apply(lambda s: int((s > 0).sum())).reindex(CHROMS)

fig, ax = plt.subplots(figsize=(FIG_WIDTH, FIG_HEIGHT), layout="constrained")
vp = ax.violinplot(groups, positions=POSITIONS, widths=VIOLIN_WIDTH, showmeans=False, showmedians=True,
                   showextrema=False)
for body in vp["bodies"]:
    body.set_facecolor(HPRC_COLOR); body.set_edgecolor("black")
    body.set_linewidth(VIOLIN_EDGE_WIDTH); body.set_alpha(VIOLIN_ALPHA); body.set_zorder(2)
vp["cmedians"].set_color("black"); vp["cmedians"].set_linewidth(VIOLIN_MEDIAN_WIDTH); vp["cmedians"].set_zorder(3)
ax.scatter(POSITIONS - REF_OFFSET, grch38, marker="^", s=GRCH38_MARKER_SIZE, facecolor=GRCH38_COLOR,
           edgecolor="white", linewidth=REF_EDGE_WIDTH, zorder=5)
ax.scatter(POSITIONS + REF_OFFSET, chm13, marker="o", s=CHM13_MARKER_SIZE, facecolor=CHM13_COLOR,
           edgecolor="white", linewidth=REF_EDGE_WIDTH, zorder=6)
ax.set_xticks(POSITIONS)
ax.set_xticklabels([str(i) for i in range(1, 23)] + ["X", "Y"], rotation=0, ha="center")
ax.set_xlim(0.35, 24.65)
ax.set_xlabel("Chromosome", fontsize=FONT_LABEL, labelpad=5)
ax.set_ylabel("Z-DNA density (bp/Mb)", fontsize=FONT_LABEL, labelpad=5)
ax.yaxis.set_major_locator(MaxNLocator(nbins=8))
ymin, ymax = ax.get_ylim()
padding = (ymax - ymin) * 0.05
ax.set_ylim(ymin, ymax + padding * 2)

for p, values in zip(POSITIONS, groups):
    ax.text(p, ymax + padding * 0.5, f"n={len(values)}",
            ha="center", va="bottom", rotation=45, fontsize=FONT_ANNOTATION,
            color="black", zorder=7)
ax.tick_params(axis="both", which="major", labelsize=FONT_TICK, width=TICK_WIDTH, length=TICK_LENGTH,
               direction="out", pad=3)
ax.grid(True, axis="both", which="major", linestyle="--", linewidth=0.4, color="#B0B0B0", alpha=0.5, zorder=0)
ax.set_axisbelow(True)
fig.get_layout_engine().set(w_pad=0.03, h_pad=0.03, wspace=0.02, hspace=0.02)
fig.savefig(OUT_FILE, dpi=600, bbox_inches=None)
fig.savefig(OUT_FILE.with_suffix(".pdf"), bbox_inches=None)
plt.show()
plt.close(fig)

legend_handles = [
    Patch(facecolor=HPRC_COLOR, edgecolor="black", linewidth=0.9, label="HPRC"),
    Line2D([0], [0], marker="^", linestyle="None", color="black", markerfacecolor=GRCH38_COLOR,
           markeredgecolor="black", markeredgewidth=REF_EDGE_WIDTH, markersize=6, label="GRCh38"),
    Line2D([0], [0], marker="o", linestyle="None", color="black", markerfacecolor=CHM13_COLOR,
           markeredgecolor="black", markeredgewidth=REF_EDGE_WIDTH, markersize=6, label="CHM13"),
]
legend_fig, legend_ax = plt.subplots(figsize=(2.0, 1.1))
legend_ax.axis("off")
legend_ax.legend(handles=legend_handles, loc="center", frameon=False, fontsize=FONT_BASE, ncol=1,
                 labelspacing=0.45, handlelength=1.5, handletextpad=0.6)
legend_fig.savefig(LEGEND_FILE, dpi=600, bbox_inches="tight", pad_inches=0.05, transparent=True)
legend_fig.savefig(LEGEND_FILE.with_suffix(".pdf"), bbox_inches="tight", pad_inches=0.05, transparent=True)
plt.close(legend_fig)

print("Density = (longest primary contig + assigned chr_random) Z-DNA bp / ACGT bp (min1)")
print(f"HPRC haplotypes: {df['filename'].nunique()}; haplotype-chromosome pairs: {len(df):,} "
      f"(with chr_random: {(df['n_chr_random'] > 0).sum():,})")
```

#### Suppl. Fig. 1b

```python
"""Suppl. Fig. 1b - chromosome GC% vs Z-DNA density: HPRC haplotype-chromosomes (grey), GRCh38 and CHM13 chromosomes
with per-reference least-squares fits; Spearman correlations + legend.
"""
from __future__ import annotations

import re
from pathlib import Path

import matplotlib.pyplot as plt
from matplotlib.lines import Line2D
import numpy as np
import pandas as pd
from scipy.stats import linregress, spearmanr

BASE = Path('/path/to/pangenomes_zdna')
PLOT_DIR = BASE / 'plots'
TEST_DIR = BASE / 'tests'
STEM = 'suppl_fig1b_gc_vs_zdna_density'
READABLE = {
    "dataset": "Assembly", "filename": "Assembly", "source_filename": "Assembly", "assembly": "Reference genome",
    "dataset_type": "Assembly type", "sample_id": "Sample", "superpopulation": "Superpopulation",
    "pat_or_mat": "Parental origin", "chr": "Chromosome", "ucsc_chr": "Chromosome", "n_contigs": "Contigs (n)",
    "n_chr_random": "chr_random contigs (n)", "zdna_bp": "Z-DNA (bp)", "acgt_bp_min1": "ACGT (bp)", "genome_bp": "ACGT (bp)",
    "zdna_density_bp_per_Mb": "Z-DNA density (bp/Mb)", "density_bp_per_Mb": "Z-DNA density (bp/Mb)",
    "density_bp_per_Mb_min1": "Z-DNA density (bp/Mb)", "gc_pct": "GC (%)", "rank": "Rank",
    "n_unique_zdna_sequences": "Distinct Z-DNA sequences (n)",
    "n_private_zdna_sequences": "Assembly-specific Z-DNA sequences (n)", "fraction_private": "Fraction assembly-specific",
    "length": "Z-DNA length (bp)", "score": "ZSeeker score", "metric_value": "Value", "transformed_coordinate": "Plot coordinate",
    "kde_percent_per_visual_bin": "KDE (% per visual bin)", "n_sampled_predictions": "Sampled predictions (n)",
    "n_in_display_range": "Predictions in display range (n)", "percent_in_display_range": "Predictions in display range (%)",
    "scott_bandwidth_multiplier": "Scott bandwidth multiplier",
    "visual_bin_width_custom_axis_units": "Visual bin width (axis units)", "n_autosomes": "Autosomes (n)",
    "paternal_bp_per_Mb": "Paternal density (bp/Mb)", "maternal_bp_per_Mb": "Maternal density (bp/Mb)",
    "delta_pat_minus_mat": "Paternal - maternal (bp/Mb)", "absolute_delta": "Absolute difference (bp/Mb)",
    "n_chromosomes": "Chromosomes (n)", "spearman_rho": "Spearman rho", "spearman_p": "Spearman P value",
    "ols_slope_bp_per_Mb_per_pct_GC": "OLS slope (bp/Mb per % GC)", "ols_intercept_bp_per_Mb": "OLS intercept (bp/Mb)",
    "ols_r_squared": "OLS R squared", "ols_p": "OLS P value", "test": "Test", "n_pairs": "Pairs (n)",
    "density_denominator": "Density denominator", "contig_rule": "Contig rule", "n_pat_only": "Paternal only (n)",
    "n_mat_only": "Maternal only (n)", "median_autosomes_per_pair": "Median autosomes per pair",
    "min_autosomes_per_pair": "Minimum autosomes per pair",
    "median_delta_pat_minus_mat_bp_per_Mb": "Median paternal - maternal (bp/Mb)",
    "mean_delta_bp_per_Mb": "Mean paternal - maternal (bp/Mb)", "sd_delta_bp_per_Mb": "SD paternal - maternal (bp/Mb)",
    "mean_absolute_delta_bp_per_Mb": "Mean absolute difference (bp/Mb)",
    "mean_absolute_delta_pct_grand_median": "Mean absolute difference (% of grand median)",
    "wilcoxon_W": "Wilcoxon W", "p_value": "P value",
}
CHROM_TABLE = BASE / '3_hprc_chromosome_tables.final.tsv'
CONTIG_GC = BASE / '5_hprc_chromosome_gc_min1.tsv'
REF_GC = BASE / '5_reference_chromosome_gc.tsv'
REF_DENSITY = {
    'CHM13': BASE / '2_CHM13_zdna_density_per_chr.tsv',
    'GRCh38': BASE / '2_GRCh38_zdna_density_per_chr.tsv',
}

FILTER_HPRC_WITHOUT_PRIMARY = True
CHROMS = [f'chr{i}' for i in range(1, 23)] + ['chrX', 'chrY']
KEEP_CATEGORIES = {'primary_chr', 'chr_random'}
CHR_RE = re.compile(r'^(chr(?:[1-9]|1[0-9]|2[0-2]|X|Y))(?=_|$)')

FIG_WIDTH, FIG_HEIGHT = 3.9, 2
DPI = 600
FONT_BASE, FONT_LABEL, FONT_TICK = 10, 11.5, 10
AXIS_WIDTH, TICK_WIDTH, TICK_LENGTH = 1.0, 0.8, 3
HPRC_COLOR = '#BDBDBD'
REF_COLORS = {'GRCh38': '#547853', 'CHM13': '#536078'}
HPRC_SIZE, REF_SIZE = 9, 24
HPRC_ALPHA, REF_ALPHA = 0.5, 0.9
HPRC_EDGE_LW, REF_EDGE_LW = 0.25, 0.4
POINT_EDGE_COLOR = 'black'
SHOW_REFERENCE_FITS = True
FIT_LINEWIDTH, FIT_LINESTYLE = 1.2, '--'
ANNOTATE_REFERENCE_CHROMS = {'chr13', 'chr21'}
ANNOTATION_SIZE = 10
X_LIMITS = None
Y_LIMITS = None
LEGEND_FIGSIZE = (2.3, 1.1)

plt.rcParams.update({
    'font.family': 'Nimbus Sans',
    'font.size': FONT_BASE,
    'axes.labelsize': FONT_LABEL,
    'xtick.labelsize': FONT_TICK,
    'ytick.labelsize': FONT_TICK,
    'axes.linewidth': AXIS_WIDTH,
    'axes.spines.top': False,
    'axes.spines.right': False,
    'pdf.fonttype': 42,
    'ps.fonttype': 42,
    'savefig.facecolor': 'white',
})


def canonical_chrom(value):
    match = CHR_RE.match(str(value))
    return match.group(1) if match else None


def load_hprc():
    for path in (CHROM_TABLE, CONTIG_GC):
        if not path.is_file():
            raise FileNotFoundError(f'Missing input: {path}')
    ct = pd.read_csv(CHROM_TABLE, sep='\t')
    required = {'filename', 'assembly_contig', 'ucsc_name', 'category',
                'genome_bp_raw', 'zdna_bp'}
    if required - set(ct):
        raise ValueError(f'Missing chromosome columns: {sorted(required - set(ct))}')
    ct = ct.loc[ct['category'].isin(KEEP_CATEGORIES)].copy()
    ct['chr'] = ct['ucsc_name'].map(canonical_chrom)
    ct = ct.loc[ct['chr'].isin(CHROMS)].copy()
    for col in ('genome_bp_raw', 'zdna_bp'):
        ct[col] = pd.to_numeric(ct[col], errors='raise').astype('int64')

    key = ['filename', 'assembly_contig']
    if (ct.groupby(key)[['genome_bp_raw', 'zdna_bp']].nunique() > 1).any(axis=None):
        raise RuntimeError('Conflicting duplicate filename/contig records')
    ct = ct.drop_duplicates(key).copy()

    primary = (ct.loc[ct['category'].eq('primary_chr')]
               .sort_values('genome_bp_raw', ascending=False, kind='mergesort')
               .drop_duplicates(['filename', 'chr'], keep='first'))
    random = ct.loc[ct['category'].eq('chr_random')]
    if FILTER_HPRC_WITHOUT_PRIMARY:
        pairs = primary[['filename', 'chr']]
        random = random.merge(pairs, on=['filename', 'chr'], how='inner')
        n_random_pairs = random[['filename', 'chr']].drop_duplicates().shape[0]
        print(f'Primary-chromosome QC: {len(pairs):,} primary pairs; '
              f'chr_random contigs retained for {n_random_pairs:,} pairs')
    ct = pd.concat([primary, random], ignore_index=True)

    gc = pd.read_csv(CONTIG_GC, sep='\t')
    if gc.duplicated(key).any():
        raise RuntimeError('Duplicated per-contig GC entries')
    merged = ct.merge(gc, on=key, how='left', validate='one_to_one', indicator=True)
    if not merged['_merge'].eq('both').all():
        raise RuntimeError(f'{merged["_merge"].ne("both").sum()} contigs missing GC data')

    perchr = merged.groupby(['filename', 'chr'], as_index=False).agg(
        n_contigs=('assembly_contig', 'size'),
        acgt_bp_min1=('acgt_bp_min1', 'sum'),
        gc_bp_min1=('gc_bp_min1', 'sum'),
        zdna_bp=('zdna_bp', 'sum'),
    )
    perchr = perchr.loc[perchr['acgt_bp_min1'] > 0].copy()
    perchr['gc_fraction_min1'] = perchr['gc_bp_min1'] / perchr['acgt_bp_min1']
    perchr['gc_pct'] = 100 * perchr['gc_fraction_min1']
    perchr['density_bp_per_Mb_min1'] = 1e6 * perchr['zdna_bp'] / perchr['acgt_bp_min1']
    perchr['assembly'] = 'HPRC'
    return perchr.rename(columns={'chr': 'ucsc_chr'})


def load_reference():
    gc = pd.read_csv(REF_GC, sep='\t')
    rows = []
    for assembly, path in REF_DENSITY.items():
        g = gc.loc[gc['assembly'].eq(assembly) & gc['ucsc_chr'].isin(CHROMS),
                   ['ucsc_chr', 'gc_fraction_min1']].copy()
        d = pd.read_csv(path, sep='\t')
        d = d.loc[d['ucsc_chr'].isin(CHROMS),
                  ['ucsc_chr', 'density_bp_per_Mb_min1']].copy()
        if g['ucsc_chr'].duplicated().any() or d['ucsc_chr'].duplicated().any():
            raise RuntimeError(f'{assembly}: duplicated chromosome')
        out = g.merge(d, on='ucsc_chr', how='inner', validate='one_to_one')
        if len(out) != 24 or set(out['ucsc_chr']) != set(CHROMS):
            raise RuntimeError(f'{assembly}: expected 24 chromosomes, found {len(out)}')
        out['gc_pct'] = pd.to_numeric(out['gc_fraction_min1'], errors='raise') * 100
        out['density_bp_per_Mb_min1'] = pd.to_numeric(
            out['density_bp_per_Mb_min1'], errors='raise')
        out['assembly'] = assembly
        out['filename'] = assembly
        out['n_contigs'] = 1
        rows.append(out)
    return pd.concat(rows, ignore_index=True)


def save_separate_legend():
    """Export a standalone legend without altering the main plot canvas."""
    handles = [
        Line2D([0], [0], marker='o', linestyle='None',
               markerfacecolor=HPRC_COLOR, markeredgecolor=POINT_EDGE_COLOR,
               markeredgewidth=0.5,
               markersize=5, alpha=HPRC_ALPHA, label='HPRC haplotypes'),
    ]
    for name in ('GRCh38', 'CHM13'):
        handles.append(Line2D(
            [0], [0], marker='o',
            linestyle=FIT_LINESTYLE if SHOW_REFERENCE_FITS else 'None',
            color=REF_COLORS[name], linewidth=FIT_LINEWIDTH,
            markerfacecolor=REF_COLORS[name], markeredgecolor=POINT_EDGE_COLOR,
            markeredgewidth=0.5,
            markersize=5.5, label=name,
        ))

    legend_fig, legend_ax = plt.subplots(figsize=LEGEND_FIGSIZE)
    legend_ax.axis('off')
    legend_ax.legend(
        handles=handles, loc='center', ncol=1, frameon=False,
        fontsize=FONT_BASE, labelspacing=0.45,
        handlelength=1.7, handletextpad=0.6,
    )
    legend_png = PLOT_DIR / f'{STEM}_legend.png'
    legend_pdf = PLOT_DIR / f'{STEM}_legend.pdf'
    legend_fig.savefig(legend_png, dpi=DPI, bbox_inches='tight',
                       pad_inches=0.05, transparent=True)
    legend_fig.savefig(legend_pdf, bbox_inches='tight',
                       pad_inches=0.05, transparent=True)
    plt.close(legend_fig)
    return legend_png, legend_pdf


def main():
    PLOT_DIR.mkdir(parents=True, exist_ok=True)
    TEST_DIR.mkdir(parents=True, exist_ok=True)
    hprc = load_hprc().sort_values(['filename', 'ucsc_chr'])
    references = load_reference().sort_values(['assembly', 'ucsc_chr'])
    df = pd.concat([hprc, references], ignore_index=True)
    if not np.isfinite(df[['gc_pct', 'density_bp_per_Mb_min1']]
                       .to_numpy(dtype=float)).all():
        raise ValueError('Non-finite GC or density values')
    df.rename(columns=READABLE).to_csv(TEST_DIR / f'{STEM}.tsv', sep='\t', index=False,
              float_format='%.10g')

    fig, ax = plt.subplots(figsize=(FIG_WIDTH, FIG_HEIGHT), layout='constrained')
    ax.scatter(
        hprc['gc_pct'], hprc['density_bp_per_Mb_min1'],
        s=HPRC_SIZE,
        color=HPRC_COLOR,
        alpha=HPRC_ALPHA,
        edgecolors=POINT_EDGE_COLOR,
        linewidths=HPRC_EDGE_LW,
        rasterized=True,
        zorder=2
    )
    for name in ('GRCh38', 'CHM13'):
        sub = references.loc[references['assembly'].eq(name)]
        ax.scatter(sub['gc_pct'], sub['density_bp_per_Mb_min1'],
                   s=REF_SIZE, color=REF_COLORS[name], alpha=REF_ALPHA,
                   edgecolors='none', zorder=4)
        ax.scatter(
        sub['gc_pct'], sub['density_bp_per_Mb_min1'],
        s=REF_SIZE,
        color=REF_COLORS[name],
        alpha=REF_ALPHA,
        edgecolors=POINT_EDGE_COLOR,
        linewidths=REF_EDGE_LW,
        zorder=4
    )

    statistics = []
    for name in ('CHM13', 'GRCh38'):
        sub = references.loc[references['assembly'].eq(name)]
        x = sub['gc_pct'].to_numpy(dtype=float)
        y = sub['density_bp_per_Mb_min1'].to_numpy(dtype=float)
        rho, p = spearmanr(x, y)
        fit = linregress(x, y)
        statistics.append({
            'assembly': name, 'n_chromosomes': len(sub),
            'spearman_rho': rho, 'spearman_p': p,
            'ols_slope_bp_per_Mb_per_pct_GC': fit.slope,
            'ols_intercept_bp_per_Mb': fit.intercept,
            'ols_r_squared': fit.rvalue ** 2, 'ols_p': fit.pvalue,
        })
        if SHOW_REFERENCE_FITS:
            xx = np.linspace(x.min(), x.max(), 150)
            ax.plot(xx, fit.intercept + fit.slope * xx,
                    color=REF_COLORS[name], linewidth=FIT_LINEWIDTH,
                    linestyle=FIT_LINESTYLE, zorder=3)
        for row in sub.itertuples(index=False):
            if row.ucsc_chr in ANNOTATE_REFERENCE_CHROMS:
                ax.annotate(row.ucsc_chr.replace('chr', ''),
                            (row.gc_pct, row.density_bp_per_Mb_min1),
                            xytext=(3, 3 if name == 'CHM13' else -12),
                            textcoords='offset points', fontsize=ANNOTATION_SIZE,
                            color=REF_COLORS[name], zorder=5)
        print(f'{name}: Spearman rho={rho:.4g}, p={p:.4g}; R²={fit.rvalue**2:.4g}')

    ax.set_xlabel('GC content (%)', fontsize=FONT_LABEL, labelpad=5)
    ax.set_ylabel('Z-DNA density (bp/Mb)', fontsize=FONT_LABEL, labelpad=5)
    ax.tick_params(axis='both', which='major', labelsize=FONT_TICK,
                   width=TICK_WIDTH, length=TICK_LENGTH, direction='out', pad=3)
    ax.minorticks_off()
    if X_LIMITS is not None:
        ax.set_xlim(*X_LIMITS)
    if Y_LIMITS is not None:
        ax.set_ylim(*Y_LIMITS)
    ax.grid(True, axis='both', which='major', linestyle='--',
            linewidth=0.4, color='#B0B0B0', alpha=0.5, zorder=0)
    ax.set_axisbelow(True)
    fig.get_layout_engine().set(w_pad=0.03, h_pad=0.03, wspace=0.02, hspace=0.02)

    png = PLOT_DIR / f'{STEM}.png'
    pdf = PLOT_DIR / f'{STEM}.pdf'
    fig.savefig(png, dpi=DPI, bbox_inches=None)
    fig.savefig(pdf, dpi=DPI, bbox_inches=None)
    plt.show()
    plt.close(fig)

    legend_png, legend_pdf = save_separate_legend()

    pd.DataFrame(statistics).rename(columns=READABLE).to_csv(
        TEST_DIR / f'{STEM}_statistics.tsv', sep='\t', index=False,
        float_format='%.10g')
    print(f'HPRC observations: {len(hprc):,}; haplotypes: {hprc["filename"].nunique()}')
    print(f'Reference observations: {len(references)}')
    print(f'Saved:\n  {png}\n  {pdf}')
    print(f'Legend:\n  {legend_png}\n  {legend_pdf}')


if __name__ == '__main__':
    main()
```

#### Suppl. Fig. 1c

```python
"""Suppl. Fig. 1c - paternal vs maternal autosomal Z-DNA density for trio-phased individuals (autosomes assembled in both
haplotypes); two-sided paired Wilcoxon signed-rank test.
"""
import re
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from matplotlib.ticker import MaxNLocator
from scipy.stats import wilcoxon

BASE = Path('/path/to/pangenomes_zdna')
PLOT_DIR, TEST_DIR = BASE / 'plots', BASE / 'tests'
CHROM_TABLE = BASE / '3_hprc_chromosome_tables.final.tsv'
META_PATH = BASE / '1_hprc_metadata.tsv'
STEM = 'suppl_fig1c_paternal_vs_maternal_density'
READABLE = {
    "dataset": "Assembly", "filename": "Assembly", "source_filename": "Assembly", "assembly": "Reference genome",
    "dataset_type": "Assembly type", "sample_id": "Sample", "superpopulation": "Superpopulation",
    "pat_or_mat": "Parental origin", "chr": "Chromosome", "ucsc_chr": "Chromosome", "n_contigs": "Contigs (n)",
    "n_chr_random": "chr_random contigs (n)", "zdna_bp": "Z-DNA (bp)", "acgt_bp_min1": "ACGT (bp)", "genome_bp": "ACGT (bp)",
    "zdna_density_bp_per_Mb": "Z-DNA density (bp/Mb)", "density_bp_per_Mb": "Z-DNA density (bp/Mb)",
    "density_bp_per_Mb_min1": "Z-DNA density (bp/Mb)", "gc_pct": "GC (%)", "rank": "Rank",
    "n_unique_zdna_sequences": "Distinct Z-DNA sequences (n)",
    "n_private_zdna_sequences": "Assembly-specific Z-DNA sequences (n)", "fraction_private": "Fraction assembly-specific",
    "length": "Z-DNA length (bp)", "score": "ZSeeker score", "metric_value": "Value", "transformed_coordinate": "Plot coordinate",
    "kde_percent_per_visual_bin": "KDE (% per visual bin)", "n_sampled_predictions": "Sampled predictions (n)",
    "n_in_display_range": "Predictions in display range (n)", "percent_in_display_range": "Predictions in display range (%)",
    "scott_bandwidth_multiplier": "Scott bandwidth multiplier",
    "visual_bin_width_custom_axis_units": "Visual bin width (axis units)", "n_autosomes": "Autosomes (n)",
    "paternal_bp_per_Mb": "Paternal density (bp/Mb)", "maternal_bp_per_Mb": "Maternal density (bp/Mb)",
    "delta_pat_minus_mat": "Paternal - maternal (bp/Mb)", "absolute_delta": "Absolute difference (bp/Mb)",
    "n_chromosomes": "Chromosomes (n)", "spearman_rho": "Spearman rho", "spearman_p": "Spearman P value",
    "ols_slope_bp_per_Mb_per_pct_GC": "OLS slope (bp/Mb per % GC)", "ols_intercept_bp_per_Mb": "OLS intercept (bp/Mb)",
    "ols_r_squared": "OLS R squared", "ols_p": "OLS P value", "test": "Test", "n_pairs": "Pairs (n)",
    "density_denominator": "Density denominator", "contig_rule": "Contig rule", "n_pat_only": "Paternal only (n)",
    "n_mat_only": "Maternal only (n)", "median_autosomes_per_pair": "Median autosomes per pair",
    "min_autosomes_per_pair": "Minimum autosomes per pair",
    "median_delta_pat_minus_mat_bp_per_Mb": "Median paternal - maternal (bp/Mb)",
    "mean_delta_bp_per_Mb": "Mean paternal - maternal (bp/Mb)", "sd_delta_bp_per_Mb": "SD paternal - maternal (bp/Mb)",
    "mean_absolute_delta_bp_per_Mb": "Mean absolute difference (bp/Mb)",
    "mean_absolute_delta_pct_grand_median": "Mean absolute difference (% of grand median)",
    "wilcoxon_W": "Wilcoxon W", "p_value": "P value",
}
AUTOSOMES = [f'chr{i}' for i in range(1, 23)]
CHR_RE = re.compile(r'^(chr(?:[1-9]|1[0-9]|2[0-2]|X|Y))(?=_|$)')          # same rule as Fig. 1f / Suppl. Fig. 1b

FIG_WIDTH, FIG_HEIGHT = 3.3, 2.4
STATS_FIGSIZE = (2.1, 1.05)
DPI = 600
FONT_BASE, FONT_LABEL, FONT_TICK = 10, 11.5, 10
AXIS_WIDTH, TICK_WIDTH, TICK_LENGTH = 1.0, 0.8, 3
HPRC_COLOR = '#BDBDBD'
POINT_SIZE, POINT_ALPHA = 18, 0.8
IDENTITY_LINEWIDTH = 1.2

plt.rcParams.update({
    'font.family': 'Nimbus Sans', 'font.size': FONT_BASE, 'axes.labelsize': FONT_LABEL,
    'xtick.labelsize': FONT_TICK, 'ytick.labelsize': FONT_TICK, 'axes.linewidth': AXIS_WIDTH,
    'axes.spines.top': False, 'axes.spines.right': False, 'pdf.fonttype': 42, 'ps.fonttype': 42,
    'savefig.facecolor': 'white',
})


def autosome_table():
    """one row per haplotype x autosome (longest primary + assigned chr_random)."""
    ct = pd.read_csv(CHROM_TABLE, sep='\t', dtype={'filename': str, 'assembly_contig': str,
                                                    'ucsc_name': str, 'category': str})
    ct = ct[ct['category'].isin(['primary_chr', 'chr_random'])].copy()
    for col in ('genome_bp_raw', 'genome_bp_effective_min1', 'zdna_bp'):
        ct[col] = pd.to_numeric(ct[col], errors='raise').astype('int64')
    ct = ct.drop_duplicates(['filename', 'assembly_contig'])
    ct['chr'] = ct['ucsc_name'].fillna('').str.extract(CHR_RE.pattern, expand=False)
    ct = ct[ct['chr'].isin(AUTOSOMES)]
    primary = (ct[ct['category'].eq('primary_chr')]
               .sort_values('genome_bp_raw', ascending=False, kind='mergesort')
               .drop_duplicates(['filename', 'chr'], keep='first'))
    random = ct[ct['category'].eq('chr_random')].merge(primary[['filename', 'chr']], on=['filename', 'chr'])
    sel = pd.concat([primary, random], ignore_index=True)
    return (sel.groupby(['filename', 'chr'], as_index=False)
            .agg(zdna_bp=('zdna_bp', 'sum'), genome_bp=('genome_bp_effective_min1', 'sum')))


def main():
    PLOT_DIR.mkdir(parents=True, exist_ok=True)
    TEST_DIR.mkdir(parents=True, exist_ok=True)
    meta = pd.read_csv(META_PATH, sep='\t', dtype=str)
    if not meta['filename'].is_unique:
        raise ValueError('Duplicate filenames in metadata')
    trio = meta[meta['pat_or_mat'].isin(['pat', 'mat']) & meta['sample_id'].fillna('').ne('')]
    if trio.duplicated(['sample_id', 'pat_or_mat']).any():
        raise ValueError('Multiple haplotypes per sample_id and parental origin')
    origin = trio.groupby('sample_id')['pat_or_mat'].agg(set)
    pat_only = int(origin.apply(lambda s: s == {'pat'}).sum())
    mat_only = int(origin.apply(lambda s: s == {'mat'}).sum())

    per_chr = autosome_table().merge(trio[['filename', 'sample_id', 'pat_or_mat']], on='filename', how='inner')
    both = per_chr.groupby(['sample_id', 'chr'])['pat_or_mat'].transform('nunique').eq(2)
    matched = per_chr[both]
    per_hap = (matched.groupby(['filename', 'sample_id', 'pat_or_mat'], as_index=False)
               .agg(n_autosomes=('chr', 'nunique'), zdna_bp=('zdna_bp', 'sum'), genome_bp=('genome_bp', 'sum')))
    per_hap['density_bp_per_Mb'] = per_hap['zdna_bp'] * 1e6 / per_hap['genome_bp']
    per_hap.rename(columns=READABLE).to_csv(TEST_DIR / f'{STEM}_per_haplotype.tsv', sep='\t', index=False, float_format='%.10g')

    wide = per_hap.pivot(index='sample_id', columns='pat_or_mat', values='density_bp_per_Mb')
    n_auto = per_hap.pivot(index='sample_id', columns='pat_or_mat', values='n_autosomes')
    paired = wide.dropna(subset=['pat', 'mat']).copy().sort_index()
    if paired.empty:
        raise ValueError('No paternal/maternal pairs found')
    paired['n_autosomes'] = n_auto.loc[paired.index, 'pat'].astype(int)
    paired['delta_pat_minus_mat'] = paired['pat'] - paired['mat']
    paired['absolute_delta'] = paired['delta_pat_minus_mat'].abs()
    for k, lab in (('zdna_bp', 'Z-DNA (bp)'), ('genome_bp', 'ACGT (bp)')):
        w = per_hap.pivot(index='sample_id', columns='pat_or_mat', values=k)
        paired[f'Paternal {lab}'], paired[f'Maternal {lab}'] = w.loc[paired.index, 'pat'], w.loc[paired.index, 'mat']
    paired.reset_index().rename(columns={'pat': 'paternal_bp_per_Mb', 'mat': 'maternal_bp_per_Mb'}).rename(
        columns=READABLE).to_csv(
        TEST_DIR / f'{STEM}.tsv', sep='\t', index=False, float_format='%.10g')

    pat, mat = paired['pat'].to_numpy(float), paired['mat'].to_numpy(float)
    delta = pat - mat
    if np.all(delta == 0):
        w_stat, p_val = 0.0, 1.0
    else:
        test = wilcoxon(delta, alternative='two-sided', zero_method='wilcox')
        w_stat, p_val = float(test.statistic), float(test.pvalue)
    median_delta, mean_abs = float(np.median(delta)), float(np.mean(np.abs(delta)))
    grand_median = (np.median(pat) + np.median(mat)) / 2
    summary = pd.DataFrame([{
        'test': 'Wilcoxon signed-rank, two-sided', 'density_denominator': 'genome_bp_effective_min1',
        'contig_rule': 'longest primary + assigned chr_random; autosomes present in both haplotypes',
        'n_pairs': len(paired), 'n_pat_only': pat_only, 'n_mat_only': mat_only,
        'median_autosomes_per_pair': float(paired['n_autosomes'].median()),
        'min_autosomes_per_pair': int(paired['n_autosomes'].min()),
        'median_delta_pat_minus_mat_bp_per_Mb': median_delta, 'mean_delta_bp_per_Mb': float(np.mean(delta)),
        'sd_delta_bp_per_Mb': float(np.std(delta, ddof=1)) if len(delta) > 1 else np.nan,
        'mean_absolute_delta_bp_per_Mb': mean_abs,
        'mean_absolute_delta_pct_grand_median': 100 * mean_abs / grand_median if grand_median else np.nan,
        'wilcoxon_W': w_stat, 'p_value': p_val,
    }])
    summary.rename(columns=READABLE).to_csv(TEST_DIR / f'{STEM}_statistics.tsv', sep='\t', index=False, float_format='%.10g')

    fig, ax = plt.subplots(figsize=(FIG_WIDTH, FIG_HEIGHT), layout='constrained')
    ax.scatter(mat, pat, s=POINT_SIZE, alpha=POINT_ALPHA, color=HPRC_COLOR, edgecolors='none', zorder=3,
               rasterized=True)
    lo, hi = float(min(mat.min(), pat.min())), float(max(mat.max(), pat.max()))
    pad = max((hi - lo) * 0.05, max(abs(hi), 1.0) * 0.005)
    lo, hi = lo - pad, hi + pad
    ax.plot([lo, hi], [lo, hi], color='black', linewidth=IDENTITY_LINEWIDTH, zorder=2)
    ax.set_xlim(lo, hi)
    ax.set_ylim(lo, hi)
    ax.set_xlabel('Maternal autosomal\nZ-DNA density (bp/Mb)', fontsize=FONT_LABEL, labelpad=5)
    ax.set_ylabel('Paternal autosomal\nZ-DNA density (bp/Mb)', fontsize=FONT_LABEL, labelpad=5)
    ax.xaxis.set_major_locator(MaxNLocator(nbins=6))
    ax.yaxis.set_major_locator(MaxNLocator(nbins=6))
    ax.tick_params(axis='both', which='major', labelsize=FONT_TICK, width=TICK_WIDTH, length=TICK_LENGTH,
                   direction='out', pad=3)
    ax.minorticks_off()
    ax.grid(True, axis='both', which='major', linestyle='--', linewidth=0.4, color='#B0B0B0', alpha=0.5, zorder=0)
    ax.set_axisbelow(True)
    fig.get_layout_engine().set(w_pad=0.03, h_pad=0.03, wspace=0.02, hspace=0.02)
    output = PLOT_DIR / f'{STEM}.png'
    fig.savefig(output, dpi=DPI, bbox_inches=None)
    fig.savefig(output.with_suffix('.pdf'), dpi=DPI, bbox_inches=None)
    plt.show()
    plt.close(fig)

    stats_fig, stats_ax = plt.subplots(figsize=STATS_FIGSIZE)
    stats_ax.axis('off')
    stats_ax.text(0.5, 0.5, f'n = {len(paired)} pairs\nMedian Δ = {median_delta:+.2f} bp/Mb\n'
                  f'Mean |Δ| = {mean_abs:.2f} bp/Mb\nWilcoxon p = {p_val:.3g}', ha='center', va='center',
                  fontsize=FONT_BASE, bbox=dict(facecolor='white', edgecolor='black', boxstyle='round,pad=0.35',
                                                linewidth=AXIS_WIDTH))
    stats_fig.savefig(PLOT_DIR / f'{STEM}_stats.png', dpi=DPI, bbox_inches='tight', pad_inches=0.05)
    stats_fig.savefig(PLOT_DIR / f'{STEM}_stats.pdf', bbox_inches='tight', pad_inches=0.05)
    plt.close(stats_fig)

    print('Autosomes per haplotype: longest primary + assigned chr_random; matched within each pair')
    print(summary.drop(columns=['test', 'density_denominator', 'contig_rule']).T.to_string(header=False))


if __name__ == '__main__':
    main()
```

## Results 2 – Z-DNA across T2T-CHM13 (Fig. 2, Suppl. Fig. 1d–f, Suppl. Fig. 2a, Table 2, Supplementary Table 2)

### Analysis

#### Protein-coding introns (input to 6.3)

```python
"""Protein-coding transcript introns (CHM13 RefSeq Liftoff v5.2) -> pc_transcript_introns.bed (input to 6.3).

Introns = gaps between consecutive exons of every transcript of protein_coding genes (BED3, isoform duplicates kept).
If the BED already exists it is validated (exact multiset match) instead of overwritten.
"""
import gzip
from collections import Counter, defaultdict
from pathlib import Path

BASE = Path("/path/to/pangenomes_zdna")
ANN = BASE / "tmp/chm13/chm13_marbl_ann_files/annotation"
GFF = ANN / "chm13v2.0_RefSeq_Liftoff_v5.2.gff3.gz"
OUT_BED = ANN / "pc_transcript_introns.bed"


def attrs(s):
    return dict(item.split("=", 1) for item in s.split(";") if "=" in item)


def main():
    if not GFF.is_file():
        raise FileNotFoundError(GFF)
    genes, transcripts, exons = set(), {}, defaultdict(list)
    with gzip.open(GFF, "rt") as fh:
        rows = [line.rstrip("\n").split("\t") for line in fh if line.strip() and not line.startswith("#")]
    rows = [x for x in rows if len(x) >= 9]
    for x in rows:
        if x[2] == "gene" and attrs(x[8]).get("gene_biotype") == "protein_coding":
            genes.add(attrs(x[8])["ID"])
    for x in rows:
        if x[2] == "transcript":
            a = attrs(x[8])
            if any(p in genes for p in a.get("Parent", "").split(",")):
                transcripts[a["ID"]] = (x[0], x[6])
    for x in rows:
        if x[2] == "exon":
            for p in attrs(x[8]).get("Parent", "").split(","):
                if p in transcripts:
                    exons[p].append((int(x[3]) - 1, int(x[4])))
    introns = []
    for tid, iv in exons.items():
        iv = sorted(set(iv))
        chrom = transcripts[tid][0]
        introns += [(chrom, e0, s1) for (_, e0), (s1, _) in zip(iv, iv[1:]) if s1 > e0]
    introns.sort()

    if OUT_BED.is_file():
        ref = Counter()
        with OUT_BED.open() as fh:
            for line in fh:
                if line.strip() and not line.startswith("#"):
                    c, s, e = line.split("\t")[:3]
                    ref[(c, int(s), int(e))] += 1
        if ref != Counter(introns):
            raise AssertionError(f"Recomputed introns differ from {OUT_BED.name}")
        print(f"{OUT_BED.name} validated: {len(introns):,} introns (exact match)")
    else:
        with OUT_BED.open("w") as out:
            out.writelines(f"{c}\t{s}\t{e}\n" for c, s, e in introns)
        print(f"Wrote {OUT_BED.name}: {len(introns):,} introns")
    print(f"Protein-coding genes {len(genes):,}; transcripts {len(transcripts):,}")


if __name__ == "__main__":
    main()
```

#### 6.1 · Density per cytoband, CenSat array and chrX/Y class → Fig. 2b, 2c, Suppl. Fig. 1d

```python
"""6.1 - Z-DNA density per cytoband, CenSat array and chrX/Y sequence-class segment (Fig. 2b, 2c, Suppl. Fig. 1d).

Density = merged Z-DNA bp in the interval / interval bp. Group comparisons: two-sided Mann-Whitney U, BH within family,
Cliff's delta. chrY decomposition by merged sequence class (text values).
"""
import bisect
from pathlib import Path
import numpy as np
import pandas as pd
from scipy.stats import mannwhitneyu
from statsmodels.stats.multitest import multipletests

BASE = Path("/path/to/pangenomes_zdna")
ANN_DIR = BASE / "tmp/chm13/chm13_marbl_ann_files"
ANN = ANN_DIR / "annotation"
CYTO_BED = ANN_DIR / "chm13v2.0_cytobands_allchrs.bed"
SEQ_BED = ANN_DIR / "chm13v2.0_chrXY_sequence_class_v1.bed"
CENSAT_BED = ANN / "chm13v2.0_censat_v2.1.bed"
ZDNA_BED = ANN / "chm13_zdna_all.bed"
CHROMSIZES = ANN / "chm13_chrom_sizes.txt"

OUT_CYTO = BASE / "6_chm13_zdna_by_cytoband.tsv"
OUT_CENSAT = BASE / "6_chm13_zdna_by_censat.tsv"
OUT_XY = BASE / "6_chm13_zdna_by_xy_sequence_class.tsv"
OUT_CYTO_TESTS = BASE / "6_chm13_cytoband_comparisons.tsv"
OUT_CENSAT_TESTS = BASE / "6_chm13_censat_comparisons.tsv"
OUT_CHRY = BASE / "6_chm13_chrY_decomposition.tsv"
OUT_CHRY_HET = BASE / "6_chm13_chrY_het_censat.tsv"

CHROM_ORDER = [f"chr{i}" for i in range(1, 23)] + ["chrX", "chrY"]
CENSAT_CATEGORIES = ["hor", "dhor", "mon", "hsat1A", "hsat1B", "hsat2", "hsat3", "bsat", "gsat", "censat", "ct", "rDNA"]
SEQ_COLLAPSE = {
    "PAR1": "PAR", "PAR2": "PAR", "X-DEG": "X-degenerate", "XTR": "X-transposed", "XTR1": "X-transposed",
    "XTR2": "X-transposed", "AMPL": "Ampliconic", "SAT": "Satellite", "DYZ17": "Satellite", "DYZ19": "Satellite",
    "CEN": "Centromere", "HET": "Heterochromatin", "OTHER": "Other",
}
SPLIT_BY_CHROM = ["PAR", "X-transposed"]                 # plotted separately for chrX and chrY
MSY_EUCHROMATIN = ["X-degenerate", "Ampliconic", "X-transposed", "Other"]
_CYTO_ALL = ["gneg", "gpos25", "gpos50", "gpos75", "gpos100", "acen", "gvar"]
_SAT = ["hor", "dhor", "hsat1A", "hsat1B", "hsat2", "hsat3", "bsat", "gsat"]
CYTO_COMPARISONS = [                                      # (group 1, categories, group 2, categories)
    ("euchromatin", ["gneg", "gpos25"], "heterochromatin", ["gpos75", "gpos100"]),
    ("stalk", ["stalk"], "other", _CYTO_ALL),
    ("acen", ["acen"], "non_acen", [c for c in _CYTO_ALL if c != "acen"]),
]
CENSAT_COMPARISONS = [
    ("rDNA", ["rDNA"], "non_rDNA", _SAT + ["censat", "ct", "mon"]),
    ("censat", ["censat"], "other_satellites", _SAT + ["ct", "mon"]),
    ("hsat1B", ["hsat1B"], "other_hsat", ["hsat1A", "hsat2", "hsat3"]),
]


def read_bed(path, cols, names):
    df = pd.read_csv(path, sep="\t", header=None, usecols=cols, names=names, comment="#")
    df = df[df.chrom.isin(CHROM_ORDER)].copy()
    df[["start", "end"]] = df[["start", "end"]].astype("int64")
    return df[df.end > df.start].reset_index(drop=True)


def merge(iv):
    out = []
    for s, e in sorted(iv):
        if out and s <= out[-1][1]:
            out[-1][1] = max(out[-1][1], e)
        else:
            out.append([s, e])
    return out


def merged_by_chrom(df):
    return {c: merge(zip(g.start, g.end)) for c, g in df.groupby("chrom")}


def overlap_bp(start, end, merged, ends):
    """bp of [start, end) covered by sorted, non-overlapping intervals (ends = their end coordinates)."""
    tot, i = 0, bisect.bisect_right(ends, start)
    while i < len(merged) and merged[i][0] < end:
        tot += min(end, merged[i][1]) - max(start, merged[i][0])
        i += 1
    return tot


def densities(ann, zm):
    ends = {c: [e for _, e in iv] for c, iv in zm.items()}
    z = [overlap_bp(s, e, zm.get(c, []), ends.get(c, [])) for c, s, e in zip(ann.chrom, ann.start, ann.end)]
    out = ann.copy()
    out["length_bp"] = out.end - out.start
    out["zdna_bp"] = z
    out["density_bp_per_kb"] = out.zdna_bp / out.length_bp * 1e3
    out["density_bp_per_Mb"] = out.zdna_bp / out.length_bp * 1e6
    return out


def cliffs_delta(x, y):
    x, y = np.asarray(x), np.sort(np.asarray(y))
    gt = np.searchsorted(y, x, side="left").sum()
    lt = (len(y) - np.searchsorted(y, x, side="right")).sum()
    return (gt - lt) / (len(x) * len(y))


def run_tests(df, comparisons):
    rows = []
    for g1, c1, g2, c2 in comparisons:
        x = df.loc[df.category.isin(c1), "density_bp_per_kb"].to_numpy()
        y = df.loc[df.category.isin(c2), "density_bp_per_kb"].to_numpy()
        u, p = mannwhitneyu(x, y, alternative="two-sided")
        rows.append({"group1": g1, "group2": g2, "n1": len(x), "n2": len(y), "median1_bp_per_kb": np.median(x),
                     "median2_bp_per_kb": np.median(y), "mean1_bp_per_kb": x.mean(), "mean2_bp_per_kb": y.mean(),
                     "cliffs_delta": cliffs_delta(x, y), "u_statistic": u, "p_value": p})
    out = pd.DataFrame(rows)
    out["p_adj_bh"] = multipletests(out.p_value, method="fdr_bh")[1]
    return out


def isect(a, b):
    i = j = tot = 0
    while i < len(a) and j < len(b):
        tot += max(0, min(a[i][1], b[j][1]) - max(a[i][0], b[j][0]))
        if a[i][1] < b[j][1]:
            i += 1
        else:
            j += 1
    return tot


def bp(m):
    return sum(e - s for s, e in m)


def main():
    for p in (CYTO_BED, SEQ_BED, CENSAT_BED, ZDNA_BED, CHROMSIZES):
        if not p.exists():
            raise FileNotFoundError(p)
    zdna = read_bed(ZDNA_BED, [0, 1, 2], ["chrom", "start", "end"])
    zm = merged_by_chrom(zdna)
    sizes = pd.read_csv(CHROMSIZES, sep="\t", header=None, usecols=[0, 1], names=["chrom", "length"])
    genome_density = sum(bp(v) for v in zm.values()) / sizes.length.sum() * 1e6

    cyto = read_bed(CYTO_BED, [0, 1, 2, 3, 4], ["chrom", "start", "end", "name", "category"])
    cen = read_bed(CENSAT_BED, [0, 1, 2, 3], ["chrom", "start", "end", "name"])
    cen["category"] = cen.name.astype(str).str.split("_").str[0]
    cen = cen[cen.category.isin(CENSAT_CATEGORIES)].reset_index(drop=True)
    xy = read_bed(SEQ_BED, [0, 1, 2, 3], ["chrom", "start", "end", "original_category"])
    xy = xy[xy.chrom.isin(["chrX", "chrY"])].reset_index(drop=True)
    xy["category"] = xy.original_category.map(SEQ_COLLAPSE)
    if xy.category.isna().any():
        raise RuntimeError(f"Unmapped chrX/Y classes: {sorted(xy.original_category[xy.category.isna()].unique())}")

    cyto_d, cen_d, xy_d = densities(cyto, zm), densities(cen, zm), densities(xy, zm)
    xy_d["plot_group"] = np.where(xy_d.category.isin(SPLIT_BY_CHROM), xy_d.category + " " + xy_d.chrom, xy_d.category)
    cyto_d.to_csv(OUT_CYTO, sep="\t", index=False)
    cen_d.to_csv(OUT_CENSAT, sep="\t", index=False)
    xy_d.to_csv(OUT_XY, sep="\t", index=False)
    cyto_t, cen_t = run_tests(cyto_d, CYTO_COMPARISONS), run_tests(cen_d, CENSAT_COMPARISONS)
    cyto_t.to_csv(OUT_CYTO_TESTS, sep="\t", index=False)
    cen_t.to_csv(OUT_CENSAT_TESTS, sep="\t", index=False)

    cls = {(c, k): merge(zip(g.start, g.end)) for (c, k), g in xy.groupby(["chrom", "category"])}
    rows = [{"chrom": c, "group": k, "n_segments": int(((xy.chrom == c) & (xy.category == k)).sum()),
             "bp": bp(m), "zdna_bp": isect(m, zm.get(c, []))} for (c, k), m in cls.items()]
    y_all = merge([iv for (c, _), m in cls.items() if c == "chrY" for iv in m])
    for name, keep in [("chrY, all classes", None), ("chrY without heterochromatin", "noHET"),
                       ("chrY male-specific euchromatin", "MSY")]:
        m = y_all if keep is None else merge([iv for (c, k), mm in cls.items() if c == "chrY" and
                                               (k != "Heterochromatin" if keep == "noHET" else k in MSY_EUCHROMATIN)
                                               for iv in mm])
        rows.append({"chrom": "chrY", "group": name, "n_segments": len(m), "bp": bp(m), "zdna_bp": isect(m, zm.get("chrY", []))})
    dec = pd.DataFrame(rows)
    dec["zdna_bp_per_Mb"] = dec.zdna_bp / dec.bp * 1e6
    dec["frac_of_chrY"] = (dec.bp / bp(y_all)).where(dec.chrom == "chrY")
    dec["ratio_to_genome"] = dec.zdna_bp_per_Mb / genome_density
    dec.to_csv(OUT_CHRY, sep="\t", index=False)
    het = cls[("chrY", "Heterochromatin")]
    cy = cen[cen.chrom == "chrY"]
    hrows = []
    for k, g in cy.groupby("category"):
        m = merge([(max(s, hs), min(e, he)) for s, e in merge(zip(g.start, g.end)) for hs, he in het if s < he and e > hs])
        if bp(m):
            hrows.append({"censat_category": k, "bp_in_het": bp(m), "frac_of_het": bp(m) / bp(het),
                          "zdna_bp": isect(m, zm.get("chrY", [])), "zdna_bp_per_Mb": isect(m, zm.get("chrY", [])) / bp(m) * 1e6})
    het_t = pd.DataFrame(hrows).sort_values("bp_in_het", ascending=False)
    het_t.to_csv(OUT_CHRY_HET, sep="\t", index=False)

    print(f"Z-DNA: {len(zdna):,} predictions; genome-wide density {genome_density:,.1f} bp/Mb")
    for title, t in [("Cytoband comparisons", cyto_t), ("CenSat comparisons", cen_t)]:
        print(f"{title}:\n" + t[["group1", "group2", "median1_bp_per_kb", "median2_bp_per_kb", "cliffs_delta", "p_adj_bh"]]
              .to_string(index=False, float_format=lambda v: f"{v:.3g}"))
    y = dec.set_index("group").zdna_bp_per_Mb
    print(f"chrY: all classes {y['chrY, all classes']:,.0f} bp/Mb; male-specific euchromatin "
          f"{y['chrY male-specific euchromatin']:,.0f} bp/Mb; Yq12 heterochromatin {isect(het, zm.get('chrY', [])):,} Z-DNA bp")


if __name__ == "__main__":
    main()
```

#### 6.2 · Karyogram bins and labels → Fig. 2a, Suppl. Fig. 1d, Table 2

```python
"""6.2 - Karyogram tables: Z-DNA bp per 100-kb bin, bin labels, telomere labels, chromosome-end enrichment
(Fig. 2a, Suppl. Fig. 1d ideogram, Table 2).

Bin label = annotation source holding the largest share of the bin's Z-DNA bp; per chromosome the N_LABEL_PER_CHROM
highest bins with distinct labels are labelled, plus terminal telomere arrays with Z-DNA. Chromosome ends: terminal
END_WINDOW_BP vs rest of the arm, paired two-sided Wilcoxon over the 48 arm ends.
"""
import gzip
import re
import shutil
from pathlib import Path

import numpy as np
import pandas as pd
import pybedtools
from pybedtools import BedTool
from scipy.stats import wilcoxon

BASE = Path("/path/to/pangenomes_zdna")
ANN_DIR = BASE / "tmp/chm13/chm13_marbl_ann_files"
ANN = ANN_DIR / "annotation"
CYTO_BED = ANN_DIR / "chm13v2.0_cytobands_allchrs.bed"
CENSAT_BED = ANN / "chm13v2.0_censat_v2.1.bed"
ZDNA_BED = ANN / "chm13_zdna_all.bed"
GFF = ANN / "chm13v2.0_RefSeq_Liftoff_v5.2.gff3.gz"
RMSK = ANN / "chm13v2.0_RepeatMasker_4.1.2p1.2022Apr14.bed"
TELO = ANN / "chm13v2.0_telomere.bed"

OUT_BINS = BASE / "6_chm13_karyogram_bins.tsv"
OUT_LABELS = BASE / "6_chm13_karyogram_labels.tsv"
OUT_TRACKS = BASE / "6_chm13_karyogram_tracks.tsv"
OUT_TELO = BASE / "6_chm13_karyogram_telomeres.tsv"
OUT_ENDS = BASE / "6_chm13_karyogram_arm_ends_all48.tsv"
OUT_END_STATS = BASE / "6_chm13_karyogram_arm_ends_all48_stats.tsv"
OUT_EVIDENCE = BASE / "6_chm13_karyogram_label_evidence.tsv"
OUT_SUMMARY = BASE / "6_chm13_karyogram_summary.tsv"
OUT_CLASSES = BASE / "6_chm13_karyogram_label_classes.tsv"
OUT_TOP = BASE / "6_chm13_karyogram_top_bins.tsv"
OUT_TABLE2 = BASE / "Table_2.tsv"
TMPDIR = BASE / "6_2_tmp"

CHROM_ORDER = [f"chr{i}" for i in range(1, 23)] + ["chrX", "chrY"]

BIN = 100_000                 # bin width (bp)
CAP_PCT = 100.0               # genome-wide percentile used as the top of the ridge scale (100 = highest bin, nothing clipped)

N_LABEL_PER_CHROM = 2         # labelled bins per chromosome
DISTINCT_LABELS = True        # each labelled bin of a chromosome must carry a different label text
MIN_LABEL_SHARE = 0.0         # a bin is eligible only if its label holds at least this share of its Z-DNA bp
LABEL_UNANNOTATED = False     # False: bins whose Z-DNA overlaps no label source are skipped when choosing labels
MAJORITY_SHARE = 0.5          # share at which a label is flagged as majority in the tables
LABEL_RENAME = {}             # applied to label texts after checking the evidence, e.g. {"SST": "SST1", "LSAU": "D4Z4"}
CENSAT_DISPLAY = {"hor": "HOR", "dhor": "dHOR", "mon": "mon. alpha-sat", "hsat1A": "HSat1A", "hsat1B": "HSat1B",
                  "hsat2": "HSat2", "hsat3": "HSat3", "bsat": "BSat", "gsat": "GSat"}
RMSK_LABEL_CLASSES = {"Satellite": "RM Satellite", "Retroposon": "Retroposon", "Simple_repeat": "Simple repeat"}
TRACK_CATEGORIES = ["hor", "dhor", "mon", "hsat1A", "hsat1B", "hsat2", "hsat3", "bsat", "gsat", "censat", "ct", "rDNA"]

MIN_ELEMENT_BP = 0            # elements shorter than this are left out of the class distribution
ENRICHMENT_QUANTILES = [0.10, 0.25, 0.50, 0.75, 0.90]
FOLD_THRESHOLDS = [1, 2]      # per-element enrichment thresholds reported

TERMINAL_MAX_BP = 10_000      # array is terminal if it lies within this distance of a chromosome end
TELOMERE_MIN_ZDNA_BP = 1      # terminal arrays with at least this much Z-DNA are labelled "telomere"

EVIDENCE_LABELS = ["LSAU"]    # bins with these labels are checked for genes matching GENE_REGEX
GENE_REGEX = r"^DUX4"
SST_REGEX = r"\(SST"          # CenSat names checked for their RepeatMasker composition

END_WINDOW_BP = 1_000_000                          # main terminal window
END_SENSITIVITY_BP = [500_000, 2_000_000, 5_000_000]


def merge(iv):
    out = []
    for s, e in sorted(iv):
        if out and s <= out[-1][1]:
            out[-1][1] = max(out[-1][1], e)
        else:
            out.append([s, e])
    return out


def ov(merged, s, e):
    return sum(max(0, min(e, b) - max(s, a)) for a, b in merged if b > s and a < e)


def bin_coverage(merged, length):
    edges = np.append(np.arange(0, length, BIN, dtype=np.int64), length)
    cov = np.zeros(len(edges) - 1)
    for s, e in merged:
        i, j = s // BIN, (e - 1) // BIN
        if i == j:
            cov[i] += e - s
        else:
            cov[i] += (i + 1) * BIN - s
            cov[j] += e - j * BIN
            cov[i + 1:j] += BIN
    return edges, cov


def censat_category(names):
    return names.astype(str).str.replace(r"\(.*$", "", regex=True).str.split("_").str[0]


def read_gff_genes(sizes):
    rows = []
    with gzip.open(GFF, "rt") as fh:
        for line in fh:
            if line.startswith("#"):
                continue
            f = line.rstrip("\n").split("\t")
            if len(f) < 9 or f[2] != "gene" or f[0] not in sizes:
                continue
            att = dict(kv.split("=", 1) for kv in f[8].split(";") if "=" in kv)
            rows.append((f[0], int(f[3]) - 1, int(f[4]), att.get("Name", att.get("gene", "")), att.get("gene_biotype", "")))
    return pd.DataFrame(rows, columns=["chrom", "start", "end", "name", "biotype"])


def label_sources(sizes, genes, rm, cen):
    """Label sources as 'group|label' intervals, merged per name (same array definitions as 6.3)."""
    motif = re.compile(r"^\(?[ACGTN]+\)?n$", re.IGNORECASE)

    def censat_label(cat, sub):
        if cat == "censat":
            sf = re.sub(r"_[Cc]omposite$", "", sub if isinstance(sub, str) else "")
            return "censat" if (not sf or sf.startswith("rnd-") or motif.match(sf)) else sf
        return CENSAT_DISPLAY[cat]

    src = [genes[genes.biotype == "rRNA"].assign(name="rDNA|rDNA"), rm[rm.rclass == "rRNA"].assign(name="rDNA|rDNA"),
           cen[cen.category == "rDNA"].assign(name="rDNA|rDNA"),
           genes[genes.biotype == "tRNA"].assign(name="tRNA|tRNA"), rm[rm.rclass == "tRNA"].assign(name="tRNA|tRNA"),
           cen[cen.category == "ct"].assign(name="CenSat CT|CT")]
    sat = cen[cen.category.isin(CENSAT_DISPLAY) | (cen.category == "censat")].copy()
    sat["name"] = ["CenSat|" + censat_label(c, s) for c, s in zip(sat.category, sat.subfamily)]
    src.append(sat)
    for rclass, group in RMSK_LABEL_CLASSES.items():
        sub = rm[rm.rclass == rclass].copy()
        if rclass == "Satellite":
            lab = sub.rep.where(~sub.rep.str.match(motif), "satellite")
        elif rclass == "Retroposon":
            lab = sub.rep.where(~sub.rep.str.startswith("SVA"), "SVA")
        else:
            lab = pd.Series("simple repeat", index=sub.index)
        src.append(sub.assign(name=group + "|" + lab))
    df = pd.concat([d[["chrom", "start", "end", "name"]] for d in src], ignore_index=True)
    df = df[df.chrom.isin(sizes) & (df.end > df.start)]
    rows = []
    for name, g in df.groupby("name"):
        for c, gg in g.groupby("chrom"):
            rows += [(c, s, e, name) for s, e in merge(zip(gg.start, gg.end))]
    return BedTool.from_dataframe(pd.DataFrame(rows, columns=["chrom", "start", "end", "name"])).sort()


def main():
    for p in (CYTO_BED, CENSAT_BED, ZDNA_BED, GFF, RMSK, TELO):
        if not p.exists():
            raise FileNotFoundError(p)
    if TMPDIR.exists():
        shutil.rmtree(TMPDIR)
    TMPDIR.mkdir(parents=True)
    pybedtools.helpers.set_tempdir(str(TMPDIR))

    cyto = pd.read_csv(CYTO_BED, sep="\t", header=None, usecols=[0, 1, 2, 3, 4], names=["chrom", "start", "end", "name", "category"])
    cyto = cyto[cyto.chrom.isin(CHROM_ORDER)].copy()
    sizes = cyto.groupby("chrom").end.max().astype(np.int64).to_dict()
    chroms = [c for c in CHROM_ORDER if c in sizes]
    z = pd.read_csv(ZDNA_BED, sep="\t", header=None, usecols=[0, 1, 2], names=["chrom", "start", "end"])
    zm = {c: merge(zip(g.start, g.end)) for c, g in z[z.chrom.isin(sizes)].groupby("chrom")}
    cen = pd.read_csv(CENSAT_BED, sep="\t", header=None, usecols=[0, 1, 2, 3, 8], names=["chrom", "start", "end", "name", "itemRgb"])
    cen = cen[cen.chrom.isin(sizes)].copy()
    cen["category"] = censat_category(cen.name)
    cen["subfamily"] = cen.name.astype(str).str.extract(r"\((.*)\)\s*$", expand=False)
    rm = pd.read_csv(RMSK, sep="\t", header=None, usecols=[0, 1, 2, 3, 6], names=["chrom", "start", "end", "rep", "rclass"],
                     dtype={"chrom": str, "rep": str, "rclass": str})
    rm = rm[rm.chrom.isin(sizes)].copy()
    genes = read_gff_genes(sizes)

    trk = cen[cen.category.isin(TRACK_CATEGORIES)]
    rgb = lambda s: "#%02X%02X%02X" % tuple(int(x) for x in str(s).split(","))
    modal = {c: rgb(g.itemRgb.value_counts().index[0]) for c, g in trk.groupby("category")}
    cols = ["track", "chrom", "start", "end", "name", "category", "color_hex"]
    pd.concat([cyto.assign(track="cytoband", color_hex="")[cols], trk.assign(track="censat", color_hex=trk.category.map(modal))[cols]],
              ignore_index=True).to_csv(OUT_TRACKS, sep="\t", index=False)

    rows = []
    for c in chroms:
        edges, cv = bin_coverage(zm.get(c, []), int(sizes[c]))
        rows.append(pd.DataFrame({"chrom": c, "bin_start": edges[:-1], "bin_end": edges[1:], "zdna_bp": cv.astype(np.int64)}))
    bins = pd.concat(rows, ignore_index=True)
    bins.insert(0, "bin_id", np.arange(len(bins)))
    cap = float(np.percentile(bins.zdna_bp, CAP_PCT))
    bins["bin_len"] = bins.bin_end - bins.bin_start
    bins["zdna_bp_per_kb"] = bins.zdna_bp / bins.bin_len * 1e3
    bins["zdna_bp_capped"] = np.minimum(bins.zdna_bp, cap)
    bins["is_capped"] = bins.zdna_bp > cap

    src = label_sources(sizes, genes, rm, cen)
    b_bt = BedTool.from_dataframe(bins[["chrom", "bin_start", "bin_end", "bin_id"]]).sort()
    z_bt = BedTool.from_dataframe(pd.DataFrame([(c, s, e) for c in chroms for s, e in zm.get(c, [])],
                                               columns=["chrom", "start", "end"])).sort()
    zb = b_bt.intersect(z_bt).sort()
    o = zb.intersect(src, wo=True).to_dataframe(disable_auto_names=True, header=None,
                                                names=["c", "s", "e", "bin_id", "c2", "s2", "e2", "name", "ov"])
    lab = o.groupby(["bin_id", "name"]).ov.sum().reset_index()
    lab[["group", "label"]] = lab.name.str.split("|", n=1, expand=True)
    tot = bins.set_index("bin_id").zdna_bp
    lab["share"] = lab.ov / lab.bin_id.map(tot)
    best = lab.sort_values(["bin_id", "ov", "name"], ascending=[True, False, True]).drop_duplicates("bin_id")
    bins = bins.merge(best[["bin_id", "group", "label", "share"]].rename(
        columns={"group": "label_group", "label": "label_text", "share": "label_share"}), on="bin_id", how="left")
    gshare = lab.groupby(["bin_id", "group"]).share.sum().unstack(fill_value=0.0)
    gshare.columns = [f"share_{g.replace(' ', '_')}" for g in gshare.columns]
    bins = bins.merge(gshare.reset_index(), on="bin_id", how="left")
    bins[["label_group", "label_text"]] = bins[["label_group", "label_text"]].fillna("")
    bins = bins.fillna({c: 0.0 for c in bins.columns if c.startswith("share_") or c == "label_share"})
    bins["label_text_raw"] = bins.label_text
    bins["label_text"] = bins.label_text.replace(LABEL_RENAME)
    bins["majority"] = bins.label_share >= MAJORITY_SHARE

    bins["labelled"] = False
    for c in chroms:
        sub = bins[(bins.chrom == c) & (bins.zdna_bp > 0)].sort_values(["zdna_bp", "bin_start"], ascending=[False, True])
        used, n = set(), 0
        for r in sub.itertuples():
            if n == N_LABEL_PER_CHROM:
                break
            if (not r.label_text and not LABEL_UNANNOTATED) or r.label_share < MIN_LABEL_SHARE:
                continue
            if DISTINCT_LABELS and r.label_text in used:
                continue
            bins.loc[bins.bin_id == r.bin_id, "labelled"] = True
            used.add(r.label_text)
            n += 1
    bins.to_csv(OUT_BINS, sep="\t", index=False)

    top = (bins.sort_values(["zdna_bp", "bin_start"], ascending=[False, True]).drop_duplicates("chrom")
           .set_index("chrom").loc[chroms].reset_index())
    top["region_1based"] = [f"{c}:{s + 1:,}-{e:,}" for c, s, e in zip(top.chrom, top.bin_start, top.bin_end)]
    top["label_text"] = top.label_text.replace("", "none")
    top = top[["chrom", "region_1based", "bin_start", "bin_end", "zdna_bp", "zdna_bp_per_kb", "label_group", "label_text",
               "label_share", "majority", "labelled"]]
    top.to_csv(OUT_TOP, sep="\t", index=False)
    pd.DataFrame({"Chromosome": top.chrom, "Z-DNA density (bp/kb)": top.zdna_bp_per_kb.round(2),
                  "Feature": top.label_text}).to_csv(OUT_TABLE2, sep="\t", index=False)

    telo = pd.read_csv(TELO, sep="\t", header=None, usecols=[0, 1, 2], names=["chrom", "start", "end"])
    telo = telo[telo.chrom.isin(sizes)].copy()
    L = telo.chrom.map(sizes)
    telo["arm"] = np.where(telo.start <= L - telo.end, "p", "q")
    telo["dist_to_end_bp"] = np.minimum(telo.start, L - telo.end)
    telo["type"] = np.where(telo.dist_to_end_bp <= TERMINAL_MAX_BP, "terminal", "interstitial")
    telo["array_bp"] = telo.end - telo.start
    telo["zdna_bp"] = [ov(zm.get(c, []), s, e) for c, s, e in zip(telo.chrom, telo.start, telo.end)]
    telo["labelled"] = (telo.type == "terminal") & (telo.zdna_bp >= TELOMERE_MIN_ZDNA_BP)
    telo["x_plot"] = np.where(telo.arm == "p", 0, L)
    telo.to_csv(OUT_TELO, sep="\t", index=False)

    lb = bins[bins.labelled]
    labels = pd.concat([
        pd.DataFrame({"record_type": "bin", "chrom": lb.chrom, "x": (lb.bin_start + lb.bin_end) // 2, "align": "center",
                      "bin_start": lb.bin_start, "bin_end": lb.bin_end, "zdna_bp": lb.zdna_bp, "bin_id": lb.bin_id,
                      "label_group": lb.label_group, "label_text": lb.label_text, "label_text_raw": lb.label_text_raw,
                      "label_share": lb.label_share}),
        pd.DataFrame({"record_type": "telomere", "chrom": telo[telo.labelled].chrom, "x": telo[telo.labelled].x_plot,
                      "align": np.where(telo[telo.labelled].arm == "p", "left", "right"),
                      "bin_start": telo[telo.labelled].start, "bin_end": telo[telo.labelled].end,
                      "zdna_bp": telo[telo.labelled].zdna_bp, "label_group": "Telomere", "label_text": "telomere",
                      "label_text_raw": "telomere", "label_share": np.nan})], ignore_index=True)
    labels["chrom_order"] = labels.chrom.map({c: i for i, c in enumerate(chroms)})
    labels = labels.sort_values(["chrom_order", "x"]).drop(columns="chrom_order").reset_index(drop=True)

    genome_density = sum(e - s for v in zm.values() for s, e in v) / sum(sizes.values())
    labels["class_name"] = labels.label_group + "|" + labels.label_text_raw
    el = src.to_dataframe(disable_auto_names=True, header=None, names=["chrom", "start", "end", "name"])
    el = pd.concat([el[el.name.isin(set(labels.class_name))],
                    telo.assign(name="Telomere|telomere")[["chrom", "start", "end", "name"]]], ignore_index=True)
    el["element_id"] = np.arange(len(el))
    zo = BedTool.from_dataframe(el[["chrom", "start", "end", "element_id"]]).sort().intersect(z_bt, wao=True).to_dataframe(
        disable_auto_names=True, header=None, names=["c", "s", "e", "element_id", "c2", "s2", "e2", "ov"])
    el["zdna_bp"] = el.element_id.map(zo.groupby("element_id").ov.sum()).fillna(0).astype(np.int64)
    el["length_bp"] = el.end - el.start
    el["enrichment"] = el.zdna_bp / el.length_bp / genome_density
    el = el[el.length_bp >= MIN_ELEMENT_BP]
    crows, pooled = [], {}
    for name, g in el.groupby("name"):
        grp, txt = name.split("|", 1)
        pooled[name] = g.zdna_bp.sum() / g.length_bp.sum() / genome_density
        r = {"label_group": grp, "label_text": LABEL_RENAME.get(txt, txt), "n_elements": len(g),
             "class_bp": int(g.length_bp.sum()), "class_zdna_bp": int(g.zdna_bp.sum()), "pooled_enrichment": pooled[name],
             "median_element_bp": g.length_bp.median(),
             **{f"element_enrichment_q{int(round(q * 100))}": v for q, v in g.enrichment.quantile(ENRICHMENT_QUANTILES).items()},
             "frac_elements_no_zdna": (g.zdna_bp == 0).mean()}
        for t in FOLD_THRESHOLDS:
            m = g.enrichment >= t
            r.update({f"frac_elements_ge{t}x": m.mean(), f"frac_class_bp_in_ge{t}x": g.length_bp[m].sum() / g.length_bp.sum(),
                      f"frac_class_zdna_in_ge{t}x": g.zdna_bp[m].sum() / max(g.zdna_bp.sum(), 1)})
        r["n_labels_in_figure"] = int((labels.class_name == name).sum())
        crows.append(r)
    classes = pd.DataFrame(crows).sort_values("n_labels_in_figure", ascending=False)
    classes.to_csv(OUT_CLASSES, sep="\t", index=False)
    el_by = {k: g for k, g in el.groupby(["name", "chrom"])}
    lab_ov = lab.set_index(["bin_id", "name"]).ov
    loc = []
    for r in labels.itertuples():
        if r.record_type == "bin":
            g = el_by.get((r.class_name, r.chrom))
            cbp = ov(list(zip(g.start, g.end)), r.bin_start, r.bin_end) if g is not None else 0
            czb = int(lab_ov.get((r.bin_id, r.class_name), 0))
        else:
            cbp, czb = r.bin_end - r.bin_start, r.zdna_bp
        local = czb / cbp / genome_density if cbp else np.nan
        allv = el.loc[el.name == r.class_name, "enrichment"]
        loc.append({"class_bp_in_bin": cbp, "class_zdna_bp_in_bin": czb, "class_enrichment_in_bin": local,
                    "percentile_in_class": (allv <= local).mean() * 100 if len(allv) else np.nan,
                    "class_pooled_enrichment": pooled.get(r.class_name, np.nan)})
    labels = pd.concat([labels, pd.DataFrame(loc)], axis=1)
    labels.to_csv(OUT_LABELS, sep="\t", index=False)

    ev = []
    rm_by = {c: g for c, g in rm.groupby("chrom")}
    for r in cen[cen.name.astype(str).str.contains(SST_REGEX, regex=True)].itertuples():
        g0 = rm_by.get(r.chrom, rm.iloc[:0])
        sub = g0[(g0.end > r.start) & (g0.start < r.end)]
        o2 = (np.minimum(sub.end, r.end) - np.maximum(sub.start, r.start)).groupby(sub.rep).sum().sort_values(ascending=False)
        ev.append({"check": "SST array RepeatMasker composition", "subject": r.name,
                   "region": f"{r.chrom}:{r.start + 1:,}-{r.end:,}", "length_bp": r.end - r.start,
                   "result": "; ".join(f"{k} {v / (r.end - r.start):.0%}" for k, v in o2.head(3).items())})
    for r in bins[bins.label_text_raw.isin(EVIDENCE_LABELS) & (bins.zdna_bp > 0)].itertuples():
        g = genes[(genes.chrom == r.chrom) & (genes.end > r.bin_start) & (genes.start < r.bin_end)
                  & genes.name.str.contains(GENE_REGEX, regex=True)]
        ev.append({"check": f"genes matching {GENE_REGEX} in {r.label_text_raw} bins", "subject": r.label_text_raw,
                   "region": f"{r.chrom}:{r.bin_start + 1:,}-{r.bin_end:,}", "length_bp": r.bin_len,
                   "result": f"{len(g)} genes: {', '.join(sorted(g.name.unique())[:10])}" if len(g) else "none"})
    evidence = pd.DataFrame(ev, columns=["check", "subject", "region", "length_bp", "result"])
    evidence.to_csv(OUT_EVIDENCE, sep="\t", index=False)

    cen_pos = cyto[cyto.category == "acen"].groupby("chrom").agg(
        cs=("start", "min"), ce=("end", "max")
    )
    erows = []
    for w in [END_WINDOW_BP] + END_SENSITIVITY_BP:
        for c in chroms:
            Lc = int(sizes[c])
            cs, ce = int(cen_pos.at[c, "cs"]), int(cen_pos.at[c, "ce"])
            m = zm.get(c, [])
            arms = [
                ("p", (0, w), (w, cs)),
                ("q", (Lc - w, Lc), (ce, Lc - w)),
            ]
            for arm, term, rest in arms:
                if rest[1] - rest[0] < w:
                    if w == END_WINDOW_BP:
                        raise ValueError(
                            f"Insufficient remaining arm for {c}{arm} at main window {w:,} bp"
                        )
                    continue
                terminal = ov(m, *term) / (term[1] - term[0]) * 1e6
                remainder = ov(m, *rest) / (rest[1] - rest[0]) * 1e6
                erows.append({
                    "window_bp": w, "chrom": c, "arm": arm,
                    "terminal_bp_per_Mb": terminal,
                    "rest_of_arm_bp_per_Mb": remainder,
                })

    ends = pd.DataFrame(erows)
    ends["ratio"] = ends.terminal_bp_per_Mb / ends.rest_of_arm_bp_per_Mb
    ends.to_csv(OUT_ENDS, sep="\t", index=False)

    end_stats = []
    for w, g in ends.groupby("window_bp", sort=False):
        st = wilcoxon(
            g.terminal_bp_per_Mb, g.rest_of_arm_bp_per_Mb,
            alternative="two-sided"
        )
        end_stats.append({
            "window_bp": w,
            "n_arm_ends": len(g),
            "median_terminal": g.terminal_bp_per_Mb.median(),
            "median_rest": g.rest_of_arm_bp_per_Mb.median(),
            "median_ratio": g.ratio.median(),
            "n_terminal_higher": int(
                (g.terminal_bp_per_Mb > g.rest_of_arm_bp_per_Mb).sum()
            ),
            "W": st.statistic,
            "p_value": st.pvalue,
        })
    end_stats = pd.DataFrame(end_stats)
    end_stats.to_csv(OUT_END_STATS, sep="\t", index=False)

    main_ends = ends.loc[ends.window_bp == END_WINDOW_BP]
    main_stats = end_stats.loc[end_stats.window_bp == END_WINDOW_BP].iloc[0]
    assert len(main_ends) == 48, "Expected all 48 chromosome-arm ends"
    assert int(main_stats.n_terminal_higher) == 41, (
        "Expected 41 enriched ends; check inputs and annotations"
    )
    assert np.isclose(main_stats.median_ratio, 2.428, atol=0.001), (
        "Median ratio differs from manuscript"
    )
    assert np.isclose(main_stats.p_value, 2.25783e-6, rtol=1e-4), (
        "Wilcoxon p-value differs from manuscript"
    )

    summ = {"bin_bp": BIN, "n_bins": len(bins), "cap_percentile": CAP_PCT,
            "cap_zdna_bp_per_bin": cap,
            "n_bins_above_cap": int(bins.is_capped.sum()),
            "n_label_per_chrom": N_LABEL_PER_CHROM,
            "distinct_labels": DISTINCT_LABELS,
            "min_label_share": MIN_LABEL_SHARE,
            "label_rename": str(LABEL_RENAME),
            "n_labelled_bins": int(bins.labelled.sum()),
            "n_telomere_labels": int(telo.labelled.sum()),
            "end_window_bp": END_WINDOW_BP,
            "max_chrom_bp": int(max(sizes.values()))}
    for k, v in main_stats.items():
        summ[f"arm_end_{k}"] = v
    pd.Series(summ, name="value").rename_axis("parameter").to_csv(
        OUT_SUMMARY, sep="\t"
    )

    print(f"Bins: {len(bins):,} x {BIN:,} bp; labelled bins {int(bins.labelled.sum())}; telomere labels "
          f"{int(telo.labelled.sum())} ({', '.join(c + a for c, a in zip(telo[telo.labelled].chrom, telo[telo.labelled].arm))})")
    print(f"Arm ends (terminal {END_WINDOW_BP / 1e6:g} Mb vs rest of arm): {int(main_stats.n_terminal_higher)}/{len(main_ends)} "
          f"higher; median ratio {main_stats.median_ratio:.2f}; Wilcoxon p = {main_stats.p_value:.2g}")
    pybedtools.cleanup(remove_all=True)
    shutil.rmtree(TMPDIR, ignore_errors=True)


if __name__ == "__main__":
    main()
```

#### 6.3 · Feature-class enrichment → Suppl. Fig. 1e

```python
"""6.3 - Z-DNA density and enrichment across genomic feature classes (Suppl. Fig. 1e).

Enrichment = (Z-DNA bp in the merged class / class bp) / genome-wide Z-DNA density; classes are not mutually exclusive.
Also writes the merged BED of every class to 6_chm13_features/ (input to 7.1, 7.5, 18.1).
"""
import gzip
import math
import shutil
from pathlib import Path

import numpy as np
import pandas as pd
import pybedtools
from pybedtools import BedTool

BASE = Path("/path/to/pangenomes_zdna")
ANN = BASE / "tmp/chm13/chm13_marbl_ann_files/annotation"
GFF = ANN / "chm13v2.0_RefSeq_Liftoff_v5.2.gff3.gz"
ZDNA = ANN / "chm13_zdna_all.bed"
CENSAT = ANN / "chm13v2.0_censat_v2.1.bed"
TELO = ANN / "chm13v2.0_telomere.bed"
RMSK = ANN / "chm13v2.0_RepeatMasker_4.1.2p1.2022Apr14.bed"
INTRONS = ANN / "pc_transcript_introns.bed"
CHROMSIZES = ANN / "chm13_chrom_sizes.txt"
OUT = BASE / "6_chm13_zdna_feature_enrichment.tsv"
TMPDIR = BASE / "6_3_tmp"
FEATDIR = BASE / "6_chm13_features"
GENE_BIOTYPES = ["protein_coding", "lncRNA", "miRNA", "snoRNA"]          # one class each (gene_<biotype>)
PSEUDOGENE_BIOTYPES = ["pseudogene", "transcribed_pseudogene"]           # union -> gene_pseudogene
EXPLICIT_GENE_CLASSES = GENE_BIOTYPES + PSEUDOGENE_BIOTYPES + ["rRNA", "tRNA"]   # everything else -> gene_other
CENSAT_CLASSES = ["ct", "censat", "hor", "dhor", "mon", "hsat1A", "hsat1B", "hsat2", "hsat3", "gsat", "bsat", "rDNA"]
RMSK_CLASSES = ["LINE", "SINE", "LTR", "Simple_repeat", "Satellite", "Low_complexity", "Retroposon", "RC", "Beta",
                "snRNA", "srpRNA", "scRNA", "Unknown"]
COLS = ["chrom", "start", "end"]


def feature_frames(valid):
    """Raw (unmerged) intervals of every feature class, BED coordinates (GFF start - 1)."""
    genes, exons, cds = [], [], []
    with gzip.open(GFF, "rt") as fh:
        for line in fh:
            if line.startswith("#"):
                continue
            f = line.rstrip("\n").split("\t")
            if len(f) < 9 or f[0] not in valid:
                continue
            s, e = int(f[3]) - 1, int(f[4])
            if e <= s:
                continue
            if f[2] == "gene":
                att = dict(kv.split("=", 1) for kv in f[8].split(";") if "=" in kv)
                if att.get("gene_biotype") is not None:
                    genes.append((f[0], s, e, att["gene_biotype"]))
            elif f[2] == "exon":
                exons.append((f[0], s, e))
            elif f[2] == "CDS":
                cds.append((f[0], s, e))
    genes = pd.DataFrame(genes, columns=COLS + ["biotype"])
    bed3 = lambda p: pd.read_csv(p, sep="\t", header=None, usecols=[0, 1, 2], names=COLS, comment="#")
    cen = pd.read_csv(CENSAT, sep="\t", header=None, usecols=[0, 1, 2, 3], names=COLS + ["name"])
    cen["category"] = cen.name.astype(str).str.replace(r"\(.*$", "", regex=True).str.split("_").str[0]
    rm = pd.read_csv(RMSK, sep="\t", header=None, dtype=str).iloc[:, :7]
    rm.columns = COLS + ["col4", "col5", "col6", "class"]

    fr = {f"gene_{b}": genes[genes.biotype == b] for b in GENE_BIOTYPES}
    fr["gene_pseudogene"] = genes[genes.biotype.isin(PSEUDOGENE_BIOTYPES)]
    fr["gene_other"] = genes[~genes.biotype.isin(EXPLICIT_GENE_CLASSES)]
    fr["gene_body"] = genes[~genes.biotype.isin(["rRNA", "tRNA"])]
    fr["pc_exon"] = pd.DataFrame(exons, columns=COLS)
    fr["pc_CDS"] = pd.DataFrame(cds, columns=COLS)
    fr["pc_intron"] = bed3(INTRONS)
    for c in CENSAT_CLASSES:
        fr[f"censat_{c}"] = cen[cen.category == c]
    fr["rmsk_DNA_transposon"] = rm[rm["class"] == "DNA"]
    fr["rmsk_DNA_other"] = rm[rm["class"].str.startswith("DNA", na=False) & ~rm["class"].isin(["DNA", "DNA?"])]
    for c in RMSK_CLASSES:
        fr[f"rmsk_{c}"] = rm[rm["class"] == c]
    fr["rmsk_all"] = rm
    fr["telomere"] = bed3(TELO)
    fr["rDNA_arrays"] = pd.concat([genes[genes.biotype == "rRNA"], rm[rm["class"] == "rRNA"], cen[cen.category == "rDNA"]])
    fr["tRNA_arrays"] = pd.concat([genes[genes.biotype == "tRNA"], rm[rm["class"] == "tRNA"]])
    return fr


def merged_bed(df, valid, path):
    """Sort + merge one class, save it as BED, return its bp (0 and no file if empty)."""
    df = df[COLS].copy()
    df["start"] = pd.to_numeric(df["start"]).astype("int64")
    df["end"] = pd.to_numeric(df["end"]).astype("int64")
    df = df[df.chrom.isin(valid) & (df.end > df.start)]
    if df.empty:
        return 0
    bt = BedTool.from_dataframe(df.sort_values(COLS)).sort().merge()
    bt.saveas(str(path))
    return sum(int(i.end) - int(i.start) for i in bt)


def main():
    for p in (GFF, ZDNA, CENSAT, TELO, RMSK, INTRONS, CHROMSIZES):
        if not p.exists():
            raise FileNotFoundError(p)
    if TMPDIR.exists():
        shutil.rmtree(TMPDIR)
    (TMPDIR / "features").mkdir(parents=True)
    if FEATDIR.exists():
        shutil.rmtree(FEATDIR)
    FEATDIR.mkdir()
    pybedtools.helpers.set_tempdir(str(TMPDIR))

    sizes = pd.read_csv(CHROMSIZES, sep="\t", header=None, usecols=[0, 1], names=["chrom", "length"])
    valid, genome_bp = set(sizes.chrom), int(sizes.length.sum())
    z_bp = merged_bed(pd.read_csv(ZDNA, sep="\t", header=None, usecols=[0, 1, 2], names=COLS), valid, TMPDIR / "zdna.bed")
    genome_frac = z_bp / genome_bp
    zbt = BedTool(str(TMPDIR / "zdna.bed"))

    rows = []
    for name, df in sorted(feature_frames(valid).items()):
        path = FEATDIR / f"{name}.bed"
        fbp = merged_bed(df, valid, path)
        ovl = sum(int(i.fields[-1]) for i in zbt.intersect(BedTool(str(path)), wo=True)) if fbp else 0
        if ovl > fbp:
            raise RuntimeError(f"{name}: Z-DNA overlap exceeds feature length")
        frac = ovl / fbp if fbp else np.nan
        rel = frac / genome_frac if fbp else np.nan
        rows.append({"feature": name, "feature_bp": fbp, "zdna_overlap_bp": ovl, "genome_bp": genome_bp, "zdna_total_bp": z_bp,
                     "feature_zdna_fraction": frac, "feature_zdna_bp_per_Mb": frac * 1e6 if fbp else np.nan,
                     "genome_zdna_fraction": genome_frac, "genome_zdna_bp_per_Mb": genome_frac * 1e6,
                     "relative_zdna_density": rel,
                     "log2_relative_zdna_density": math.log2(rel) if fbp and rel > 0 else np.nan})
    out = pd.DataFrame(rows)
    out.to_csv(OUT, sep="\t", index=False)

    print(f"Genome-wide Z-DNA density {genome_frac * 1e6:,.1f} bp/Mb; {len(out)} feature classes -> {OUT.name}, {FEATDIR.name}/")
    pybedtools.cleanup(remove_all=True)
    shutil.rmtree(TMPDIR, ignore_errors=True)


if __name__ == "__main__":
    main()
```

#### 7.1 · Predictions with compartment membership

```python
"""7.1 - CHM13 ZSeeker predictions with sequence and compartment membership (input to 7.2-7.5).

Compartments = merged 6.3 feature BEDs (6_chm13_features/); a prediction belongs to every compartment it overlaps by
>= MIN_OVERLAP_BP. Sequences are checked against the reference FASTA when VERIFY_FASTA is True.
"""
import shutil
from pathlib import Path

import numpy as np
import pandas as pd
import pybedtools
from pybedtools import BedTool

BASE = Path("/path/to/pangenomes_zdna")
ANN = BASE / "tmp/chm13/chm13_marbl_ann_files/annotation"
CSV = BASE / "tmp/chm13/chm13_zdna_files/GCF_009914755.1_T2T-CHM13v2.0_genomic_zdna_score.csv"
FASTA = BASE / "tmp/chm13/chm13_refseq/GCF_009914755.1_T2T-CHM13v2.0_genomic_ucsc.fna"   # needs .fai
ZDNA_BED = ANN / "chm13_zdna_all.bed"
CHROMSIZES = ANN / "chm13_chrom_sizes.txt"
ENRICH = BASE / "6_chm13_zdna_feature_enrichment.tsv"
FEATDIR = BASE / "6_chm13_features"
OUT_HITS = BASE / "7_chm13_zdna_hits.tsv.gz"
OUT_COMP = BASE / "7_chm13_zdna_compartments.tsv"
TMPDIR = BASE / "7_1_tmp"

MIN_OVERLAP_BP = 1             # bp a prediction must share with a compartment to belong to it
VERIFY_FASTA = True            # compare every ZSeeker sequence with the reference FASTA
GENOME = "genome_all"
COMPARTMENTS = [               # (6.3 feature, label, panel group) in Suppl. Fig. 1e order
    ("censat_hsat1A", "HSAT1A", "tandem"), ("censat_hsat1B", "HSAT1B", "tandem"), ("censat_hsat2", "HSAT2", "tandem"),
    ("censat_hsat3", "HSAT3", "tandem"), ("censat_hor", "HOR", "tandem"), ("censat_dhor", "dHOR", "tandem"),
    ("censat_mon", "Monomers", "tandem"), ("censat_gsat", "GSAT", "tandem"), ("censat_bsat", "BSAT", "tandem"),
    ("censat_ct", "CT", "tandem"), ("censat_censat", "CenSat", "tandem"), ("telomere", "Telomeres", "tandem"),
    ("tRNA_arrays", "tRNA", "tandem"), ("rDNA_arrays", "rDNA", "tandem"),
    ("rmsk_LINE", "LINEs", "interspersed"), ("rmsk_SINE", "SINEs", "interspersed"), ("rmsk_LTR", "LTR retr.", "interspersed"),
    ("rmsk_DNA_transposon", "DNA transp.", "interspersed"), ("rmsk_Retroposon", "Retroposons", "interspersed"),
    ("rmsk_RC", "Rolling-circle", "interspersed"), ("rmsk_Simple_repeat", "Simple repeats", "interspersed"),
    ("rmsk_Beta", "Beta repeats", "interspersed"), ("rmsk_snRNA", "snRNA repeats", "interspersed"),
    ("rmsk_srpRNA", "srpRNA repeats", "interspersed"), ("rmsk_scRNA", "scRNA repeats", "interspersed"),
    ("gene_protein_coding", "Protein-coding", "biotypes"), ("gene_lncRNA", "lncRNA", "biotypes"),
    ("gene_pseudogene", "Pseudogenes", "biotypes"), ("gene_miRNA", "miRNA", "biotypes"), ("gene_snoRNA", "snoRNA", "biotypes"),
    ("gene_body", "Gene body", "gene_parts"), ("pc_intron", "Introns", "gene_parts"), ("pc_exon", "Exons", "gene_parts"),
    ("pc_CDS", "CDS", "gene_parts"),
]
COLS = ["chrom", "start", "end"]


def compartment_beds():
    """Merged 6.3 BED and bp of every compartment (bp 0 and no file if the class is empty)."""
    paths, bps = {}, {}
    for key, _, _ in COMPARTMENTS:
        paths[key] = FEATDIR / f"{key}.bed"
        if paths[key].exists():
            b = pd.read_csv(paths[key], sep="\t", header=None, usecols=[1, 2])
            bps[key] = int((b[2] - b[1]).sum())
        else:
            bps[key] = 0
    return paths, bps


class Fasta:
    """Minimal .fai-indexed FASTA reader (0-based, half-open, upper-case)."""

    def __init__(self, path):
        self.path, self.idx = str(path), {}
        with open(self.path + ".fai") as fh:
            for line in fh:
                n, length, off, lb, lw = line.rstrip("\n").split("\t")[:5]
                self.idx[n] = (int(length), int(off), int(lb), int(lw))

    def fetch(self, name, start, end):
        length, off, lb, lw = self.idx[name]
        start, end = max(0, start), min(end, length)
        first = off + start // lb * lw + start % lb
        last = off + (end - 1) // lb * lw + (end - 1) % lb + 1
        with open(self.path, "rb") as fh:
            fh.seek(first)
            raw = fh.read(last - first)
        return raw.replace(b"\n", b"").replace(b"\r", b"").decode("ascii").upper()


def main():
    for p in (CSV, FASTA, Path(str(FASTA) + ".fai"), ZDNA_BED, CHROMSIZES, ENRICH, FEATDIR):
        if not p.exists():
            raise FileNotFoundError(p)
    if TMPDIR.exists():
        shutil.rmtree(TMPDIR)
    TMPDIR.mkdir(parents=True)
    pybedtools.helpers.set_tempdir(str(TMPDIR))
    sizes = pd.read_csv(CHROMSIZES, sep="\t", header=None, usecols=[0, 1], names=["chrom", "length"])
    sizes = dict(zip(sizes.chrom, sizes.length.astype(np.int64)))

    raw = pd.read_csv(CSV).dropna(subset=["Start", "End", "Sequence"])
    hits = pd.DataFrame({"chrom": raw.Chromosome.astype(str), "start": pd.to_numeric(raw.Start).astype(np.int64),
                         "end": pd.to_numeric(raw.End).astype(np.int64) + 1,
                         "score": pd.to_numeric(raw["Z-DNA Score"]).astype(float), "sequence": raw.Sequence.astype(str).str.upper()})
    hits = hits[hits.chrom.isin(sizes)].reset_index(drop=True)
    hits["length"] = hits.end - hits.start
    if (hits.sequence.str.len() != hits.length).any():
        raise RuntimeError("len(Sequence) != End - Start + 1 for some predictions")
    zb = pd.read_csv(ZDNA_BED, sep="\t", header=None, usecols=[0, 1, 2], names=COLS)
    zb = zb[zb.chrom.isin(sizes)].astype({"start": "int64", "end": "int64"}).sort_values(COLS).reset_index(drop=True)
    if not zb.equals(hits[COLS].sort_values(COLS).reset_index(drop=True)):
        raise RuntimeError("ZSeeker CSV (Start, End + 1) and the Z-DNA BED differ")
    if VERIFY_FASTA:
        fa, bad = Fasta(FASTA), 0
        for c, g in hits.groupby("chrom", sort=False):
            seq = fa.fetch(c, 0, int(sizes[c]))
            bad += sum(seq[s:e] != q for s, e, q in zip(g.start, g.end, g.sequence))
        if bad:
            raise RuntimeError(f"{bad} ZSeeker sequences differ from the reference FASTA")

    paths, feat_bp = compartment_beds()
    e = pd.read_csv(ENRICH, sep="\t").set_index("feature")
    bad = [(k, feat_bp[k], int(e.at[k, "feature_bp"])) for k, _, _ in COMPARTMENTS if feat_bp[k] != int(e.at[k, "feature_bp"])]
    if bad:
        raise RuntimeError(f"Compartment bp differ from 6.3 (feature, here, 6.3): {bad}")

    hbt = BedTool.from_dataframe(hits[COLS].assign(hit_id=np.arange(len(hits)))).sort()
    memb = [[] for _ in range(len(hits))]
    for key, _, _ in COMPARTMENTS:
        if not paths[key].exists():
            continue
        o = hbt.intersect(BedTool(str(paths[key])), wo=True).to_dataframe(
            disable_auto_names=True, header=None, names=["c", "s", "e", "hit_id", "c2", "s2", "e2", "ov"])
        if o.empty:
            continue
        for i in o.groupby("hit_id").ov.sum().loc[lambda v: v >= MIN_OVERLAP_BP].index:
            memb[int(i)].append(key)
    hits["compartments"] = [";".join(m) for m in memb]
    hits[COLS + ["length", "score", "sequence", "compartments"]].to_csv(OUT_HITS, sep="\t", index=False)

    rows = [{"compartment": GENOME, "label": "All Z-DNA", "panel": "reference", "n_hits": len(hits),
             "hit_bp": int(hits.length.sum()), "feature_bp": int(sum(sizes.values()))}]
    for key, label, panel in COMPARTMENTS:
        sel = hits.compartments.str.contains(rf"(?:^|;){key}(?:;|$)", regex=True)
        rows.append({"compartment": key, "label": label, "panel": panel, "n_hits": int(sel.sum()),
                     "hit_bp": int(hits.length[sel].sum()), "feature_bp": feat_bp[key]})
    comp = pd.DataFrame(rows)
    comp.to_csv(OUT_COMP, sep="\t", index=False)

    print(f"Predictions: {len(hits):,} (CSV = BED{' = FASTA' if VERIFY_FASTA else ''}); {len(COMPARTMENTS)} compartments "
          f"(bp identical to 6.3); in no compartment: {(hits.compartments == '').sum():,}")
    pybedtools.cleanup(remove_all=True)
    shutil.rmtree(TMPDIR, ignore_errors=True)


if __name__ == "__main__":
    main()
```

#### 7.2 · Telomere arrays → Suppl. Fig. 1f

```python
"""7.2 - Telomere arrays: (TTAGGG)n tracts, the 29-bp telomere-associated repeat and Z-DNA predictions (Suppl. Fig. 1f).

Scans every annotated array (terminal arrays + FLANK_BP of subtelomere). Core tract: >= MIN_UNITS in-phase hexamers
within MAX_MISMATCHES of TTAGGG/CCCTAA; 29-bp repeat tracts as in Alaguponniah et al. 2020. Per prediction: overlap with
both tract types, closest hexamer match, RepeatMasker/CenSat element.
"""
from pathlib import Path

import numpy as np
import pandas as pd
from numpy.lib.stride_tricks import sliding_window_view

BASE = Path("/path/to/pangenomes_zdna")
ANN = BASE / "tmp/chm13/chm13_marbl_ann_files/annotation"
TELO = ANN / "chm13v2.0_telomere.bed"
CHROMSIZES = ANN / "chm13_chrom_sizes.txt"
FASTA = BASE / "tmp/chm13/chm13_refseq/GCF_009914755.1_T2T-CHM13v2.0_genomic_ucsc.fna"   # needs .fai
RMSK = ANN / "chm13v2.0_RepeatMasker_4.1.2p1.2022Apr14.bed"
CENSAT = ANN / "chm13v2.0_censat_v2.1.bed"
HITS = BASE / "7_chm13_zdna_hits.tsv.gz"
OUT_ARRAYS = BASE / "7_chm13_telomere_arrays.tsv"
OUT_FEATURES = BASE / "7_chm13_telomere_repeats_zdna.tsv"

CORE_MOTIF = "TTAGGG"          # scanned together with its reverse complement (CCCTAA)
MAX_MISMATCHES = 1             # per hexamer unit
MIN_UNITS = 2                  # consecutive in-phase units forming a tract
ANCHOR_EXACT = True            # at least one unit of a tract must be an exact match
LONG_REPEAT = "TCTCTGCGCCTGCGCCGGCGCGGCGCGCC"   # 29-bp consensus, scanned with its reverse complement
LONG_MAX_MISMATCH_FRAC = 0.10  # per copy: mismatches <= floor(frac * 29)
LONG_MIN_UNITS = 2             # copies forming a tract
LONG_SPACING_TOLERANCE = 3     # copies may start this many bp earlier than one unit length after the previous copy
LONG_MAX_GAP_BP = 3            # copies whose windows are at most this far apart join the same tract
TERMINAL_MAX_BP = 10_000       # array is terminal if it lies within this distance of a chromosome end
FLANK_BP = 1_000               # scan distance past the inner end of terminal arrays
CHROM_ORDER = [f"chr{i}" for i in range(1, 23)] + ["chrX", "chrY"]
RC = str.maketrans("ACGT", "TGCA")
LUT = np.full(256, 4, np.uint8)
for _i, _b in enumerate(b"ACGT"):
    LUT[_b] = _i


class Fasta:
    """Minimal .fai-indexed FASTA reader (0-based, half-open, upper-case)."""

    def __init__(self, path):
        self.path, self.idx = str(path), {}
        with open(self.path + ".fai") as fh:
            for line in fh:
                n, length, off, lb, lw = line.rstrip("\n").split("\t")[:5]
                self.idx[n] = (int(length), int(off), int(lb), int(lw))

    def fetch(self, name, start, end):
        length, off, lb, lw = self.idx[name]
        start, end = max(0, start), min(end, length)
        first = off + start // lb * lw + start % lb
        last = off + (end - 1) // lb * lw + (end - 1) % lb + 1
        with open(self.path, "rb") as fh:
            fh.seek(first)
            raw = fh.read(last - first)
        return raw.replace(b"\n", b"").replace(b"\r", b"").decode("ascii").upper()


def codes(s):
    return LUT[np.frombuffer(s.encode("ascii"), np.uint8)]


def revcomp(s):
    return s.translate(RC)[::-1]


def merge(iv):
    out = []
    for s, e in sorted(iv):
        if out and s <= out[-1][1]:
            out[-1][1] = max(out[-1][1], e)
        else:
            out.append([s, e])
    return out


def ovl(a, s, e):
    return sum(max(0, min(e, y) - max(s, x)) for x, y in a if y > s and x < e)


def core_tracts(arr):
    """[(start, end, motif, n_units, n_exact)] in region coordinates."""
    out = []
    for motif in (CORE_MOTIF, revcomp(CORE_MOTIF)):
        m = codes(motif)
        for phase in range(6):
            n = (len(arr) - phase) // 6
            if n < MIN_UNITS:
                continue
            mm = (arr[phase:phase + 6 * n].reshape(n, 6) != m).sum(1)
            d = np.diff(np.r_[0, (mm <= MAX_MISMATCHES).astype(np.int8), 0])
            for s, e in zip(np.flatnonzero(d == 1), np.flatnonzero(d == -1)):
                n_exact = int((mm[s:e] == 0).sum())
                if e - s >= MIN_UNITS and (n_exact >= 1 or not ANCHOR_EXACT):
                    out.append((phase + 6 * s, phase + 6 * e, motif, int(e - s), n_exact))
    return out


def long_tracts(arr):
    """[(start, end, orientation, n_copies, mean_mismatches)] in region coordinates."""
    L, out = len(LONG_REPEAT), []
    if len(arr) < L:
        return out
    max_mm = int(np.floor(LONG_MAX_MISMATCH_FRAC * L))
    win = sliding_window_view(arr, L)
    for orient, unit in (("forward", LONG_REPEAT), ("revcomp", revcomp(LONG_REPEAT))):
        mm = (win != codes(unit)).sum(1)
        copies, last = [], -L
        for i in np.flatnonzero(mm <= max_mm):
            if i - last >= L - LONG_SPACING_TOLERANCE:
                copies.append((int(i), int(mm[i])))
                last = i
        tract = []
        for i, m in copies + [(None, None)]:
            if i is not None and tract and i <= tract[-1][0] + L + LONG_MAX_GAP_BP:
                tract.append((i, m))
                continue
            if len(tract) >= LONG_MIN_UNITS:
                out.append((tract[0][0], tract[-1][0] + L, orient, len(tract), float(np.mean([x[1] for x in tract]))))
            tract = [(i, m)] if i is not None else []
    return out


def min_hex_mismatch(seq):
    a = codes(seq)
    if len(a) < 6:
        return 6
    w = sliding_window_view(a, 6)
    return int(min((w != codes(m)).sum(1).min() for m in (CORE_MOTIF, revcomp(CORE_MOTIF))))


def long_repeat_mismatch(seq):
    """Fewest mismatches between seq and any same-length window of the tandem 29-bp repeat (both strands)."""
    a, best = codes(seq), len(seq)
    reps = LONG_REPEAT * (len(seq) // len(LONG_REPEAT) + 2)
    for t in (reps, revcomp(reps)):
        w = sliding_window_view(codes(t), len(a))[:len(LONG_REPEAT)]
        best = min(best, int((w != a).sum(1).min()))
    return best


def best_name(df, s, e, col):
    sub = df[(df.end > s) & (df.start < e)]
    if sub.empty:
        return ""
    return str(sub[col].iloc[int(np.argmax((np.minimum(sub.end, e) - np.maximum(sub.start, s)).to_numpy()))])


def main():
    for p in (TELO, CHROMSIZES, FASTA, Path(str(FASTA) + ".fai"), RMSK, CENSAT, HITS):
        if not p.exists():
            raise FileNotFoundError(p)
    fa = Fasta(FASTA)
    sizes = pd.read_csv(CHROMSIZES, sep="\t", header=None, usecols=[0, 1], names=["chrom", "length"])
    sizes = dict(zip(sizes.chrom, sizes.length.astype(np.int64)))

    telo = pd.read_csv(TELO, sep="\t", header=None, usecols=[0, 1, 2], names=["chrom", "start", "end"])
    telo = telo[telo.chrom.isin(CHROM_ORDER)].copy()
    telo["chrom_len"] = telo.chrom.map(sizes)
    telo["arm"] = np.where(telo.start <= telo.chrom_len - telo.end, "p", "q")
    telo["dist_to_end_bp"] = np.minimum(telo.start, telo.chrom_len - telo.end)
    telo["type"] = np.where(telo.dist_to_end_bp <= TERMINAL_MAX_BP, "terminal", "interstitial")
    telo["end_id"] = telo.chrom + telo.arm
    telo["array_bp"] = telo.end - telo.start
    telo["scan_start"] = np.where((telo.type == "terminal") & (telo.arm == "q"), np.maximum(0, telo.start - FLANK_BP), telo.start)
    telo["scan_end"] = np.where((telo.type == "terminal") & (telo.arm == "p"), np.minimum(telo.chrom_len, telo.end + FLANK_BP), telo.end)
    telo = telo.sort_values(["chrom", "start"]).reset_index(drop=True)

    hits = pd.read_csv(HITS, sep="\t", usecols=["chrom", "start", "end", "length", "score", "sequence"])
    hits = hits[hits.chrom.isin(set(telo.chrom))]
    rm = pd.read_csv(RMSK, sep="\t", header=None, usecols=[0, 1, 2, 3, 6], names=["chrom", "start", "end", "name", "rclass"],
                     dtype={"chrom": str, "name": str, "rclass": str})
    cen = pd.read_csv(CENSAT, sep="\t", header=None, usecols=[0, 1, 2, 3], names=["chrom", "start", "end", "name"])
    win = lambda d, r: d[(d.chrom == r.chrom) & (d.end > r.scan_start) & (d.start < r.scan_end)]

    feats, arrays = [], []
    for r in telo.itertuples():
        to_x = (lambda s, e: (s, e)) if r.arm == "p" else (lambda s, e: (r.chrom_len - e, r.chrom_len - s))
        base = {"chrom": r.chrom, "end_id": r.end_id, "array_type": r.type}
        seq = fa.fetch(r.chrom, r.scan_start, r.scan_end)
        arr = codes(seq)
        core = [(r.scan_start + s, r.scan_start + e, m, n, x) for s, e, m, n, x in core_tracts(arr)]
        lng = [(r.scan_start + s, r.scan_start + e, o, n, mm) for s, e, o, n, mm in long_tracts(arr)]
        core_m, long_m = merge([(s, e) for s, e, *_ in core]), merge([(s, e) for s, e, *_ in lng])
        for ft, s, e, extra in ([("scan_region", r.scan_start, r.scan_end, {}), ("telomere_array", r.start, r.end, {})]
                                + [("core_tract", s, e, {"motif": m, "n_units": n, "n_exact_units": x}) for s, e, m, n, x in core]
                                + [("long_repeat_tract", s, e, {"motif": o, "n_units": n, "mean_mismatches": mm})
                                   for s, e, o, n, mm in lng]):
            xs, xe = to_x(s, e)
            feats.append({**base, "feature_type": ft, "start": s, "end": e, "length_bp": e - s, "x_start": xs, "x_end": xe, **extra})
        h = win(hits, r)
        rmw, cenw = win(rm, r), win(cen, r)
        inner = r.end if r.arm == "p" else r.start
        zrows = []
        for z in h.itertuples():
            xs, xe = to_x(z.start, z.end)
            gap = (inner - z.end) if r.arm == "p" else (z.start - inner)
            zrows.append({**base, "feature_type": "zdna_prediction", "start": z.start, "end": z.end, "length_bp": z.length,
                          "x_start": xs, "x_end": xe, "score": z.score, "bp_in_array": ovl([(r.start, r.end)], z.start, z.end),
                          "bp_overlap_core": ovl(core_m, z.start, z.end), "bp_overlap_long_repeat": ovl(long_m, z.start, z.end),
                          "min_core_hexamer_mismatches": min_hex_mismatch(z.sequence),
                          "long_repeat_mismatches": long_repeat_mismatch(z.sequence),
                          "gap_to_inner_end_bp": int(gap) if r.type == "terminal" else np.nan,
                          "rmsk_name": best_name(rmw, z.start, z.end, "name"), "censat_name": best_name(cenw, z.start, z.end, "name"),
                          "sequence": z.sequence})
        zdf = pd.DataFrame(zrows)
        if len(zdf):
            zdf = zdf.sort_values("start")
            zdf["spacing_to_previous_bp"] = zdf.start.diff()
            feats += zdf.to_dict("records")
        in_arr = zdf[zdf.bp_in_array > 0] if len(zdf) else zdf
        arrays.append({"chrom": r.chrom, "start": r.start, "end": r.end, "end_id": r.end_id, "type": r.type,
                       "dist_to_end_bp": r.dist_to_end_bp, "array_bp": r.array_bp, "scan_start": r.scan_start, "scan_end": r.scan_end,
                       "core_bp_in_array": ovl(core_m, r.start, r.end), "core_frac_of_array": ovl(core_m, r.start, r.end) / r.array_bp,
                       "n_core_tracts": len(core), "long_repeat_bp_scanned": sum(e - s for s, e in long_m),
                       "long_repeat_bp_in_array": ovl(long_m, r.start, r.end),
                       "n_long_repeat_copies": int(sum(n for _, _, _, n, _ in lng)),
                       "n_zdna_scanned": len(zdf), "n_zdna_in_array": len(in_arr),
                       "zdna_bp_in_array": ovl(merge(zip(in_arr.start, in_arr.end)) if len(in_arr) else [], r.start, r.end),
                       "n_zdna_overlapping_core": int((zdf.bp_overlap_core > 0).sum()) if len(zdf) else 0,
                       "n_zdna_overlapping_long_repeat": int((zdf.bp_overlap_long_repeat > 0).sum()) if len(zdf) else 0,
                       "innermost_gap_bp": zdf.gap_to_inner_end_bp.min() if len(zdf) and r.type == "terminal" else np.nan,
                       "zdna_cluster_span_bp": int(zdf.end.max() - zdf.start.min()) if len(zdf) else 0})
    feat = pd.DataFrame(feats)
    feat.to_csv(OUT_FEATURES, sep="\t", index=False)
    arr_t = pd.DataFrame(arrays)
    arr_t.to_csv(OUT_ARRAYS, sep="\t", index=False)

    z = feat[feat.feature_type == "zdna_prediction"]
    za = z[z.bp_in_array > 0]
    term = arr_t[arr_t.type == "terminal"]
    on29 = (z.bp_overlap_long_repeat > 0) | (z.long_repeat_mismatches <= LONG_MAX_MISMATCH_FRAC * z.length_bp)
    print(f"Arrays: {len(arr_t)} ({len(term)} terminal; median terminal length {term.array_bp.median():,.0f} bp); "
          f"(TTAGGG)n tracts cover {arr_t.core_bp_in_array.sum() / arr_t.array_bp.sum():.1%} of array bp")
    print(f"Z-DNA predictions: {len(z)} scanned, {len(za)} in arrays at {za.end_id.nunique()} ends; overlapping core tracts "
          f"{(z.bp_overlap_core > 0).sum()}; with a perfect TTAGGG/CCCTAA {(z.min_core_hexamer_mismatches == 0).sum()}; "
          f"in TAR1 {(z.rmsk_name == 'TAR1').sum()}; on the 29-bp repeat {on29.sum()}")


if __name__ == "__main__":
    main()
```

#### 7.3 · Composition per compartment (text)

```python
"""7.3 - Length, score and dinucleotide-step composition of Z-DNA predictions per compartment; recurrent lengths
(e.g. the 64-bp L1 prediction) and compartment co-membership (text values for the Suppl. Fig. 2a section).

Membership as in 7.1. Steps (GC/CG, CA/TG, AT/TA, non-alternating) pooled over all adjacent bases per compartment.
Recurrent length: count >= SPIKE_RATIO x median of the +/-SPIKE_NEIGHBOURS lengths and >= SPIKE_MIN_N.
"""
import shutil
from pathlib import Path

import numpy as np
import pandas as pd
import pybedtools
from pybedtools import BedTool

BASE = Path("/path/to/pangenomes_zdna")
ANN = BASE / "tmp/chm13/chm13_marbl_ann_files/annotation"
HITS = BASE / "7_chm13_zdna_hits.tsv.gz"
COMP = BASE / "7_chm13_zdna_compartments.tsv"
ENRICH = BASE / "6_chm13_zdna_feature_enrichment.tsv"
RMSK = ANN / "chm13v2.0_RepeatMasker_4.1.2p1.2022Apr14.bed"
OUT_LENGTH = BASE / "7_chm13_zdna_length_by_feature.tsv"
OUT_COMPOSITION = BASE / "7_chm13_zdna_composition_by_compartment.tsv"
OUT_RECURRENT = BASE / "7_chm13_zdna_recurrent_segments.tsv"
OUT_COMEMBERSHIP = BASE / "7_chm13_zdna_comembership.tsv"
TMPDIR = BASE / "7_3_tmp"

GENOME = "genome_all"
TOP_N_LENGTHS = 3              # most frequent lengths reported per compartment
SPIKE_RATIO = 3.0              # count at length L / median count of neighbouring lengths
SPIKE_NEIGHBOURS = 3           # neighbouring lengths on each side
SPIKE_MIN_N = 500              # minimum predictions of length L
TOP_SEQ = 10                   # most common sequences reported per recurrent length
TOP_RMSK = 5                   # RepeatMasker elements reported per recurrent length and compartment
FOCUS = ["rmsk_LINE", "rmsk_SINE", "censat_hor", "censat_dhor", "censat_mon", "censat_ct", "rDNA_arrays"]
TEXT_ROWS = ["genome_all", "rmsk_Simple_repeat", "pc_CDS", "rDNA_arrays", "rmsk_LINE", "telomere"]
STEP_CLASSES = ["GC_CG", "CA_TG", "AT_TA", "non_alternating"]
RC = str.maketrans("ACGT", "TGCA")


def step_counts(seqs):
    """(n, 5) steps per class (last column: steps with a non-ACGT base)."""
    tab = np.full(25, 4, np.int64)
    for i, a in enumerate("ACGT"):
        for j, b in enumerate("ACGT"):
            d = a + b
            tab[i * 5 + j] = 0 if d in ("GC", "CG") else 1 if d in ("CA", "TG", "AC", "GT") else 2 if d in ("AT", "TA") else 3
    lut = np.full(256, 4, np.int64)
    for i, b in enumerate(b"ACGT"):
        lut[b] = i
    lens = np.fromiter((len(s) for s in seqs), np.int64, len(seqs))
    a = lut[np.frombuffer("|".join(seqs).encode("ascii"), np.uint8)]
    sep = np.zeros(len(a), bool)
    sep[np.cumsum(lens + 1)[:-1] - 1] = True
    keep = ~(sep[:-1] | sep[1:])
    cls = tab[a[:-1] * 5 + a[1:]][keep]
    hid = np.repeat(np.arange(len(seqs)), lens + 1)[: len(a) - 1][keep]
    return np.bincount(hid * 5 + cls, minlength=len(seqs) * 5).reshape(len(seqs), 5)


def canonical(s):
    return min(s, s.translate(RC)[::-1])


def main():
    for p in (HITS, COMP, ENRICH, RMSK):
        if not p.exists():
            raise FileNotFoundError(p)
    h = pd.read_csv(HITS, sep="\t", keep_default_na=False, dtype={"chrom": str, "sequence": str, "compartments": str})
    h["hit_id"] = np.arange(len(h))
    comp = pd.read_csv(COMP, sep="\t")
    labels = comp.set_index("compartment").label.to_dict()
    e = pd.read_csv(ENRICH, sep="\t").set_index("feature")
    cnt = step_counts(h.sequence.tolist())
    for k, c in enumerate(STEP_CLASSES + ["non_ACGT"]):
        h[f"steps_{c}"] = cnt[:, k]
    h["score_per_bp"] = h.score / h.length
    memb = h.compartments.str.split(";").apply(lambda m: [c for c in m if c])
    ex = h.assign(compartment=memb).explode("compartment")
    long = pd.concat([ex[ex.compartment.notna()], h.assign(compartment=GENOME)], ignore_index=True)
    groups = {k: g for k, g in long.groupby("compartment")}

    lrows, crows = [], []
    for r in comp.itertuples():
        g = groups.get(r.compartment)
        if g is None:
            continue
        a = g.length.to_numpy()
        vc = pd.Series(a).value_counts().sort_index().sort_values(ascending=False, kind="stable")
        q1, med, q3 = np.percentile(a, [25, 50, 75])
        lr = {"compartment": r.compartment, "label": r.label, "panel": r.panel, "n_hits": len(a), "length_min": int(a.min()),
              "length_q1": q1, "length_median": med, "length_q3": q3, "length_max": int(a.max())}
        for i in range(TOP_N_LENGTHS):
            ln, n = (int(vc.index[i]), int(vc.iloc[i])) if i < len(vc) else (np.nan, 0)
            lr.update({f"top{i + 1}_length": ln, f"top{i + 1}_count": n, f"top{i + 1}_fraction": n / len(a)})
        lrows.append(lr)
        tot = g[[f"steps_{c}" for c in STEP_CLASSES]].to_numpy().sum()
        cr = {"compartment": r.compartment, "label": r.label, "panel": r.panel, "n_hits": len(g),
              "score_median": g.score.median(), "score_q1": g.score.quantile(.25), "score_q3": g.score.quantile(.75),
              "score_max": g.score.max(), "score_per_bp_median": g.score_per_bp.median(),
              **{f"frac_steps_{c}": g[f"steps_{c}"].sum() / tot for c in STEP_CLASSES},
              "n_steps_non_ACGT": int(g.steps_non_ACGT.sum())}
        if r.compartment in e.index and r.hit_bp > 0:
            cr["frac_hit_bp_inside"] = float(e.at[r.compartment, "zdna_overlap_bp"]) / r.hit_bp
        crows.append(cr)
    lt, ct = pd.DataFrame(lrows), pd.DataFrame(crows)
    lt.to_csv(OUT_LENGTH, sep="\t", index=False)
    ct.to_csv(OUT_COMPOSITION, sep="\t", index=False)

    vc = h.length.value_counts().sort_index()
    full = vc.reindex(range(vc.index.min(), vc.index.max() + 1), fill_value=0)
    spikes = []
    for L, n in full.items():
        nb = full.reindex([L + d for d in range(-SPIKE_NEIGHBOURS, SPIKE_NEIGHBOURS + 1) if d]).dropna()
        base = float(np.median(nb)) if len(nb) else 0.0
        if n >= SPIKE_MIN_N and n >= SPIKE_RATIO * max(base, 1.0):
            spikes.append((L, int(n), base))
    rec = []
    if spikes:
        if TMPDIR.exists():
            shutil.rmtree(TMPDIR)
        TMPDIR.mkdir(parents=True)
        pybedtools.helpers.set_tempdir(str(TMPDIR))
        sp = h[h.length.isin([s[0] for s in spikes])]
        rm = pd.read_csv(RMSK, sep="\t", header=None, usecols=[0, 1, 2, 3, 6], names=["chrom", "start", "end", "name", "rclass"],
                         dtype={"chrom": str, "name": str, "rclass": str})
        o = BedTool.from_dataframe(sp[["chrom", "start", "end", "hit_id"]]).sort().intersect(
            BedTool.from_dataframe(rm[rm.chrom.isin(set(sp.chrom))]).sort(), wao=True).to_dataframe(
            disable_auto_names=True, header=None, names=["c", "s", "e", "hit_id", "c2", "s2", "e2", "rm_name", "rm_class", "ov"])
        o = o.sort_values("ov", ascending=False).drop_duplicates("hit_id")
        o.loc[o.ov == 0, ["rm_name", "rm_class"]] = ["(none)", "(none)"]
        sp = sp.merge(o[["hit_id", "rm_name", "rm_class"]], on="hit_id", how="left")
        for L, n, base in spikes:
            g = sp[sp.length == L].copy()
            g["canon"] = g.sequence.map(canonical)
            share = pd.Series([c for m in memb[g.hit_id] for c in m]).value_counts() / len(g)
            rec.append({"record_type": "length", "length": L, "n": n, "neighbour_median": base, "frac_of_all_hits": n / len(h),
                        "n_distinct_sequences": g.canon.nunique(),
                        "detail": "; ".join(f"{labels.get(c, c)} {v:.0%}" for c, v in share.head(8).items())})
            for k, (seq, m) in enumerate(g.canon.value_counts().head(TOP_SEQ).items(), 1):
                rec.append({"record_type": "sequence", "length": L, "rank": k, "n": int(m), "frac_of_length": m / len(g),
                            "score_median": g[g.canon == seq].score.median(), "detail": seq})
            for comp_key in ["all"] + FOCUS:
                gg = g if comp_key == "all" else g[g.compartments.str.contains(rf"(?:^|;){comp_key}(?:;|$)", regex=True)]
                for (nm, cl), m in gg.groupby(["rm_name", "rm_class"]).size().sort_values(ascending=False).head(TOP_RMSK).items():
                    rec.append({"record_type": "rmsk", "length": L, "compartment": labels.get(comp_key, comp_key),
                                "n_in_compartment": len(gg), "n": int(m), "frac": m / len(gg), "detail": f"{nm} ({cl})"})
        pybedtools.cleanup(remove_all=True)
        shutil.rmtree(TMPDIR, ignore_errors=True)
    rt = pd.DataFrame(rec)
    for col in ("rank", "n", "n_distinct_sequences", "n_in_compartment"):
        if col in rt:
            rt[col] = rt[col].astype("Int64")
    rt.to_csv(OUT_RECURRENT, sep="\t", index=False)

    keys = [k for k in comp.compartment if k != GENOME]
    idx = {k: i for i, k in enumerate(keys)}
    m = np.zeros((len(h), len(keys)), bool)
    for r, cs in enumerate(memb):
        for c in cs:
            m[r, idx[c]] = True
    n_a = m.sum(0)
    co = pd.DataFrame((m.T.astype(np.int64) @ m.astype(np.int64)) / np.maximum(n_a, 1)[:, None],
                      index=[labels[k] for k in keys], columns=[labels[k] for k in keys])
    co.insert(0, "n_hits", n_a)
    co.to_csv(OUT_COMEMBERSHIP, sep="\t")

    cols = ["label", "n_hits", "frac_steps_GC_CG", "frac_steps_CA_TG"]
    print("Dinucleotide steps:\n" + ct[ct.compartment.isin(TEXT_ROWS)][cols].round(3).to_string(index=False))
    if len(rt):
        print("Recurrent lengths:\n" + rt[rt.record_type == "length"][["length", "n", "frac_of_all_hits"]]
              .round(3).to_string(index=False))
    for k in ("censat_hor", "censat_dhor", "censat_mon"):
        if labels.get(k) in co.index:
            print(f"{labels[k]} predictions overlapping LINEs: {co.at[labels[k], labels['rmsk_LINE']]:.2f}")


if __name__ == "__main__":
    main()
```

#### 7.4 · Sequence-logo counts → Suppl. Fig. 2a

```python
"""7.4 - Sequence-logo counts per compartment (Suppl. Fig. 2a).

One LOGO_LEN window per prediction >= LOGO_LEN bp (centre), shifted <= 1 bp and strand-flipped to fit the alternating
pyrimidine/purine register; counts, probabilities and information content per position. Window choice is parallel.
"""
import os
from multiprocessing import get_context
from pathlib import Path

import numpy as np
import pandas as pd

BASE = Path("/path/to/pangenomes_zdna")
HITS = BASE / "7_chm13_zdna_hits.tsv.gz"
COMP = BASE / "7_chm13_zdna_compartments.tsv"
OUT = BASE / "7_chm13_zdna_logo_counts.tsv"

LOGO_LEN = 20                   # minimum prediction length = logo width (bp)
ALIGN = "center"                # "center" or "left"
REGISTER_ALIGN = True           # align windows to the YRYR... register (1-bp shift + strand choice)
REQUIRE_YR_REGISTER = False     # drop windows that do not start with a pyrimidine after orientation
SMALL_SAMPLE_CORRECTION = True  # subtract the small-sample error from the information content
GENOME = "genome_all"
N_WORKERS = min(32, os.cpu_count() or 1)
RC = str.maketrans("ACGT", "TGCA")
YR = str.maketrans("ACGT", "RYRY")


def yr_fit(w):
    y = w.translate(YR)
    return y[0::2].count("Y") + y[1::2].count("R")


def pick(seq, off, L):
    offs = [off] + ([o for o in (off + 1, off - 1) if 0 <= o <= len(seq) - L] if REGISTER_ALIGN else [])
    cands = []
    for k, o in enumerate(offs):
        w = seq[o:o + L]
        for v in (w, w.translate(RC)[::-1]):
            fit = yr_fit(v) if REGISTER_ALIGN else 2 * L * (v[0] in "CT")
            cands.append((-fit, k, -v[0::2].count("C"), v))
    best = min(cands)
    return best[3], yr_fit(best[3]) / L


def pick_chunk(args):
    seqs, offs = args
    return [pick(s, int(o), LOGO_LEN) for s, o in zip(seqs, offs)]


def members(df):
    ex = df.assign(compartment=df.compartments.str.split(";")).explode("compartment")
    return pd.concat([ex[ex.compartment != ""], df.assign(compartment=GENOME)], ignore_index=True)


def main():
    if ALIGN not in ("center", "left"):
        raise SystemExit('ALIGN must be "center" or "left".')
    L = int(LOGO_LEN)
    comp = pd.read_csv(COMP, sep="\t")
    h = pd.read_csv(HITS, sep="\t", usecols=["length", "sequence", "compartments"], keep_default_na=False)
    h = h[h.length >= L].copy()
    off = (h.length - L) // 2 if ALIGN == "center" else 0 * h.length
    seqs, offs = h.sequence.tolist(), off.tolist()
    step = -(-len(seqs) // (N_WORKERS * 4))
    chunks = [(seqs[i:i + step], offs[i:i + step]) for i in range(0, len(seqs), step)]
    with get_context("fork").Pool(N_WORKERS) as pool:
        res = [r for part in pool.map(pick_chunk, chunks) for r in part]
    h["seq"], h["fit"] = [r[0] for r in res], [r[1] for r in res]
    h["yr"] = h.seq.str[0].isin(["C", "T"])
    n_len = members(h).groupby("compartment").size()
    grp = {k: g for k, g in members(h[h.yr] if REQUIRE_YR_REGISTER else h).groupby("compartment")}

    rows = []
    for r in comp.itertuples():
        g = grp.get(r.compartment)
        if g is None:
            continue
        arr = np.frombuffer("".join(g.seq).encode("ascii"), np.uint8).reshape(len(g), L)
        cnt = np.stack([(arr == ord(b)).sum(0) for b in "ACGT"], axis=1)
        tot = cnt.sum(1, keepdims=True)
        p = cnt / tot
        ent = -np.where(p > 0, p * np.log2(np.where(p > 0, p, 1)), 0).sum(1)
        ic = np.maximum(0, 2 - ent - (3 / (2 * np.log(2) * tot[:, 0]) if SMALL_SAMPLE_CORRECTION else 0))
        for pos in range(L):
            rows.append({"compartment": r.compartment, "label": r.label, "panel": r.panel, "logo_length": L,
                         "n_hits_ge_length": int(n_len[r.compartment]), "n_used": len(g),
                         "frac_YR_register": round(float(g.yr.mean()), 4), "mean_YR_fit": round(float(g.fit.mean()), 4),
                         "position": pos + 1, "n": int(tot[pos, 0]),
                         **{f"count_{b}": int(cnt[pos, i]) for i, b in enumerate("ACGT")},
                         **{f"prob_{b}": round(p[pos, i], 6) for i, b in enumerate("ACGT")},
                         "information_bits": round(ic[pos], 6),
                         **{f"bits_{b}": round(p[pos, i] * ic[pos], 6) for i, b in enumerate("ACGT")}})
    out = pd.DataFrame(rows)
    out.to_csv(OUT, sep="\t", index=False)
    print(f"Predictions >= {L} bp: {len(h):,}; logo counts for {out.compartment.nunique()} compartments -> {OUT.name}")


if __name__ == "__main__":
    main()
```

#### 7.5 · Example predictions → Supplementary Table 2

```python
"""7.5 - Example Z-DNA predictions per compartment -> Supplementary_Table_2.tsv.

Candidates: member predictions with >= MIN_FRAC_INSIDE of their bp inside the compartment (6_chm13_features/ BEDs).
Per compartment: highest score, the N_PER_BP highest score-per-bp predictions (>= MIN_LEN bp, distinct sequences) and
the most frequent sequence (reverse complements collapsed).
"""
import shutil
from pathlib import Path

import numpy as np
import pandas as pd
import pybedtools
from pybedtools import BedTool

BASE = Path("/path/to/pangenomes_zdna")
HITS = BASE / "7_chm13_zdna_hits.tsv.gz"
COMP = BASE / "7_chm13_zdna_compartments.tsv"
FEATDIR = BASE / "6_chm13_features"
OUT_TSV = BASE / "Supplementary_Table_2.tsv"
TMPDIR = BASE / "7_5_tmp"

MIN_FRAC_INSIDE = 0.5          # share of a prediction's bp that must lie inside the compartment
MIN_LEN = 20                   # minimum length for the score-per-bp selection
N_PER_BP = 3                   # predictions selected by score per bp
DISPLAY_BP = 60                # sequence_display column truncates longer sequences
PANEL_NAME = {"tandem": "Satellites and tandem repeats", "interspersed": "Interspersed and other repeats",
              "biotypes": "Gene biotypes", "gene_parts": "Gene parts"}
RC = str.maketrans("ACGT", "TGCA")
TABLE_COLUMNS = {"panel": "Category", "compartment": "Compartment", "selection": "Selection", "chrom": "Chromosome",
                 "start_1based": "Start (1-based)", "end": "End", "length_bp": "Length (bp)",
                 "zseeker_score": "ZSeeker score", "score_per_bp": "Score per bp",
                 "frac_in_compartment": "Fraction in compartment", "identical_copies_genome": "Identical copies in genome",
                 "member_predictions": "Predictions in compartment", "also_in": "Also in",
                 "sequence_display": "Sequence (display)", "sequence": "Sequence"}


def canonical(s):
    return min(s, s.translate(RC)[::-1])


def main():
    for p in (HITS, COMP, FEATDIR):
        if not p.exists():
            raise FileNotFoundError(p)
    comps = pd.read_csv(COMP, sep="\t")
    comps = comps[comps.compartment != "genome_all"]
    compartments = list(zip(comps.compartment, comps.label, comps.panel))
    paths = {k: FEATDIR / f"{k}.bed" for k, _, _ in compartments}
    if TMPDIR.exists():
        shutil.rmtree(TMPDIR)
    TMPDIR.mkdir(parents=True)
    pybedtools.helpers.set_tempdir(str(TMPDIR))

    h = pd.read_csv(HITS, sep="\t", keep_default_na=False, dtype={"chrom": str, "sequence": str, "compartments": str})
    h["hit_id"] = np.arange(len(h))
    h["canon"] = h.sequence.map(canonical)
    h["identical_copies_genome"] = h.canon.map(h.canon.value_counts())
    h["score_per_bp"] = h.score / h.length
    hbt = BedTool.from_dataframe(h[["chrom", "start", "end", "hit_id"]]).sort()
    label = {k: lab for k, lab, _ in compartments}

    rows, skipped = [], []
    for key, lab, panel in compartments:
        if not paths[key].exists():
            skipped.append(f"{lab} (no intervals)")
            continue
        o = hbt.intersect(BedTool(str(paths[key])), wo=True).to_dataframe(
            disable_auto_names=True, header=None, names=["c", "s", "e", "hit_id", "c2", "s2", "e2", "ov"])
        if o.empty:
            skipped.append(f"{lab} (no prediction)")
            continue
        inside = o.groupby("hit_id").ov.sum()
        g = h.loc[inside.index].assign(bp_inside=inside.to_numpy())
        g["frac_inside"] = g.bp_inside / g.length
        cand = g[g.frac_inside >= MIN_FRAC_INSIDE]
        if cand.empty:
            skipped.append(f"{lab} (no prediction >= {MIN_FRAC_INSIDE:.0%} inside)")
            continue
        pick = {}
        pick.setdefault(cand.sort_values(["score", "length"], ascending=False).hit_id.iloc[0], []).append("highest score")
        for i in (cand[cand.length >= MIN_LEN].sort_values(["score_per_bp", "score"], ascending=False)
                  .drop_duplicates("canon").head(N_PER_BP).hit_id):
            pick.setdefault(i, []).append("highest score per bp")
        fc = cand.groupby("canon").agg(n=("hit_id", "size"), best=("score", "max")).sort_values(["n", "best"], ascending=False)
        rep = cand[cand.canon == fc.index[0]].sort_values("score", ascending=False).hit_id.iloc[0]
        pick.setdefault(rep, []).append(f"most frequent sequence (n = {int(fc.n.iloc[0])} in compartment)")
        for i, why in pick.items():
            r = g[g.hit_id == i].iloc[0]
            rows.append({"panel": PANEL_NAME[panel], "compartment": lab, "selection": "; ".join(why), "chrom": r.chrom,
                         "start_1based": int(r.start) + 1, "end": int(r.end), "length_bp": int(r.length),
                         "zseeker_score": float(r.score), "score_per_bp": round(float(r.score_per_bp), 3),
                         "frac_in_compartment": round(float(r.frac_inside), 3),
                         "identical_copies_genome": int(r.identical_copies_genome), "member_predictions": len(g),
                         "also_in": ", ".join(label[c] for c in r.compartments.split(";") if c and c != key),
                         "sequence_display": r.sequence if len(r.sequence) <= DISPLAY_BP else f"{r.sequence[:DISPLAY_BP]}... ({len(r.sequence)} bp)",
                         "sequence": r.sequence})
    out = pd.DataFrame(rows).rename(columns=TABLE_COLUMNS)
    out.to_csv(OUT_TSV, sep="\t", index=False)
    pybedtools.cleanup(remove_all=True)
    shutil.rmtree(TMPDIR, ignore_errors=True)
    print(f"{len(out)} examples for {out.Compartment.nunique()} compartments -> {OUT_TSV.name}; "
          f"skipped: {', '.join(skipped) or 'none'}")


if __name__ == "__main__":
    main()
```

```python
"""Fig. 2a (karyogram + legend), Fig. 2b (cytobands), Fig. 2c (CenSat), Suppl. Fig. 1d (chrX/Y ideogram + legend,
sequence-class densities). Reads the 6.1 and 6.2 tables.
"""
from pathlib import Path

import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from matplotlib.colors import to_rgb
from matplotlib.patches import FancyBboxPatch, Patch, Rectangle

BASE = Path("/path/to/pangenomes_zdna")
PLOTS, TESTS = BASE / "plots", BASE / "tests"
BINS = BASE / "6_chm13_karyogram_bins.tsv"
LABELS = BASE / "6_chm13_karyogram_labels.tsv"
TRACKS = BASE / "6_chm13_karyogram_tracks.tsv"
SUMMARY = BASE / "6_chm13_karyogram_summary.tsv"
CYTO_TSV = BASE / "6_chm13_zdna_by_cytoband.tsv"
CENSAT_TSV = BASE / "6_chm13_zdna_by_censat.tsv"
CYTO_STATS = BASE / "6_chm13_cytoband_comparisons.tsv"
CENSAT_STATS = BASE / "6_chm13_censat_comparisons.tsv"
XY_TSV = BASE / "6_chm13_zdna_by_xy_sequence_class.tsv"
STEM_KARYO, STEM_CYTO, STEM_CENSAT = "fig2a_zdna_karyogram", "fig2b_cytoband_zdna_density", "fig2c_censat_zdna_density"
STEM_XY_KARYO, STEM_XY_CLASS = "suppl_fig1d_xy_karyogram", "suppl_fig1d_xy_sequence_class_density"
COLUMNS = {"chrom": "Chromosome", "start": "Start (0-based)", "end": "End", "name": "Name", "category": "Category",
           "original_category": "Sequence class (annotation)", "plot_group": "Plotted group", "length_bp": "Length (bp)",
           "zdna_bp": "Z-DNA (bp)", "density_bp_per_kb": "Z-DNA density (bp/kb)",
           "density_bp_per_Mb": "Z-DNA density (bp/Mb)"}
STATS_COLUMNS = {"group1": "Group 1", "group2": "Group 2", "n1": "n (group 1)", "n2": "n (group 2)",
                 "median1_bp_per_kb": "Median, group 1 (bp/kb)", "median2_bp_per_kb": "Median, group 2 (bp/kb)",
                 "mean1_bp_per_kb": "Mean, group 1 (bp/kb)", "mean2_bp_per_kb": "Mean, group 2 (bp/kb)",
                 "cliffs_delta": "Cliff's delta", "u_statistic": "Mann-Whitney U", "p_value": "P value",
                 "p_adj_bh": "BH-adjusted q value"}

FIG_WIDTH = 23.4
KARYO_FIGSIZE = (8.26, 11.68)
XY_FIGSIZE = (8, 2.5)
ROW_GAP = 3
BAR_H = 0.3
TRACK_GAP = 0.02
STRIP_H = 1
RIDGE_H = 1
RIDGE_COLOR = "#513AB5"
RIDGE_ALPHA = 0.90
RIDGE_LW = 0.35
BAR_OUTLINE_LW = 1
CHROM_LABEL_FS = 13
PEAK_LABEL_FS = 9
LABEL_COLOR = "#222222"
LABEL_PAD = 0.05
LABEL_LEVEL_STEP = 0.22
MAX_LABEL_LEVELS = 2
LABEL_MIN_GAP_IN = 0.10
CHAR_WIDTH_EM = 0.58
SHOW_LEADERS = True
LEADER_COLOR = "#777777"
LEADER_LW = 0.5
TELO_LABEL_OFFSET = 0.004
LEFT_MARGIN, RIGHT_MARGIN = 0.06, 0.07
CHROM_LABEL_X = 0.012
YSCALE_FS = 10
SCALE_BAR_MB = 12
SCALE_FS = 12
XY_CHROM_LABEL_FS, XY_PEAK_LABEL_FS, XY_YSCALE_FS, XY_SCALE_FS = 10, 7, 7, 8
XY_ROW_GAP = 3.8
XY_LABEL_LEVEL_STEP = 0.55
CHROM_ROWS = [f"chr{i}" for i in range(1, 8)] + ["chrX"] + [f"chr{i}" for i in range(8, 23)] + ["chrY"]
XY_ROWS = ["chrX", "chrY"]
KARYO_LEGEND_FIGSIZE = (2.62, 3.59)
LEGEND_FS, LEGEND_TITLE_FS, LEGEND_LABELSPACING = 5.5, 6.5, 0.2
XY_LEGEND_FIGSIZE = (3.3, 1.4)
XY_LEGEND_FS, XY_LEGEND_TITLE_FS, XY_LEGEND_LABELSPACING = 7, 8, 0.5
LEGEND_BLOCK_GAP_IN = 0.06
CYTO_FIGSIZE = (4.45, 2.36)
CENSAT_FIGSIZE = (5.6, 2.5)
XY_BOX_FIGSIZE = (4.8, 2.8)
FS_AXIS, FS_TICK, FS_COUNT = 9.5, 8, 9
TICK_ROTATION = 35
BOX_WIDTH, BOX_LW, BOX_COLOR, BOX_FILL_ALPHA = 0.45, 1.0, "black", 0.85
DARK_LUMINANCE = 0.30
MEDIAN_ON_DARK = "white"
POINT_COLOR, POINT_SIZE, POINT_ALPHA, POINT_EDGE_LW = "black", 4, 0.3, 0
LIGHT_EDGE, LIGHT_LUMINANCE = "#555555", 0.75
JITTER, SEED = 0.15, 42
AXIS_WIDTH, TICK_WIDTH, TICK_LENGTH = 1, 0.8, 3
N_Y = 1.02
RASTERIZE_POINTS = True
CYTO_TICKS = [0, 2, 4, 8, 16, 32, 64, 128, 256, 512, 1024]
CENSAT_TICKS = [0, 1, 10, 100, 1000]
XY_TICKS = [0, 1, 10, 100, 1000, 10000, 100000, 1000000]
XY_POINT_SIZE, XY_POINT_ALPHA, XY_POINT_EDGE_COLOR, XY_POINT_EDGE_LW = 60, 0.95, "black", 0.4
DPI = 600
FONT = "Nimbus Sans"

STAIN = {"gneg": "#FFFFFF", "gpos25": "#C0C0C0", "gpos50": "#808080", "gpos75": "#404040", "gpos100": "#000000",
         "acen": "#B22222", "gvar": "#4682B4", "stalk": "#32CD32"}
STAIN_DESC = {"gneg": "gneg: Giemsa-negative (light) band", "gpos25": "gpos25: Giemsa-positive, 25% stain",
              "gpos50": "gpos50: Giemsa-positive, 50% stain", "gpos75": "gpos75: Giemsa-positive, 75% stain",
              "gpos100": "gpos100: Giemsa-positive, 100% stain", "acen": "acen: centromere",
              "gvar": "gvar: variable heterochromatin", "stalk": "stalk: acrocentric stalk (NOR/rDNA)"}
CEN_ORDER = ["hor", "dhor", "mon", "hsat1A", "hsat1B", "hsat2", "hsat3", "bsat", "gsat", "censat", "ct", "rDNA"]
CEN_DESC = {"hor": "HOR: alpha-satellite higher-order repeat", "dhor": "dHOR: divergent alpha-satellite HOR",
            "mon": "mon: monomeric alpha-satellite", "hsat1A": "HSat1A: human satellite 1A",
            "hsat1B": "HSat1B: human satellite 1B", "hsat2": "HSat2: human satellite 2", "hsat3": "HSat3: human satellite 3",
            "bsat": "BSat: beta-satellite", "gsat": "GSat: gamma-satellite", "censat": "censat: other satellites (e.g. SST1)",
            "ct": "CT: centromeric transition", "rDNA": "rDNA: ribosomal DNA array"}
CENSAT_NAME = {"hor": "HOR", "dhor": "dHOR", "mon": "mon", "hsat1A": "HSat1A", "hsat1B": "HSat1B", "hsat2": "HSat2",
               "hsat3": "HSat3", "bsat": "BSat", "gsat": "GSat", "censat": "censat", "ct": "CT", "rDNA": "rDNA"}
CYTO_ORDER = ["gneg", "gpos25", "gpos50", "gpos75", "gpos100", "acen", "gvar", "stalk"]
SEQ_COLORS = {"PAR": "#E6B95E", "X-degenerate": "#ABA737", "X-transposed": "#56B4E9", "Ampliconic": "#009E73",
              "Satellite": "#CC79A7", "Centromere": "#B22222", "Heterochromatin": "#999999", "Other": "#D1C7C7"}
XY_GROUP_ORDER = ["PAR chrX", "PAR chrY", "X-transposed chrX", "X-transposed chrY", "X-degenerate", "Ampliconic",
                  "Satellite", "Centromere", "Heterochromatin", "Other"]

plt.rcParams.update({"font.family": FONT, "font.size": 15, "pdf.fonttype": 42, "ps.fonttype": 42})
PLOTS.mkdir(exist_ok=True)
TESTS.mkdir(exist_ok=True)
bins = pd.read_csv(BINS, sep="\t", keep_default_na=False)
labels = pd.read_csv(LABELS, sep="\t", keep_default_na=False)
tracks = pd.read_csv(TRACKS, sep="\t", keep_default_na=False)
CAP = float(pd.read_csv(SUMMARY, sep="\t", index_col=0)["value"]["cap_zdna_bp_per_bin"])
sizes = bins.groupby("chrom").bin_end.max().to_dict()
cyto_trk = tracks[tracks.track == "cytoband"]
cen = tracks[tracks.track == "censat"]
CEN_COL = cen.drop_duplicates("category").set_index("category")["color_hex"].to_dict()


def ideogram(rows, strips, figsize, label_fig_width, out, chrom_fs=CHROM_LABEL_FS, peak_fs=PEAK_LABEL_FS,
             yscale_fs=YSCALE_FS, scale_fs=SCALE_FS, row_gap=ROW_GAP, label_step=LABEL_LEVEL_STEP):
    """Cytoband bar + Z-DNA ridge (genome-wide capped scale) + annotation strip per row; labels from 6.2."""
    n = len(rows)
    maxbp = max(sizes[c] for c in rows)
    x0, x1 = -maxbp * LEFT_MARGIN, maxbp * (1 + RIGHT_MARGIN)
    bp_per_inch = (x1 - x0) / label_fig_width
    text_w = lambda t: len(t) * CHAR_WIDTH_EM * peak_fs / 72 * bp_per_inch
    fig, ax = plt.subplots(figsize=figsize)
    unplaced = []
    for i, ch in enumerate(rows):
        L, y = sizes[ch], (n - 1 - i) * row_gap
        y0 = y + BAR_H + TRACK_GAP
        for r in cyto_trk[cyto_trk.chrom == ch].itertuples():
            ax.add_patch(Rectangle((r.start, y - BAR_H), r.end - r.start, 2 * BAR_H,
                                   facecolor=STAIN.get(r.category, "#FFFFFF"), edgecolor="none"))
        ax.add_patch(FancyBboxPatch((0, y - BAR_H), L, 2 * BAR_H, boxstyle=f"round,pad=0,rounding_size={BAR_H * .7}",
                                    fill=False, edgecolor="black", linewidth=BAR_OUTLINE_LW))
        for r in strips[strips.chrom == ch].itertuples():
            ax.add_patch(Rectangle((r.start, y - BAR_H - TRACK_GAP - STRIP_H), r.end - r.start, STRIP_H,
                                   facecolor=r.color_hex, edgecolor="none"))
        b = bins[bins.chrom == ch]
        mids = ((b.bin_start + b.bin_end) / 2).to_numpy()
        h = (b.zdna_bp_capped / CAP).to_numpy() * RIDGE_H
        ax.fill_between(mids, y0, y0 + h, color=RIDGE_COLOR, alpha=RIDGE_ALPHA, linewidth=0)
        if RIDGE_LW > 0:
            ax.plot(mids, y0 + h, color=RIDGE_COLOR, linewidth=RIDGE_LW)
        ax.text(-maxbp * CHROM_LABEL_X, y, ch[3:], ha="right", va="center", fontsize=chrom_fs)
        items = []
        for r in labels[labels.chrom == ch].itertuples():
            if r.record_type == "bin":
                hh = min(r.zdna_bp, CAP) / CAP * RIDGE_H
            else:
                nb = b.iloc[0] if r.align == "left" else b.iloc[-1]
                hh = min(nb.zdna_bp, CAP) / CAP * RIDGE_H
            items.append((float(r.x), hh, r.label_text, r.align))
        placed = {lv: [] for lv in range(MAX_LABEL_LEVELS)}
        gap_bp, off = LABEL_MIN_GAP_IN * bp_per_inch, TELO_LABEL_OFFSET * maxbp
        for x, hh, txt, ha in sorted(items, key=lambda t: t[0]):
            w = text_w(txt)
            lo, hi = {"center": (x - w / 2, x + w / 2), "left": (x + off, x + off + w), "right": (x - off - w, x - off)}[ha]
            level = next((lv for lv in range(MAX_LABEL_LEVELS)
                          if all(hi + gap_bp < a or lo - gap_bp > c for a, c in placed[lv])), None)
            if level is None:
                level = MAX_LABEL_LEVELS - 1
                unplaced.append(f"{ch}:{x:,.0f} {txt}")
            placed[level].append((lo, hi))
            ty = y0 + RIDGE_H + LABEL_PAD + level * label_step
            tx = {"center": x, "left": x + off, "right": x - off}[ha]
            if SHOW_LEADERS:
                ax.plot([x, x], [y0 + hh, ty - 0.02], color=LEADER_COLOR, linewidth=LEADER_LW, zorder=1)
            ax.text(tx, ty, txt, ha=ha, va="bottom", fontsize=peak_fs, color=LABEL_COLOR)
    ytop = (n - 1) * row_gap + BAR_H + TRACK_GAP
    xs = maxbp * 1.012
    ax.plot([xs, xs], [ytop, ytop + RIDGE_H], color="black", lw=1.2)
    for yy, txt in [(ytop, "0"), (ytop + RIDGE_H, f"{CAP / 1e3:.1f} kb")]:
        ax.plot([xs, xs + maxbp * .004], [yy, yy], color="black", lw=1.2)
        ax.text(xs + maxbp * .007, yy, txt, ha="left", va="center", fontsize=yscale_fs)
    sy = -row_gap * .75
    ax.plot([0, SCALE_BAR_MB * 1e6], [sy, sy], color="black", lw=1.4)
    ax.text(SCALE_BAR_MB * 1e6 / 2, sy - .22, f"{SCALE_BAR_MB} Mb", ha="center", va="top", fontsize=scale_fs)
    ax.set_xlim(x0, x1)
    ax.set_ylim(-row_gap * 1.25, n * row_gap)
    ax.axis("off")
    fig.subplots_adjust(left=0, right=1, top=1, bottom=0)
    for ext in ("pdf", "png"):
        fig.savefig(PLOTS / f"{out}.{ext}", dpi=DPI if ext == "png" else None, bbox_inches="tight")
    plt.close(fig)
    if unplaced:
        print(f"WARNING ({out}): overlapping labels (raise MAX_LABEL_LEVELS / ROW_GAP or lower PEAK_LABEL_FS): " + "; ".join(unplaced))


def ideogram_plot_data(rows, out):
    b = bins[bins.chrom.isin(rows)]
    t = labels[labels.chrom.isin(rows) & (labels.record_type == "telomere")]
    pd.concat([
        pd.DataFrame({"Record": "100-kb bin", "Chromosome": b.chrom, "Start (0-based)": b.bin_start, "End": b.bin_end,
                      "Length (bp)": b.bin_len, "Z-DNA (bp)": b.zdna_bp, "Z-DNA density (bp/kb)": b.zdna_bp_per_kb,
                      "Feature label": b.label_text, "Feature share of Z-DNA bp": b.label_share,
                      "Labelled in figure": b.labelled}),
        pd.DataFrame({"Record": "Terminal telomere array", "Chromosome": t.chrom, "Start (0-based)": t.bin_start,
                      "End": t.bin_end, "Length (bp)": t.bin_end - t.bin_start, "Z-DNA (bp)": t.zdna_bp,
                      "Z-DNA density (bp/kb)": t.zdna_bp / (t.bin_end - t.bin_start) * 1e3, "Feature label": "telomere",
                      "Labelled in figure": True}),
    ], ignore_index=True).to_csv(TESTS / f"{out}.tsv", sep="\t", index=False, float_format="%.6g")


def legend_figure(figsize, blocks, out, fs, title_fs):
    """blocks: (title, handles, labelspacing or None[, ncol]), stacked top-down; saved at exactly figsize."""
    lf = plt.figure(figsize=figsize)
    renderer = lf.canvas.get_renderer()
    y, x_max, y_min = 1.0, 0.0, 1.0
    for title, handles, spacing, *opt in blocks:
        kw = {"labelspacing": spacing} if spacing is not None else {}
        leg = lf.legend(handles=handles, title=title, loc="upper left", bbox_to_anchor=(0, y), frameon=False, fontsize=fs,
                        title_fontsize=title_fs, alignment="left", borderaxespad=0, ncol=opt[0] if opt else 1, **kw)
        bb = leg.get_window_extent(renderer).transformed(lf.transFigure.inverted())
        x_max, y_min = max(x_max, bb.x1), min(y_min, bb.y0)
        y = bb.y0 - LEGEND_BLOCK_GAP_IN / figsize[1]
    for ext in ("pdf", "png"):
        lf.savefig(PLOTS / f"{out}.{ext}", dpi=DPI if ext == "png" else None)
    plt.close(lf)
    if x_max > 1 or y_min < 0:
        print(f"WARNING ({out}): legend does not fit {figsize[0]} x {figsize[1]} in; lower the legend font sizes")


zdna_patch = Patch(facecolor=RIDGE_COLOR, alpha=RIDGE_ALPHA, edgecolor=RIDGE_COLOR, linewidth=RIDGE_LW, label="Z-DNA density")
patch = lambda col, lab: Patch(facecolor=col, edgecolor="black", linewidth=0.5, label=lab)


def luminance(hex_color):
    r, g, b = to_rgb(hex_color)
    return 0.2126 * r + 0.7152 * g + 0.0722 * b


def boxplot(df, order, colors, names, transform, ticks, xlabel, figsize, out, by="category", col="density_bp_per_kb",
            ylabel="Z-DNA density (bp per kb)", show_box=True, point_color=POINT_COLOR, point_edge_color=None,
            point_size=POINT_SIZE, point_alpha=POINT_ALPHA, point_edge_lw=POINT_EDGE_LW, axis_fs=FS_AXIS):
    cats = [c for c in order if (df[by] == c).any()]
    vals = [df.loc[df[by] == c, col].to_numpy(float) for c in cats]
    tvals = [transform(v) for v in vals]
    pos = np.arange(1, len(cats) + 1)
    fig, ax = plt.subplots(figsize=figsize)
    if show_box:
        bp = ax.boxplot(tvals, positions=pos, widths=BOX_WIDTH, patch_artist=True, showfliers=False, zorder=2)
        for box, c in zip(bp["boxes"], cats):
            box.set(facecolor=colors[c], edgecolor=BOX_COLOR, linewidth=BOX_LW, alpha=BOX_FILL_ALPHA, zorder=2)
        for k in ("whiskers", "caps"):
            for a in bp[k]:
                a.set(color=BOX_COLOR, linewidth=BOX_LW, zorder=2)
        for med, c in zip(bp["medians"], cats):
            med.set(color=MEDIAN_ON_DARK if luminance(colors[c]) < DARK_LUMINANCE else BOX_COLOR, linewidth=BOX_LW, zorder=4)
    rng = np.random.default_rng(SEED)
    for p, c, v in zip(pos, cats, tvals):
        color = point_color if point_color is not None else colors[c]
        ec = "none" if point_edge_lw == 0 else (point_edge_color if point_edge_color is not None else
                                                (LIGHT_EDGE if luminance(color) > LIGHT_LUMINANCE else color))
        ax.scatter(p + rng.uniform(-JITTER, JITTER, len(v)), v, s=point_size, color=color, edgecolors=ec,
                   linewidths=point_edge_lw, alpha=point_alpha, zorder=3, rasterized=RASTERIZE_POINTS)
    ymax = max(float(v.max()) for v in tvals if len(v))
    shown = [t for t in ticks if transform(np.array([t]))[0] <= ymax * 1.08 or t == ticks[0]]
    yt = shown + [t for t in ticks if t not in shown][:1]
    ax.set_yticks(transform(np.array(yt, float)), [f"{t:g}" for t in yt])
    ax.set_ylim(0, max(transform(np.array([yt[-1]], float))[0], ymax * 1.03))
    ax.set_xticks(pos, [names.get(c, c) for c in cats], rotation=TICK_ROTATION, ha="right" if TICK_ROTATION else "center",
                  rotation_mode="anchor")
    ax.set_xlim(0.4, len(cats) + 0.6)
    ax.set_ylabel(ylabel, fontsize=axis_fs)
    ax.set_xlabel(xlabel, fontsize=axis_fs)
    ax.tick_params(
        axis="both", which="major",
        labelsize=FS_TICK,
        width=TICK_WIDTH,
        length=TICK_LENGTH,
        direction="out",
        pad=3
    )

    for s in ("top", "right"):
        ax.spines[s].set_visible(False)

    for s in ("left", "bottom"):
        ax.spines[s].set_linewidth(AXIS_WIDTH)

    ax.grid(
        True,
        axis="both",
        which="major",
        linestyle="--",
        linewidth=0.4,
        color="#B0B0B0",
        alpha=0.5,
        zorder=0
    )
    ax.set_axisbelow(True)
    for p, v in zip(pos, vals):
        ax.text(p, N_Y, f"n={len(v)}", transform=ax.get_xaxis_transform(), ha="center", va="bottom", rotation=45,
                rotation_mode="anchor", fontsize=FS_COUNT)
    fig.tight_layout()
    fig.set_size_inches(*figsize)
    for ext in ("pdf", "png"):
        fig.savefig(PLOTS / f"{out}.{ext}", dpi=DPI if ext == "png" else None)
    plt.close(fig)
    d = df[df[by].isin(cats)]
    d[[c for c in COLUMNS if c in d.columns and c != ("density_bp_per_Mb" if col.endswith("kb") else "density_bp_per_kb")]] \
        .rename(columns=COLUMNS).to_csv(TESTS / f"{out}.tsv", sep="\t", index=False, float_format="%.6g")
    return dict(zip(cats, [len(v) for v in vals]))


ideogram(CHROM_ROWS, cen, KARYO_FIGSIZE, FIG_WIDTH, STEM_KARYO)
ideogram_plot_data(CHROM_ROWS, STEM_KARYO)
legend_figure(KARYO_LEGEND_FIGSIZE, [
    ("Z-DNA (upper track)", [zdna_patch], None),
    ("Cytobands, Giemsa stain (middle track)", [patch(STAIN[k], STAIN_DESC[k]) for k in STAIN], LEGEND_LABELSPACING),
    ("CenSat v2.1 annotation (lower track)", [patch(CEN_COL[k], CEN_DESC[k]) for k in CEN_ORDER if k in CEN_COL], LEGEND_LABELSPACING),
], f"{STEM_KARYO}_legend", LEGEND_FS, LEGEND_TITLE_FS)

n_cyto = boxplot(pd.read_csv(CYTO_TSV, sep="\t"), CYTO_ORDER, STAIN, {}, lambda x: np.log2(np.asarray(x) + 1), CYTO_TICKS,
                 "Cytoband stain category", CYTO_FIGSIZE, STEM_CYTO)
censat = pd.read_csv(CENSAT_TSV, sep="\t")
missing = set(censat.category) & set(CEN_ORDER) - set(CEN_COL)
if missing:
    raise RuntimeError(f"No karyogram colour for CenSat categories {missing}; rerun 6.2")
n_cen = boxplot(censat, CEN_ORDER, CEN_COL, CENSAT_NAME, lambda x: np.log10(np.asarray(x) + 1), CENSAT_TICKS,
                "CenSat category", CENSAT_FIGSIZE, STEM_CENSAT)

xy = pd.read_csv(XY_TSV, sep="\t", keep_default_na=False)
xy["color_hex"] = xy.category.map(SEQ_COLORS)
if xy.color_hex.isna().any():
    raise RuntimeError(f"No colour for sequence classes: {sorted(xy.category[xy.color_hex.isna()].unique())}")
ideogram(XY_ROWS, xy, XY_FIGSIZE, XY_FIGSIZE[0], STEM_XY_KARYO, chrom_fs=XY_CHROM_LABEL_FS, peak_fs=XY_PEAK_LABEL_FS,
         yscale_fs=XY_YSCALE_FS, scale_fs=XY_SCALE_FS, row_gap=XY_ROW_GAP, label_step=XY_LABEL_LEVEL_STEP)
ideogram_plot_data(XY_ROWS, STEM_XY_KARYO)
legend_figure(XY_LEGEND_FIGSIZE, [
    ("Z-DNA (upper track)", [zdna_patch], None),
    ("chrX/Y sequence class (lower track)", [patch(c, k) for k, c in SEQ_COLORS.items()], XY_LEGEND_LABELSPACING, 2),
], f"{STEM_XY_KARYO}_legend", XY_LEGEND_FS, XY_LEGEND_TITLE_FS)
n_xy = boxplot(xy, XY_GROUP_ORDER, {g: SEQ_COLORS[g.split()[0]] for g in XY_GROUP_ORDER}, {},
               lambda x: np.log10(np.asarray(x) + 1), XY_TICKS, "Sequence class", XY_BOX_FIGSIZE, STEM_XY_CLASS,
               by="plot_group", col="density_bp_per_Mb", ylabel="Z-DNA density (bp per Mb)", show_box=False, point_color=None,
               point_edge_color=XY_POINT_EDGE_COLOR, point_size=XY_POINT_SIZE, point_alpha=XY_POINT_ALPHA,
               point_edge_lw=XY_POINT_EDGE_LW)

for path, stem in ((CYTO_STATS, STEM_CYTO), (CENSAT_STATS, STEM_CENSAT)):
    pd.read_csv(path, sep="\t").rename(columns=STATS_COLUMNS).to_csv(TESTS / f"{stem}_statistics.tsv", sep="\t",
                                                                     index=False, float_format="%.6g")
print(f"n per category: Fig. 2b {n_cyto}; Fig. 2c {n_cen}; Suppl. Fig. 1d {n_xy}")
```

#### Suppl. Fig. 1e

```python
"""Suppl. Fig. 1e - log2 Z-DNA enrichment per feature class (bars by category) + legend. Reads 6.3.
"""
from pathlib import Path

import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from matplotlib.patches import Patch

BASE = Path("/path/to/pangenomes_zdna")
IN_TSV = BASE / "6_chm13_zdna_feature_enrichment.tsv"
PLOTS, TESTS = BASE / "plots", BASE / "tests"
STEM = "suppl_fig1e_feature_enrichment"
COLUMNS = {"label": "Feature", "category": "Category", "feature_bp": "Feature (bp)", "zdna_overlap_bp": "Z-DNA in feature (bp)",
           "feature_zdna_bp_per_Mb": "Z-DNA density in feature (bp/Mb)",
           "genome_zdna_bp_per_Mb": "Genome-wide Z-DNA density (bp/Mb)", "enrichment": "Enrichment",
           "log2_enrichment": "log2 enrichment", "bar_height_log2": "Bar height (log2, floored)"}

FIGSIZE = (8, 3)
LEGEND_FIGSIZE = (3, 1.5)
FONT, DPI = "Nimbus Sans", 600
ENRICHMENT_FLOOR = 0.06
LOG2_YMIN, LOG2_YMAX = -4.2, 4.2
Y_BOTTOM_PAD, Y_TOP_PAD = 1.9, 0.6
BAR_WIDTH, BAR_EDGE_COLOR, BAR_EDGE_LW = 0.80, "black", 0.5
FS_TICK, FS_YTICK, FS_LABEL, FS_FOLD = 9, 9, 10, 7.5
FOLD_ROTATION, FOLD_PAD_PT = -45, 1.5
AXIS_WIDTH, TICK_WIDTH, TICK_LENGTH = 1, 0.8, 3
LEGEND_FS, LEGEND_TITLE_FS, LEGEND_TITLE = 9, 10, "Feature category"
PANEL_ORDER = ["tandem_repeats", "interspersed_repeats", "gene_biotypes", "gene_parts"]
PANEL_COLORS = {"tandem_repeats": "#B4CFC7", "interspersed_repeats": "#5FBFB5", "gene_biotypes": "#7B7BC9",
                "gene_parts": "#B9A4D8"}
PANEL_TITLES = {"tandem_repeats": "Satellites & tandem repeats", "interspersed_repeats": "Interspersed & other repeats",
                "gene_biotypes": "Gene biotypes", "gene_parts": "Gene parts"}
FEATURE_MAP = {
    "censat_hsat1A": ("HSAT1A", "tandem_repeats"), "censat_hsat1B": ("HSAT1B", "tandem_repeats"),
    "censat_hsat2": ("HSAT2", "tandem_repeats"), "censat_hsat3": ("HSAT3", "tandem_repeats"),
    "censat_hor": ("HOR", "tandem_repeats"), "censat_dhor": ("dHOR", "tandem_repeats"),
    "censat_mon": ("Monomers", "tandem_repeats"), "censat_gsat": ("GSAT", "tandem_repeats"),
    "censat_bsat": ("BSAT", "tandem_repeats"), "censat_ct": ("CT", "tandem_repeats"),
    "censat_censat": ("CenSat", "tandem_repeats"), "telomere": ("Telomeres", "tandem_repeats"),
    "tRNA_arrays": ("tRNA", "tandem_repeats"), "rDNA_arrays": ("rDNA", "tandem_repeats"),
    "rmsk_LINE": ("LINEs", "interspersed_repeats"), "rmsk_SINE": ("SINEs", "interspersed_repeats"),
    "rmsk_LTR": ("LTR retr.", "interspersed_repeats"), "rmsk_DNA_transposon": ("DNA transp.", "interspersed_repeats"),
    "rmsk_Retroposon": ("Retroposons", "interspersed_repeats"), "rmsk_RC": ("Rolling-circle", "interspersed_repeats"),
    "rmsk_Simple_repeat": ("Simple repeats", "interspersed_repeats"), "rmsk_Beta": ("Beta repeats", "interspersed_repeats"),
    "rmsk_snRNA": ("snRNA repeats", "interspersed_repeats"), "rmsk_srpRNA": ("srpRNA repeats", "interspersed_repeats"),
    "rmsk_scRNA": ("scRNA repeats", "interspersed_repeats"),
    "gene_protein_coding": ("Protein-coding", "gene_biotypes"), "gene_lncRNA": ("lncRNA", "gene_biotypes"),
    "gene_pseudogene": ("Pseudogenes", "gene_biotypes"), "gene_miRNA": ("miRNA", "gene_biotypes"),
    "gene_snoRNA": ("snoRNA", "gene_biotypes"),
    "gene_body": ("Gene body", "gene_parts"), "pc_intron": ("Introns", "gene_parts"), "pc_exon": ("Exons", "gene_parts"),
    "pc_CDS": ("CDS", "gene_parts"),
}

plt.rcParams.update({
    "font.family": FONT,
    "mathtext.default": "regular",
    "axes.linewidth": AXIS_WIDTH,
    "axes.spines.top": False,
    "axes.spines.right": False,
    "pdf.fonttype": 42,
    "ps.fonttype": 42,
})
PLOTS.mkdir(exist_ok=True)
TESTS.mkdir(exist_ok=True)

df = pd.read_csv(IN_TSV, sep="\t")
df["enrichment"] = df.relative_zdna_density
df["log2_enrichment"] = np.where(df.enrichment > 0, np.log2(df.enrichment.where(df.enrichment > 0)), np.nan)
d = df[df.feature.isin(FEATURE_MAP)].copy()
d["label"] = d.feature.map(lambda f: FEATURE_MAP[f][0])
d["category"] = d.feature.map(lambda f: FEATURE_MAP[f][1])
d = pd.concat([d[d.category == p].assign(_k=d.log2_enrichment.fillna(-np.inf)).sort_values("_k") for p in PANEL_ORDER])
d = d.drop(columns="_k").reset_index(drop=True)
d["x"] = np.arange(len(d), dtype=float)
d["bar_height_log2"] = d.log2_enrichment.fillna(LOG2_YMIN).clip(LOG2_YMIN, LOG2_YMAX)
d.assign(category=d.category.map(PANEL_TITLES))[list(COLUMNS)].rename(columns=COLUMNS).to_csv(
    TESTS / f"{STEM}.tsv", sep="\t", index=False, float_format="%.6g")


def fold_label(enr):
    if enr is None or np.isnan(enr) or enr == 0:
        return "None"
    return f"<{ENRICHMENT_FLOOR:.2f}x" if enr < ENRICHMENT_FLOOR else f"{enr:.2f}x"


fig, ax = plt.subplots(figsize=FIGSIZE)
for r in d.itertuples():
    ax.bar(r.x, r.bar_height_log2, width=BAR_WIDTH, color=PANEL_COLORS[r.category], edgecolor=BAR_EDGE_COLOR,
           linewidth=BAR_EDGE_LW, zorder=2)
    kw = dict(xytext=(0, FOLD_PAD_PT), ha="right", va="bottom") if r.bar_height_log2 >= 0 else \
        dict(xytext=(0, -FOLD_PAD_PT), ha="left", va="top")
    ax.annotate(fold_label(r.enrichment), xy=(r.x, r.bar_height_log2), textcoords="offset points", rotation=FOLD_ROTATION,
                rotation_mode="anchor", fontsize=FS_FOLD, annotation_clip=False, **kw)


ax.axhline(0, color="black", lw=AXIS_WIDTH, linestyle="--", zorder=1)

ax.set_xticks(d.x.to_list())
ax.set_xticklabels(
    d.label.to_list(),
    rotation=45,
    ha="right",
    rotation_mode="anchor",
    fontsize=FS_TICK
)

ax.set_xlim(-0.6, len(d) - 0.4)
ax.set_ylabel(r"$\log_{2}$ enrichment (Z-DNA / genome)", fontsize=FS_LABEL)
ax.set_ylim(LOG2_YMIN - Y_BOTTOM_PAD, LOG2_YMAX + Y_TOP_PAD)
ax.set_yticks(np.arange(int(LOG2_YMIN), int(LOG2_YMAX) + 1, 2))

ax.tick_params(
    axis="both",
    which="major",
    width=TICK_WIDTH,
    length=TICK_LENGTH,
    direction="out",
    pad=3
)
ax.tick_params(axis="y", labelsize=FS_YTICK)
ax.tick_params(axis="x", labelsize=FS_TICK)

for s in ("top", "right"):
    ax.spines[s].set_visible(False)

for s in ("left", "bottom"):
    ax.spines[s].set_linewidth(AXIS_WIDTH)

ax.grid(
    True,
    axis="y",
    which="major",
    linestyle="--",
    linewidth=0.4,
    color="#B0B0B0",
    alpha=0.5,
    zorder=0
)
ax.set_axisbelow(True)
fig.tight_layout(pad=0.3)
for ext in ("png", "pdf"):
    fig.savefig(PLOTS / f"{STEM}.{ext}", dpi=DPI if ext == "png" else None)
plt.close(fig)

lf = plt.figure(figsize=LEGEND_FIGSIZE)
handles = [Patch(facecolor=PANEL_COLORS[p], edgecolor=BAR_EDGE_COLOR, linewidth=BAR_EDGE_LW, label=PANEL_TITLES[p])
           for p in PANEL_ORDER]
leg = lf.legend(handles=handles, title=LEGEND_TITLE, loc="upper left", bbox_to_anchor=(0, 1), frameon=False,
                fontsize=LEGEND_FS, title_fontsize=LEGEND_TITLE_FS, alignment="left", borderaxespad=0)
bb = leg.get_window_extent(lf.canvas.get_renderer()).transformed(lf.transFigure.inverted())
if bb.x1 > 1 or bb.y0 < 0:
    print(f"WARNING: legend does not fit {LEGEND_FIGSIZE} in; lower LEGEND_FS")
for ext in ("png", "pdf"):
    lf.savefig(PLOTS / f"{STEM}_legend.{ext}", dpi=DPI if ext == "png" else None)
plt.close(lf)

print(f"Genome-wide density {d.genome_zdna_bp_per_Mb.iloc[0]:,.1f} bp/Mb; {len(d)} features drawn")
```

#### Suppl. Fig. 1f

```python
"""Suppl. Fig. 1f - terminal telomere arrays with Z-DNA, aligned by distance from the chromosome end: Z-DNA
predictions, 29-bp repeat tracts and (TTAGGG)n tracts, scanned 1-kb subtelomeric flank shaded + legend. Reads 7.2.
"""
from pathlib import Path

import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from matplotlib.patches import Patch, Rectangle

BASE = Path("/path/to/pangenomes_zdna")
IN_TSV = BASE / "7_chm13_telomere_repeats_zdna.tsv"
PLOTS, TESTS = BASE / "plots", BASE / "tests"
STEM = "suppl_fig1f_telomere_arrays"
COLUMNS = {"chrom": "Chromosome", "end_id": "Chromosome end", "feature_type": "Track", "start": "Start (0-based)",
           "end": "End", "length_bp": "Length (bp)", "x_start": "Distance from chromosome end, start (bp)",
           "x_end": "Distance from chromosome end, end (bp)", "score": "ZSeeker score", "motif": "Motif / orientation",
           "n_units": "Repeat units (n)", "sequence": "Sequence"}
TRACK_NAMES = {"scan_region": "Scanned region", "telomere_array": "Telomere array", "core_tract": "(TTAGGG)n tract",
               "long_repeat_tract": "29-bp repeat tract", "zdna_prediction": "Z-DNA prediction"}

ROWS = "array"
FIGSIZE = (4.6, 2.6)
ROW_HEIGHT, EXTRA_HEIGHT = 0.42, 0.8
OUTLINE_H = 0.78
TRACK_H = 0.22
TRACK_GAP = 0.03
FLANK_COLOR = "#EDEDED"
OUTLINE_COLOR, OUTLINE_LW = "black", 0.7
ZDNA_COLOR = "#513AB5"
LONG_COLOR = "#E69F00"
CORE_COLOR = "#555555"
MIN_FEATURE_LW = 0.4
FS_TICK, FS_AXIS, FS_YTICK_MAX, FS_LEGEND = 9, 10, 9, 9
YTICK_ROW_FRAC = 0.8
SPINE_LW = 0.8
FONT, DPI = "Nimbus Sans", 600
CHROM_ORDER = [f"chr{i}" for i in range(1, 23)] + ["chrX", "chrY"]

plt.rcParams.update({"font.family": FONT, "pdf.fonttype": 42, "ps.fonttype": 42})
PLOTS.mkdir(exist_ok=True)
TESTS.mkdir(exist_ok=True)
f = pd.read_csv(IN_TSV, sep="\t", low_memory=False)
f = f[f.array_type == "terminal"]
zd = f[f.feature_type == "zdna_prediction"]
keep = {"core": zd.bp_overlap_core > 0, "array": zd.bp_in_array > 0, "scan": zd.length_bp > 0}
ends = sorted(f.end_id.unique() if ROWS == "all" else zd[keep[ROWS]].end_id.unique(),
              key=lambda e: (CHROM_ORDER.index(e[:-1]), e[-1]))
if not ends:
    raise SystemExit(f'No chromosome end has a Z-DNA prediction for ROWS = "{ROWS}"; nothing to draw.')
f = f[f.end_id.isin(ends)]
f.assign(feature_type=f.feature_type.map(TRACK_NAMES))[[c for c in COLUMNS if c in f.columns]].rename(columns=COLUMNS) \
    .to_csv(TESTS / f"{STEM}.tsv", sep="\t", index=False, float_format="%.6g")
ypos = {e: i for i, e in enumerate(ends[::-1])}
n = len(ends)
figsize = FIGSIZE if ROWS != "all" else (FIGSIZE[0], ROW_HEIGHT * n + EXTRA_HEIGHT)
fs_ytick = min(FS_YTICK_MAX, np.floor(YTICK_ROW_FRAC * (figsize[1] - 0.5) * 72 / max(n, 1) * 2) / 2)

fig, ax = plt.subplots(figsize=figsize)
for e, y in ypos.items():
    g = f[f.end_id == e]
    sc = g[g.feature_type == "scan_region"].iloc[0]
    ar = g[g.feature_type == "telomere_array"].iloc[0]
    ax.add_patch(Rectangle((sc.x_start / 1e3, y - OUTLINE_H / 2), (sc.x_end - sc.x_start) / 1e3, OUTLINE_H,
                           facecolor=FLANK_COLOR, edgecolor="none", zorder=0))
    ax.add_patch(Rectangle((ar.x_start / 1e3, y - OUTLINE_H / 2), (ar.x_end - ar.x_start) / 1e3, OUTLINE_H, fill=False,
                           edgecolor=OUTLINE_COLOR, linewidth=OUTLINE_LW, zorder=4))
    for k, (ft, col) in enumerate([("zdna_prediction", ZDNA_COLOR), ("long_repeat_tract", LONG_COLOR), ("core_tract", CORE_COLOR)]):
        t = g[g.feature_type == ft]
        lo = y + TRACK_H / 2 + TRACK_GAP - k * (TRACK_H + TRACK_GAP)
        ax.broken_barh([(a / 1e3, (b - a) / 1e3) for a, b in zip(t.x_start, t.x_end)], (lo, TRACK_H),
                       facecolors=col, edgecolor=col, linewidth=MIN_FEATURE_LW, zorder=2)
ax.set_yticks(list(ypos.values()), [f"{e[:-1]} {e[-1]}" for e in ypos], fontsize=fs_ytick)
ax.set_xlim(0, f[f.feature_type == "scan_region"].x_end.max() / 1e3 * 1.02)
ax.set_ylim(-0.6, n - 0.4)
ax.set_xlabel("Distance from chromosome end (kb)", fontsize=FS_AXIS)
ax.tick_params(axis="x", labelsize=FS_TICK, width=SPINE_LW, length=3)
ax.tick_params(axis="y", width=SPINE_LW, length=3)
for s in ("top", "right"):
    ax.spines[s].set_visible(False)
for s in ("left", "bottom"):
    ax.spines[s].set_linewidth(SPINE_LW)
fig.tight_layout(pad=0.3)
for ext in ("pdf", "png"):
    fig.savefig(PLOTS / f"{STEM}.{ext}", dpi=DPI if ext == "png" else None)
plt.close(fig)

handles = [Patch(facecolor="none", edgecolor=OUTLINE_COLOR, label="Telomere array"),
           Patch(facecolor=FLANK_COLOR, label="Scanned subtelomeric flank"),
           Patch(facecolor=ZDNA_COLOR, label="Z-DNA predictions"),
           Patch(facecolor=LONG_COLOR, label="29-bp telomere-associated repeat"),
           Patch(facecolor=CORE_COLOR, label="(TTAGGG)n / (CCCTAA)n")]
lf, lax = plt.subplots(figsize=(3, 1.1))
lax.axis("off")
lax.legend(handles=handles, loc="center", frameon=False, fontsize=FS_LEGEND, handlelength=1.4)
for ext in ("pdf", "png"):
    lf.savefig(PLOTS / f"{STEM}_legend.{ext}", dpi=DPI if ext == "png" else None, bbox_inches="tight")
plt.close(lf)
print(f"{n} chromosome ends drawn: {', '.join(ends)}")
```

#### Suppl. Fig. 2a

```python
"""Suppl. Fig. 2a - stacked sequence logos (probability) for selected compartments. Reads 7.4; requires logomaker.
"""
from pathlib import Path

import logomaker
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

BASE = Path("/path/to/pangenomes_zdna")
TABLE = BASE / "7_chm13_zdna_logo_counts.tsv"
PLOTS, TESTS = BASE / "plots", BASE / "tests"
STEM = "suppl_fig2a_zdna_sequence_logos"
COLUMNS = {"label": "Compartment", "n_used": "Predictions (n)", "position": "Position", **{f"count_{b}": f"Count {b}" for b in "ACGT"},
           **{f"prob_{b}": f"Probability {b}" for b in "ACGT"}, "information_bits": "Information (bits)"}

FEATURES = ["All Z-DNA", "Simple repeats", "CDS", "rDNA", "LINEs", "Telomeres"]
ROW_NAMES = {"All Z-DNA": "ALL"}
LOGO_MODE = "probability"
PSEUDO, ENRICH_YLIM = 0.5, 2.0
FIGSIZE = (5, 3.6)
H_PAD, H_SPACE = 0.03, 0.04
SPINE_LW, TICK_W, TICK_LEN = 1.0, 1.0, 3
FS_TICK, FS_AXIS, FS_LABEL = 8, 9.5, 8.5
LABEL_PAD_PT, X_STEP = 1.5, 5
SHOW_N = True
COLOR_SCHEME = "classic"
FONT, DPI = "Nimbus Sans", 600

plt.rcParams.update({"font.family": FONT, "mathtext.default": "regular", "pdf.fonttype": 42, "ps.fonttype": 42})
PLOTS.mkdir(exist_ok=True)
TESTS.mkdir(exist_ok=True)
tab = pd.read_csv(TABLE, sep="\t")
missing = [f for f in FEATURES if f not in set(tab.label)]
if missing:
    raise SystemExit(f"Not in the 7.4 table: {missing}. Available: {sorted(tab.label.unique())}")
L = int(tab.logo_length.iloc[0])
ACGT = list("ACGT")
ref = tab[tab.compartment == "genome_all"].sort_values("position")[[f"prob_{b}" for b in ACGT]].to_numpy()
ylim, yticks, ylab = {"bits": ((0, 2), [0, 1, 2], "Bits"), "probability": ((0, 1), [0, 0.5, 1], "Probability"),
                      "enrichment": ((-ENRICH_YLIM, ENRICH_YLIM), [-ENRICH_YLIM, 0, ENRICH_YLIM], "log$_2$ enrichment")}[LOGO_MODE]
tab[tab.label.isin(FEATURES)][list(COLUMNS)].rename(columns=COLUMNS).to_csv(TESTS / f"{STEM}.tsv", sep="\t", index=False)

fig, axs = plt.subplots(len(FEATURES), 1, figsize=FIGSIZE, sharex=True, layout="constrained")
fig.get_layout_engine().set(h_pad=H_PAD, hspace=H_SPACE, w_pad=0.03)
for i, (ax, feat) in enumerate(zip(axs, FEATURES)):
    d = tab[tab.label == feat].sort_values("position")
    if LOGO_MODE == "enrichment":
        c = d[[f"count_{b}" for b in ACGT]].to_numpy() + PSEUDO
        val = np.log2(c / c.sum(1, keepdims=True) / ref)
    else:
        val = d[[f'{"bits" if LOGO_MODE == "bits" else "prob"}_{b}' for b in ACGT]].to_numpy()
    logomaker.Logo(pd.DataFrame(val, index=d.position.to_numpy(), columns=ACGT), ax=ax, color_scheme=COLOR_SCHEME,
                   show_spines=False, flip_below=False)
    if LOGO_MODE == "enrichment":
        ax.axhline(0, color="black", lw=SPINE_LW * 0.5)
    name = ROW_NAMES.get(feat, feat) + (f" (n = {int(d.n_used.iloc[0]):,})" if SHOW_N else "")
    ax.annotate(name, xy=(0, 1), xycoords="axes fraction", xytext=(0, LABEL_PAD_PT), textcoords="offset points",
                ha="left", va="bottom", fontsize=FS_LABEL)
    ax.set_ylim(*ylim)
    ax.set_yticks(yticks, [f"{t:g}" for t in yticks])
    ax.set_xlim(0.5, L + 0.5)
    for s in ("top", "right"):
        ax.spines[s].set_visible(False)
    for s in ("left", "bottom"):
        ax.spines[s].set_visible(True)
        ax.spines[s].set_linewidth(SPINE_LW)
    ax.tick_params(labelsize=FS_TICK, width=TICK_W, length=TICK_LEN, pad=2)
    ax.tick_params(axis="x", bottom=(i == len(FEATURES) - 1), labelbottom=(i == len(FEATURES) - 1))
axs[-1].set_xticks(np.r_[1, np.arange(X_STEP, L + 1, X_STEP)] if X_STEP > 1 else np.arange(1, L + 1))
axs[-1].set_xlabel("Position", fontsize=FS_AXIS)
fig.supylabel(ylab, fontsize=FS_AXIS)
for ext in ("pdf", "png"):
    fig.savefig(PLOTS / f"{STEM}.{ext}", dpi=DPI if ext == "png" else None)
plt.close(fig)
info = tab[tab.label.isin(FEATURES)].drop_duplicates("label").set_index("label").n_used
print(f"Window {L} bp; mode {LOGO_MODE}; rows: " + ", ".join(f"{f} (n = {int(info[f]):,})" for f in FEATURES))
```

### Analysis

#### 8.2 · Matched vs unmatched loci → Suppl. Fig. 2c

```python
"""8.2 - Matched vs unmatched Z-DNA loci and their controls (CHM13, TOL / DINUC_MODE; Suppl. Fig. 2c inputs).

Per set (Z_matched, Z_unmatched, control): composition (length, GC, CpG+GpC, score), control matching error, genomic
occupancy by repeat context, and variant density inside each set (rate ratios with chromosome-bootstrap CIs).
"""
import time
from pathlib import Path

import numpy as np
import pandas as pd
import pyarrow as pa
import pyarrow.compute as pc
import pyarrow.csv as pacsv
from scipy.stats import binomtest, mannwhitneyu, wilcoxon

BASE = Path("/path/to/pangenomes_zdna")
MASTER = BASE / "8_CHM13_controls_MASTER.tsv"
MUT_META = BASE / "9_hprc-v2.0-mc-chm13.wave.normalized.mutations.meta.tsv.gz"
FAI = BASE / "tmp/chm13/chm13_refseq/GCF_009914755.1_T2T-CHM13v2.0_genomic_ucsc.fna.fai"
FEATDIR = BASE / "tmp/chm13/chm13_marbl_ann_files/annotation/enrichment_investigation/features"   # as 16 / 17
OUT_LOCI = BASE / "8_CHM13_matching_loci.tsv.gz"
OUT_COMP = BASE / "8_CHM13_matching_composition.tsv"
OUT_ERR = BASE / "8_CHM13_matching_error.tsv"
OUT_OCC = BASE / "8_CHM13_matching_occupancy.tsv"
OUT_VAR = BASE / "8_CHM13_matching_variant_density.tsv"

TOL, DINUC_MODE = "10%", "CpG_GpC"
ERR_BIN = 0.01                 # histogram bin width of the signed relative error (fraction)
LENGTH_CLASSES = [(1, 20), (21, 50), (51, 10**9)]   # bp, inclusive
DEDUP_NORMALIZED = True
N_BOOTS = 2000
SEED = 42
SETS = ["Z_matched", "Z_unmatched", "Z_non_acgt", "control"]
ACTIVE = list(SETS)            # sets present in the data (Z_non_acgt is empty for CHM13, which has no non-ACGT bases)
FEATURES = ["pc_CDS", "pc_exon", "pc_intron", "gene_body", "censat_ct", "censat_censat", "rmsk_LINE", "rmsk_SINE", "rmsk_LTR",
            "rmsk_DNA_transposon", "rmsk_Simple_repeat", "rmsk_Low_complexity", "rmsk_Satellite", "rDNA_arrays",
            "unannotated_intergenic"]
CONTEXTS = {"simple_repeat": ["rmsk_Simple_repeat"], "low_complexity": ["rmsk_Low_complexity"],
            "satellite": ["rmsk_Satellite", "censat_censat", "telomere"],
            "transposable_element": ["rmsk_LINE", "rmsk_SINE", "rmsk_LTR", "rmsk_DNA_transposon", "rmsk_Retroposon"]}
CTX_NAMES = list(CONTEXTS) + ["non_repeat"]
CLASSES = ["SNV", "MNV", "small_del", "small_ins", "large_indel", "complex_del", "complex_ins"]
CHROMS = [f"chr{i}" for i in range(1, 23)] + ["chrX", "chrY"]
STRUCT = ["snp", "mnp", "substitution", "del", "ins", "complex"]
DOM = ["snp", "mnp", "substitution", "del", "ins", "del_like", "ins_like", "balanced"]
ARROW_THREADS, BLOCK_SIZE = 16, 1 << 26


def merge(s, e):
    if len(s) == 0:
        return np.empty(0, np.int64), np.empty(0, np.int64)
    o = np.argsort(s, kind="stable"); s, e = np.asarray(s)[o], np.asarray(e)[o]
    new = np.ones(len(s), bool); new[1:] = s[1:] > np.maximum.accumulate(e)[:-1]
    return s[new].astype(np.int64), np.maximum.reduceat(e, np.flatnonzero(new)).astype(np.int64)


def intersect(a, b):
    (as_, ae), (bs, be) = a, b
    if len(as_) == 0 or len(bs) == 0:
        return np.empty(0, np.int64), np.empty(0, np.int64)
    lo = np.searchsorted(be, as_, side="right")
    n = np.maximum(np.searchsorted(bs, ae, side="left") - lo, 0)
    i = np.repeat(np.arange(len(as_)), n)
    j = np.arange(n.sum()) + np.repeat(lo - (np.cumsum(n) - n), n)
    s, e = np.maximum(as_[i], bs[j]), np.minimum(ae[i], be[j])
    return s[e > s], e[e > s]


def subtract(a, b):
    (as_, ae), (bs, be) = a, b
    if len(bs) == 0:
        return as_, ae
    pts = np.r_[as_, ae, bs, be]
    w = np.r_[np.ones(len(as_)), -np.ones(len(ae)), -2 * np.ones(len(bs)), 2 * np.ones(len(be))]
    o = np.lexsort((w, pts)); pts, w = pts[o], w[o]
    cov = np.cumsum(w)
    seg = (cov[:-1] == 1) & (pts[1:] > pts[:-1])
    return merge(pts[:-1][seg], pts[1:][seg])


def inside(p, iv):
    s, e = iv
    if len(s) == 0:
        return np.zeros(len(p), bool)
    i = np.searchsorted(s, p, side="right") - 1
    return (i >= 0) & (p < e[np.maximum(i, 0)])


def bp(iv):
    return int((iv[1] - iv[0]).sum())


def load_bed(name):
    p = FEATDIR / f"{name}.merged.bed"
    if not p.exists():
        print(f"  [missing] {p}")
        return {}
    d = pd.read_csv(p, sep="\t", header=None, usecols=[0, 1, 2], names=["chrom", "s", "e"], dtype={"chrom": str})
    return {c: merge(g.s.to_numpy(np.int64), g.e.to_numpy(np.int64)) for c, g in d.groupby("chrom") if c in CHROMS}


def union(beds):
    out = {}
    for c in CHROMS:
        parts = [b[c] for b in beds if b and c in b]
        if parts:
            out[c] = merge(np.concatenate([p[0] for p in parts]), np.concatenate([p[1] for p in parts]))
    return out


def empty():
    return np.empty(0, np.int64), np.empty(0, np.int64)


def cliffs_delta(x, y):
    u = mannwhitneyu(x, y, alternative="two-sided").statistic
    return 2 * u / (len(x) * len(y)) - 1


def mh_rr(n1, n0, e1, e0):
    """Mantel-Haenszel rate ratio (set 1 vs set 0) over the last axis (contexts)."""
    T = e1 + e0
    ok = T > 0
    num, den = (n1 * e0 / np.where(ok, T, 1))[ok].sum(), (n0 * e1 / np.where(ok, T, 1))[ok].sum()
    return num / den if num > 0 and den > 0 else np.nan


def crude_rr(n1, n0, e1, e0):
    return (n1.sum() / e1.sum()) / (n0.sum() / e0.sum()) if e1.sum() > 0 and e0.sum() > 0 and n0.sum() > 0 else np.nan


def boot_ci(f, arrays, rng):
    """arrays: (n_chrom, n_ctx); resample chromosomes with replacement."""
    k = arrays[0].shape[0]
    idx = rng.integers(0, k, size=(N_BOOTS, k))
    v = np.array([f(*[a[i].sum(0) for a in arrays]) for i in idx])
    v = v[np.isfinite(v)]
    return tuple(np.percentile(v, [2.5, 97.5])) if len(v) else (np.nan, np.nan)


def load_master():
    m = pd.read_csv(MASTER, sep="\t", low_memory=False, dtype={"tol_level": str, "dinuc_mode": str, "Chromosome": str,
                                                                "has_control": str, "search_stage": str})
    m = m[(m.tol_level == TOL) & (m.dinuc_mode == DINUC_MODE) & m.Chromosome.isin(CHROMS)].copy()
    if m.duplicated(["Chromosome", "Start0", "End0"]).any():
        raise RuntimeError("MASTER: duplicated Z-DNA locus within one condition")
    hc = m.has_control.str.strip() == "True"
    st = m.search_stage.str.strip()
    m["set"] = np.select([hc, st == "exhausted_no_match", st == "target_non_acgt"],
                         ["Z_matched", "Z_unmatched", "Z_non_acgt"], default="other")
    if (m.set == "other").any():
        raise RuntimeError(f"MASTER: loci without control and unexpected search_stage: "
                           f"{m.loc[m.set == 'other', 'search_stage'].value_counts().to_dict()}")
    m["length"] = (m.End0 - m.Start0).astype(np.int64)
    m["score"] = m["Z-DNA Score"].astype(float)
    m["score_per_bp"] = m.score / m.length
    m["gc"] = pd.to_numeric(m.GC_pct, errors="coerce")
    m["dinuc"] = pd.to_numeric(m.Dinuc_density, errors="coerce")
    z = m[["Chromosome", "Start0", "End0", "set", "length", "score", "score_per_bp", "gc", "dinuc", "search_stage"]].rename(
        columns={"Chromosome": "chrom", "Start0": "start0", "End0": "end0"})
    c = m[hc].copy()
    c = pd.DataFrame({"chrom": c.Chromosome, "start0": c.Ctrl_Start.astype(float).astype(np.int64),
                      "end0": c.Ctrl_End.astype(float).astype(np.int64), "set": "control", "length": c.length,
                      "score": np.nan, "score_per_bp": np.nan, "gc": pd.to_numeric(c.Ctrl_GC_pct, errors="coerce"),
                      "dinuc": pd.to_numeric(c.Ctrl_Dinuc_density, errors="coerce"), "search_stage": c.search_stage,
                      "target_gc": c.gc, "target_dinuc": c.dinuc, "z_start0": c.Start0})
    if ((c.end0 - c.start0) != c.length).any():
        raise RuntimeError("MASTER: control length != Z-DNA length")
    loci = pd.concat([z, c], ignore_index=True)
    print(f"[loci] {TOL} {DINUC_MODE}: " + ", ".join(f"{s} {int((loci.set == s).sum()):,}" for s in SETS))
    return loci


def load_variants():
    pa.set_cpu_count(ARROW_THREADS)
    types = {"seqID": pa.string(), "normalized_start0": pa.int64(), "normalized_alt": pa.string(), "struct_class": pa.string(),
             "dominant_op": pa.string(), "ref_span": pa.int64(), "event_size": pa.int64()}
    reader = pacsv.open_csv(MUT_META, read_options=pacsv.ReadOptions(block_size=BLOCK_SIZE),
                            parse_options=pacsv.ParseOptions(delimiter="\t"),
                            convert_options=pacsv.ConvertOptions(include_columns=list(types), column_types=types,
                                                                 strings_can_be_null=False))
    keys = ["ch", "s0", "rs", "L", "sc", "dom", "h"]
    parts, t0 = {k: [] for k in keys}, time.time()
    for b in reader:
        chi = pc.index_in(b.column("seqID"), value_set=pa.array(CHROMS))
        b, chi = b.filter(pc.is_valid(chi)), chi.filter(pc.is_valid(chi))
        if b.num_rows == 0:
            continue
        sc = pc.index_in(b.column("struct_class"), value_set=pa.array(STRUCT))
        dom = pc.index_in(b.column("dominant_op"), value_set=pa.array(DOM))
        if sc.null_count or dom.null_count:
            raise RuntimeError("meta: unexpected struct_class / dominant_op values")
        col = lambda n: b.column(n).to_numpy()
        for k, v in zip(keys, (chi.to_numpy(zero_copy_only=False).astype(np.int8), col("normalized_start0"), col("ref_span"),
                               col("event_size"), sc.to_numpy(zero_copy_only=False).astype(np.int8),
                               dom.to_numpy(zero_copy_only=False).astype(np.int8),
                               pd.util.hash_array(b.column("normalized_alt").to_numpy(zero_copy_only=False)))):
            parts[k].append(v)
    v = {k: np.concatenate(x) for k, x in parts.items()}
    if DEDUP_NORMALIZED:
        o = np.lexsort((v["h"], v["rs"], v["s0"], v["ch"]))
        v = {k: x[o] for k, x in v.items()}
        new = np.ones(len(o), bool)
        new[1:] = ((v["ch"][1:] != v["ch"][:-1]) | (v["s0"][1:] != v["s0"][:-1]) | (v["rs"][1:] != v["rs"][:-1])
                   | (v["h"][1:] != v["h"][:-1]))
        v = {k: x[new] for k, x in v.items()}
    sc, dom, L = v["sc"], v["dom"], v["L"]
    is_ = lambda x: sc == STRUCT.index(x)
    cdel = is_("complex") & (dom == DOM.index("del_like"))
    cls = np.full(len(sc), -1, np.int8)
    for name, mask in (("SNV", is_("snp")), ("MNV", is_("mnp")), ("small_del", is_("del") & (L <= 50)),
                       ("small_ins", is_("ins") & (L <= 50)), ("large_indel", (is_("del") | is_("ins")) & (L > 50)),
                       ("complex_del", cdel), ("complex_ins", is_("complex") & (dom == DOM.index("ins_like")))):
        cls[mask] = CLASSES.index(name)
    pos = np.where(is_("mnp") | ((is_("substitution") | is_("del") | cdel) & (L <= 50)), v["s0"] + (v["rs"] - 1) // 2, v["s0"])
    keep = cls >= 0
    print(f"[variants] {int(keep.sum()):,} distinct normalized alleles in the {len(CLASSES)} classes ({time.time() - t0:.0f}s)")
    return v["ch"][keep], pos[keep], cls[keep]


def composition(loci):
    rows = []
    for var in ("length", "gc", "dinuc", "score", "score_per_bp"):
        for s in ACTIVE + ["Z_all"]:
            x = loci.loc[loci.set.isin(["Z_matched", "Z_unmatched", "Z_non_acgt"]) if s == "Z_all" else loci.set == s, var].dropna()
            if len(x):
                rows.append({"record": "summary", "variable": var, "set": s, "n": len(x), "median": x.median(),
                             "q1": x.quantile(.25), "q3": x.quantile(.75), "mean": x.mean(), "min": x.min(), "max": x.max()})
        a = loci.loc[loci.set == "Z_matched", var].dropna()
        b = loci.loc[loci.set == "Z_unmatched", var].dropna()
        if len(a) and len(b):
            rows.append({"record": "test", "variable": var, "set": "Z_unmatched_vs_Z_matched", "n": len(a) + len(b),
                         "test": "Mann-Whitney U, two-sided", "p": mannwhitneyu(b, a, alternative="two-sided").pvalue,
                         "cliffs_delta": cliffs_delta(b, a)})
    return pd.DataFrame(rows)


def length_class(L):
    lab = np.full(len(L), "", dtype=object)
    for lo, hi in LENGTH_CLASSES:
        lab[(L >= lo) & (L <= hi)] = f"{lo}-{hi}bp" if hi < 10**9 else f">{lo - 1}bp"
    return lab


def matching_error(loci):
    c = loci[loci.set == "control"].copy()
    tol = float(TOL.rstrip("%")) / 100
    rows = []
    for var, tgt in (("gc", "target_gc"), ("dinuc", "target_dinuc")):
        e = np.where(c[tgt] > 0, (c[var] - c[tgt]) / c[tgt].where(c[tgt] > 0), 0.0)
        c[f"err_{var}"] = e
        for lab, sel in [("all", np.ones(len(c), bool))] + [(k, length_class(c.length.to_numpy()) == k)
                                                          for k in pd.unique(length_class(c.length.to_numpy()))]:
            x = e[sel]
            x = x[np.isfinite(x)]
            nz = x[np.abs(x) > 1e-12]
            below, above = int((nz < 0).sum()), int((nz > 0).sum())
            rows.append({"record": "summary", "variable": var, "length_class": lab, "n": len(x),
                         "frac_below": below / len(x) if len(x) else np.nan,
                         "frac_equal": 1 - len(nz) / len(x) if len(x) else np.nan,
                         "frac_above": above / len(x) if len(x) else np.nan,
                         "median_error": np.median(x) if len(x) else np.nan, "mean_error": x.mean() if len(x) else np.nan,
                         "max_abs_error": np.abs(x).max() if len(x) else np.nan,
                         "p_wilcoxon": wilcoxon(nz).pvalue if len(nz) > 10 else np.nan,
                         "p_sign": binomtest(below, below + above).pvalue if below + above else np.nan})
        edges = np.round(np.arange(-tol, tol + ERR_BIN / 2, ERR_BIN), 10)
        nzv = e[np.isfinite(e) & (np.abs(e) > 1e-12)]
        h, _ = np.histogram(np.clip(nzv, -tol, tol), bins=edges)
        rows += [{"record": "histogram", "variable": var, "length_class": "all", "bin_lo": lo, "bin_hi": hi, "n": int(k)}
                 for lo, hi, k in zip(edges[:-1], edges[1:], h)]
        rows.append({"record": "histogram", "variable": var, "length_class": "all", "bin_lo": 0.0, "bin_hi": 0.0,
                     "n": int((np.abs(e[np.isfinite(e)]) <= 1e-12).sum())})
    return pd.DataFrame(rows), c[["err_gc", "err_dinuc"]]


def set_intervals(loci):
    return {s: {c: merge(g.start0.to_numpy(np.int64), g.end0.to_numpy(np.int64)) for c, g in loci[loci.set == s].groupby("chrom")}
            for s in ACTIVE}


def occupancy(iv, clen):
    genome = {c: (np.array([0], np.int64), np.array([L], np.int64)) for c, L in clen.items()}
    allsets = dict(iv, genome=genome)
    tot = {s: sum(bp(x) for x in v.values()) for s, v in allsets.items()}
    beds = {f: load_bed(f) for f in set(FEATURES) | {x for v in CONTEXTS.values() for x in v}}
    ctx_iv, claimed = {}, {}
    for ctx, files in CONTEXTS.items():
        u = union([beds.get(f) for f in files])
        ctx_iv[ctx] = {c: subtract(u[c], claimed[c]) if c in claimed else u[c] for c in u}
        claimed = union([claimed, u])
    ctx_iv["non_repeat"] = {c: subtract(genome[c], claimed[c]) if c in claimed else genome[c] for c in genome}
    rows = []
    for section, regions in (("context", ctx_iv), ("feature", {f: beds[f] for f in FEATURES if beds.get(f)})):
        for name, reg in regions.items():
            for s, v in allsets.items():
                b = sum(bp(intersect(v[c], reg[c])) for c in v if c in reg)
                rows.append({"section": section, "name": name, "set": s, "bp": b, "set_bp": tot[s],
                             "frac_of_set_bp": b / tot[s] if tot[s] else np.nan})
    return pd.DataFrame(rows), ctx_iv


def locus_context(loci, ctx_iv):
    mid = ((loci.start0 + loci.end0) // 2).to_numpy()
    out = np.full(len(loci), "non_repeat", dtype=object)
    ch = loci.chrom.to_numpy()
    for name in CONTEXTS:
        hit = np.zeros(len(loci), bool)
        for c, iv in ctx_iv[name].items():
            m = (ch == c) & (out == "non_repeat") & ~hit
            hit[m] = inside(mid[m], iv)
        out[hit & (out == "non_repeat")] = name
    return out


def variant_density(iv, ctx_iv, ch, pos, cls, rng):
    n = {s: np.zeros((len(CHROMS), len(CTX_NAMES), len(CLASSES))) for s in ACTIVE}
    e = {s: np.zeros((len(CHROMS), len(CTX_NAMES))) for s in ACTIVE}
    for ci, c in enumerate(CHROMS):
        vm = ch == ci
        p, k = pos[vm], cls[vm]
        for s in ACTIVE:
            if c not in iv[s]:
                continue
            for xi, ctx in enumerate(CTX_NAMES):
                if c not in ctx_iv[ctx]:
                    continue
                piece = merge(*intersect(iv[s][c], ctx_iv[ctx][c]))
                e[s][ci, xi] = bp(piece)
                if len(piece[0]):
                    np.add.at(n[s][ci, xi], k[inside(p, piece)], 1)
    rows = []
    for s in ACTIVE:
        for k, c in enumerate(CLASSES):
            rows.append({"record": "density", "set": s, "class": c, "n_variants": int(n[s][..., k].sum()),
                         "bp": int(e[s].sum()), "per_kb": n[s][..., k].sum() / e[s].sum() * 1e3 if e[s].sum() else np.nan})
    for a, b in (("Z_unmatched", "Z_matched"), ("Z_matched", "control"), ("Z_unmatched", "control")):
        for k, c in enumerate(CLASSES):
            arrs = [n[a][..., k], n[b][..., k], e[a], e[b]]
            for adj, f in (("crude", crude_rr), ("MH_context", mh_rr)):
                est = f(*[x.sum(0) for x in arrs])
                lo, hi = boot_ci(f, arrs, rng)
                rows.append({"record": "rate_ratio", "set": f"{a}_vs_{b}", "class": c, "adjustment": adj, "estimate": est,
                             "ci_lo": lo, "ci_hi": hi, "n_variants": int(arrs[0].sum() + arrs[1].sum())})
    return pd.DataFrame(rows)


def main():
    t0 = time.time()
    rng = np.random.default_rng(SEED)
    fai = pd.read_csv(FAI, sep="\t", header=None, usecols=[0, 1], names=["chrom", "len"])
    clen = {c: int(l) for c, l in zip(fai.chrom, fai.len) if c in CHROMS}
    loci = load_master()
    ACTIVE[:] = [s for s in SETS if (loci.set == s).any()]
    if len(ACTIVE) < len(SETS):
        print(f"[sets] empty and left out: {', '.join(x for x in SETS if x not in ACTIVE)}")

    comp = composition(loci)
    comp.to_csv(OUT_COMP, sep="\t", index=False, float_format="%.6g")
    err, errcols = matching_error(loci)
    err.to_csv(OUT_ERR, sep="\t", index=False, float_format="%.6g")
    loci.loc[errcols.index, ["err_gc", "err_dinuc"]] = errcols.to_numpy()

    iv = set_intervals(loci)
    occ, ctx_iv = occupancy(iv, clen)
    occ.to_csv(OUT_OCC, sep="\t", index=False, float_format="%.6g")
    loci["context"] = locus_context(loci, ctx_iv)
    loci["z_start0"] = loci.z_start0.astype("Int64")
    loci.drop(columns=["target_gc", "target_dinuc"]).to_csv(OUT_LOCI, sep="\t", index=False, float_format="%.6g")

    ch, pos, cls = load_variants()
    dens = variant_density(iv, ctx_iv, ch, pos, cls, rng)
    dens.to_csv(OUT_VAR, sep="\t", index=False, float_format="%.6g")

    t = comp[comp.record == "test"]
    print("Z_unmatched vs Z_matched (Mann-Whitney U):\n" + t[["variable", "p", "cliffs_delta"]]
          .to_string(index=False, float_format=lambda x: f"{x:.3g}"))
    print(f"Outputs: {', '.join(p.name for p in (OUT_LOCI, OUT_COMP, OUT_ERR, OUT_OCC, OUT_VAR))} [{time.time() - t0:.0f}s]")


if __name__ == "__main__":
    main()
```

#### 8.3 · Composition tests → Suppl. Fig. 2c

```python
"""8.3 - Composition tests of matched Z-DNA vs controls (paired Wilcoxon, rank-biserial r; Mann-Whitney U, Cliff's
delta) and unmatched vs matched Z-DNA (Mann-Whitney U, Cliff's delta) for GC fraction, CpG+GpC density, length
(Suppl. Fig. 2c statistics).
"""
from pathlib import Path

import numpy as np
import pandas as pd
from scipy.stats import mannwhitneyu, rankdata, wilcoxon

BASE = Path("/path/to/pangenomes_zdna")
LOCI = BASE / "8_CHM13_matching_loci.tsv.gz"             # unmatched loci (8.2)
MASTER = BASE / "8_CHM13_controls_MASTER.tsv"              # matched pairs (both members at full precision)
OUT_TESTS = BASE / "8_CHM13_matching_tests.tsv"

TOL, DINUC_MODE = "10%", "CpG_GpC"
METRICS = {"gc": "GC fraction", "dinuc": "CpG+GpC density", "length": "Length (bp)"}
BENCH = [(0.147, "negligible"), (0.33, "small"), (0.474, "medium"), (np.inf, "large")]
CHROMS = [f"chr{i}" for i in range(1, 23)] + ["chrX", "chrY"]


def label(x):
    return next(t for lim, t in BENCH if abs(x) < lim) if np.isfinite(x) else ""


def summary(x):
    return {"median": np.median(x), "q1": np.percentile(x, 25), "q3": np.percentile(x, 75)}


def paired(z, c):
    d = c - z
    nz = d[d != 0]
    out = {"n_pairs": len(d), "frac_identical": 1 - len(nz) / len(d), "median_diff_control_minus_Z": np.median(d)}
    if len(nz) < 10:
        return {**out, "test": "identical by design (no test)", "p": np.nan, "effect_size": 0.0}
    r = rankdata(np.abs(nz))
    return {**out, "test": "Wilcoxon signed-rank, two-sided", "p": wilcoxon(nz).pvalue,
            "effect_size": (r[nz > 0].sum() - r[nz < 0].sum()) / r.sum(), "effect_size_name": "matched-pairs rank-biserial r"}


def independent(a, b):
    u = mannwhitneyu(b, a, alternative="two-sided")
    return {"test": "Mann-Whitney U, two-sided", "p": u.pvalue, "effect_size": 2 * u.statistic / (len(a) * len(b)) - 1,
            "effect_size_name": "rank-biserial r = Cliff's delta"}


def load_pairs():
    """One row per matched Z-DNA locus with its own control, from the controls master (GC_pct holds a fraction)."""
    m = pd.read_csv(MASTER, sep="\t", low_memory=False,
                    usecols=["tol_level", "dinuc_mode", "Chromosome", "Start0", "End0", "Ctrl_Start", "Ctrl_End", "GC_pct",
                             "Dinuc_density", "Ctrl_GC_pct", "Ctrl_Dinuc_density", "has_control"],
                    dtype={"tol_level": str, "dinuc_mode": str, "Chromosome": str, "has_control": str})
    m = m[(m.tol_level == TOL) & (m.dinuc_mode == DINUC_MODE) & (m.has_control.str.strip() == "True")
          & m.Chromosome.isin(CHROMS)].drop_duplicates(["Chromosome", "Start0", "End0"])
    num = lambda x: pd.to_numeric(x, errors="coerce").to_numpy(float)
    return pd.DataFrame({"gc_Z": num(m.GC_pct), "gc_C": num(m.Ctrl_GC_pct), "dinuc_Z": num(m.Dinuc_density),
                         "dinuc_C": num(m.Ctrl_Dinuc_density), "length_Z": (m.End0 - m.Start0).to_numpy(float),
                         "length_C": num(m.Ctrl_End) - num(m.Ctrl_Start)}).dropna()


def main():
    loci = pd.read_csv(LOCI, sep="\t", low_memory=False)
    n_matched = int((loci.set == "Z_matched").sum())
    un = loci[loci.set == "Z_unmatched"]
    pairs = load_pairs()
    if len(pairs) != n_matched:
        raise RuntimeError(f"{len(pairs):,} matched pairs in the master vs {n_matched:,} matched loci in {LOCI.name}")
    rows = []
    for m, lab in METRICS.items():
        a, b = pairs[f"{m}_Z"].to_numpy(float), pairs[f"{m}_C"].to_numpy(float)
        u = un[m].dropna().to_numpy(float)
        rows.append({"comparison": "Z_matched_vs_control_paired", "metric": m, "metric_label": lab,
                     **{f"Z_matched_{k}": v for k, v in summary(a).items()},
                     **{f"other_{k}": v for k, v in summary(b).items()}, "n_other": len(b), **paired(a, b)})
        rows.append({"comparison": "Z_matched_vs_control_independent", "metric": m, "metric_label": lab,
                     "n_pairs": len(a), **independent(a, b), "effect_size_name": "Cliff's delta (overlap of distributions)"})
        rows.append({"comparison": "Z_unmatched_vs_Z_matched", "metric": m, "metric_label": lab,
                     **{f"Z_matched_{k}": v for k, v in summary(a).items()},
                     **{f"other_{k}": v for k, v in summary(u).items()}, "n_other": len(u), **independent(a, u)})
    tests = pd.DataFrame(rows)
    tests["magnitude"] = tests.effect_size.map(label)
    tests.to_csv(OUT_TESTS, sep="\t", index=False, float_format="%.6g")

    print(f"n: matched Z-DNA {n_matched:,} (= controls), unmatched Z-DNA {len(un):,}")
    print(tests[["comparison", "metric", "p", "effect_size", "magnitude"]].to_string(index=False, float_format=lambda x: f"{x:.3g}"))


if __name__ == "__main__":
    main()
```

#### 11 · Variant composition → Suppl. Fig. 2b

```python
"""11 - Composition of the normalized HPRC v2 (CHM13) variants per class (Suppl. Fig. 2b).

One row per ALT allele of the 9_ meta table: SNV types (pyrimidine-collapsed) or size bins (event_size).
"""
from pathlib import Path

import polars as pl

BASE = Path("/path/to/pangenomes_zdna")
META = BASE / "9_hprc-v2.0-mc-chm13.wave.normalized.mutations.meta.tsv.gz"
OUT = BASE / "11_mutation_composition_counts.tsv"

CLASS_ORDER = ["SNP", "MNP", "SUB", "DEL", "INS", "CINS", "CDEL"]
SNP_ORDER = ["C>A", "C>G", "C>T", "T>A", "T>C", "T>G"]
SIZE_ORDER = ["1bp", "2_5bp", "6_50bp", "51_200bp", "gt200bp"]   # DEL, INS, CDEL, CINS
SUB_ORDER = ["21_50bp", "51_200bp", "gt200bp"]                   # substitutions (> 20 bp)
MNP_ORDER = ["2_20bp"]                                            # MNVs (2-20 bp) kept as one category
COMP = {"A": "T", "C": "G", "G": "C", "T": "A"}


def main():
    lf = pl.scan_csv(META, separator="\t",
                     schema_overrides={"normalized_ref": pl.Utf8, "normalized_alt": pl.Utf8, "struct_class": pl.Utf8,
                                       "dominant_op": pl.Utf8, "event_size": pl.Int64}
                     ).select("struct_class", "dominant_op", "event_size", pl.col("normalized_ref", "normalized_alt").fill_null(""))
    sc, dom, L = pl.col("struct_class"), pl.col("dominant_op"), pl.col("event_size")
    ref, alt = pl.col("normalized_ref").str.to_uppercase(), pl.col("normalized_alt").str.to_uppercase()
    flip = ref.is_in(["A", "G"])
    snp_type = (pl.when(flip).then(ref.replace(COMP)).otherwise(ref) + pl.lit(">")
                + pl.when(flip).then(alt.replace(COMP)).otherwise(alt))
    class_key = (pl.when(sc == "snp").then(pl.lit("SNP")).when(sc == "mnp").then(pl.lit("MNP"))
                 .when(sc == "substitution").then(pl.lit("SUB")).when(sc == "ins").then(pl.lit("INS"))
                 .when(sc == "del").then(pl.lit("DEL"))
                 .when((sc == "complex") & (dom == "ins_like")).then(pl.lit("CINS"))
                 .when((sc == "complex") & (dom == "del_like")).then(pl.lit("CDEL")).otherwise(pl.lit("OTHER")))
    size = (pl.when(L == 1).then(pl.lit("1bp")).when(L <= 5).then(pl.lit("2_5bp")).when(L <= 50).then(pl.lit("6_50bp"))
            .when(L <= 200).then(pl.lit("51_200bp")).otherwise(pl.lit("gt200bp")))
    sub = pl.when(L <= 50).then(pl.lit("21_50bp")).when(L <= 200).then(pl.lit("51_200bp")).otherwise(pl.lit("gt200bp"))
    lf = lf.with_columns(class_key=class_key).with_columns(
        category=pl.when(pl.col("class_key") == "SNP").then(snp_type).when(pl.col("class_key") == "MNP").then(pl.lit("2_20bp"))
        .when(pl.col("class_key") == "SUB").then(sub).otherwise(size))
    counts = lf.group_by("class_key", "category").len().collect(engine="streaming")
    got = {(k, c): n for k, c, n in counts.iter_rows()}

    other = {k: n for k, n in got.items() if k[0] == "OTHER" or (k[0] == "SNP" and k[1] not in SNP_ORDER)}
    if other:
        print(f"[NOTE] excluded (SNVs with an ambiguous N allele, or unclassified): {other}")
    total = sum(n for k, n in got.items() if k not in other)
    order = {"SNP": SNP_ORDER, "MNP": MNP_ORDER, "SUB": SUB_ORDER}
    rows = []
    for k in CLASS_ORDER:
        cats = order.get(k, SIZE_ORDER)
        ct = sum(got.get((k, c), 0) for c in cats)
        unexpected = {c: n for (kk, c), n in got.items() if kk == k and c not in cats and (kk, c) not in other}
        if unexpected:
            raise RuntimeError(f"{k}: categories outside the expected bins: {unexpected}")
        rows += [{"class_key": k, "category": c, "count": got.get((k, c), 0), "class_total": ct,
                  "frac_within_class": got.get((k, c), 0) / ct if ct else 0.0,
                  "frac_of_all": got.get((k, c), 0) / total} for c in cats]
    out = pl.DataFrame(rows)
    out.with_columns(pl.col("frac_within_class", "frac_of_all").round(6)).write_csv(OUT, separator="\t")

    print(out.group_by("class_key", maintain_order=True).agg(pl.col("class_total").first())
          .with_columns((100 * pl.col("class_total") / total).round(2).alias("pct_of_all")))
    print(f"Total ALT alleles: {total:,} -> {OUT.name}")


if __name__ == "__main__":
    main()
```

#### 12.1 · Variant-centred metaprofiles → Fig. 3a, 3b, Suppl. Fig. 2e

```python
"""12.1 - Z-DNA and matched-control metaprofiles (+/-WINDOW_RADIUS bp) around normalized variant anchors, per variant
stratum and anchor (CHM13, HPRC v2; Fig. 3a, 3b, Suppl. Fig. 2e inputs).

Anchors: site (SNV, insertions), midpoint (MNV, substitutions and deletions <= 50 bp), left/right (> 50 bp). Each
distinct normalized allele counted once; 95% chromosome-block bootstrap bands. Strata run in parallel.
"""
import os
import time
import multiprocessing as mp
from concurrent.futures import ProcessPoolExecutor, as_completed
from pathlib import Path

import numpy as np
import pandas as pd
import pyarrow as pa
import pyarrow.compute as pc
import pyarrow.csv as pacsv

BASE = Path("/path/to/pangenomes_zdna")
MASTER = BASE / "8_CHM13_controls_MASTER.tsv"
MUT_META = BASE / "9_hprc-v2.0-mc-chm13.wave.normalized.mutations.meta.tsv.gz"
OUT_PROF = BASE / "12_zdna_mutation_profiles.tsv.gz"
OUT_CHR = BASE / "12_zdna_mutation_profiles_bychrom.tsv.gz"

TOL = "10%"                 # control matching tolerance: "5%" | "10%" | "15%"
DINUC_MODE = "CpG_GpC"      # "CpG_GpC" | "CpG_only"
DEDUP_NORMALIZED = True     # count each distinct normalized allele once (False = once per VCF record)
WINDOW_RADIUS = 500
N_BOOTS = 500               # chromosome-block bootstrap for the profile bands
CI = (2.5, 97.5)
MIN_EVENTS = 500            # strata with fewer variants are skipped
SEED = 42
ARROW_THREADS, BLOCK_SIZE = 16, 1 << 26   # pyarrow parse threads / 64 MB text blocks (> longest line)

CHROMS = [f"chr{i}" for i in range(1, 23)] + ["chrX", "chrY"]
OFFSETS = np.arange(-WINDOW_RADIUS, WINDOW_RADIUS + 1, dtype=np.int64)
STRATA = {
    "SNP": "site", "MNP_2_20": "midpoint",
    "DEL_1bp": "midpoint", "DEL_2_5bp": "midpoint", "DEL_6_50bp": "midpoint", "DEL_gt50bp": "dual",
    "INS_1bp": "site", "INS_2_5bp": "site", "INS_6_50bp": "site", "INS_51_200bp": "site", "INS_gt200bp": "site",
    "CDEL_2_5bp": "midpoint", "CDEL_6_50bp": "midpoint", "CDEL_gt50bp": "dual",
    "CINS_2_5bp": "site", "CINS_6_50bp": "site", "CINS_51_200bp": "site", "CINS_gt200bp": "site",
    "SUB_21_50": "midpoint", "SUB_51_200": "dual", "SUB_gt200": "dual",
}
SNAMES = list(STRATA)
SCODE = {s: i for i, s in enumerate(SNAMES)}
MODES = {"site": ["site"], "midpoint": ["midpoint"], "dual": ["left", "right"]}
STRUCT = ["snp", "mnp", "substitution", "del", "ins", "complex"]
DOM = ["snp", "mnp", "substitution", "del", "ins", "del_like", "ins_like", "balanced"]


def merge_intervals(s, e):
    """Sorted, disjoint union of half-open intervals."""
    if len(s) == 0:
        return np.empty(0, np.int64), np.empty(0, np.int64)
    o = np.argsort(s, kind="stable"); s, e = s[o], e[o]
    new = np.ones(len(s), bool); new[1:] = s[1:] > np.maximum.accumulate(e)[:-1]
    return s[new].astype(np.int64), np.maximum.reduceat(e, np.flatnonzero(new)).astype(np.int64)


def sweep_counts(anchors, iv_s, iv_e, R=WINDOW_RADIUS, chunk=200_000):
    """out[k + R] = number of (sorted) anchors a with a + k inside one of the disjoint intervals."""
    W = 2 * R + 1
    diff = np.zeros(W + 1, np.int64)
    if len(anchors) == 0 or len(iv_s) == 0:
        return diff[:W]
    for i0 in range(0, len(iv_s), chunk):
        s, e = iv_s[i0:i0 + chunk], iv_e[i0:i0 + chunk]
        lo = np.searchsorted(anchors, s - R, side="left")
        n = np.searchsorted(anchors, e - 1 + R, side="right") - lo
        tot = int(n.sum())
        if tot == 0:
            continue
        iv = np.repeat(np.arange(len(s)), n)
        a = anchors[np.arange(tot) + np.repeat(lo - (np.cumsum(n) - n), n)]
        diff += np.bincount(np.maximum(s[iv] - a, -R) + R, minlength=W + 1)
        diff -= np.bincount(np.minimum(e[iv] - 1 - a, R) + R + 1, minlength=W + 1)
    return np.cumsum(diff)[:W]


def codes(arr, values, name):
    idx = pc.index_in(arr, value_set=pa.array(values))
    if idx.null_count:
        raise RuntimeError(f"unexpected {name}: {pc.unique(arr.filter(pc.is_null(idx))).to_pylist()[:10]}")
    return idx.to_numpy(zero_copy_only=False).astype(np.int8)


def load_intervals():
    cols = ["tol_level", "dinuc_mode", "Chromosome", "Start", "End", "Start0", "End0", "Ctrl_Start", "Ctrl_End", "has_control"]
    m = pd.read_csv(MASTER, sep="\t", usecols=cols, low_memory=False,
                    dtype={"tol_level": str, "dinuc_mode": str, "Chromosome": str, "has_control": str})
    if ((m.Start0 != m.Start) | (m.End0 != m.End + 1)).any():
        raise RuntimeError("MASTER: Start0/End0 do not follow ZSeeker Start / End + 1")
    m = m[(m.tol_level == TOL) & (m.dinuc_mode == DINUC_MODE) & (m.has_control.str.strip() == "True")].copy()
    if m.duplicated(["Chromosome", "Start0", "End0"]).any():
        raise RuntimeError("MASTER: duplicated Z-DNA locus within one condition")
    m["Ctrl_Start"] = m.Ctrl_Start.astype(float).astype(np.int64)
    m["Ctrl_End"] = m.Ctrl_End.astype(float).astype(np.int64)
    if ((m.Ctrl_End - m.Ctrl_Start) != (m.End0 - m.Start0)).any():
        raise RuntimeError("MASTER: control length != Z-DNA length")
    Z, C = {}, {}
    for chrom, d in m.groupby("Chromosome", sort=False):
        Z[chrom] = merge_intervals(d.Start0.to_numpy(np.int64), d.End0.to_numpy(np.int64))
        C[chrom] = merge_intervals(d.Ctrl_Start.to_numpy(np.int64), d.Ctrl_End.to_numpy(np.int64))
    zbp = {c: int((e - s).sum()) for c, (s, e) in Z.items()}
    cbp = {c: int((e - s).sum()) for c, (s, e) in C.items()}
    print(f"[intervals] {TOL} {DINUC_MODE}: {len(m):,} matched Z-DNA loci | Z-DNA {sum(zbp.values()):,} bp | "
          f"control {sum(cbp.values()):,} bp")
    return Z, C, zbp, cbp


def load_variants():
    """Stream the meta table into compact arrays; collapse identical normalized alleles."""
    pa.set_cpu_count(ARROW_THREADS)
    types = {"seqID": pa.string(), "start": pa.int64(), "normalized_start0": pa.int64(), "normalized_ref": pa.string(),
             "normalized_alt": pa.string(), "struct_class": pa.string(), "dominant_op": pa.string(),
             "ref_span": pa.int64(), "event_size": pa.int64()}
    reader = pacsv.open_csv(MUT_META, read_options=pacsv.ReadOptions(block_size=BLOCK_SIZE),
                            parse_options=pacsv.ParseOptions(delimiter="\t"),
                            convert_options=pacsv.ConvertOptions(include_columns=list(types), column_types=types,
                                                                 strings_can_be_null=False))
    parts = {k: [] for k in ("ch", "s0", "rs", "L", "sc", "dom", "h")}
    n_raw, t0 = 0, time.time()
    for b in reader:
        n_raw += b.num_rows
        chi = pc.index_in(b.column("seqID"), value_set=pa.array(CHROMS))
        keep = pc.is_valid(chi)
        b, chi = b.filter(keep), chi.filter(keep)
        if b.num_rows == 0:
            continue
        s0, rs, st = (b.column(c).to_numpy() for c in ("normalized_start0", "ref_span", "start"))
        sc, dom = codes(b.column("struct_class"), STRUCT, "struct_class"), codes(b.column("dominant_op"), DOM, "dominant_op")
        mid = (sc == STRUCT.index("mnp")) | (sc == STRUCT.index("substitution"))
        if (np.where(mid, s0 + (rs - 1) // 2, s0) != st).any():
            raise RuntimeError("meta: 'start' disagrees with the normalized_start0 anchor rule")
        if not pc.all(pc.equal(pc.utf8_length(b.column("normalized_ref")).cast(pa.int64()), b.column("ref_span"))).as_py():
            raise RuntimeError("meta: ref_span != len(normalized_ref)")
        for k, v in (("ch", chi.to_numpy(zero_copy_only=False).astype(np.int8)), ("s0", s0), ("rs", rs.astype(np.int32)),
                     ("L", b.column("event_size").to_numpy().astype(np.int32)), ("sc", sc), ("dom", dom),
                     ("h", pd.util.hash_array(b.column("normalized_alt").to_numpy(zero_copy_only=False)))):
            parts[k].append(v)
    a = {k: np.concatenate(v) for k, v in parts.items()}
    n_kept = len(a["ch"])
    o = np.lexsort((a["h"], a["rs"], a["s0"], a["ch"]))
    a = {k: v[o] for k, v in a.items()}
    new = np.ones(n_kept, bool)
    new[1:] = ((a["ch"][1:] != a["ch"][:-1]) | (a["s0"][1:] != a["s0"][:-1]) | (a["rs"][1:] != a["rs"][:-1])
               | (a["h"][1:] != a["h"][:-1]))
    if DEDUP_NORMALIZED:
        a = {k: v[new] for k, v in a.items()}
    print(f"[variants] {n_raw:,} ALT rows | {n_kept:,} on primary chromosomes | {int(new.sum()):,} distinct normalized "
          f"alleles | using {len(a['ch']):,} ({time.time() - t0:.0f}s)")
    return a


def assign_strata(a):
    sc, dom, L = a["sc"], a["dom"], a["L"]
    st = np.full(len(sc), -1, np.int16)
    is_ = lambda x: sc == STRUCT.index(x)
    cdel = is_("complex") & (dom == DOM.index("del_like"))
    cins = is_("complex") & (dom == DOM.index("ins_like"))
    rules = [("SNP", is_("snp")), ("MNP_2_20", is_("mnp")),
             ("SUB_21_50", is_("substitution") & (L <= 50)), ("SUB_51_200", is_("substitution") & (L > 50) & (L <= 200)),
             ("SUB_gt200", is_("substitution") & (L > 200)),
             ("DEL_1bp", is_("del") & (L == 1)), ("DEL_2_5bp", is_("del") & (L >= 2) & (L <= 5)),
             ("DEL_6_50bp", is_("del") & (L >= 6) & (L <= 50)), ("DEL_gt50bp", is_("del") & (L > 50)),
             ("INS_1bp", is_("ins") & (L == 1)), ("INS_2_5bp", is_("ins") & (L >= 2) & (L <= 5)),
             ("INS_6_50bp", is_("ins") & (L >= 6) & (L <= 50)), ("INS_51_200bp", is_("ins") & (L >= 51) & (L <= 200)),
             ("INS_gt200bp", is_("ins") & (L > 200)),
             ("CDEL_2_5bp", cdel & (L <= 5)), ("CDEL_6_50bp", cdel & (L >= 6) & (L <= 50)), ("CDEL_gt50bp", cdel & (L > 50)),
             ("CINS_2_5bp", cins & (L <= 5)), ("CINS_6_50bp", cins & (L >= 6) & (L <= 50)),
             ("CINS_51_200bp", cins & (L >= 51) & (L <= 200)), ("CINS_gt200bp", cins & (L > 200))]
    for name, mask in rules:
        st[mask] = SCODE[name]
    if (st < 0).any():
        print(f"[WARN] {int((st < 0).sum()):,} alleles in no stratum (excluded)")
    return st


def run_job(stratum, mode):
    idx = ORDER[BOUNDS[SCODE[stratum]]:BOUNDS[SCODE[stratum] + 1]]
    ch, s0, rs = CH[idx], S0[idx], RS[idx]
    anc = {"site": s0, "left": s0, "midpoint": s0 + (rs - 1) // 2, "right": s0 + rs - 1}[mode]
    if mode in ("midpoint", "right") and (rs < 1).any():
        raise RuntimeError(f"{stratum}/{mode}: ref_span < 1")
    o = np.lexsort((anc, ch)); ch, anc = ch[o], anc[o]
    cuts = np.searchsorted(ch, np.arange(len(CHROMS) + 1))
    rows = []
    for ci, chrom in enumerate(CHROMS):
        a = anc[cuts[ci]:cuts[ci + 1]]
        if len(a) and chrom in Z_IV:
            rows.append((chrom, len(a), sweep_counts(a, *Z_IV[chrom]), sweep_counts(a, *C_IV[chrom])))
    chroms = [r[0] for r in rows]
    nev = np.array([r[1] for r in rows])
    Zm, Cm = np.stack([r[2] for r in rows]), np.stack([r[3] for r in rows])
    Zt, Ct = Zm.sum(0).astype(float), Cm.sum(0).astype(float)
    rng = np.random.default_rng(SEED)
    bi = rng.integers(0, len(chroms), size=(N_BOOTS, len(chroms)))
    with np.errstate(divide="ignore", invalid="ignore"):
        zb, cb = Zm[bi].sum(1).astype(float), Cm[bi].sum(1).astype(float)
        zb /= zb.mean(1, keepdims=True); cb /= cb.mean(1, keepdims=True)
        Zn, Cn = Zt / Zt.mean(), Ct / Ct.mean()
    zlo, zhi = np.nanpercentile(zb, CI, axis=0)
    clo, chi = np.nanpercentile(cb, CI, axis=0)
    zbp, cbp = sum(ZBP[c] for c in chroms), sum(CBP[c] for c in chroms)
    prof = pd.DataFrame({"stratum": stratum, "anchor": mode, "offset": OFFSETS, "Z_mean": Zn, "Z_lo": zlo, "Z_hi": zhi,
                         "C_mean": Cn, "C_lo": clo, "C_hi": chi, "Z_raw": Zt.astype(np.int64), "C_raw": Ct.astype(np.int64),
                         "n_events": int(nev.sum()), "n_chromosomes": len(chroms), "Z_bp": zbp, "C_bp": cbp})
    W = len(OFFSETS)
    bych = pd.DataFrame({"stratum": stratum, "anchor": mode, "chrom": np.repeat(chroms, W),
                         "offset": np.tile(OFFSETS, len(chroms)), "Z_raw": Zm.ravel(), "C_raw": Cm.ravel(),
                         "n_events": np.repeat(nev, W), "Z_bp": np.repeat([ZBP[c] for c in chroms], W),
                         "C_bp": np.repeat([CBP[c] for c in chroms], W)})
    return prof, bych


if __name__ == "__main__":
    t_all = time.time()
    Z_IV, C_IV, ZBP, CBP = load_intervals()
    v = load_variants()
    ST = assign_strata(v)
    CH, S0, RS = v["ch"].astype(np.int16), v["s0"].astype(np.int64), v["rs"].astype(np.int64)
    del v
    ORDER = np.argsort(ST, kind="stable")
    BOUNDS = np.searchsorted(ST[ORDER], np.arange(len(SNAMES) + 1), side="left")
    n_by = {s: int(BOUNDS[SCODE[s] + 1] - BOUNDS[SCODE[s]]) for s in SNAMES}
    jobs = [(s, m) for s in SNAMES for m in MODES[STRATA[s]] if n_by[s] >= MIN_EVENTS]
    skipped = [s for s in SNAMES if n_by[s] < MIN_EVENTS]
    if skipped:
        print(f"[strata] skipped (< {MIN_EVENTS} variants): {skipped}")

    profs, bychs = [], []
    with ProcessPoolExecutor(max_workers=min(len(jobs), os.cpu_count() or 1), mp_context=mp.get_context("fork")) as ex:
        futs = {ex.submit(run_job, s, m): (s, m) for s, m in jobs}
        for f in as_completed(futs):
            p, b = f.result()
            profs.append(p); bychs.append(b)
    key = lambda d: d.assign(_o=d.stratum.map(SCODE), _a=d.anchor.map({"site": 0, "midpoint": 0, "left": 0, "right": 1}))
    prof = key(pd.concat(profs)).sort_values(["_o", "_a", "offset"]).drop(columns=["_o", "_a"])
    bych = key(pd.concat(bychs)).sort_values(["_o", "_a", "chrom", "offset"]).drop(columns=["_o", "_a"])
    prof.to_csv(OUT_PROF, sep="\t", index=False, float_format="%.6g")
    bych.to_csv(OUT_CHR, sep="\t", index=False)
    print(f"{len(jobs)} profiles -> {OUT_PROF.name}, {OUT_CHR.name} [{time.time() - t_all:.0f}s]")
```

#### 12.2 · Window statistics → Fig. 3a, 3b, Suppl. Fig. 2e

```python
"""12.2 - Z-DNA vs control enrichment within +/-r bp of variant anchors, per stratum and anchor (Fig. 3a, 3b, Suppl. Fig. 2e statistics).

Relative density = mean count in +/-r / mean count over +/-R (Z-DNA and controls); fold = Z_rel / C_rel with 95%
chromosome-block bootstrap CI; paired two-sided Wilcoxon over chromosomes and bootstrap test of log(fold) = 0;
Benjamini-Hochberg across stratum/anchor rows within each window.
"""
from pathlib import Path

import numpy as np
import pandas as pd
from scipy.stats import wilcoxon
from statsmodels.stats.multitest import multipletests

BASE = Path("/path/to/pangenomes_zdna")
BYCHROM = BASE / "12_zdna_mutation_profiles_bychrom.tsv.gz"
OUT = BASE / "12_zdna_mutation_window_stats.tsv"

WINDOWS = (10, 25, 50)        # half-widths (bp) of the windows tested
ALTERNATIVE = "two-sided"       # "greater" (Z-DNA > control, as in the manuscript) or "two-sided"
N_BOOTS = 50_000
MIN_CHROMS = 6                # minimum chromosomes with variants for the Wilcoxon test
SEED = 42
CI = (2.5, 97.5)
STARS = [(1e-3, "***"), (1e-2, "**"), (5e-2, "*")]


def stars(q, positive):
    if pd.isna(q) or (ALTERNATIVE == "greater" and not positive):
        return "ns"
    return next((s for t, s in STARS if q < t), "ns")


def boot_p(boots, observed):
    """Bootstrap test of log ratio = 0 (null distribution = bootstrap centred on zero)."""
    null = boots - boots.mean()
    if ALTERNATIVE == "greater":
        return (np.sum(null >= observed) + 1) / (len(null) + 1)
    return (np.sum(np.abs(null) >= abs(observed)) + 1) / (len(null) + 1)


def stats_one(g, r, rng):
    Z = g.pivot(index="chrom", columns="offset", values="Z_raw").sort_index(axis=1)
    C = g.pivot(index="chrom", columns="offset", values="C_raw").reindex(index=Z.index, columns=Z.columns)
    off, chroms = Z.columns.to_numpy(), Z.index
    win, at0 = np.abs(off) <= r, off == 0
    Z, C = Z.to_numpy(float), C.to_numpy(float)
    nw, na = win.sum(), len(off)
    zw, zt, cw, ct = Z[:, win].sum(1), Z.sum(1), C[:, win].sum(1), C.sum(1)
    meta = g.drop_duplicates("chrom").set_index("chrom").loc[chroms]
    zbp, cbp, nev = meta.Z_bp.to_numpy(float), meta.C_bp.to_numpy(float), meta.n_events.to_numpy()
    zin, cin = Z[:, at0].ravel(), C[:, at0].ravel()

    rel = lambda w, t: (w / nw) / (t / na)
    with np.errstate(divide="ignore", invalid="ignore"):
        z_rel, c_rel = rel(zw.sum(), zt.sum()), rel(cw.sum(), ct.sum())
        fold = z_rel / c_rel
        dens = (zin.sum() / zbp.sum()) / (cin.sum() / cbp.sum())
        bi = rng.integers(0, len(zw), size=(N_BOOTS, len(zw)))
        zb, cb = rel(zw[bi].sum(1), zt[bi].sum(1)), rel(cw[bi].sum(1), ct[bi].sum(1))
        lb = np.log(zb / cb)
        db = (zin[bi].sum(1) / zbp[bi].sum(1)) / (cin[bi].sum(1) / cbp[bi].sum(1))
    ok = np.isfinite(lb)
    q = lambda x: np.nanpercentile(x[np.isfinite(x)], CI) if np.isfinite(x).any() else (np.nan, np.nan)
    use = (zt > 0) & (ct > 0)
    zc, cc = rel(zw[use], zt[use]), rel(cw[use], ct[use])
    if use.sum() < MIN_CHROMS:
        w_stat = w_p = np.nan
    elif np.allclose(zc, cc):
        w_stat, w_p = 0.0, 1.0
    else:
        w_stat, w_p = wilcoxon(zc, cc, alternative=ALTERNATIVE)
    return {"Z_rel": z_rel, "Z_rel_lo": q(zb)[0], "Z_rel_hi": q(zb)[1], "C_rel": c_rel, "C_rel_lo": q(cb)[0],
            "C_rel_hi": q(cb)[1], "fold_enrichment": fold, "fold_lo": np.exp(q(lb)[0]), "fold_hi": np.exp(q(lb)[1]),
            "p_bootstrap": boot_p(lb[ok], np.log(fold)) if ok.any() and np.isfinite(fold) else np.nan,
            "wilcoxon_stat": w_stat, "p_wilcoxon": w_p, "n_chromosomes_tested": int(use.sum()),
            "n_events": int(nev.sum()), "n_inside_Z": int(zin.sum()), "n_inside_C": int(cin.sum()),
            "Z_per_kb": 1e3 * zin.sum() / zbp.sum(), "C_per_kb": 1e3 * cin.sum() / cbp.sum(),
            "inside_density_ratio": dens, "inside_ratio_lo": q(db)[0], "inside_ratio_hi": q(db)[1],
            "Z_count_window": int(zw.sum()), "Z_count_profile": int(zt.sum()), "C_count_window": int(cw.sum()),
            "C_count_profile": int(ct.sum()), "n_offsets_window": int(nw), "n_offsets_profile": int(na)}


def main():
    d = pd.read_csv(BYCHROM, sep="\t")
    rng = np.random.default_rng(SEED)
    keys = list(dict.fromkeys(zip(d.stratum, d.anchor)))
    groups = {k: g for k, g in d.groupby(["stratum", "anchor"], sort=False)}
    rows = []
    for r in WINDOWS:
        tab = pd.DataFrame([{"window_bp": r, "stratum": s, "anchor": a, **stats_one(groups[(s, a)], r, rng)}
                            for s, a in keys])
        for p in ("p_wilcoxon", "p_bootstrap"):
            m = tab[p].notna()
            tab[p.replace("p_", "q_")] = np.nan
            tab.loc[m, p.replace("p_", "q_")] = multipletests(tab.loc[m, p], method="fdr_bh")[1]
        pos = tab.fold_enrichment > 1
        tab["sig_wilcoxon"] = [stars(q, s) for q, s in zip(tab.q_wilcoxon, pos)]
        tab["sig_bootstrap"] = [stars(q, s) for q, s in zip(tab.q_bootstrap, pos)]
        rows.append(tab)
    out = pd.concat(rows, ignore_index=True)
    out.insert(3, "label", out.stratum + np.where(out.anchor.isin(["left", "right"]), "_" + out.anchor, ""))
    out.to_csv(OUT, sep="\t", index=False, float_format="%.6g")

    t = out[out.window_bp == WINDOWS[0]]
    print(f"+/-{WINDOWS[0]} bp:\n" + t[["label", "n_events", "fold_enrichment", "q_wilcoxon", "sig_wilcoxon"]]
          .to_string(index=False, float_format=lambda x: f"{x:.3g}"))
    print(f"-> {OUT.name}")


if __name__ == "__main__":
    main()
```

#### Original vs normalized VCF parsing (response to reviewers)

```python
"""Fast standalone comparison of original vs normalized VCF ALT-row classifications.

Reads the two gzip TSVs, decompresses each to a temporary TSV on disk, then
uses Polars' parallel CSV scanner to aggregate counts. Temporary files are
removed automatically. Does not change either input or any Fig. 3a outputs.

Requires: polars. Optional: pigz (otherwise gzip -dc).
Set POLARS_MAX_THREADS in the shell to control Polars CPU use.
"""
import os
import shutil
import subprocess
import tempfile
from pathlib import Path

import polars as pl

BASE = Path('/path/to/pangenomes_zdna')
OLD = BASE / 'vcf_mutations/hprcv2_chm13/hprc-v2.0-mc-chm13.wave.mutations.meta.tsv.gz'
NEW = BASE / '9_hprc-v2.0-mc-chm13.wave.normalized.mutations.meta.tsv.gz'
OUT = BASE / 'vcf_parsing_rtr_classification_counts.tsv'
CHROMS = [f'chr{i}' for i in range(1, 23)] + ['chrX', 'chrY']


def decompress(src, dst):
    executable = shutil.which('pigz') or shutil.which('gzip')
    if executable is None:
        raise RuntimeError('Neither pigz nor gzip is installed')
    print(f'Decompressing {src.name} using {Path(executable).name} ...', flush=True)
    with dst.open('wb') as handle:
        subprocess.run([executable, '-dc', str(src)], stdout=handle, check=True)
    print(f'  Temporary TSV: {dst.stat().st_size / 1e9:.2f} GB', flush=True)


def summarize(src, *, normalized, tmpdir):
    tmp = tmpdir / ('normalized.tsv' if normalized else 'original.tsv')
    decompress(src, tmp)
    cols = ['seqID', 'struct_class', 'size_bin']
    if normalized:
        cols += ['raw_struct_class', 'classification_changed', 'normalization_changed']
    schema = {c: pl.String for c in cols}
    frame = pl.scan_csv(
        tmp, separator='\t', has_header=True, schema_overrides=schema,
        infer_schema_length=0, low_memory=False,
    ).select(cols).filter(pl.col('seqID').is_in(CHROMS))

    print(f'Aggregating {"normalized" if normalized else "original"} records ...', flush=True)
    classes = frame.group_by('struct_class').agg(pl.len().alias('n')).collect(engine='streaming')
    bins = frame.group_by('size_bin').agg(pl.len().alias('n')).collect(engine='streaming')
    pairs = frame.group_by(['struct_class', 'size_bin']).agg(pl.len().alias('n')).collect(engine='streaming')
    if normalized:
        stats = frame.select(
            pl.len().alias('n'),
            (pl.col('raw_struct_class') != pl.col('struct_class')).sum().alias('changed'),
            pl.col('classification_changed').str.to_lowercase().eq('true').sum().alias('flag_changed'),
            pl.col('normalization_changed').str.to_lowercase().eq('true').sum().alias('normalization_changed'),
        ).collect(engine='streaming').row(0, named=True)
        if stats['changed'] != stats['flag_changed']:
            raise RuntimeError('classification_changed flag disagrees with raw_struct_class')
    else:
        stats = frame.select(pl.len().alias('n')).collect(engine='streaming').row(0, named=True)
    tmp.unlink()
    return {
        'n': stats['n'],
        'changed': stats.get('changed', 0),
        'normalization_changed': stats.get('normalization_changed', 0),
        'classes': dict(classes.iter_rows()),
        'bins': dict(bins.iter_rows()),
        'pairs': {(a, b): n for a, b, n in pairs.iter_rows()},
    }


def main():
    for p in (OLD, NEW):
        if not p.is_file():
            raise FileNotFoundError(p)
    tmp_root = Path(os.environ.get('TMPDIR', str(BASE)))
    with tempfile.TemporaryDirectory(prefix='vcf_parsing_rtr_', dir=tmp_root) as d:
        temp = Path(d)
        old = summarize(OLD, normalized=False, tmpdir=temp)
        new = summarize(NEW, normalized=True, tmpdir=temp)
    if old['n'] != new['n']:
        raise RuntimeError(f'ALT-row totals differ: old={old["n"]:,}, new={new["n"]:,}')
    rows = []
    categories = [('all', None)]
    for dimension, field in [('class', 'classes'), ('size_bin', 'bins'), ('class_size', 'pairs')]:
        categories.extend((dimension, k) for k in sorted(set(old[field]) | set(new[field])))
    for dimension, key in categories:
        if dimension == 'all':
            a, b, label = old['n'], new['n'], 'ALL'
        else:
            field = {'class': 'classes', 'size_bin': 'bins', 'class_size': 'pairs'}[dimension]
            a, b = old[field].get(key, 0), new[field].get(key, 0)
            label = '|'.join(key) if isinstance(key, tuple) else str(key)
        rows.append({
            'dimension': dimension, 'category': label, 'old_n': a, 'normalized_n': b,
            'delta_n': b-a, 'pct_change': 100*(b-a)/a if a else float('nan'),
            'old_pct_all': 100*a/old['n'], 'normalized_pct_all': 100*b/new['n'],
        })
    pl.DataFrame(rows).write_csv(OUT, separator='\t')
    print(f'Wrote {OUT}', flush=True)
    print(f'ALT rows: {old["n"]:,}', flush=True)
    print(f'Class changed: {new["changed"]:,} ({100*new["changed"]/new["n"]:.3f}%)', flush=True)
    print(f'Normalization changed: {new["normalization_changed"]:,} '
          f'({100*new["normalization_changed"]/new["n"]:.3f}%)', flush=True)
    print('Counts are per original ALT row, not deduplicated Fig. 3a events.', flush=True)


if __name__ == '__main__':
    main()
```

```python
"""Standalone old-vs-normalized Fig. 3a enrichment reconstruction.
No external project scripts are imported or executed.
Inputs and outputs are specified below in each stage.
"""

"""Analysis 12.1: Z-DNA and matched-control metaprofiles around normalized variant anchors (CHM13, HPRC v2).

For every variant stratum (class x size) and anchor, counts at each offset k in [-R, R] how many variant anchors a
have a+k inside a Z-DNA interval, and inside a matched-control interval (one control per Z-DNA locus, condition
TOL / DINUC_MODE). Each distinct normalized allele is counted once.

Coordinates (0-based, half-open):
  Z-DNA   [Start0, End0)  = [ZSeeker Start, ZSeeker End + 1)        control [Ctrl_Start, Ctrl_End)
  anchors site/left = normalized_start0 (SNV, insertions: first base right of the junction)
          midpoint  = normalized_start0 + (ref_span - 1) // 2   (MNV, substitutions and deletions <= 50 bp)
          right     = normalized_start0 + ref_span - 1          (last changed base; > 50 bp deletion-like events)
Outputs (used by 12_2 and 12_4):
  12_zdna_mutation_profiles.tsv.gz          stratum, anchor, offset, Z/C relative density (+ 95% chromosome-bootstrap
                                            band), raw counts, n_events, Z_bp, C_bp
  12_zdna_mutation_profiles_bychrom.tsv.gz  stratum, anchor, chrom, offset, Z_raw, C_raw, n_events, Z_bp, C_bp
"""
import os
import time
import shutil
import subprocess
import tempfile
import multiprocessing as mp
import polars as pl
from concurrent.futures import ProcessPoolExecutor, as_completed
from pathlib import Path

import numpy as np
import pandas as pd
import pyarrow as pa
import pyarrow.compute as pc
import pyarrow.csv as pacsv

BASE = Path("/path/to/pangenomes_zdna")
MASTER = BASE / "8_CHM13_controls_MASTER.tsv"
MUT_META = BASE / "vcf_mutations/hprcv2_chm13/hprc-v2.0-mc-chm13.wave.mutations.meta.tsv.gz"
OUT_PROF = BASE / "vcf_parsing_rtr_OLD_profiles.tsv.gz"
OUT_CHR = BASE / "vcf_parsing_rtr_OLD_profiles_bychrom.tsv.gz"

TOL = "10%"                 # control matching tolerance: "5%" | "10%" | "15%"
DINUC_MODE = "CpG_GpC"      # "CpG_GpC" | "CpG_only"
DEDUP_NORMALIZED = True     # deduplicate identical original alleles
WINDOW_RADIUS = 500
N_BOOTS = 0                 # profile bands unnecessary for Fig. 3a; stats bootstrap is retained
CI = (2.5, 97.5)
MIN_EVENTS = 500            # strata with fewer variants are skipped
SEED = 42
ARROW_THREADS, BLOCK_SIZE = 16, 1 << 26   # pyarrow parse threads / 64 MB text blocks (> longest line)

CHROMS = [f"chr{i}" for i in range(1, 23)] + ["chrX", "chrY"]
OFFSETS = np.arange(-WINDOW_RADIUS, WINDOW_RADIUS + 1, dtype=np.int64)
STRATA = {
    "SNP": "site", "MNP_2_20": "midpoint",
    "DEL_1bp": "midpoint", "DEL_2_5bp": "midpoint", "DEL_6_50bp": "midpoint", "DEL_gt50bp": "dual",
    "INS_1bp": "site", "INS_2_5bp": "site", "INS_6_50bp": "site", "INS_51_200bp": "site", "INS_gt200bp": "site",
    "CDEL_2_5bp": "midpoint", "CDEL_6_50bp": "midpoint", "CDEL_gt50bp": "dual",
    "CINS_2_5bp": "site", "CINS_6_50bp": "site", "CINS_51_200bp": "site", "CINS_gt200bp": "site",
    "SUB_21_50": "midpoint", "SUB_51_200": "dual", "SUB_gt200": "dual",
}
SNAMES = list(STRATA)
SCODE = {s: i for i, s in enumerate(SNAMES)}
MODES = {"site": ["site"], "midpoint": ["midpoint"], "dual": ["left", "right"]}
STRUCT = ["snp", "mnp", "substitution", "del", "ins", "complex"]
DOM = ["snp", "mnp", "substitution", "del", "ins", "del_like", "ins_like", "balanced"]


def merge_intervals(s, e):
    """Sorted, disjoint union of half-open intervals."""
    if len(s) == 0:
        return np.empty(0, np.int64), np.empty(0, np.int64)
    o = np.argsort(s, kind="stable"); s, e = s[o], e[o]
    new = np.ones(len(s), bool); new[1:] = s[1:] > np.maximum.accumulate(e)[:-1]
    return s[new].astype(np.int64), np.maximum.reduceat(e, np.flatnonzero(new)).astype(np.int64)


def sweep_counts(anchors, iv_s, iv_e, R=WINDOW_RADIUS, chunk=200_000):
    """out[k + R] = number of (sorted) anchors a with a + k inside one of the disjoint intervals."""
    W = 2 * R + 1
    diff = np.zeros(W + 1, np.int64)
    if len(anchors) == 0 or len(iv_s) == 0:
        return diff[:W]
    for i0 in range(0, len(iv_s), chunk):
        s, e = iv_s[i0:i0 + chunk], iv_e[i0:i0 + chunk]
        lo = np.searchsorted(anchors, s - R, side="left")
        n = np.searchsorted(anchors, e - 1 + R, side="right") - lo
        tot = int(n.sum())
        if tot == 0:
            continue
        iv = np.repeat(np.arange(len(s)), n)
        a = anchors[np.arange(tot) + np.repeat(lo - (np.cumsum(n) - n), n)]
        diff += np.bincount(np.maximum(s[iv] - a, -R) + R, minlength=W + 1)
        diff -= np.bincount(np.minimum(e[iv] - 1 - a, R) + R + 1, minlength=W + 1)
    return np.cumsum(diff)[:W]


def codes(arr, values, name):
    idx = pc.index_in(arr, value_set=pa.array(values))
    if idx.null_count:
        raise RuntimeError(f"unexpected {name}: {pc.unique(arr.filter(pc.is_null(idx))).to_pylist()[:10]}")
    return idx.to_numpy(zero_copy_only=False).astype(np.int8)


def load_intervals():
    cols = ["tol_level", "dinuc_mode", "Chromosome", "Start", "End", "Start0", "End0", "Ctrl_Start", "Ctrl_End", "has_control"]
    m = pd.read_csv(MASTER, sep="\t", usecols=cols, low_memory=False,
                    dtype={"tol_level": str, "dinuc_mode": str, "Chromosome": str, "has_control": str})
    if ((m.Start0 != m.Start) | (m.End0 != m.End + 1)).any():
        raise RuntimeError("MASTER: Start0/End0 do not follow ZSeeker Start / End + 1")
    m = m[(m.tol_level == TOL) & (m.dinuc_mode == DINUC_MODE) & (m.has_control.str.strip() == "True")].copy()
    if m.duplicated(["Chromosome", "Start0", "End0"]).any():
        raise RuntimeError("MASTER: duplicated Z-DNA locus within one condition")
    m["Ctrl_Start"] = m.Ctrl_Start.astype(float).astype(np.int64)
    m["Ctrl_End"] = m.Ctrl_End.astype(float).astype(np.int64)
    if ((m.Ctrl_End - m.Ctrl_Start) != (m.End0 - m.Start0)).any():
        raise RuntimeError("MASTER: control length != Z-DNA length")
    Z, C = {}, {}
    for chrom, d in m.groupby("Chromosome", sort=False):
        Z[chrom] = merge_intervals(d.Start0.to_numpy(np.int64), d.End0.to_numpy(np.int64))
        C[chrom] = merge_intervals(d.Ctrl_Start.to_numpy(np.int64), d.Ctrl_End.to_numpy(np.int64))
    zbp = {c: int((e - s).sum()) for c, (s, e) in Z.items()}
    cbp = {c: int((e - s).sum()) for c, (s, e) in C.items()}
    print(f"[intervals] {TOL} {DINUC_MODE}: {len(m):,} matched Z-DNA loci | Z-DNA {sum(zbp.values()):,} bp | "
          f"control {sum(cbp.values()):,} bp")
    return Z, C, zbp, cbp


def load_variants():
    """Original per-ALT records, original class/size and original parser anchors.
    Deduplication uses original REF/ALT identity, not normalized allele identity.
    """
    pa.set_cpu_count(ARROW_THREADS)
    types = {"seqID": pa.string(), "start": pa.int64(), "reference": pa.string(),
             "variant": pa.string(), "struct_class": pa.string(), "dominant_op": pa.string(),
             "ref_span": pa.int64(), "alt_span": pa.int64()}
    reader = pacsv.open_csv(MUT_META, read_options=pacsv.ReadOptions(block_size=BLOCK_SIZE),
                            parse_options=pacsv.ParseOptions(delimiter="\t"),
                            convert_options=pacsv.ConvertOptions(include_columns=list(types), column_types=types,
                                                                 strings_can_be_null=False))
    parts = {k: [] for k in ("ch", "s0", "rs", "L", "sc", "dom", "h")}
    n_raw, t0 = 0, time.time()
    for b in reader:
        n_raw += b.num_rows
        chi = pc.index_in(b.column("seqID"), value_set=pa.array(CHROMS))
        keep = pc.is_valid(chi)
        b, chi = b.filter(keep), chi.filter(keep)
        if not b.num_rows: continue
        sc = codes(b.column("struct_class"), STRUCT, "struct_class")
        dom = codes(b.column("dominant_op"), DOM, "dominant_op")
        st = b.column("start").to_numpy().astype(np.int64)
        rs = b.column("ref_span").to_numpy().astype(np.int64)
        al = b.column("alt_span").to_numpy().astype(np.int64)
        is_sub = np.isin(sc, [STRUCT.index(x) for x in ("snp", "mnp", "substitution")])
        is_ins = sc == STRUCT.index("ins")
        is_del = sc == STRUCT.index("del")
        is_cins = dom == DOM.index("ins_like")
        is_cdel = dom == DOM.index("del_like")
        L = np.where(is_sub | is_del | is_cdel, rs,
                     np.where(is_ins | is_cins, al, np.maximum(rs, al)))
        is_mid = is_sub & (sc != STRUCT.index("snp"))
        s0 = np.where(is_mid, st - (rs - 1) // 2, st)
        refs = b.column("reference").to_numpy(zero_copy_only=False)
        alts = b.column("variant").to_numpy(zero_copy_only=False)
        h = pl.DataFrame({"ref": refs, "alt": alts}).hash_rows().to_numpy()
        for k, v in (("ch", chi.to_numpy(zero_copy_only=False).astype(np.int8)),
                     ("s0", s0), ("rs", rs.astype(np.int32)), ("L", L.astype(np.int32)),
                     ("sc", sc), ("dom", dom), ("h", h)):
            parts[k].append(v)
    a = {k: np.concatenate(v) for k, v in parts.items()}
    n_kept = len(a["ch"])
    o = np.lexsort((a["h"], a["rs"], a["s0"], a["ch"]))
    a = {k: v[o] for k, v in a.items()}
    new = np.ones(n_kept, bool)
    new[1:] = ((a["ch"][1:] != a["ch"][:-1]) | (a["s0"][1:] != a["s0"][:-1]) |
               (a["rs"][1:] != a["rs"][:-1]) | (a["h"][1:] != a["h"][:-1]))
    if DEDUP_NORMALIZED:
        a = {k: v[new] for k, v in a.items()}
    print(f"[old variants] {n_raw:,} ALT rows | {n_kept:,} primary | {int(new.sum()):,} distinct raw alleles | using {len(a['ch']):,} ({time.time()-t0:.0f}s)")
    return a


def assign_strata(a):
    sc, dom, L = a["sc"], a["dom"], a["L"]
    st = np.full(len(sc), -1, np.int16)
    is_ = lambda x: sc == STRUCT.index(x)
    cdel = is_("complex") & (dom == DOM.index("del_like"))
    cins = is_("complex") & (dom == DOM.index("ins_like"))
    rules = [("SNP", is_("snp")), ("MNP_2_20", is_("mnp")),
             ("SUB_21_50", is_("substitution") & (L <= 50)), ("SUB_51_200", is_("substitution") & (L > 50) & (L <= 200)),
             ("SUB_gt200", is_("substitution") & (L > 200)),
             ("DEL_1bp", is_("del") & (L == 1)), ("DEL_2_5bp", is_("del") & (L >= 2) & (L <= 5)),
             ("DEL_6_50bp", is_("del") & (L >= 6) & (L <= 50)), ("DEL_gt50bp", is_("del") & (L > 50)),
             ("INS_1bp", is_("ins") & (L == 1)), ("INS_2_5bp", is_("ins") & (L >= 2) & (L <= 5)),
             ("INS_6_50bp", is_("ins") & (L >= 6) & (L <= 50)), ("INS_51_200bp", is_("ins") & (L >= 51) & (L <= 200)),
             ("INS_gt200bp", is_("ins") & (L > 200)),
             ("CDEL_2_5bp", cdel & (L <= 5)), ("CDEL_6_50bp", cdel & (L >= 6) & (L <= 50)), ("CDEL_gt50bp", cdel & (L > 50)),
             ("CINS_2_5bp", cins & (L <= 5)), ("CINS_6_50bp", cins & (L >= 6) & (L <= 50)),
             ("CINS_51_200bp", cins & (L >= 51) & (L <= 200)), ("CINS_gt200bp", cins & (L > 200))]
    for name, mask in rules:
        st[mask] = SCODE[name]
    if (st < 0).any():
        print(f"[WARN] {int((st < 0).sum()):,} alleles in no stratum (excluded)")
    return st


def run_job(stratum, mode):
    idx = ORDER[BOUNDS[SCODE[stratum]]:BOUNDS[SCODE[stratum] + 1]]
    ch, s0, rs = CH[idx], S0[idx], RS[idx]
    anc = {"site": s0, "left": s0, "midpoint": s0 + (rs - 1) // 2, "right": s0 + rs - 1}[mode]
    if mode in ("midpoint", "right") and (rs < 1).any():
        raise RuntimeError(f"{stratum}/{mode}: ref_span < 1")
    o = np.lexsort((anc, ch)); ch, anc = ch[o], anc[o]
    cuts = np.searchsorted(ch, np.arange(len(CHROMS) + 1))
    rows = []
    for ci, chrom in enumerate(CHROMS):
        a = anc[cuts[ci]:cuts[ci + 1]]
        if len(a) and chrom in Z_IV:
            rows.append((chrom, len(a), sweep_counts(a, *Z_IV[chrom]), sweep_counts(a, *C_IV[chrom])))
    chroms = [r[0] for r in rows]
    nev = np.array([r[1] for r in rows])
    Zm, Cm = np.stack([r[2] for r in rows]), np.stack([r[3] for r in rows])
    Zt, Ct = Zm.sum(0).astype(float), Cm.sum(0).astype(float)
    rng = np.random.default_rng(SEED)
    with np.errstate(divide="ignore", invalid="ignore"):
        Zn, Cn = Zt / Zt.mean(), Ct / Ct.mean()
    zlo, zhi = (np.full(WINDOW_RADIUS * 2 + 1, np.nan),) * 2
    clo, chi = (np.full(WINDOW_RADIUS * 2 + 1, np.nan),) * 2
    zbp, cbp = sum(ZBP[c] for c in chroms), sum(CBP[c] for c in chroms)
    prof = pd.DataFrame({"stratum": stratum, "anchor": mode, "offset": OFFSETS, "Z_mean": Zn, "Z_lo": zlo, "Z_hi": zhi,
                         "C_mean": Cn, "C_lo": clo, "C_hi": chi, "Z_raw": Zt.astype(np.int64), "C_raw": Ct.astype(np.int64),
                         "n_events": int(nev.sum()), "n_chromosomes": len(chroms), "Z_bp": zbp, "C_bp": cbp})
    W = len(OFFSETS)
    bych = pd.DataFrame({"stratum": stratum, "anchor": mode, "chrom": np.repeat(chroms, W),
                         "offset": np.tile(OFFSETS, len(chroms)), "Z_raw": Zm.ravel(), "C_raw": Cm.ravel(),
                         "n_events": np.repeat(nev, W), "Z_bp": np.repeat([ZBP[c] for c in chroms], W),
                         "C_bp": np.repeat([CBP[c] for c in chroms], W)})
    return prof, bych


if True:
    t_all = time.time()
    Z_IV, C_IV, ZBP, CBP = load_intervals()
    compressed_meta = MUT_META
    temp_root = Path(os.environ.get("TMPDIR", str(BASE)))
    with tempfile.TemporaryDirectory(prefix="rtr_enrichment_", dir=temp_root) as d:
        plain = Path(d) / "original_metadata.tsv"
        unzip = shutil.which("pigz") or shutil.which("gzip")
        if unzip is None:
            raise RuntimeError("Install pigz or ensure gzip is on PATH")
        print(f"[input] decompressing {compressed_meta} using {unzip}", flush=True)
        with plain.open("wb") as fh:
            subprocess.run([unzip, "-dc", str(compressed_meta)], stdout=fh, check=True)
        MUT_META = plain
        v = load_variants()
    MUT_META = compressed_meta
    ST = assign_strata(v)
    CH, S0, RS = v["ch"].astype(np.int16), v["s0"].astype(np.int64), v["rs"].astype(np.int64)
    del v
    ORDER = np.argsort(ST, kind="stable")
    BOUNDS = np.searchsorted(ST[ORDER], np.arange(len(SNAMES) + 1), side="left")
    n_by = {s: int(BOUNDS[SCODE[s] + 1] - BOUNDS[SCODE[s]]) for s in SNAMES}
    print("[strata] " + ", ".join(f"{s}: {n:,}" for s, n in n_by.items()))
    jobs = [(s, m) for s in SNAMES for m in MODES[STRATA[s]] if n_by[s] >= MIN_EVENTS]
    skipped = [s for s in SNAMES if n_by[s] < MIN_EVENTS]
    if skipped:
        print(f"[strata] skipped (< {MIN_EVENTS} variants): {skipped}")

    profs, bychs = [], []
    with ProcessPoolExecutor(max_workers=min(len(jobs), int(os.environ.get("RTR_WORKERS", "16"))), mp_context=mp.get_context("fork")) as ex:
        futs = {ex.submit(run_job, s, m): (s, m) for s, m in jobs}
        for f in as_completed(futs):
            p, b = f.result()
            profs.append(p); bychs.append(b)
            z0, c0 = p.Z_raw.iloc[WINDOW_RADIUS], p.C_raw.iloc[WINDOW_RADIUS]
            print(f"  {futs[f][0]:14s} {futs[f][1]:8s} n={p.n_events.iloc[0]:>11,}  inside Z/C = {z0:,}/{c0:,}", flush=True)
    key = lambda d: d.assign(_o=d.stratum.map(SCODE), _a=d.anchor.map({"site": 0, "midpoint": 0, "left": 0, "right": 1}))
    prof = key(pd.concat(profs)).sort_values(["_o", "_a", "offset"]).drop(columns=["_o", "_a"])
    bych = key(pd.concat(bychs)).sort_values(["_o", "_a", "chrom", "offset"]).drop(columns=["_o", "_a"])
    bych.to_csv(OUT_CHR, sep="\t", index=False, compression="gzip")
    print(f"Wrote: {OUT_CHR}\n[DONE] {len(jobs)} profiles in {time.time() - t_all:.0f}s")


"""Analysis 12.2: Z-DNA vs matched-control enrichment around variant anchors, per stratum/anchor and window.

Reads 12_zdna_mutation_profiles_bychrom.tsv.gz (12.1). For a window of +/- r bp around the anchors:
  relative density  Z_rel = mean Z count in the window / mean Z count over +/- R   (same for the control, C_rel)
  fold enrichment   Z_rel / C_rel  (pooled over chromosomes; 95% CI from a chromosome-block bootstrap)
  tests             paired Wilcoxon over chromosomes (per-chromosome Z_rel vs C_rel; primary) and the bootstrap
                    test of log(fold) = 0; Benjamini-Hochberg across all stratum/anchor rows within each window
  inside density    variants per kb inside Z-DNA / variants per kb inside the controls (offset 0; window-independent)
Writes ONE table used by 12_3: 12_zdna_mutation_window_stats.tsv
"""
from pathlib import Path

import numpy as np
import pandas as pd
from scipy.stats import wilcoxon
from statsmodels.stats.multitest import multipletests

BASE = Path("/path/to/pangenomes_zdna")
BYCHROM = BASE / "vcf_parsing_rtr_OLD_profiles_bychrom.tsv.gz"
OUT = BASE / "vcf_parsing_rtr_OLD_window_stats.tsv"

WINDOWS = (10, 25, 50)
ALTERNATIVE = "two-sided"
N_BOOTS = 50_000
MIN_CHROMS = 6
SEED = 42
CI = (2.5, 97.5)
STARS = [(1e-3, "***"), (1e-2, "**"), (5e-2, "*")]


def stars(q, positive):
    if pd.isna(q) or (ALTERNATIVE == "greater" and not positive):
        return "ns"
    return next((s for t, s in STARS if q < t), "ns")


def boot_p(boots, observed):
    """Bootstrap test of log ratio = 0 (null distribution = bootstrap centred on zero)."""
    null = boots - boots.mean()
    if ALTERNATIVE == "greater":
        return (np.sum(null >= observed) + 1) / (len(null) + 1)
    return (np.sum(np.abs(null) >= abs(observed)) + 1) / (len(null) + 1)


def stats_one(g, r, rng):
    Z = g.pivot(index="chrom", columns="offset", values="Z_raw").sort_index(axis=1)
    C = g.pivot(index="chrom", columns="offset", values="C_raw").reindex(index=Z.index, columns=Z.columns)
    off, chroms = Z.columns.to_numpy(), Z.index
    win, at0 = np.abs(off) <= r, off == 0
    Z, C = Z.to_numpy(float), C.to_numpy(float)
    nw, na = win.sum(), len(off)
    zw, zt, cw, ct = Z[:, win].sum(1), Z.sum(1), C[:, win].sum(1), C.sum(1)
    meta = g.drop_duplicates("chrom").set_index("chrom").loc[chroms]
    zbp, cbp, nev = meta.Z_bp.to_numpy(float), meta.C_bp.to_numpy(float), meta.n_events.to_numpy()
    zin, cin = Z[:, at0].ravel(), C[:, at0].ravel()

    rel = lambda w, t: (w / nw) / (t / na)
    with np.errstate(divide="ignore", invalid="ignore"):
        z_rel, c_rel = rel(zw.sum(), zt.sum()), rel(cw.sum(), ct.sum())
        fold = z_rel / c_rel
        dens = (zin.sum() / zbp.sum()) / (cin.sum() / cbp.sum())
        bi = rng.integers(0, len(zw), size=(N_BOOTS, len(zw)))
        zb, cb = rel(zw[bi].sum(1), zt[bi].sum(1)), rel(cw[bi].sum(1), ct[bi].sum(1))
        lb = np.log(zb / cb)
        db = (zin[bi].sum(1) / zbp[bi].sum(1)) / (cin[bi].sum(1) / cbp[bi].sum(1))
    ok = np.isfinite(lb)
    q = lambda x: np.nanpercentile(x[np.isfinite(x)], CI) if np.isfinite(x).any() else (np.nan, np.nan)
    use = (zt > 0) & (ct > 0)
    zc, cc = rel(zw[use], zt[use]), rel(cw[use], ct[use])
    if use.sum() < MIN_CHROMS:
        w_stat = w_p = np.nan
    elif np.allclose(zc, cc):
        w_stat, w_p = 0.0, 1.0
    else:
        w_stat, w_p = wilcoxon(zc, cc, alternative=ALTERNATIVE)
    return {"Z_rel": z_rel, "Z_rel_lo": q(zb)[0], "Z_rel_hi": q(zb)[1], "C_rel": c_rel, "C_rel_lo": q(cb)[0],
            "C_rel_hi": q(cb)[1], "fold_enrichment": fold, "fold_lo": np.exp(q(lb)[0]), "fold_hi": np.exp(q(lb)[1]),
            "p_bootstrap": boot_p(lb[ok], np.log(fold)) if ok.any() and np.isfinite(fold) else np.nan,
            "wilcoxon_stat": w_stat, "p_wilcoxon": w_p, "n_chromosomes_tested": int(use.sum()),
            "n_events": int(nev.sum()), "n_inside_Z": int(zin.sum()), "n_inside_C": int(cin.sum()),
            "Z_per_kb": 1e3 * zin.sum() / zbp.sum(), "C_per_kb": 1e3 * cin.sum() / cbp.sum(),
            "inside_density_ratio": dens, "inside_ratio_lo": q(db)[0], "inside_ratio_hi": q(db)[1]}


def main():
    d = pd.read_csv(BYCHROM, sep="\t")
    rng = np.random.default_rng(SEED)
    keys = list(dict.fromkeys(zip(d.stratum, d.anchor)))
    groups = {k: g for k, g in d.groupby(["stratum", "anchor"], sort=False)}
    rows = []
    for r in WINDOWS:
        tab = pd.DataFrame([{"window_bp": r, "stratum": s, "anchor": a, **stats_one(groups[(s, a)], r, rng)}
                            for s, a in keys])
        for p in ("p_wilcoxon", "p_bootstrap"):
            m = tab[p].notna()
            tab[p.replace("p_", "q_")] = np.nan
            tab.loc[m, p.replace("p_", "q_")] = multipletests(tab.loc[m, p], method="fdr_bh")[1]
        pos = tab.fold_enrichment > 1
        tab["sig_wilcoxon"] = [stars(q, s) for q, s in zip(tab.q_wilcoxon, pos)]
        tab["sig_bootstrap"] = [stars(q, s) for q, s in zip(tab.q_bootstrap, pos)]
        rows.append(tab)
    out = pd.concat(rows, ignore_index=True)
    out.insert(3, "label", out.stratum + np.where(out.anchor.isin(["left", "right"]), "_" + out.anchor, ""))
    out.to_csv(OUT, sep="\t", index=False, float_format="%.6g")

    pd.set_option("display.width", 220)
    for r in WINDOWS:
        t = out[out.window_bp == r]
        print(f"\n=== +/-{r} bp ===")
        print(t[["label", "n_events", "Z_rel", "C_rel", "fold_enrichment", "fold_lo", "fold_hi", "q_wilcoxon",
                 "sig_wilcoxon", "q_bootstrap", "sig_bootstrap", "inside_density_ratio"]]
              .to_string(index=False, float_format=lambda x: f"{x:.3g}"))
    print(f"\nWrote: {OUT}")


if True:
    main()


"""RTR table 2: exact before/after Fig. 3a bar heights at +/-10 bp.
Requires BOTH old and normalized 12.2 statistics computed with identical settings.
"""
from pathlib import Path
import numpy as np
import pandas as pd

BASE = Path('/path/to/pangenomes_zdna')
OLD_STATS = BASE / 'vcf_parsing_rtr_OLD_window_stats.tsv'
NEW_STATS = BASE / '12_zdna_mutation_window_stats.tsv'
OUT = BASE / 'vcf_parsing_rtr_enrichment_10bp.tsv'
WINDOW = 10
COLUMNS = ['label', 'n_events', 'Z_rel', 'C_rel', 'fold_enrichment',
           'fold_lo', 'fold_hi', 'q_wilcoxon']


def read(path):
    if not path.is_file():
        raise FileNotFoundError(f'Missing {path}. Run the original 12.1/12.2 '
                                'with old variant metadata, save its 12.2 output '
                                'at OLD_STATS, then rerun this script.')
    t = pd.read_csv(path, sep='\t')
    required = set(COLUMNS + ['window_bp'])
    if required - set(t.columns):
        raise ValueError(f'{path}: missing {sorted(required - set(t.columns))}')
    t = t.loc[t.window_bp == WINDOW, COLUMNS].copy()
    if t.label.duplicated().any():
        raise ValueError(f'{path}: duplicate stratum labels')
    return t.set_index('label')


def main():
    if OLD_STATS.resolve() == NEW_STATS.resolve():
        raise ValueError('OLD_STATS and NEW_STATS must be different files')
    old, new = read(OLD_STATS), read(NEW_STATS)
    labels = old.index.union(new.index, sort=False)
    out = pd.DataFrame({'label': labels})
    for c in COLUMNS[1:]:
        out[f'old_{c}'] = old[c].reindex(labels).to_numpy()
        out[f'normalized_{c}'] = new[c].reindex(labels).to_numpy()
    out['delta_n_events'] = out.normalized_n_events - out.old_n_events
    for c in ['Z_rel', 'C_rel', 'fold_enrichment']:
        out[f'delta_{c}'] = out[f'normalized_{c}'] - out[f'old_{c}']
        out[f'pct_change_{c}'] = 100 * out[f'delta_{c}'] / out[f'old_{c}'].replace(0, np.nan)
    out.insert(1, 'window_bp', WINDOW)
    out.to_csv(OUT, sep='\t', index=False, float_format='%.6g')
    print(f'Wrote {OUT}: {len(out)} strata (old/new union).')
    missing = out.loc[out.old_n_events.isna() | out.normalized_n_events.isna(), 'label'].tolist()
    if missing:
        print('WARNING: strata missing from one analysis:', missing)
    print('Z_rel and C_rel are bar heights; fold_enrichment = Z_rel / C_rel.')
    print('Bootstrap CIs and BH-adjusted q-values are shown separately for old and new.')


if True:
    main()
```

#### 13.1 · SNV-type profiles → Suppl. Fig. 2d

```python
"""13.1 - Z-DNA and matched-control profiles (+/-WINDOW_RADIUS bp) around SNVs per pyrimidine-collapsed SNV type
(CHM13, HPRC v2, normalized alleles; Suppl. Fig. 2d input). Same intervals and counting as 12.1; types in parallel.
"""
import os
import time
import multiprocessing as mp
from concurrent.futures import ProcessPoolExecutor
from pathlib import Path

import numpy as np
import pandas as pd
import pyarrow as pa
import pyarrow.compute as pc
import pyarrow.csv as pacsv

BASE = Path("/path/to/pangenomes_zdna")
MASTER = BASE / "8_CHM13_controls_MASTER.tsv"
MUT_META = BASE / "9_hprc-v2.0-mc-chm13.wave.normalized.mutations.meta.tsv.gz"
OUT = BASE / "13_zdna_snv_type_profiles_bychrom.tsv.gz"

TOL = "10%"                 # control matching tolerance: "5%" | "10%" | "15%"
DINUC_MODE = "CpG_GpC"      # "CpG_GpC" | "CpG_only"
WINDOW_RADIUS = 500
ARROW_THREADS, BLOCK_SIZE = 16, 1 << 26

CHROMS = [f"chr{i}" for i in range(1, 23)] + ["chrX", "chrY"]
OFFSETS = np.arange(-WINDOW_RADIUS, WINDOW_RADIUS + 1, dtype=np.int64)
BASES = "ACGT"
COMP = np.array([3, 2, 1, 0])               # A<->T, C<->G on BASES indices
TYPES = ["C>A", "C>G", "C>T", "T>A", "T>C", "T>G"]


def merge_intervals(s, e):
    if len(s) == 0:
        return np.empty(0, np.int64), np.empty(0, np.int64)
    o = np.argsort(s, kind="stable"); s, e = s[o], e[o]
    new = np.ones(len(s), bool); new[1:] = s[1:] > np.maximum.accumulate(e)[:-1]
    return s[new].astype(np.int64), np.maximum.reduceat(e, np.flatnonzero(new)).astype(np.int64)


def sweep_counts(anchors, iv_s, iv_e, R=WINDOW_RADIUS, chunk=200_000):
    """out[k + R] = number of (sorted) anchors a with a + k inside one of the disjoint intervals."""
    W = 2 * R + 1
    diff = np.zeros(W + 1, np.int64)
    if len(anchors) == 0 or len(iv_s) == 0:
        return diff[:W]
    for i0 in range(0, len(iv_s), chunk):
        s, e = iv_s[i0:i0 + chunk], iv_e[i0:i0 + chunk]
        lo = np.searchsorted(anchors, s - R, side="left")
        n = np.searchsorted(anchors, e - 1 + R, side="right") - lo
        tot = int(n.sum())
        if tot == 0:
            continue
        iv = np.repeat(np.arange(len(s)), n)
        a = anchors[np.arange(tot) + np.repeat(lo - (np.cumsum(n) - n), n)]
        diff += np.bincount(np.maximum(s[iv] - a, -R) + R, minlength=W + 1)
        diff -= np.bincount(np.minimum(e[iv] - 1 - a, R) + R + 1, minlength=W + 1)
    return np.cumsum(diff)[:W]


def load_intervals():
    cols = ["tol_level", "dinuc_mode", "Chromosome", "Start", "End", "Start0", "End0", "Ctrl_Start", "Ctrl_End", "has_control"]
    m = pd.read_csv(MASTER, sep="\t", usecols=cols, low_memory=False,
                    dtype={"tol_level": str, "dinuc_mode": str, "Chromosome": str, "has_control": str})
    if ((m.Start0 != m.Start) | (m.End0 != m.End + 1)).any():
        raise RuntimeError("MASTER: Start0/End0 do not follow ZSeeker Start / End + 1")
    m = m[(m.tol_level == TOL) & (m.dinuc_mode == DINUC_MODE) & (m.has_control.str.strip() == "True")].copy()
    m["Ctrl_Start"] = m.Ctrl_Start.astype(float).astype(np.int64)
    m["Ctrl_End"] = m.Ctrl_End.astype(float).astype(np.int64)
    if ((m.Ctrl_End - m.Ctrl_Start) != (m.End0 - m.Start0)).any():
        raise RuntimeError("MASTER: control length != Z-DNA length")
    Z = {c: merge_intervals(d.Start0.to_numpy(np.int64), d.End0.to_numpy(np.int64)) for c, d in m.groupby("Chromosome")}
    C = {c: merge_intervals(d.Ctrl_Start.to_numpy(np.int64), d.Ctrl_End.to_numpy(np.int64)) for c, d in m.groupby("Chromosome")}
    print(f"[intervals] {TOL} {DINUC_MODE}: {len(m):,} matched Z-DNA loci")
    return Z, C, {c: int((e - s).sum()) for c, (s, e) in Z.items()}, {c: int((e - s).sum()) for c, (s, e) in C.items()}


def load_snvs():
    pa.set_cpu_count(ARROW_THREADS)
    types = {"seqID": pa.string(), "normalized_start0": pa.int64(), "normalized_ref": pa.string(),
             "normalized_alt": pa.string(), "struct_class": pa.string()}
    reader = pacsv.open_csv(MUT_META, read_options=pacsv.ReadOptions(block_size=BLOCK_SIZE),
                            parse_options=pacsv.ParseOptions(delimiter="\t"),
                            convert_options=pacsv.ConvertOptions(include_columns=list(types), column_types=types,
                                                                 strings_can_be_null=False))
    ch, s0, code, n_snv, n_n, t0 = [], [], [], 0, 0, time.time()
    bases = pa.array(list(BASES))
    for b in reader:
        b = b.filter(pc.equal(b.column("struct_class"), "snp"))
        chi = pc.index_in(b.column("seqID"), value_set=pa.array(CHROMS))
        b, chi = b.filter(pc.is_valid(chi)), chi.filter(pc.is_valid(chi))
        r, a = pc.index_in(pc.utf8_upper(b.column("normalized_ref")), value_set=bases), \
            pc.index_in(pc.utf8_upper(b.column("normalized_alt")), value_set=bases)
        ok = pc.and_(pc.is_valid(r), pc.is_valid(a))
        n_snv += b.num_rows; n_n += b.num_rows - pc.sum(ok.cast(pa.int64())).as_py()
        r, a = (x.filter(ok).to_numpy(zero_copy_only=False).astype(np.int8) for x in (r, a))
        flip = (r == 0) | (r == 2)
        code.append((np.where(flip, COMP[r], r) * 4 + np.where(flip, COMP[a], a)).astype(np.int8))
        ch.append(chi.filter(ok).to_numpy(zero_copy_only=False).astype(np.int8))
        s0.append(b.column("normalized_start0").filter(ok).to_numpy())
    ch, s0, code = np.concatenate(ch), np.concatenate(s0), np.concatenate(code)
    o = np.lexsort((code, s0, ch)); ch, s0, code = ch[o], s0[o], code[o]
    new = np.ones(len(ch), bool)
    new[1:] = (ch[1:] != ch[:-1]) | (s0[1:] != s0[:-1]) | (code[1:] != code[:-1])
    ch, s0, code = ch[new], s0[new], code[new]
    print(f"[snvs] {n_snv:,} SNV rows | {n_n:,} with an N allele dropped | {len(ch):,} distinct SNVs "
          f"({time.time() - t0:.0f}s)")
    return ch, s0, code


def run_type(t):
    k = BASES.index(t[0]) * 4 + BASES.index(t[2])
    sel = CODE == k
    ch, anc = CH[sel], S0[sel]
    o = np.lexsort((anc, ch)); ch, anc = ch[o], anc[o]
    cuts = np.searchsorted(ch, np.arange(len(CHROMS) + 1))
    parts = []
    for ci, chrom in enumerate(CHROMS):
        a = anc[cuts[ci]:cuts[ci + 1]]
        if len(a) and chrom in Z_IV:
            parts.append(pd.DataFrame({"snv_type": t, "chrom": chrom, "offset": OFFSETS,
                                       "Z_raw": sweep_counts(a, *Z_IV[chrom]), "C_raw": sweep_counts(a, *C_IV[chrom]),
                                       "n_events": len(a), "Z_bp": ZBP[chrom], "C_bp": CBP[chrom]}))
    return pd.concat(parts)


if __name__ == "__main__":
    t_all = time.time()
    Z_IV, C_IV, ZBP, CBP = load_intervals()
    CH, S0, CODE = load_snvs()
    with ProcessPoolExecutor(max_workers=min(len(TYPES), os.cpu_count() or 1), mp_context=mp.get_context("fork")) as ex:
        out = pd.concat(list(ex.map(run_type, TYPES)), ignore_index=True)
    out.to_csv(OUT, sep="\t", index=False)
    print(f"{out.snv_type.nunique()} SNV types -> {OUT.name} [{time.time() - t_all:.0f}s]")
```

#### 13.2 · SNV-type window statistics → Suppl. Fig. 2d

```python
"""13.2 - Z-DNA vs control enrichment within +/-r bp of SNVs, per SNV type (Suppl. Fig. 2d statistics).

Relative density = mean count in +/-r / mean count over +/-R (Z-DNA and controls); fold = Z_rel / C_rel with 95%
chromosome-block bootstrap CI; paired two-sided Wilcoxon over chromosomes and bootstrap test of log(fold) = 0;
Benjamini-Hochberg across the six SNV types within each window.
"""
from pathlib import Path

import numpy as np
import pandas as pd
from scipy.stats import wilcoxon
from statsmodels.stats.multitest import multipletests

BASE = Path("/path/to/pangenomes_zdna")
BYCHROM = BASE / "13_zdna_snv_type_profiles_bychrom.tsv.gz"
OUT = BASE / "13_zdna_snv_type_window_stats.tsv"

WINDOWS = (10, 25, 50)        # half-widths (bp) of the windows tested
ALTERNATIVE = "two-sided"       # "greater" (Z-DNA > control) or "two-sided"
N_BOOTS = 50_000
MIN_CHROMS = 6
SEED = 42
CI = (2.5, 97.5)
STARS = [(1e-3, "***"), (1e-2, "**"), (5e-2, "*")]


def stars(q, positive):
    if pd.isna(q) or (ALTERNATIVE == "greater" and not positive):
        return "ns"
    return next((s for t, s in STARS if q < t), "ns")


def boot_p(boots, observed):
    null = boots - boots.mean()
    if ALTERNATIVE == "greater":
        return (np.sum(null >= observed) + 1) / (len(null) + 1)
    return (np.sum(np.abs(null) >= abs(observed)) + 1) / (len(null) + 1)


def stats_one(g, r, rng):
    Zp = g.pivot(index="chrom", columns="offset", values="Z_raw").sort_index(axis=1)
    Cp = g.pivot(index="chrom", columns="offset", values="C_raw").reindex(index=Zp.index, columns=Zp.columns)
    off = Zp.columns.to_numpy()
    meta = g.drop_duplicates("chrom").set_index("chrom").loc[Zp.index]
    Z, C = Zp.to_numpy(float), Cp.to_numpy(float)
    win, at0, nw, na = np.abs(off) <= r, off == 0, (np.abs(off) <= r).sum(), len(off)
    zw, zt, cw, ct = Z[:, win].sum(1), Z.sum(1), C[:, win].sum(1), C.sum(1)
    zin, cin = Z[:, at0].ravel(), C[:, at0].ravel()
    zbp, cbp = meta.Z_bp.to_numpy(float), meta.C_bp.to_numpy(float)
    rel = lambda w, t: (w / nw) / (t / na)
    with np.errstate(divide="ignore", invalid="ignore"):
        z_rel, c_rel = rel(zw.sum(), zt.sum()), rel(cw.sum(), ct.sum())
        fold = z_rel / c_rel
        bi = rng.integers(0, len(zw), size=(N_BOOTS, len(zw)))
        zb, cb = rel(zw[bi].sum(1), zt[bi].sum(1)), rel(cw[bi].sum(1), ct[bi].sum(1))
        lb = np.log(zb / cb)
        db = (zin[bi].sum(1) / zbp[bi].sum(1)) / (cin[bi].sum(1) / cbp[bi].sum(1))
        dens = (zin.sum() / zbp.sum()) / (cin.sum() / cbp.sum())
    q = lambda x: np.nanpercentile(x[np.isfinite(x)], CI) if np.isfinite(x).any() else (np.nan, np.nan)
    use = (zt > 0) & (ct > 0)
    zc, cc = rel(zw[use], zt[use]), rel(cw[use], ct[use])
    if use.sum() < MIN_CHROMS:
        w_stat = w_p = np.nan
    elif np.allclose(zc, cc):
        w_stat, w_p = 0.0, 1.0
    else:
        w_stat, w_p = wilcoxon(zc, cc, alternative=ALTERNATIVE)
    ok = np.isfinite(lb)
    return {"Z_rel": z_rel, "Z_rel_lo": q(zb)[0], "Z_rel_hi": q(zb)[1], "C_rel": c_rel, "C_rel_lo": q(cb)[0],
            "C_rel_hi": q(cb)[1], "fold_enrichment": fold, "fold_lo": np.exp(q(lb)[0]), "fold_hi": np.exp(q(lb)[1]),
            "p_bootstrap": boot_p(lb[ok], np.log(fold)) if ok.any() and np.isfinite(fold) else np.nan,
            "wilcoxon_stat": w_stat, "p_wilcoxon": w_p, "n_chromosomes_tested": int(use.sum()),
            "n_events": int(meta.n_events.sum()), "n_inside_Z": int(zin.sum()), "n_inside_C": int(cin.sum()),
            "Z_per_kb": 1e3 * zin.sum() / zbp.sum(), "C_per_kb": 1e3 * cin.sum() / cbp.sum(),
            "inside_density_ratio": dens, "inside_ratio_lo": q(db)[0], "inside_ratio_hi": q(db)[1],
            "Z_count_window": int(zw.sum()), "Z_count_profile": int(zt.sum()), "C_count_window": int(cw.sum()),
            "C_count_profile": int(ct.sum()), "n_offsets_window": int(nw), "n_offsets_profile": int(na)}


def main():
    d = pd.read_csv(BYCHROM, sep="\t")
    rng = np.random.default_rng(SEED)
    groups = {t: g for t, g in d.groupby("snv_type", sort=False)}
    rows = []
    for r in WINDOWS:
        tab = pd.DataFrame([{"window_bp": r, "snv_type": t, **stats_one(g, r, rng)} for t, g in groups.items()])
        for p in ("p_wilcoxon", "p_bootstrap"):
            m = tab[p].notna()
            tab[p.replace("p_", "q_")] = np.nan
            tab.loc[m, p.replace("p_", "q_")] = multipletests(tab.loc[m, p], method="fdr_bh")[1]
        pos = tab.fold_enrichment > 1
        tab["sig_wilcoxon"] = [stars(q, s) for q, s in zip(tab.q_wilcoxon, pos)]
        tab["sig_bootstrap"] = [stars(q, s) for q, s in zip(tab.q_bootstrap, pos)]
        rows.append(tab)
    out = pd.concat(rows, ignore_index=True)
    out.to_csv(OUT, sep="\t", index=False, float_format="%.6g")
    t = out[out.window_bp == WINDOWS[0]]
    print(f"+/-{WINDOWS[0]} bp:\n" + t[["snv_type", "n_events", "fold_enrichment", "q_wilcoxon", "sig_wilcoxon"]]
          .to_string(index=False, float_format=lambda x: f"{x:.3g}"))
    print(f"-> {OUT.name}")


if __name__ == "__main__":
    main()
```

#### 14.1 · denovo-db parsing → Fig. 3c

```python
"""14.1 - Parse and normalize denovo-db (non-SSC, GRCh37) de novo variants from whole-genome studies (Fig. 3c input).

Each ALT allele is trimmed and left-aligned against GRCh37, then classified with the CHM13 parser rules (9_).
"""
import gzip
import json
from collections import Counter
from pathlib import Path

import pandas as pd

BASE = Path("/path/to/pangenomes_zdna")
VCF = BASE / "denovo_db/denovo-db.non-ssc-samples.variants.vcf.gz"
REFERENCE_FASTA = BASE / "tmp/grch37/grch37_refseq/GCF_000001405.13_GRCh37_genomic_only_primary.fna"     # GRCh37/hg19 FASTA (+ .fai); missing -> no left-alignment / REF check
OUT = BASE / "14_denovo_hg19_normalized.tsv.gz"

SEQUENCE_TYPES = {"genome"}    # STUDIES SequenceType values kept ("exome" / "targeted" excluded)
PHENOTYPES = None              # None = all; or e.g. {"control"} / {"autism", "developmental_disorder"}
LEFT_ALIGN = True
CHROMS = [f"chr{i}" for i in range(1, 23)] + ["chrX", "chrY"]


def raw_class(ref, alt):
    if len(ref) == len(alt) == 1:
        return "snp"
    if len(ref) == len(alt):
        return "mnp" if len(ref) <= 20 else "substitution"
    if len(ref) < len(alt):
        return "ins" if len(ref) == 1 else "complex"
    return "del" if len(alt) == 1 else "complex"


def trim(pos0, ref, alt):
    while ref and alt and ref[-1] == alt[-1]:
        ref, alt = ref[:-1], alt[:-1]
    while ref and alt and ref[0] == alt[0]:
        ref, alt, pos0 = ref[1:], alt[1:], pos0 + 1
    if not ref and not alt:
        raise RuntimeError("REF == ALT after trimming")
    return pos0, ref, alt


def left_align(pos0, ref, alt, base_at):
    shift, seq = 0, ref or alt
    while pos0 > 0 and base_at(pos0 - 1) == seq[-1]:
        seq, pos0, shift = base_at(pos0 - 1) + seq[:-1], pos0 - 1, shift + 1
    return (pos0, seq, "", shift) if ref else (pos0, "", seq, shift)


def classify(ref, alt):
    r, a = len(ref), len(alt)
    if r == a == 1:
        sc = "snp"
    elif r == a:
        sc = "mnp" if r <= 20 else "substitution"
    elif a == 0:
        sc = "del"
    elif r == 0:
        sc = "ins"
    else:
        sc = "complex"
    dom = sc if sc != "complex" else ("del_like" if r > a else "ins_like")
    size = a if sc == "ins" or dom == "ins_like" else r
    return sc, dom, r, a, size


def main():
    fa = None
    if REFERENCE_FASTA.exists():
        from pyfaidx import Fasta
        fa = Fasta(str(REFERENCE_FASTA), as_raw=True, sequence_always_upper=True)
    else:
        print(f"[WARN] {REFERENCE_FASTA} not found: trimming only (no left-alignment, no REF check)")
    fa_name = lambda c: c if fa is not None and c in fa else c.replace("chr", "")
    rows, skipped, n_lines = [], Counter(), 0
    with gzip.open(VCF, "rt") as f:
        for line in f:
            if line[0] == "#":
                continue
            n_lines += 1
            chrom, pos, _, ref, alts, _, _, info = line.rstrip("\n").split("\t", 8)[:8]
            if "SequenceType" not in info:
                skipped["no_studies"] += 1
                continue
            studies = next((json.loads(x[8:]) for x in info.split(";") if x.startswith("STUDIES=")), [])
            sel = [s for s in studies if s.get("SequenceType") in SEQUENCE_TYPES
                   and (PHENOTYPES is None or s.get("PrimaryPhenotype") in PHENOTYPES)]
            if not sel:
                skipped["sequence_type_or_phenotype"] += 1
                continue
            chrom = chrom if chrom.startswith("chr") else "chr" + chrom
            if chrom not in CHROMS:
                skipped["other_contig"] += 1
                continue
            ref, pos0 = ref.upper(), int(pos) - 1
            if set(ref) - set("ACGTN"):
                skipped["non_sequence_ref"] += 1
                continue
            ref_match = pd.NA
            if fa is not None:
                ref_match = fa[fa_name(chrom)][pos0:pos0 + len(ref)] == ref
                base_at = lambda p, c=fa_name(chrom): fa[c][p:p + 1]
            phen = ",".join(sorted({str(s.get("PrimaryPhenotype", "")) for s in sel}))
            for alt in alts.upper().split(","):
                if alt in (".", "*") or set(alt) - set("ACGTN"):
                    skipped["non_sequence_alt"] += 1
                    continue
                s0, nr, na = trim(pos0, ref, alt)
                shift = 0
                if LEFT_ALIGN and fa is not None and (not nr or not na):
                    s0, nr, na, shift = left_align(s0, nr, na, base_at)
                sc, dom, rs, as_, size = classify(nr, na)
                rc = raw_class(ref, alt)
                rows.append((chrom, int(pos), ref, alt, s0, nr, na, sc, dom, rs, as_, size, rc, rc != sc, shift, ref_match,
                             len(sel), phen))
    d = pd.DataFrame(rows, columns=["chrom", "pos1", "ref", "alt", "normalized_start0", "normalized_ref", "normalized_alt",
                                    "struct_class", "dominant_op", "ref_span", "alt_span", "event_size", "raw_struct_class",
                                    "classification_changed", "left_shift_bp", "ref_match", "n_entries", "phenotypes"])
    n0 = len(d)
    d = (d.groupby(["chrom", "normalized_start0", "normalized_ref", "normalized_alt"], sort=False, as_index=False)
          .agg({**{c: "first" for c in d.columns if c not in ("chrom", "normalized_start0", "normalized_ref", "normalized_alt",
                                                                 "n_entries")}, "n_entries": "sum"}))
    d.to_csv(OUT, sep="\t", index=False)
    print(f"[vcf] {n_lines:,} records; skipped: {dict(skipped)}")
    print(f"[alleles] {n0:,} ALT alleles -> {len(d):,} distinct normalized alleles")
    print(f"[normalization] class changed: {int(d.classification_changed.sum()):,} ({100 * d.classification_changed.mean():.2f}%); "
          f"left-shifted indels: {int((d.left_shift_bp > 0).sum()):,}; REF mismatches vs FASTA: "
          f"{int((d.ref_match == False).sum()) if fa is not None else 'not checked'}")
    print(f"-> {OUT.name}")


if __name__ == "__main__":
    main()
```

#### 14.2 · De novo profiles and rate ratios → Fig. 3c

```python
"""14.2 - De novo variants (denovo-db WGS, GRCh37) at Z-DNA vs matched controls (Fig. 3c).

Metaprofiles of relative density around anchors (+/-WINDOW_RADIUS bp, chromosome-bootstrap bands) and rate ratios
(n_Z / bp_Z) / (n_C / bp_C) within intervals +/- FLANK bp: two-sided chromosome-block bootstrap, BH over classes.
"""
from pathlib import Path

import numpy as np
import pandas as pd
from scipy.stats import binomtest
from statsmodels.stats.multitest import multipletests

BASE = Path("/path/to/pangenomes_zdna")
MASTER = BASE / "8_GRCh37_controls_MASTER.tsv"
VARIANTS = BASE / "14_denovo_hg19_normalized.tsv.gz"
OUT_PROF = BASE / "14_denovo_zdna_profiles.tsv.gz"
OUT_RATES = BASE / "14_denovo_zdna_rates.tsv"
OUT_CHROM = BASE / "14_denovo_zdna_rates_bychrom.tsv"

TOL = "10%"                 # control matching tolerance: "5%" | "10%" | "15%"
DINUC_MODE = "CpG_GpC"      # "CpG_GpC" | "CpG_only"
DROP_REF_MISMATCH = True    # drop alleles whose VCF REF disagrees with the GRCh37 FASTA (19_1 ref_match == False)
FLANKS = (0, 10, 50)        # bp added on each side of Z-DNA / control intervals for the rate ratio
WINDOW_RADIUS = 500
MIN_EVENTS_PROFILE = 200    # classes with fewer variants get no metaprofile
MIN_EVENTS_RATE = 10        # rate ratios need >= this many variants in Z + control
N_BOOTS_PROFILE = 1000
N_BOOTS = 10000
SEED = 42
CLASSES = ["all", "SNV", "MNV", "small_del", "small_ins", "large_indel", "substitution", "complex_del", "complex_ins"]
CHROMS = [f"chr{i}" for i in range(1, 23)] + ["chrX", "chrY"]
OFFSETS = np.arange(-WINDOW_RADIUS, WINDOW_RADIUS + 1)


def merge(s, e):
    if len(s) == 0:
        return np.empty(0, np.int64), np.empty(0, np.int64)
    o = np.argsort(s, kind="stable"); s, e = s[o], e[o]
    new = np.ones(len(s), bool); new[1:] = s[1:] > np.maximum.accumulate(e)[:-1]
    return s[new].astype(np.int64), np.maximum.reduceat(e, np.flatnonzero(new)).astype(np.int64)


def inside(p, iv):
    s, e = iv
    if len(s) == 0:
        return np.zeros(len(p), bool)
    i = np.searchsorted(s, p, side="right") - 1
    return (i >= 0) & (p < e[np.maximum(i, 0)])


def sweep_counts(anchors, iv_s, iv_e, R=WINDOW_RADIUS):
    """out[k + R] = number of (sorted) anchors a with a + k inside one of the disjoint intervals."""
    W = 2 * R + 1
    diff = np.zeros(W + 1, np.int64)
    if len(anchors) == 0 or len(iv_s) == 0:
        return diff[:W]
    lo = np.searchsorted(anchors, iv_s - R, side="left")
    n = np.searchsorted(anchors, iv_e - 1 + R, side="right") - lo
    tot = int(n.sum())
    if tot:
        iv = np.repeat(np.arange(len(iv_s)), n)
        a = anchors[np.arange(tot) + np.repeat(lo - (np.cumsum(n) - n), n)]
        diff += np.bincount(np.maximum(iv_s[iv] - a, -R) + R, minlength=W + 1)
        diff -= np.bincount(np.minimum(iv_e[iv] - 1 - a, R) + R + 1, minlength=W + 1)
    return np.cumsum(diff)[:W]


def load_loci():
    cols = ["tol_level", "dinuc_mode", "Chromosome", "Start", "End", "Start0", "End0", "Ctrl_Start", "Ctrl_End", "has_control"]
    m = pd.read_csv(MASTER, sep="\t", usecols=cols, low_memory=False,
                    dtype={"tol_level": str, "dinuc_mode": str, "Chromosome": str, "has_control": str})
    if ((m.Start0 != m.Start) | (m.End0 != m.End + 1)).any():
        raise RuntimeError("MASTER: Start0/End0 do not follow ZSeeker Start / End + 1")
    m = m[(m.tol_level == TOL) & (m.dinuc_mode == DINUC_MODE) & (m.has_control.str.strip() == "True")].copy()
    m["Chromosome"] = np.where(m.Chromosome.str.startswith("chr"), m.Chromosome, "chr" + m.Chromosome)
    m = m[m.Chromosome.isin(CHROMS)].drop_duplicates(["Chromosome", "Start0", "End0"])
    m["Ctrl_Start"] = m.Ctrl_Start.astype(float).astype(np.int64)
    m["Ctrl_End"] = m.Ctrl_End.astype(float).astype(np.int64)
    print(f"[loci] {TOL} {DINUC_MODE}: {len(m):,} matched Z-DNA loci, Z-DNA {int((m.End0 - m.Start0).sum()):,} bp")
    return {c: g for c, g in m.groupby("Chromosome")}


def load_variants():
    v = pd.read_csv(VARIANTS, sep="\t", dtype={"chrom": str})
    if DROP_REF_MISMATCH and "ref_match" in v and v.ref_match.notna().any():
        print(f"[variants] dropping {int((v.ref_match.astype(str) == 'False').sum()):,} REF mismatches")
        v = v[v.ref_match.astype(str) != "False"]
    sc, dom, L = v.struct_class, v.dominant_op, v.event_size
    cdel = (sc == "complex") & (dom == "del_like")
    v["cls"] = np.select([sc == "snp", sc == "mnp", (sc == "del") & (L <= 50), (sc == "ins") & (L <= 50),
                          sc.isin(["del", "ins"]) & (L > 50), sc == "substitution", cdel,
                          (sc == "complex") & (dom == "ins_like")], CLASSES[1:], "other")
    mid = (sc == "mnp") | ((sc.isin(["substitution", "del"]) | cdel) & (L <= 50))
    v["anchor"] = np.where(mid, v.normalized_start0 + (v.ref_span - 1) // 2, v.normalized_start0).astype(np.int64)
    v = v[v.chrom.isin(CHROMS)]
    print("[variants] " + ", ".join(f"{c}: {n:,}" for c, n in v.cls.value_counts().reindex(CLASSES[1:] + ["other"]).fillna(0)
                                    .astype(int).items()))
    return v


def main():
    rng = np.random.default_rng(SEED)
    loci = load_loci()
    v = load_variants()
    chroms = [c for c in CHROMS if c in loci]
    nch = len(chroms)
    W = np.zeros((N_BOOTS, nch))
    np.add.at(W, (np.repeat(np.arange(N_BOOTS), nch), rng.integers(0, nch, N_BOOTS * nch)), 1)
    sel = {c: (v.cls != "other") if c == "all" else (v.cls == c) for c in CLASSES}

    Z0 = {c: merge(g.Start0.to_numpy(np.int64), g.End0.to_numpy(np.int64)) for c, g in loci.items()}
    C0 = {c: merge(g.Ctrl_Start.to_numpy(np.int64), g.Ctrl_End.to_numpy(np.int64)) for c, g in loci.items()}
    profs = []
    for cls in CLASSES:
        d = v[sel[cls]]
        if len(d) < MIN_EVENTS_PROFILE:
            continue
        Zm = np.zeros((nch, len(OFFSETS))); Cm = np.zeros((nch, len(OFFSETS)))
        for k, c in enumerate(chroms):
            a = np.sort(d.anchor[d.chrom == c].to_numpy())
            Zm[k], Cm[k] = sweep_counts(a, *Z0[c]), sweep_counts(a, *C0[c])
        bi = rng.integers(0, nch, size=(N_BOOTS_PROFILE, nch))
        with np.errstate(divide="ignore", invalid="ignore"):
            zb, cb = Zm[bi].sum(1), Cm[bi].sum(1)
            zb /= zb.mean(1, keepdims=True); cb /= cb.mean(1, keepdims=True)
        Zt, Ct = Zm.sum(0), Cm.sum(0)
        profs.append(pd.DataFrame({"class": cls, "offset": OFFSETS, "Z_mean": Zt / Zt.mean(),
                                   "Z_lo": np.nanpercentile(zb, 2.5, 0), "Z_hi": np.nanpercentile(zb, 97.5, 0),
                                   "C_mean": Ct / Ct.mean(), "C_lo": np.nanpercentile(cb, 2.5, 0),
                                   "C_hi": np.nanpercentile(cb, 97.5, 0), "Z_raw": Zt.astype(int), "C_raw": Ct.astype(int),
                                   "n_events": len(d), "n_chrom": nch}))
    pd.concat(profs).to_csv(OUT_PROF, sep="\t", index=False, float_format="%.6g")

    def rr(nz, nc, bz, bc):
        with np.errstate(divide="ignore", invalid="ignore"):
            r = np.asarray(nz * bc / (nc * bz), float)
        return np.where(np.isfinite(r), r, np.nan)

    rows, bych = [], []
    for flank in FLANKS:
        ext = lambda s, e: merge(np.maximum(s - flank, 0), e + flank)
        Zf = {c: ext(g.Start0.to_numpy(np.int64), g.End0.to_numpy(np.int64)) for c, g in loci.items()}
        Cf = {c: ext(g.Ctrl_Start.to_numpy(np.int64), g.Ctrl_End.to_numpy(np.int64)) for c, g in loci.items()}
        bz = np.array([(Zf[c][1] - Zf[c][0]).sum() for c in chroms], float)
        bc = np.array([(Cf[c][1] - Cf[c][0]).sum() for c in chroms], float)
        for cls in CLASSES:
            d = v[sel[cls]]
            nz, nc = np.zeros(nch), np.zeros(nch)
            for k, c in enumerate(chroms):
                a = d.anchor[d.chrom == c].to_numpy()
                nz[k], nc[k] = inside(a, Zf[c]).sum(), inside(a, Cf[c]).sum()
            NZ, NC, BZ, BC = nz.sum(), nc.sum(), bz.sum(), bc.sum()
            ok = NZ + NC >= MIN_EVENTS_RATE
            est = float(rr(NZ, NC, BZ, BC)) if ok else np.nan
            bt = np.log(rr(W @ nz, W @ nc, W @ bz, W @ bc)) if ok else np.full(N_BOOTS, np.nan)
            bt = bt[np.isfinite(bt)]
            rows.append(dict(flank=flank, **{"class": cls}, n_events=len(d), n_Z=int(NZ), n_C=int(NC), bp_Z=int(BZ), bp_C=int(BC),
                             rate_Z_per_Mb=1e6 * NZ / BZ, rate_C_per_Mb=1e6 * NC / BC, rate_ratio=est,
                             ci_lo=np.exp(np.percentile(bt, 2.5)) if len(bt) else np.nan,
                             ci_hi=np.exp(np.percentile(bt, 97.5)) if len(bt) else np.nan,
                             p=min(1.0, 2 * (min((bt <= 0).sum(), (bt >= 0).sum()) + 1) / (len(bt) + 1)) if ok and len(bt) else np.nan,
                             p_binomial=binomtest(int(NZ), int(NZ + NC), BZ / (BZ + BC)).pvalue if ok else np.nan,
                             n_chrom=nch, n_boots=N_BOOTS))
            for k, c in enumerate(chroms):
                bych.append(dict(flank=flank, **{"class": cls}, chrom=c, n_Z=int(nz[k]), n_C=int(nc[k]), bp_Z=int(bz[k]),
                                 bp_C=int(bc[k]), rate_ratio=float(rr(nz[k], nc[k], bz[k], bc[k]))))
    out = pd.DataFrame(rows)
    out["q"] = np.nan
    for f in FLANKS:
        m = (out.flank == f) & (out["class"] != "all") & out.p.notna()
        out.loc[m, "q"] = multipletests(out.p[m], method="fdr_bh")[1]
    out.loc[out["class"] == "all", "q"] = out.loc[out["class"] == "all", "p"]
    out["sig"] = pd.cut(out.q, [-1, 1e-3, 1e-2, 0.05, 2], labels=["***", "**", "*", "ns"]).astype(str).replace("nan", "")
    out["test"] = "two-sided chromosome-block bootstrap (log rate ratio); BH over classes within flank"
    out.to_csv(OUT_RATES, sep="\t", index=False, float_format="%.6g")
    pd.DataFrame(bych).to_csv(OUT_CHROM, sep="\t", index=False, float_format="%.6g")

    t = out[(out.flank == 10) & out["class"].isin(["SNV", "small_del", "small_ins"])]
    print(t[["class", "n_events", "rate_ratio", "ci_lo", "ci_hi", "q"]].to_string(index=False, float_format=lambda x: f"{x:.3g}"))
    print(f"-> {OUT_PROF.name}, {OUT_RATES.name}, {OUT_CHROM.name}")


if __name__ == "__main__":
    main()
```

#### 15.1 · Z-DNA-centred profiles → Fig. 3d

```python
"""15.1 - Variant occurrence across Z-DNA intervals (left flank | length-scaled body | right flank) vs matched controls
(CHM13, HPRC v2; Fig. 3d input).

Flanks FLANK bp at 1-bp resolution; body rescaled to N_BODY_BINS (= shortest interval). Frequency = count / intervals,
95% chromosome-block bootstrap CI. Strata as in 12.1 plus SNV types; strata run in parallel.
"""
import os
import time
import multiprocessing as mp
from concurrent.futures import ProcessPoolExecutor, as_completed
from pathlib import Path

import numpy as np
import pandas as pd
import pyarrow as pa
import pyarrow.compute as pc
import pyarrow.csv as pacsv

BASE = Path("/path/to/pangenomes_zdna")
MASTER = BASE / "8_CHM13_controls_MASTER.tsv"
MUT_META = BASE / "9_hprc-v2.0-mc-chm13.wave.normalized.mutations.meta.tsv.gz"
OUT_PROF = BASE / "15_zdna_centred_profiles.tsv.gz"
OUT_CHR = BASE / "15_zdna_centred_profiles_bychrom.tsv.gz"

TOL = "10%"                 # control matching tolerance: "5%" | "10%" | "15%"
DINUC_MODE = "CpG_GpC"      # "CpG_GpC" | "CpG_only"
DEDUP_NORMALIZED = True     # count each distinct normalized allele once (False = once per VCF record)
FLANK = 100                 # flank (bp) on each side
N_BOOTS = 10_000
CI_LO, CI_HI = 2.5, 97.5
RNG_SEED = 42
MIN_MUTS = 500              # strata with fewer variants are skipped
ARROW_THREADS, BLOCK_SIZE = 16, 1 << 26

CHROMS = [f"chr{i}" for i in range(1, 23)] + ["chrX", "chrY"]
SNAMES = ["SNP", "MNP_2_20",
          "DEL_1bp", "DEL_2_5bp", "DEL_6_50bp", "DEL_gt50bp",
          "INS_1bp", "INS_2_5bp", "INS_6_50bp", "INS_51_200bp", "INS_gt200bp",
          "CDEL_2_5bp", "CDEL_6_50bp", "CDEL_gt50bp",
          "CINS_2_5bp", "CINS_6_50bp", "CINS_51_200bp", "CINS_gt200bp",
          "SUB_21_50", "SUB_51_200", "SUB_gt200"]
SCODE = {s: i for i, s in enumerate(SNAMES)}
SNV_TYPES = ["C>A", "C>G", "C>T", "T>A", "T>C", "T>G"]
STRUCT = ["snp", "mnp", "substitution", "del", "ins", "complex"]
DOM = ["snp", "mnp", "substitution", "del", "ins", "del_like", "ins_like", "balanced"]
BASES = "ACGT"
COMP = np.array([3, 2, 1, 0])


def load_intervals():
    cols = ["tol_level", "dinuc_mode", "Chromosome", "Start", "End", "Start0", "End0", "Ctrl_Start", "Ctrl_End", "has_control"]
    m = pd.read_csv(MASTER, sep="\t", usecols=cols, low_memory=False,
                    dtype={"tol_level": str, "dinuc_mode": str, "Chromosome": str, "has_control": str})
    if ((m.Start0 != m.Start) | (m.End0 != m.End + 1)).any():
        raise RuntimeError("MASTER: Start0/End0 do not follow ZSeeker Start / End + 1")
    m = m[(m.tol_level == TOL) & (m.dinuc_mode == DINUC_MODE) & (m.has_control.str.strip() == "True")]
    m = m.drop_duplicates(["Chromosome", "Start0", "End0"]).copy()
    m["Ctrl_Start"] = m.Ctrl_Start.astype(float).astype(np.int64)
    m["Ctrl_End"] = m.Ctrl_End.astype(float).astype(np.int64)
    Z = {c: d[["Start0", "End0"]].to_numpy(np.int64) for c, d in m.groupby("Chromosome")}
    C = {c: d[["Ctrl_Start", "Ctrl_End"]].to_numpy(np.int64) for c, d in m.groupby("Chromosome")}
    lens = np.concatenate([v[:, 1] - v[:, 0] for v in list(Z.values()) + list(C.values())])
    n_body = int(lens[lens > 0].min())
    print(f"[intervals] {TOL} {DINUC_MODE}: {len(m):,} Z-DNA / control pairs | N_BODY_BINS = {n_body}")
    return Z, C, n_body


def load_variants():
    pa.set_cpu_count(ARROW_THREADS)
    types = {"seqID": pa.string(), "start": pa.int64(), "normalized_start0": pa.int64(), "normalized_ref": pa.string(),
             "normalized_alt": pa.string(), "struct_class": pa.string(), "dominant_op": pa.string(),
             "ref_span": pa.int64(), "event_size": pa.int64()}
    reader = pacsv.open_csv(MUT_META, read_options=pacsv.ReadOptions(block_size=BLOCK_SIZE),
                            parse_options=pacsv.ParseOptions(delimiter="\t"),
                            convert_options=pacsv.ConvertOptions(include_columns=list(types), column_types=types,
                                                                 strings_can_be_null=False))
    parts = {k: [] for k in ("ch", "pos", "s0", "rs", "L", "sc", "dom", "snv", "h")}
    bases, n_raw, t0 = pa.array(list(BASES)), 0, time.time()
    for b in reader:
        n_raw += b.num_rows
        chi = pc.index_in(b.column("seqID"), value_set=pa.array(CHROMS))
        b, chi = b.filter(pc.is_valid(chi)), chi.filter(pc.is_valid(chi))
        if b.num_rows == 0:
            continue
        sc = pc.index_in(b.column("struct_class"), value_set=pa.array(STRUCT))
        dom = pc.index_in(b.column("dominant_op"), value_set=pa.array(DOM))
        if sc.null_count or dom.null_count:
            raise RuntimeError("meta: unexpected struct_class / dominant_op values")
        sc = sc.to_numpy(zero_copy_only=False).astype(np.int8)
        r = pc.index_in(pc.utf8_upper(b.column("normalized_ref")), value_set=bases).fill_null(-1).to_numpy().astype(np.int8)
        a = pc.index_in(pc.utf8_upper(b.column("normalized_alt")), value_set=bases).fill_null(-1).to_numpy().astype(np.int8)
        ok = (sc == STRUCT.index("snp")) & (r >= 0) & (a >= 0)
        flip = (r == 0) | (r == 2)
        snv = np.where(ok, np.where(flip, COMP[np.clip(r, 0, 3)], r) * 4 + np.where(flip, COMP[np.clip(a, 0, 3)], a), -1)
        for k, v in (("ch", chi.to_numpy(zero_copy_only=False).astype(np.int8)), ("pos", b.column("start").to_numpy()),
                     ("s0", b.column("normalized_start0").to_numpy()), ("rs", b.column("ref_span").to_numpy()),
                     ("L", b.column("event_size").to_numpy().astype(np.int32)), ("sc", sc),
                     ("dom", dom.to_numpy(zero_copy_only=False).astype(np.int8)), ("snv", snv.astype(np.int8)),
                     ("h", pd.util.hash_array(b.column("normalized_alt").to_numpy(zero_copy_only=False)))):
            parts[k].append(v)
    v = {k: np.concatenate(x) for k, x in parts.items()}
    n = len(v["ch"])
    if DEDUP_NORMALIZED:
        o = np.lexsort((v["h"], v["rs"], v["s0"], v["ch"]))
        v = {k: x[o] for k, x in v.items()}
        new = np.ones(n, bool)
        new[1:] = ((v["ch"][1:] != v["ch"][:-1]) | (v["s0"][1:] != v["s0"][:-1]) | (v["rs"][1:] != v["rs"][:-1])
                   | (v["h"][1:] != v["h"][:-1]))
        v = {k: x[new] for k, x in v.items()}
    print(f"[variants] {n_raw:,} ALT rows | {n:,} on primary chromosomes | using {len(v['ch']):,} ({time.time() - t0:.0f}s)")
    return v


def assign_strata(v):
    sc, dom, L = v["sc"], v["dom"], v["L"]
    is_ = lambda x: sc == STRUCT.index(x)
    cdel, cins = is_("complex") & (dom == DOM.index("del_like")), is_("complex") & (dom == DOM.index("ins_like"))
    st = np.full(len(sc), -1, np.int16)
    for name, mask in [("SNP", is_("snp")), ("MNP_2_20", is_("mnp")),
                       ("SUB_21_50", is_("substitution") & (L <= 50)),
                       ("SUB_51_200", is_("substitution") & (L > 50) & (L <= 200)), ("SUB_gt200", is_("substitution") & (L > 200)),
                       ("DEL_1bp", is_("del") & (L == 1)), ("DEL_2_5bp", is_("del") & (L >= 2) & (L <= 5)),
                       ("DEL_6_50bp", is_("del") & (L >= 6) & (L <= 50)), ("DEL_gt50bp", is_("del") & (L > 50)),
                       ("INS_1bp", is_("ins") & (L == 1)), ("INS_2_5bp", is_("ins") & (L >= 2) & (L <= 5)),
                       ("INS_6_50bp", is_("ins") & (L >= 6) & (L <= 50)), ("INS_51_200bp", is_("ins") & (L >= 51) & (L <= 200)),
                       ("INS_gt200bp", is_("ins") & (L > 200)),
                       ("CDEL_2_5bp", cdel & (L <= 5)), ("CDEL_6_50bp", cdel & (L >= 6) & (L <= 50)),
                       ("CDEL_gt50bp", cdel & (L > 50)),
                       ("CINS_2_5bp", cins & (L <= 5)), ("CINS_6_50bp", cins & (L >= 6) & (L <= 50)),
                       ("CINS_51_200bp", cins & (L >= 51) & (L <= 200)), ("CINS_gt200bp", cins & (L > 200))]:
        st[mask] = SCODE[name]
    return st


def count_flanks_and_body(iv, pos, flank, nb):
    """Counts of the sorted positions `pos` in [left flank | nb body bins | right flank] summed over intervals `iv`."""
    out = np.zeros(2 * flank + nb, np.int64)
    if len(pos) == 0 or len(iv) == 0:
        return out
    s, e = iv[:, 0], iv[:, 1]
    lo = np.searchsorted(pos, s - flank, side="left")
    n = np.searchsorted(pos, e + flank, side="left") - lo
    tot = int(n.sum())
    if tot == 0:
        return out
    j = np.repeat(np.arange(len(s)), n)
    p = pos[np.arange(tot) + np.repeat(lo - (np.cumsum(n) - n), n)]
    s, e = s[j], e[j]
    left, right = p < s, p >= e
    body = ~left & ~right
    idx = np.empty(tot, np.int64)
    idx[left] = p[left] - (s[left] - flank)
    idx[body] = flank + np.minimum(np.floor((p[body] - s[body]) / (e[body] - s[body]) * nb).astype(np.int64), nb - 1)
    idx[right] = flank + nb + (p[right] - e[right])
    return np.bincount(idx, minlength=len(out))


def bootstrap_frequency(mat, nvec):
    rng = np.random.default_rng(RNG_SEED)
    idx = rng.integers(0, len(nvec), size=(N_BOOTS, len(nvec)))
    with np.errstate(divide="ignore", invalid="ignore"):
        boots = mat[idx].sum(1) / nvec[idx].sum(1)[:, None]
    return np.nanpercentile(boots, CI_LO, axis=0), np.nanpercentile(boots, CI_HI, axis=0)


def run_stratum(label, sel):
    ch, pos = CH[sel], POS[sel]
    o = np.lexsort((pos, ch)); ch, pos = ch[o], pos[o]
    cuts = np.searchsorted(ch, np.arange(len(CHROMS) + 1))
    W = 2 * FLANK + NB
    region = np.array(["left_flank"] * FLANK + ["body"] * NB + ["right_flank"] * FLANK)
    position = np.r_[np.arange(-FLANK, 0), np.arange(NB), np.arange(FLANK)]
    prof, bych = [], []
    for track, IV in (("Z", Z_IV), ("C", C_IV)):
        chroms, mats, ns = [], [], []
        for ci, chrom in enumerate(CHROMS):
            if chrom in IV and len(IV[chrom]):
                chroms.append(chrom); ns.append(len(IV[chrom]))
                mats.append(count_flanks_and_body(IV[chrom], pos[cuts[ci]:cuts[ci + 1]], FLANK, NB))
        mat, nvec = np.stack(mats).astype(float), np.array(ns, float)
        total = mat.sum(0)
        lo, hi = bootstrap_frequency(mat, nvec)
        prof.append(pd.DataFrame({"stratum": label, "track": track, "region": region, "position": position,
                                  "window_idx": np.arange(W), "raw_count": total.astype(np.int64),
                                  "frequency": total / nvec.sum(),
                                  "rel_density": total / total.mean() if total.mean() > 0 else np.nan,
                                  "ci_lo": lo, "ci_hi": hi, "n_intervals": int(nvec.sum()), "n_variants": int(sel.sum())}))
        bych.append(pd.DataFrame({"stratum": label, "track": track, "chrom": np.repeat(chroms, W),
                                  "window_idx": np.tile(np.arange(W), len(chroms)),
                                  "raw_count": np.stack(mats).ravel(), "n_intervals": np.repeat(ns, W)}))
    return pd.concat(prof), pd.concat(bych)


if __name__ == "__main__":
    t_all = time.time()
    Z_IV, C_IV, NB = load_intervals()
    v = load_variants()
    ST, SNV = assign_strata(v), v["snv"]
    CH, POS = v["ch"].astype(np.int16), v["pos"].astype(np.int64)
    del v
    jobs = [(s, ST == SCODE[s]) for s in SNAMES]
    jobs += [("SNP_" + t.replace(">", "to"), SNV == BASES.index(t[0]) * 4 + BASES.index(t[2])) for t in SNV_TYPES]
    skipped = [l for l, m in jobs if m.sum() < MIN_MUTS]
    jobs = [(l, m) for l, m in jobs if m.sum() >= MIN_MUTS]
    print(f"[strata] {len(jobs)} profiles; skipped (< {MIN_MUTS} variants): {skipped}")
    order = {l: i for i, (l, _) in enumerate(jobs)}
    profs, bychs = [], []
    with ProcessPoolExecutor(max_workers=min(len(jobs), os.cpu_count() or 1), mp_context=mp.get_context("fork")) as ex:
        futs = {ex.submit(run_stratum, l, m): l for l, m in jobs}
        for f in as_completed(futs):
            p, b = f.result()
            profs.append(p); bychs.append(b)
    srt = lambda d, k: d.assign(_o=d.stratum.map(order)).sort_values(["_o"] + k).drop(columns="_o")
    srt(pd.concat(profs), ["track", "window_idx"]).to_csv(OUT_PROF, sep="\t", index=False, float_format="%.6g")
    srt(pd.concat(bychs), ["track", "chrom", "window_idx"]).to_csv(OUT_CHR, sep="\t", index=False)
    print(f"-> {OUT_PROF.name}, {OUT_CHR.name} [{time.time() - t_all:.0f}s]")
```

#### 15.2 · Z-DNA-centred summary → Fig. 3d

```python
"""15.2 - Summary metrics of the Z-DNA-centred profiles for the Fig. 3d strata: body fold enrichment (Z-DNA vs control,
chromosome bootstrap CI, paired two-sided Wilcoxon, BH), boundary vs interior and flank log2 ratios (Fig. 3d statistics).
"""
from pathlib import Path

import numpy as np
import pandas as pd
from scipy.stats import wilcoxon
from statsmodels.stats.multitest import multipletests

BASE = Path("/path/to/pangenomes_zdna")
PROF = BASE / "15_zdna_centred_profiles.tsv.gz"
BYCHROM = BASE / "15_zdna_centred_profiles_bychrom.tsv.gz"
OUT = BASE / "15_zdna_centred_summary_metrics.tsv"

STRATA = ["SNP", "MNP_2_20", "DEL_6_50bp", "INS_6_50bp", "CDEL_6_50bp", "CINS_6_50bp"]
N_BOOTS = 50_000
SEED = 42
MIN_CHROMS = 6


def chromosome_body_test(bc, n_flank, n_body):
    z_freqs, c_freqs = [], []
    for chrom in sorted(set(bc.chrom)):
        z = bc[(bc.track == "Z") & (bc.chrom == chrom)].sort_values("window_idx")
        c = bc[(bc.track == "C") & (bc.chrom == chrom)].sort_values("window_idx")
        if len(z) == 0 or len(c) == 0:
            continue
        z_n, c_n = z.n_intervals.iloc[0], c.n_intervals.iloc[0]
        if z_n == 0 or c_n == 0:
            continue
        z_freqs.append(z.raw_count.to_numpy()[n_flank:n_flank + n_body].mean() / z_n)
        c_freqs.append(c.raw_count.to_numpy()[n_flank:n_flank + n_body].mean() / c_n)
    z_freqs, c_freqs = np.array(z_freqs), np.array(c_freqs)
    n = len(z_freqs)
    if n < MIN_CHROMS:
        return np.nan, np.nan, np.nan, np.nan
    try:
        _, p_wilc = wilcoxon(z_freqs, c_freqs, alternative="two-sided", zero_method="wilcox")
    except ValueError:
        p_wilc = np.nan
    idx = np.random.default_rng(SEED).integers(0, n, size=(N_BOOTS, n))
    zb, cb = z_freqs[idx].mean(1), c_freqs[idx].mean(1)
    valid = (zb > 0) & (cb > 0)
    log2_boot = np.log2(zb[valid] / cb[valid])
    return (z_freqs.mean() / c_freqs.mean(), 2 ** np.percentile(log2_boot, 2.5), 2 ** np.percentile(log2_boot, 97.5),
            p_wilc)


def decay_distance(log2_arr):
    idx = np.where(log2_arr < 0.5)[0]
    return idx[0] + 1 if len(idx) else np.nan


def main():
    np.seterr(divide="ignore", invalid="ignore")
    prof = pd.read_csv(PROF, sep="\t")
    bych = pd.read_csv(BYCHROM, sep="\t")
    rows = []
    for stratum in STRATA:
        df = prof[prof.stratum == stratum]
        if df.empty:
            print(f"[WARN] {stratum} not in {PROF.name}")
            continue
        z = df[df.track == "Z"].sort_values("window_idx").reset_index(drop=True)
        c = df[df.track == "C"].sort_values("window_idx").reset_index(drop=True)
        z_freq, c_freq = z.frequency.to_numpy(), c.frequency.to_numpy()
        n_body, n_flank = int((z.region == "body").sum()), int((z.region == "left_flank").sum())
        body_z, body_c = z_freq[n_flank:n_flank + n_body], c_freq[n_flank:n_flank + n_body]
        fold, lo, hi, w_p = chromosome_body_test(bych[bych.stratum == stratum], n_flank, n_body)

        third = max(1, n_body // 3)
        bnd = np.r_[np.arange(third), np.arange(n_body - third, n_body)]
        mid = np.arange(third, n_body - third)
        lr = lambda a, b: np.log2(a / b) if (a > 0 and b > 0) else np.nan
        bz, iz, bc_, ic = body_z[bnd].mean(), body_z[mid].mean(), body_c[bnd].mean(), body_c[mid].mean()

        def flank_log2(zf, cf):
            return np.log2(zf / cf, where=(zf > 0) & (cf > 0), out=np.full_like(zf, np.nan))
        left_log2 = flank_log2(z_freq[:n_flank], c_freq[:n_flank])
        right_log2 = flank_log2(z_freq[n_flank + n_body:], c_freq[n_flank + n_body:])
        rows.append({
            "stratum": stratum, "n_intervals_Z": int(z.n_intervals.iloc[0]),
            "n_mutations_in_body": int(z.raw_count.iloc[n_flank:n_flank + n_body].sum()),
            "fold_enrichment_body": fold, "fold_enrichment_body_CI_lo": lo, "fold_enrichment_body_CI_hi": hi,
            "wilcoxon_body_p": w_p,
            "log2_boundary_vs_interior_Z": lr(bz, iz), "log2_boundary_vs_interior_C": lr(bc_, ic),
            "log2_boundary_vs_interior_diff": lr(bz, iz) - lr(bc_, ic),
            "log2_near_upstream": np.nanmean(left_log2[-5:]), "log2_distal_upstream": np.nanmean(left_log2[:5]),
            "log2_near_downstream": np.nanmean(right_log2[:5]), "log2_distal_downstream": np.nanmean(right_log2[-5:]),
            "decay_distance_upstream_bp": decay_distance(left_log2[::-1]),
            "decay_distance_downstream_bp": decay_distance(right_log2),
        })
    tbl = pd.DataFrame(rows)

    m = tbl.wilcoxon_body_p.notna()
    tbl["wilcoxon_body_p_bh"] = np.nan
    tbl.loc[m, "wilcoxon_body_p_bh"] = multipletests(tbl.loc[m, "wilcoxon_body_p"], method="fdr_bh")[1]
    tbl["significant_bh"] = tbl.wilcoxon_body_p_bh < 0.05
    tbl.to_csv(OUT, sep="\t", index=False, float_format="%.6g")
    print(tbl[["stratum", "fold_enrichment_body", "wilcoxon_body_p_bh"]].to_string(index=False, float_format=lambda x: f"{x:.3g}"))
    print(f"-> {OUT.name}")


if __name__ == "__main__":
    main()
```

#### Fig. 3a

```python

"""Fig. 3a - relative density of matched Z-DNA and controls within +/-WINDOW bp of variant anchors per variant category
(> 50 bp: left/right anchors); bars with 95% chromosome-bootstrap CIs, Wilcoxon BH stars, fold enrichment + legend.
Reads 12.2.
"""
from pathlib import Path

import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from matplotlib.patches import Patch

BASE = Path("/path/to/pangenomes_zdna")
TABLE = BASE / "12_zdna_mutation_window_stats.tsv"
PLOTS, TESTS = BASE / "plots", BASE / "tests"
STEM = "fig3a_variant_window_enrichment"
LEGEND_STEM = f"{STEM}_legend"
WS_COLUMNS = {
    "window_bp": "Window (+/- bp)", "stratum": "Stratum", "anchor": "Anchor", "label": "Variant category",
    "snv_type": "SNV type", "n_events": "Variants (n)",
    "Z_rel": "Z-DNA relative density", "Z_rel_lo": "Z-DNA relative density, 95% CI lower",
    "Z_rel_hi": "Z-DNA relative density, 95% CI upper", "C_rel": "Control relative density",
    "C_rel_lo": "Control relative density, 95% CI lower", "C_rel_hi": "Control relative density, 95% CI upper",
    "fold_enrichment": "Fold enrichment (Z-DNA/control)", "fold_lo": "Fold enrichment, 95% CI lower",
    "fold_hi": "Fold enrichment, 95% CI upper",
    "Z_count_window": "Z-DNA overlaps in window (n)", "n_offsets_window": "Offsets in window (n)",
    "Z_count_profile": "Z-DNA overlaps over +/-500 bp (n)", "C_count_window": "Control overlaps in window (n)",
    "C_count_profile": "Control overlaps over +/-500 bp (n)", "n_offsets_profile": "Offsets over +/-500 bp (n)",
    "wilcoxon_stat": "Wilcoxon statistic", "p_wilcoxon": "Wilcoxon P value", "q_wilcoxon": "Wilcoxon q value (BH)",
    "sig_wilcoxon": "Significance (Wilcoxon)", "p_bootstrap": "Bootstrap P value", "q_bootstrap": "Bootstrap q value (BH)",
    "sig_bootstrap": "Significance (bootstrap)", "n_chromosomes_tested": "Chromosomes tested (n)",
    "n_inside_Z": "Variants inside Z-DNA (n)", "n_inside_C": "Variants inside controls (n)",
    "Z_per_kb": "Variants per kb inside Z-DNA", "C_per_kb": "Variants per kb inside controls",
    "inside_density_ratio": "Inside density ratio (Z-DNA/control)",
    "inside_ratio_lo": "Inside density ratio, 95% CI lower", "inside_ratio_hi": "Inside density ratio, 95% CI upper",
}

WINDOW = 10
MIN_EVENTS = 1000
SIG_COLUMN = "sig_wilcoxon"
SHOW_FOLD = True
FIG_W, FIG_H = 8.2, 3.7
BAR_W = 0.38
AXIS_W, TICK_W, TICK_LEN = 1.1, 0.9, 3
ERR_LW, CAPSIZE = 0.8, 1.6
FS_TICK, FS_XTICK, FS_AXIS, FS_SIG, FS_FOLD, FS_LEGEND = 10, 9.5, 11, 9, 7.5, 9
X_ROT = 55
GRID = dict(linestyle="--", linewidth=0.4, color="#B0B0B0", alpha=0.5, zorder=0)
FONT, DPI = "Nimbus Sans", 600
Z_COLOR, C_COLOR = "#B3A7E0", "#A6A6A6"
LABELS = {
    "SNP": "SNV", "MNP_2_20": "MNV 2–20",
    "DEL_1bp": "Del 1", "DEL_2_5bp": "Del 2–5", "DEL_6_50bp": "Del 6–50",
    "DEL_gt50bp_left": "Del >50 (L)", "DEL_gt50bp_right": "Del >50 (R)",
    "INS_1bp": "Ins 1", "INS_2_5bp": "Ins 2–5", "INS_6_50bp": "Ins 6–50", "INS_51_200bp": "Ins 51–200",
    "INS_gt200bp": "Ins >200",
    "CDEL_2_5bp": "CDel 2–5", "CDEL_6_50bp": "CDel 6–50", "CDEL_gt50bp_left": "CDel >50 (L)",
    "CDEL_gt50bp_right": "CDel >50 (R)",
    "CINS_2_5bp": "CIns 2–5", "CINS_6_50bp": "CIns 6–50", "CINS_51_200bp": "CIns 51–200", "CINS_gt200bp": "CIns >200",
    "SUB_21_50": "Sub 21–50", "SUB_51_200_left": "Sub 51–200 (L)", "SUB_51_200_right": "Sub 51–200 (R)",
    "SUB_gt200_left": "Sub >200 (L)", "SUB_gt200_right": "Sub >200 (R)",
}

plt.rcParams.update({"font.family": FONT, "pdf.fonttype": 42, "ps.fonttype": 42, "axes.linewidth": AXIS_W,
                     "axes.spines.top": False, "axes.spines.right": False})
PLOTS.mkdir(exist_ok=True)
TESTS.mkdir(exist_ok=True)

tab = pd.read_csv(TABLE, sep="\t")
allw = tab[tab.window_bp == WINDOW].reset_index(drop=True)
if allw.empty:
    raise SystemExit(f"No rows for window {WINDOW} bp in {TABLE}")
t = allw[allw.n_events >= MIN_EVENTS].reset_index(drop=True)
dropped = allw.loc[allw.n_events < MIN_EVENTS, "label"].tolist()
n = len(t)
x = np.arange(n)
bc = pd.read_csv(BASE / "12_zdna_mutation_profiles_bychrom.tsv.gz", sep="\t",
                 usecols=["stratum", "anchor", "chrom", "offset", "Z_raw", "C_raw"])
bc["label"] = bc.stratum + np.where(bc.anchor.isin(["left", "right"]), "_" + bc.anchor, "")
bc = bc[bc.label.isin(t.label)].assign(win=lambda d: d.offset.abs() <= WINDOW)
nw, na = bc.loc[bc.win, "offset"].nunique(), bc.offset.nunique()
s = bc.groupby(["label", "chrom"]).apply(lambda g: pd.Series({
    "Z_rel": (g.Z_raw[g.win].sum() / nw) / (g.Z_raw.sum() / na),
    "C_rel": (g.C_raw[g.win].sum() / nw) / (g.C_raw.sum() / na)})).reset_index()
s = s.replace([np.inf, -np.inf], np.nan).dropna()
fig, ax = plt.subplots(figsize=(FIG_W, FIG_H), constrained_layout=True)
for vals, lo, hi, dx, col in [(t.Z_rel, t.Z_rel_lo, t.Z_rel_hi, -BAR_W / 2, Z_COLOR),
                              (t.C_rel, t.C_rel_lo, t.C_rel_hi, BAR_W / 2, C_COLOR)]:
    ax.bar(x + dx, vals, BAR_W, color=col, edgecolor="black", linewidth=0.6, zorder=3,
           yerr=[np.clip(vals - lo, 0, None), np.clip(hi - vals, 0, None)], capsize=CAPSIZE,
           error_kw={"linewidth": ERR_LW, "ecolor": "black", "capthick": ERR_LW, "zorder": 4})
pos = {l: i for i, l in enumerate(t.label)}
jit = np.random.default_rng(0).uniform(-BAR_W * 0.25, BAR_W * 0.25, len(s))
for col, dx in (("Z_rel", -BAR_W / 2), ("C_rel", BAR_W / 2)):
    ax.scatter(s.label.map(pos) + dx + jit, s[col], s=4, color="black", alpha=0.5, linewidths=0, zorder=5)
ax.axhline(1, ls=":", color="black", lw=0.9, zorder=2)
top = np.fmax(np.fmax(t.Z_rel_hi, t.C_rel_hi).to_numpy(),
              s.groupby("label")[["Z_rel", "C_rel"]].max().max(axis=1).reindex(t.label).to_numpy())
s.assign(label=s.label.map(LABELS)).rename(columns={"label": "Variant category", "chrom": "Chromosome",
    "Z_rel": "Z-DNA relative density", "C_rel": "Control relative density"}).to_csv(
    TESTS / f"{STEM}_per_chromosome.tsv", sep="\t", index=False, float_format="%.6g")

ticks = np.array([0, 2, 5, 10, 20, 40], dtype=float)
fwd = lambda y: np.interp(y, ticks, np.arange(len(ticks)))
inv = lambda u: np.interp(u, np.arange(len(ticks)), ticks)
ax.set_yscale("function", functions=(fwd, inv))
u_top = fwd(top)
for i, row in t.iterrows():
    sig = row[SIG_COLUMN] if isinstance(row[SIG_COLUMN], str) else "ns"
    u = u_top[i] + 0.08
    if sig == "ns":
        ax.text(i, inv(u), "n.s.", ha="center", va="bottom", fontsize=FS_FOLD, color="#555555", style="italic")
        continue
    ax.plot([i - BAR_W / 2, i - BAR_W / 2, i + BAR_W / 2, i + BAR_W / 2], inv(np.array([u, u + 0.04, u + 0.04, u])),
            lw=0.7, color="black", zorder=4)
    ax.text(i, inv(u + 0.03), sig, ha="center", va="bottom", fontsize=FS_SIG)
    if SHOW_FOLD:
        ax.text(i, inv(u + 0.2), f"{row.fold_enrichment:.1f}×", ha="center", va="bottom", fontsize=FS_FOLD,
                color="#333333")
ax.set_yticks(ticks[:-1])
ax.set_ylim(0, inv(np.nanmax(u_top) + 0.45))
ax.set_xlim(-0.6, n - 0.4)
ax.set_xticks(x)
ax.set_xticklabels([LABELS.get(l, l) for l in t.label], rotation=X_ROT, ha="right", rotation_mode="anchor",
                   fontsize=FS_XTICK)
ax.set_ylabel(f"Relative density (±{WINDOW} bp)", fontsize=FS_AXIS)
ax.tick_params(axis="y", labelsize=FS_TICK)
ax.tick_params(width=TICK_W, length=TICK_LEN)
ax.grid(True, axis="y", which="major", **GRID)
ax.set_axisbelow(True)
for ext in ("pdf", "png"):
    fig.savefig(PLOTS / f"{STEM}.{ext}", dpi=DPI if ext == "png" else None)
plt.close(fig)

q = "q" if SIG_COLUMN == "sig_wilcoxon" else "q$_{boot}$"
txt = lambda l: Patch(color="none", label=l)
handles = [Patch(color=Z_COLOR, label="Z-DNA"), Patch(color=C_COLOR, label="Matched control"),
           txt(f"***   {q} < 0.001"), txt(f"**     {q} < 0.01"), txt(f"*       {q} < 0.05"),
           txt("n.s.   not significant"), txt("1.5×   fold enrichment (Z-DNA / control)")]
lf, lax = plt.subplots(figsize=(2.6, 1.8))
lax.axis("off")
lax.legend(handles=handles, loc="center", frameon=False, fontsize=FS_LEGEND, handletextpad=0.6)
for ext in ("pdf", "png"):
    lf.savefig(PLOTS / f"{LEGEND_STEM}.{ext}", dpi=DPI if ext == "png" else None, bbox_inches="tight")
plt.close(lf)

PLOT_COLS = ["window_bp", "label", "n_events", "Z_count_window", "n_offsets_window", "Z_count_profile", "n_offsets_profile",
             "Z_rel", "Z_rel_lo", "Z_rel_hi", "C_count_window", "C_count_profile", "C_rel", "C_rel_lo", "C_rel_hi",
             "fold_enrichment", "fold_lo", "fold_hi", "q_wilcoxon", SIG_COLUMN]
t.assign(label=t.label.map(LABELS).fillna(t.label))[PLOT_COLS].rename(columns=WS_COLUMNS).to_csv(
    TESTS / f"{STEM}.tsv", sep="\t", index=False, float_format="%.6g")
allw.assign(label=allw.label.map(LABELS).fillna(allw.label))[[c for c in WS_COLUMNS if c in allw.columns]] \
    .rename(columns=WS_COLUMNS).to_csv(TESTS / f"{STEM}_statistics.tsv", sep="\t", index=False, float_format="%.6g")
print(f"{n} strata drawn; not drawn (< {MIN_EVENTS:,} variants): {dropped or 'none'}")
```

#### Fig. 3b, Suppl. Fig. 2e

```python
"""Fig. 3b and Suppl. Fig. 2e - metaprofiles of matched Z-DNA and control relative density (+/-500 bp) around variant
anchors (one figure per stratum, 95% chromosome-bootstrap bands) + shared legend. Reads 12.1 and 12.2.
"""
from pathlib import Path

import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
import pandas as pd
from matplotlib.lines import Line2D
from matplotlib.ticker import MaxNLocator

BASE = Path("/path/to/pangenomes_zdna")
TABLE = BASE / "12_zdna_mutation_profiles.tsv.gz"
WINDOW_STATS = BASE / "12_zdna_mutation_window_stats.tsv"
PLOTS, TESTS = BASE / "plots", BASE / "tests"
STEMS = {"main": "fig3b_variant_metaprofiles", "suppl": "suppl_fig2e_variant_metaprofiles"}
OUTDIR = PLOTS / "fig3b_suppl_fig2e_metaprofiles"
LEGEND_STEM = "fig3b_variant_metaprofiles_legend"
PROFILE_COLUMNS = {"stratum": "Stratum", "anchor": "Anchor", "offset": "Offset (bp)", "n_events": "Variants (n)",
                   "Z_raw": "Anchors overlapping Z-DNA at offset (n)", "Z_mean": "Z-DNA relative density",
                   "Z_lo": "Z-DNA relative density, 95% CI lower", "Z_hi": "Z-DNA relative density, 95% CI upper",
                   "C_raw": "Anchors overlapping controls at offset (n)", "C_mean": "Control relative density",
                   "C_lo": "Control relative density, 95% CI lower", "C_hi": "Control relative density, 95% CI upper",
                   "n_chromosomes": "Chromosomes (n)", "Z_bp": "Z-DNA (bp)", "C_bp": "Controls (bp)"}
WS_COLUMNS = {
    "window_bp": "Window (+/- bp)", "stratum": "Stratum", "anchor": "Anchor", "label": "Variant category",
    "snv_type": "SNV type", "n_events": "Variants (n)",
    "Z_rel": "Z-DNA relative density", "Z_rel_lo": "Z-DNA relative density, 95% CI lower",
    "Z_rel_hi": "Z-DNA relative density, 95% CI upper", "C_rel": "Control relative density",
    "C_rel_lo": "Control relative density, 95% CI lower", "C_rel_hi": "Control relative density, 95% CI upper",
    "fold_enrichment": "Fold enrichment (Z-DNA/control)", "fold_lo": "Fold enrichment, 95% CI lower",
    "fold_hi": "Fold enrichment, 95% CI upper",
    "Z_count_window": "Z-DNA overlaps in window (n)", "n_offsets_window": "Offsets in window (n)",
    "Z_count_profile": "Z-DNA overlaps over +/-500 bp (n)", "C_count_window": "Control overlaps in window (n)",
    "C_count_profile": "Control overlaps over +/-500 bp (n)", "n_offsets_profile": "Offsets over +/-500 bp (n)",
    "wilcoxon_stat": "Wilcoxon statistic", "p_wilcoxon": "Wilcoxon P value", "q_wilcoxon": "Wilcoxon q value (BH)",
    "sig_wilcoxon": "Significance (Wilcoxon)", "p_bootstrap": "Bootstrap P value", "q_bootstrap": "Bootstrap q value (BH)",
    "sig_bootstrap": "Significance (bootstrap)", "n_chromosomes_tested": "Chromosomes tested (n)",
    "n_inside_Z": "Variants inside Z-DNA (n)", "n_inside_C": "Variants inside controls (n)",
    "Z_per_kb": "Variants per kb inside Z-DNA", "C_per_kb": "Variants per kb inside controls",
    "inside_density_ratio": "Inside density ratio (Z-DNA/control)",
    "inside_ratio_lo": "Inside density ratio, 95% CI lower", "inside_ratio_hi": "Inside density ratio, 95% CI upper",
}
STATS_WINDOW = 10

MAIN = [("CINS_2_5bp", "site"), ("DEL_6_50bp", "midpoint"), ("SNP", "site")]
SUPPL = [("INS_gt200bp", "site")]
XLIM = 500
FIG_W, FIG_H = 2.9, 2.3
LINE_W, BAND_ALPHA = 1.6, 0.3
AXIS_W, TICK_W, TICK_LEN = 1.1, 0.9, 3
FS_TICK, FS_AXIS, FS_TITLE, FS_LEGEND = 9.5, 10.5, 10, 9
GRID = dict(linestyle="--", linewidth=0.4, color="#B0B0B0", alpha=0.5, zorder=0)
FONT, DPI = "Nimbus Sans", 600
Z_COLOR, C_COLOR = "#9C8BDB", "#8C8C8C"
TITLES = {
    "SNP": "SNVs", "MNP_2_20": "MNVs (2–20 bp)",
    "DEL_1bp": "Deletions (1 bp)", "DEL_2_5bp": "Deletions (2–5 bp)", "DEL_6_50bp": "Deletions (6–50 bp)",
    "DEL_gt50bp": "Deletions (>50 bp)",
    "INS_1bp": "Insertions (1 bp)", "INS_2_5bp": "Insertions (2–5 bp)", "INS_6_50bp": "Insertions (6–50 bp)",
    "INS_51_200bp": "Insertions (51–200 bp)", "INS_gt200bp": "Insertions (>200 bp)",
    "CDEL_2_5bp": "Complex deletions (2–5 bp)", "CDEL_6_50bp": "Complex deletions (6–50 bp)",
    "CDEL_gt50bp": "Complex deletions (>50 bp)", "CINS_2_5bp": "Complex insertions (2–5 bp)",
    "CINS_6_50bp": "Complex insertions (6–50 bp)", "CINS_51_200bp": "Complex insertions (51–200 bp)",
    "CINS_gt200bp": "Complex insertions (>200 bp)", "SUB_21_50": "Substitutions (21–50 bp)",
    "SUB_51_200": "Substitutions (51–200 bp)", "SUB_gt200": "Substitutions (>200 bp)",
}
ANCHOR = {"left": ", left end", "right": ", right end"}
XLABEL = {"site": "Distance from variant (bp)", "midpoint": "Distance from variant midpoint (bp)",
          "left": "Distance from left breakpoint (bp)", "right": "Distance from right breakpoint (bp)"}

plt.rcParams.update({"font.family": FONT, "pdf.fonttype": 42, "ps.fonttype": 42, "axes.linewidth": AXIS_W,
                     "axes.spines.top": False, "axes.spines.right": False})
OUTDIR.mkdir(parents=True, exist_ok=True)
TESTS.mkdir(exist_ok=True)
prof = pd.read_csv(TABLE, sep="\t")
ws = pd.read_csv(WINDOW_STATS, sep="\t")
ws = ws[ws.window_bp == STATS_WINDOW]
label_of = lambda s, a: s + (f"_{a}" if a in ("left", "right") else "")


def draw(stratum, anchor, stem):
    d = prof[(prof.stratum == stratum) & (prof.anchor == anchor) & (prof.offset.abs() <= XLIM)].sort_values("offset")
    if d.empty:
        print(f"[WARN] {stratum}/{anchor} not in {TABLE.name}")
        return None
    fig, ax = plt.subplots(figsize=(FIG_W, FIG_H), constrained_layout=True)
    for m, col in (("C", C_COLOR), ("Z", Z_COLOR)):
        ax.fill_between(d.offset, d[f"{m}_lo"], d[f"{m}_hi"], color=col, alpha=BAND_ALPHA, lw=0, zorder=2)
        ax.plot(d.offset, d[f"{m}_mean"], color=col, lw=LINE_W, zorder=3)
    ax.axvline(0, ls="--", color="black", lw=0.8, zorder=1)
    ax.axhline(1, ls=":", color="black", lw=0.8, zorder=1)
    ax.set_xlim(-XLIM, XLIM)
    ax.xaxis.set_major_locator(MaxNLocator(nbins=5, symmetric=True))
    ax.yaxis.set_major_locator(MaxNLocator(nbins=4))
    ax.set_xlabel(XLABEL[anchor], fontsize=FS_AXIS)
    ax.set_ylabel("Relative density", fontsize=FS_AXIS)
    ax.set_title(f"{TITLES.get(stratum, stratum)}{ANCHOR.get(anchor, '')}",
                 fontsize=FS_TITLE, loc="left")
    ax.tick_params(labelsize=FS_TICK, width=TICK_W, length=TICK_LEN)
    ax.grid(True, axis="both", which="major", **GRID)
    ax.set_axisbelow(True)
    for ext in ("pdf", "png"):
        fig.savefig(OUTDIR / f"{stem}_{stratum}_{anchor}.{ext}", dpi=DPI if ext == "png" else None)
    plt.close(fig)
    return d


for key, items in (("main", MAIN), ("suppl", SUPPL)):
    drawn = [d for s, a in items if (d := draw(s, a, STEMS[key])) is not None]
    d = pd.concat(drawn, ignore_index=True)
    d.assign(stratum=d.stratum.map(TITLES))[list(PROFILE_COLUMNS)].rename(columns=PROFILE_COLUMNS).to_csv(
        TESTS / f"{STEMS[key]}.tsv", sep="\t", index=False, float_format="%.6g")
    st = ws[ws.label.isin({label_of(s, a) for s, a in items})]
    st.assign(stratum=st.stratum.map(TITLES))[[c for c in WS_COLUMNS if c in st.columns and c != "label"]] \
        .rename(columns=WS_COLUMNS).to_csv(TESTS / f"{STEMS[key]}_statistics.tsv", sep="\t", index=False, float_format="%.6g")

lf, lax = plt.subplots(figsize=(1.9, 0.6))
lax.axis("off")
lax.legend(handles=[Line2D([0], [0], color=Z_COLOR, lw=LINE_W * 1.4, label="Z-DNA"),
                    Line2D([0], [0], color=C_COLOR, lw=LINE_W * 1.4, label="Matched control")],
           loc="center", frameon=False, fontsize=FS_LEGEND)
for ext in ("pdf", "png"):
    lf.savefig(PLOTS / f"{LEGEND_STEM}.{ext}", dpi=DPI if ext == "png" else None, bbox_inches="tight")
plt.close(lf)
```

#### Fig. 3c

```python
"""Fig. 3c - metaprofiles of Z-DNA and matched-control relative density around de novo insertions, deletions (<= 50 bp)
and SNVs (denovo-db WGS, GRCh37), 95% chromosome-bootstrap bands + legend. Reads 14.2.
"""
from pathlib import Path

import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
import pandas as pd
from matplotlib.lines import Line2D
from matplotlib.ticker import MaxNLocator

BASE = Path("/path/to/pangenomes_zdna")
PROFILES = BASE / "14_denovo_zdna_profiles.tsv.gz"
RATES = BASE / "14_denovo_zdna_rates.tsv"
PLOTS, TESTS = BASE / "plots", BASE / "tests"
STEM = "fig3c_denovo_metaprofiles"
OUTDIR = PLOTS / STEM
PROFILE_COLUMNS = {"class": "Variant class", "offset": "Offset (bp)", "n_events": "De novo variants (n)",
                   "Z_raw": "Anchors overlapping Z-DNA at offset (n)", "Z_mean": "Z-DNA relative density",
                   "Z_lo": "Z-DNA relative density, 95% CI lower", "Z_hi": "Z-DNA relative density, 95% CI upper",
                   "C_raw": "Anchors overlapping controls at offset (n)", "C_mean": "Control relative density",
                   "C_lo": "Control relative density, 95% CI lower", "C_hi": "Control relative density, 95% CI upper",
                   "n_chrom": "Chromosomes (n)"}
CLASS_NAMES = {"all": "All de novo variants", "SNV": "SNVs", "MNV": "MNVs", "small_del": "Deletions (<=50 bp)",
               "small_ins": "Insertions (<=50 bp)", "large_indel": "Indels (>50 bp)", "substitution": "Substitutions",
               "complex_del": "Complex deletions", "complex_ins": "Complex insertions"}
RATE_COLUMNS = {"flank": "Flank (bp)", "class": "Variant class", "n_events": "De novo variants (n)",
                "n_Z": "Variants in Z-DNA (n)", "n_C": "Variants in controls (n)", "bp_Z": "Z-DNA (bp)", "bp_C": "Controls (bp)",
                "rate_Z_per_Mb": "Rate in Z-DNA (per Mb)", "rate_C_per_Mb": "Rate in controls (per Mb)",
                "rate_ratio": "Rate ratio (Z-DNA/control)", "ci_lo": "95% CI lower", "ci_hi": "95% CI upper",
                "p": "Bootstrap P value", "q": "q value (BH)", "sig": "Significance", "p_binomial": "Binomial P value",
                "n_chrom": "Chromosomes (n)", "n_boots": "Bootstrap replicates (n)", "test": "Test"}

CLASSES = ["SNV", "small_del", "small_ins"]
XLIM = 500
FIG_W, FIG_H = 2.9, 2.3
LINE_W, BAND_ALPHA = 1.6, 0.3
AXIS_W, TICK_W, TICK_LEN = 1.1, 0.9, 3
FS_TICK, FS_AXIS, FS_TITLE, FS_LEGEND = 9.5, 10.5, 10, 9
GRID = dict(linestyle="--", linewidth=0.4, color="#B0B0B0", alpha=0.5, zorder=0)
FONT, DPI = "Nimbus Sans", 600
Z_COLOR, C_COLOR = "#7FB28E", "#8C8C8C"
TITLES = {"all": "All de novo variants", "SNV": "De novo SNVs", "MNV": "De novo MNVs",
          "small_del": "De novo deletions (≤50 bp)", "small_ins": "De novo insertions (≤50 bp)"}

plt.rcParams.update({"font.family": FONT, "pdf.fonttype": 42, "ps.fonttype": 42, "axes.linewidth": AXIS_W,
                     "axes.spines.top": False, "axes.spines.right": False})
OUTDIR.mkdir(parents=True, exist_ok=True)
TESTS.mkdir(exist_ok=True)
prof = pd.read_csv(PROFILES, sep="\t")
rates = pd.read_csv(RATES, sep="\t")

drawn = []
for cls in CLASSES:
    d = prof[(prof["class"] == cls) & (prof.offset.abs() <= XLIM)].sort_values("offset")
    if d.empty:
        print(f"[WARN] {cls} not in {PROFILES.name}")
        continue
    fig, ax = plt.subplots(figsize=(FIG_W, FIG_H), constrained_layout=True)
    for m, col in (("C", C_COLOR), ("Z", Z_COLOR)):
        ax.fill_between(d.offset, d[f"{m}_lo"], d[f"{m}_hi"], color=col, alpha=BAND_ALPHA, lw=0, zorder=2)
        ax.plot(d.offset, d[f"{m}_mean"], color=col, lw=LINE_W, zorder=3)
    ax.axvline(0, ls="--", color="black", lw=0.8, zorder=1)
    ax.axhline(1, ls=":", color="black", lw=0.8, zorder=1)
    ax.set_xlim(-XLIM, XLIM)
    ax.xaxis.set_major_locator(MaxNLocator(nbins=5, symmetric=True))
    ax.yaxis.set_major_locator(MaxNLocator(nbins=4))
    ax.set_xlabel("Distance from variant (bp)", fontsize=FS_AXIS)
    ax.set_ylabel("Relative density", fontsize=FS_AXIS)
    ax.set_title(f"{TITLES.get(cls, cls)}", fontsize=FS_TITLE, loc="left")
    ax.tick_params(labelsize=FS_TICK, width=TICK_W, length=TICK_LEN)
    ax.grid(True, axis="both", which="major", **GRID)
    ax.set_axisbelow(True)
    for ext in ("pdf", "png"):
        fig.savefig(OUTDIR / f"{STEM}_{cls}.{ext}", dpi=DPI if ext == "png" else None)
    plt.close(fig)
    drawn.append(d)

lf, lax = plt.subplots(figsize=(2.2, 0.6))
lax.axis("off")
lax.legend(handles=[Line2D([0], [0], color=Z_COLOR, lw=LINE_W * 1.4, label="Z-DNA (GRCh37)"),
                    Line2D([0], [0], color=C_COLOR, lw=LINE_W * 1.4, label="Matched control")],
           loc="center", frameon=False, fontsize=FS_LEGEND)
for ext in ("pdf", "png"):
    lf.savefig(PLOTS / f"{STEM}_legend.{ext}", dpi=DPI if ext == "png" else None, bbox_inches="tight")
plt.close(lf)

d = pd.concat(drawn, ignore_index=True)
d.assign(**{"class": d["class"].map(CLASS_NAMES)})[list(PROFILE_COLUMNS)].rename(columns=PROFILE_COLUMNS).to_csv(
    TESTS / f"{STEM}.tsv", sep="\t", index=False, float_format="%.6g")
rates.assign(**{"class": rates["class"].map(CLASS_NAMES)})[list(RATE_COLUMNS)] \
    .rename(columns=RATE_COLUMNS).to_csv(TESTS / f"{STEM}_statistics.tsv", sep="\t", index=False, float_format="%.6g")
print(f"{len(drawn)} profiles drawn: {', '.join(CLASSES)}")
```

#### Fig. 3d

```python
"""Fig. 3d - log2(Z-DNA/control) variant frequency across predicted Z-DNA intervals: upstream flank (FLANK_BIN-bp bins),
length-scaled body and downstream flank, one line per variant category + legend. Reads 15.1 and 15.2.
"""
from pathlib import Path

import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from matplotlib.gridspec import GridSpec
from matplotlib.lines import Line2D

BASE = Path("/path/to/pangenomes_zdna")
TABLE = BASE / "15_zdna_centred_profiles.tsv.gz"
SUMMARY = BASE / "15_zdna_centred_summary_metrics.tsv"
PLOTS, TESTS = BASE / "plots", BASE / "tests"
STEM = "fig3d_zdna_centred_variant_frequency"
COLUMNS = {"label": "Variant category", "segment": "Segment", "x": "Position (bp from interval or body bin)",
           "Z_count": "Variants at Z-DNA (n)", "n_intervals_Z": "Z-DNA intervals (n)", "C_count": "Variants at controls (n)",
           "n_intervals_C": "Control intervals (n)", "log2_ratio": "log2(Z-DNA/control) variant frequency"}
STATS_COLUMNS = {"stratum": "Variant category", "n_intervals_Z": "Z-DNA intervals (n)",
                 "n_mutations_in_body": "Variants in Z-DNA bodies (n)", "fold_enrichment_body": "Body fold enrichment",
                 "fold_enrichment_body_CI_lo": "Body fold enrichment, 95% CI lower",
                 "fold_enrichment_body_CI_hi": "Body fold enrichment, 95% CI upper",
                 "wilcoxon_body_p": "Wilcoxon P value", "wilcoxon_body_p_bh": "Wilcoxon q value (BH)",
                 "significant_bh": "Significant (q < 0.05)",
                 "log2_boundary_vs_interior_Z": "log2 boundary/interior, Z-DNA",
                 "log2_boundary_vs_interior_C": "log2 boundary/interior, controls",
                 "log2_boundary_vs_interior_diff": "log2 boundary/interior, difference",
                 "log2_near_upstream": "log2 Z/C, 5 bp next to body (upstream)",
                 "log2_distal_upstream": "log2 Z/C, 5 most distal bp (upstream)",
                 "log2_near_downstream": "log2 Z/C, 5 bp next to body (downstream)",
                 "log2_distal_downstream": "log2 Z/C, 5 most distal bp (downstream)",
                 "decay_distance_upstream_bp": "Decay distance upstream (bp)",
                 "decay_distance_downstream_bp": "Decay distance downstream (bp)"}

STRATA_TO_PLOT = [
    ("SNP", "SNV", "#3C5A78"),
    ("MNP_2_20", "MNV (2–20 bp)", "#8C6BB1"),
    ("DEL_6_50bp", "Deletion (6–50 bp)", "#4F6F52"),
    ("INS_6_50bp", "Insertion (6–50 bp)", "#8FB08A"),
    ("CDEL_6_50bp", "Complex deletion (6–50 bp)", "#A8926A"),
    ("CINS_6_50bp", "Complex insertion (6–50 bp)", "#B5654A"),
]
DISPLAY_FLANK = 50
FLANK_BIN = 5
MARKER_SIZE = 3
WIDTH_RATIOS = (5, 1.8, 5)
FIG_W, FIG_H = 8.26, 3.3
LINE_W = 1.8
BODY_BG = "#E9E9E9"
AXIS_W, TICK_W, TICK_LEN = 1.1, 0.9, 3
FS_TICK, FS_AXIS, FS_LEGEND = 10, 11, 9
MARGINS = (0.12, 0.985, 0.2, 0.96)
GRID = dict(linestyle="--", linewidth=0.4, color="#B0B0B0", alpha=0.5, zorder=0)
FONT, DPI = "Nimbus Sans", 600

plt.rcParams.update({"font.family": FONT, "pdf.fonttype": 42, "ps.fonttype": 42, "axes.linewidth": AXIS_W,
                     "axes.spines.top": False, "axes.spines.right": False})
PLOTS.mkdir(exist_ok=True)
TESTS.mkdir(exist_ok=True)
prof = pd.read_csv(TABLE, sep="\t")


def load_stratum(stratum):
    """log2(Z / C) per display bin: upstream FLANK_BIN-bp bins, body bins, downstream FLANK_BIN-bp bins."""
    df = prof[prof.stratum == stratum]
    if df.empty:
        return None
    z = df[df.track == "Z"].sort_values("window_idx").reset_index(drop=True)
    c = df[df.track == "C"].sort_values("window_idx").reset_index(drop=True)
    zn, cn = int(z.n_intervals.iloc[0]), int(c.n_intervals.iloc[0])
    nb, fl = int((z.region == "body").sum()), int((z.region == "left_flank").sum())
    if DISPLAY_FLANK > fl or DISPLAY_FLANK % FLANK_BIN:
        raise ValueError(f"DISPLAY_FLANK={DISPLAY_FLANK} must be <= {fl} and a multiple of FLANK_BIN={FLANK_BIN}")
    zr, cr = z.raw_count.to_numpy(float), c.raw_count.to_numpy(float)
    k = DISPLAY_FLANK // FLANK_BIN
    up = slice(fl - DISPLAY_FLANK, fl)
    down = slice(fl + nb, fl + nb + DISPLAY_FLANK)
    seg = lambda a, sl, w: a[sl].reshape(-1, w).sum(1)
    parts = {"upstream": (seg(zr, up, FLANK_BIN), seg(cr, up, FLANK_BIN),
                          -DISPLAY_FLANK + FLANK_BIN * (np.arange(k) + 0.5)),
             "body": (zr[fl:fl + nb], cr[fl:fl + nb], np.arange(1, nb + 1, dtype=float)),
             "downstream": (seg(zr, down, FLANK_BIN), seg(cr, down, FLANK_BIN), FLANK_BIN * (np.arange(k) + 0.5))}
    rows = []
    for name, (zc, cc, x) in parts.items():
        with np.errstate(divide="ignore", invalid="ignore"):
            l2 = np.where((zc > 0) & (cc > 0), np.log2((zc / zn) / (cc / cn)), np.nan)
        rows.append(pd.DataFrame({"segment": name, "x": x, "Z_count": zc, "C_count": cc, "log2_ratio": l2}))
    out = pd.concat(rows, ignore_index=True)
    out["n_intervals_Z"], out["n_intervals_C"] = zn, cn
    return out, nb


loaded = []
for stratum, label, color in STRATA_TO_PLOT:
    res = load_stratum(stratum)
    if res is None:
        print(f"[WARN] {stratum} not in {TABLE.name}")
        continue
    loaded.append((stratum, label, color, *res))
if not loaded:
    raise SystemExit("No strata loaded.")
nb = loaded[0][4]

fig = plt.figure(figsize=(FIG_W, FIG_H))
gs = GridSpec(1, 3, figure=fig, width_ratios=WIDTH_RATIOS, wspace=0.0, left=MARGINS[0], right=MARGINS[1],
              bottom=MARGINS[2], top=MARGINS[3])
axL = fig.add_subplot(gs[0, 0])
axM = fig.add_subplot(gs[0, 1], sharey=axL)
axR = fig.add_subplot(gs[0, 2], sharey=axL)
for _, _, color, d, _ in loaded:
    for ax, segname in ((axL, "upstream"), (axM, "body"), (axR, "downstream")):
        g = d[d.segment == segname]
        ax.plot(g.x, g.log2_ratio, color=color, lw=LINE_W, marker="o", ms=MARKER_SIZE, zorder=5, clip_on=False)
for ax in (axL, axM, axR):
    ax.axhline(0, ls=":", color="black", lw=0.8, zorder=2)
    ax.tick_params(labelsize=FS_TICK, width=TICK_W, length=TICK_LEN)
    ax.grid(True, axis="both", which="major", **GRID)
    ax.set_axisbelow(True)
for ax in (axM, axR):
    ax.spines["left"].set_visible(False)
    ax.tick_params(axis="y", left=False, labelleft=False)
axM.set_facecolor(BODY_BG)
axL.set_xlim(-DISPLAY_FLANK, 0); axL.set_xticks([-DISPLAY_FLANK, -DISPLAY_FLANK // 2, 0])
axL.set_xticklabels([f"\u2212{DISPLAY_FLANK}", f"\u2212{DISPLAY_FLANK // 2}", ""])
axM.set_xlim(0.5, nb + 0.5); axM.set_xticks([1, nb])
axR.set_xlim(0, DISPLAY_FLANK); axR.set_xticks([0, DISPLAY_FLANK // 2, DISPLAY_FLANK])
axR.set_xticklabels(["", str(DISPLAY_FLANK // 2), str(DISPLAY_FLANK)])
axL.set_ylabel("log$_2$(Z-DNA / control)\nvariant frequency", fontsize=FS_AXIS)
axL.set_xlabel("Upstream (bp)", fontsize=FS_AXIS)
axM.set_xlabel("Interval\n(scaled bins)", fontsize=FS_AXIS)
axR.set_xlabel("Downstream (bp)", fontsize=FS_AXIS)
n_missing = sum(int(d.log2_ratio.isna().sum()) for *_, d, _ in loaded)
if n_missing:
    print(f"[NOTE] {n_missing} bins without variants in Z-DNA or controls (gaps in the lines)")
for ext in ("pdf", "png"):
    fig.savefig(PLOTS / f"{STEM}.{ext}", dpi=DPI if ext == "png" else None)
plt.close(fig)

lf, lax = plt.subplots(figsize=(2.6, 1.4))
lax.axis("off")
lax.legend(handles=[Line2D([0], [0], color=c, lw=LINE_W * 1.4, label=l) for _, l, c, *_ in loaded],
           loc="center", frameon=False, fontsize=FS_LEGEND)
for ext in ("pdf", "png"):
    lf.savefig(PLOTS / f"{STEM}_legend.{ext}", dpi=DPI if ext == "png" else None, bbox_inches="tight")
plt.close(lf)

pd.concat([d.assign(label=label) for _, label, _, d, _ in loaded])[list(COLUMNS)].rename(columns=COLUMNS).to_csv(
    TESTS / f"{STEM}.tsv", sep="\t", index=False, float_format="%.6g")
names = {s_: lab for s_, lab, _ in STRATA_TO_PLOT}
summ = pd.read_csv(SUMMARY, sep="\t")
summ.assign(stratum=summ.stratum.map(names).fillna(summ.stratum))[[c for c in STATS_COLUMNS if c in summ.columns]] \
    .rename(columns=STATS_COLUMNS).to_csv(TESTS / f"{STEM}_statistics.tsv", sep="\t", index=False, float_format="%.6g")
```

#### Suppl. Fig. 2b

```python
"""Suppl. Fig. 2b - within-class composition of the normalized HPRC v2 (CHM13) variants (stacked fractions; n per class) + legend. Reads 11.
"""
from pathlib import Path

import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
import pandas as pd
from matplotlib.patches import Patch

BASE = Path("/path/to/pangenomes_zdna")
TABLE = BASE / "11_mutation_composition_counts.tsv"
PLOTS, TESTS = BASE / "plots", BASE / "tests"
STEM = "suppl_fig2b_variant_composition"
COLUMNS = {"class_key": "Variant class", "category": "Category", "count": "Variants (n)", "class_total": "Variants in class (n)",
           "frac_within_class": "Fraction within class", "frac_of_all": "Fraction of all variants"}

FIG_W, FIG_H = 5.4, 3.5
BAR_W, BAR_EDGE_LW = 0.7, 0.5
AXIS_W, TICK_W, TICK_LEN = 1.1, 0.9, 3
FS_TICK, FS_AXIS, FS_COUNT, FS_LEGEND, FS_LEGEND_TITLE = 10, 11, 8.5, 9, 9.5
X_ROT = 35
COUNT_STYLE = "short"
GRID = dict(linestyle="--", linewidth=0.4, color="#B0B0B0", alpha=0.5, zorder=0)
FONT, DPI = "Nimbus Sans", 600

SNP_COLORS = {"C>A": "#4B6378", "C>G": "#93A7B8", "C>T": "#9C5F57", "T>A": "#6F7F66", "T>C": "#BDB08F", "T>G": "#7D7D7D"}
SIZE_COLORS = {"1bp": "#E4E4E4", "2_5bp": "#C0C0C0", "6_50bp": "#929292", "51_200bp": "#5F5F5F", "gt200bp": "#2C2C2C"}
SUB_COLORS = {"21_50bp": "#CBD3DA", "51_200bp": "#8D9DAB", "gt200bp": "#4E6070"}
MNP_COLOR = "#B3A99A"
CLASS_LABELS = {"SNP": "SNV", "MNP": "MNV", "SUB": "Substitution", "DEL": "Deletion", "INS": "Insertion",
                "CINS": "Complex\ninsertion", "CDEL": "Complex\ndeletion"}
SIZE_LABELS = {"1bp": "1 bp", "2_5bp": "2–5 bp", "6_50bp": "6–50 bp", "51_200bp": "51–200 bp", "gt200bp": ">200 bp",
               "21_50bp": "21–50 bp", "2_20bp": "MNV (2–20 bp)"}

plt.rcParams.update({"font.family": FONT, "pdf.fonttype": 42, "ps.fonttype": 42, "axes.linewidth": AXIS_W,
                     "axes.spines.top": False, "axes.spines.right": False})
PLOTS.mkdir(exist_ok=True)
TESTS.mkdir(exist_ok=True)


def color(k, c):
    return (SNP_COLORS.get(c) if k == "SNP" else MNP_COLOR if k == "MNP" else SUB_COLORS.get(c) if k == "SUB"
            else SIZE_COLORS.get(c))


def fmt(n):
    if COUNT_STYLE == "full":
        return f"{n:,}"
    return f"{n / 1e6:.1f}M" if n >= 1e6 else f"{n / 1e3:.0f}k" if n >= 1e3 else str(n)


def save(fig, stem):
    for ext in ("pdf", "png"):
        fig.savefig(PLOTS / f"{stem}.{ext}", dpi=DPI if ext == "png" else None)
    plt.close(fig)


tab = pd.read_csv(TABLE, sep="\t")
classes = list(dict.fromkeys(tab.class_key))
fig, ax = plt.subplots(figsize=(FIG_W, FIG_H), constrained_layout=True)
for i, k in enumerate(classes):
    d = tab[tab.class_key == k]
    if k == "SNP":
        d = d.sort_values("count", ascending=False)
    bottom = 0.0
    for c, f in zip(d.category, d.frac_within_class):
        ax.bar(i, f, bottom=bottom, width=BAR_W, color=color(k, c), edgecolor="black", linewidth=BAR_EDGE_LW, zorder=3)
        bottom += f
    ax.text(i, 1.02, f"n={fmt(int(d.class_total.iloc[0]))}", ha="center", va="bottom", fontsize=FS_COUNT)
ax.set_xticks(range(len(classes)))
ax.set_xticklabels([CLASS_LABELS.get(k, k) for k in classes], rotation=X_ROT, ha="right", rotation_mode="anchor")
ax.set_xlim(-0.6, len(classes) - 0.4)
ax.set_ylim(0, 1.1)
ax.set_yticks([0, 0.25, 0.5, 0.75, 1])
ax.set_ylabel("Fraction within class", fontsize=FS_AXIS)
ax.tick_params(labelsize=FS_TICK, width=TICK_W, length=TICK_LEN)
ax.grid(True, axis="y", which="major", **GRID)
ax.set_axisbelow(True)
save(fig, STEM)

groups = [("SNV", [Patch(color=v, label=k) for k, v in SNP_COLORS.items()]),
          ("Indel size", [Patch(color=v, label=SIZE_LABELS[k]) for k, v in SIZE_COLORS.items()]),
          ("Substitution size", [Patch(color=v, label=SIZE_LABELS[k]) for k, v in SUB_COLORS.items()]
           + [Patch(color=MNP_COLOR, label=SIZE_LABELS["2_20bp"])])]
lf, lax = plt.subplots(1, 3, figsize=(6.0, 1.6))
for a, (title, handles) in zip(lax, groups):
    a.axis("off")
    a.legend(handles=handles, title=title, loc="upper left", frameon=False, fontsize=FS_LEGEND,
             title_fontsize=FS_LEGEND_TITLE, alignment="left", handlelength=1.2, handleheight=1.0)
lf.savefig(PLOTS / f"{STEM}_legend.pdf", bbox_inches="tight")
lf.savefig(PLOTS / f"{STEM}_legend.png", dpi=DPI, bbox_inches="tight")
plt.close(lf)

tab.assign(class_key=tab.class_key.map(CLASS_LABELS).str.replace("\n", " "),
           category=tab.category.map(SIZE_LABELS).fillna(tab.category)).rename(columns=COLUMNS).to_csv(
    TESTS / f"{STEM}.tsv", sep="\t", index=False, float_format="%.6g")
print(tab.groupby("class_key", sort=False).class_total.first().to_string())
```

#### Suppl. Fig. 2c

```python
"""Suppl. Fig. 2c - GC fraction, CpG+GpC density and length of controls, matched and unmatched Z-DNA (boxes: median,
IQR, 1.5 x IQR whiskers; Cliff's delta vs matched Z-DNA). Reads 8.2 and 8.3.
"""
from pathlib import Path
import numpy as np
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
import pandas as pd
from matplotlib.ticker import FixedLocator, FuncFormatter, NullLocator

BASE = Path("/path/to/pangenomes_zdna")
LOCI = BASE / "8_CHM13_matching_loci.tsv.gz"
TESTS_IN = BASE / "8_CHM13_matching_tests.tsv"
PLOTS, TESTS = BASE / "plots", BASE / "tests"
STEM = "suppl_fig2c_matching_composition"
COLUMNS = {"set": "Set", "chrom": "Chromosome", "start0": "Start (0-based)", "end0": "End", "length": "Length (bp)",
           "gc": "GC fraction", "dinuc": "CpG+GpC density"}
STATS_COLUMNS = {"comparison": "Comparison", "metric_label": "Metric", "Z_matched_median": "Median, matched Z-DNA",
                 "Z_matched_q1": "Q1, matched Z-DNA", "Z_matched_q3": "Q3, matched Z-DNA", "other_median": "Median, other set",
                 "other_q1": "Q1, other set", "other_q3": "Q3, other set", "n_pairs": "Matched loci (n)",
                 "n_other": "Other set (n)", "frac_identical": "Fraction identical pairs",
                 "median_diff_control_minus_Z": "Median paired difference (control - Z-DNA)", "test": "Test",
                 "p": "P value", "effect_size": "Effect size", "effect_size_name": "Effect size measure",
                 "magnitude": "Magnitude"}
SET_NAMES = {"control": "Control", "Z_matched": "Matched Z-DNA", "Z_unmatched": "Unmatched Z-DNA"}
COMPARISONS = {"Z_matched_vs_control_paired": "Matched Z-DNA vs own controls (paired)",
               "Z_matched_vs_control_independent": "Matched Z-DNA vs controls (independent)",
               "Z_unmatched_vs_Z_matched": "Unmatched vs matched Z-DNA"}

SETS = ["control", "Z_matched", "Z_unmatched"]
SET_LABELS = {"control": "Controls", "Z_matched": "Matched\nZ-DNA", "Z_unmatched": "Unmatched\nZ-DNA"}
COLORS = {"control": "#A6A6A6", "Z_matched": "#B3A7E0", "Z_unmatched": "#513AB5"}
METRICS = [("gc", "GC fraction", False), ("dinuc", "CpG+GpC density", False), ("length", "Length (bp)", True)]
LENGTH_TICKS = [10, 20, 50, 100, 200, 500]
FIG_W, FIG_H = 2.8, 6.0
BOX_W, BOX_LW = 0.6, 0.9
AXIS_W, TICK_W, TICK_LEN = 1.1, 0.9, 3
FS_TICK, FS_AXIS, FS_NOTE = 9, 10, 8.5
GRID = dict(linestyle="--", linewidth=0.4, color="#B0B0B0", alpha=0.5, zorder=0)
FONT, DPI = "Nimbus Sans", 600

plt.rcParams.update({"font.family": FONT, "pdf.fonttype": 42, "ps.fonttype": 42, "axes.linewidth": AXIS_W,
                     "axes.spines.top": False, "axes.spines.right": False})
PLOTS.mkdir(exist_ok=True)
TESTS.mkdir(exist_ok=True)


def bracket(ax, x1, x2, y, text, logy):
    lo, hi = ax.get_ylim()
    h = y * 0.07 if logy else (hi - lo) * 0.03
    ax.plot([x1, x1, x2, x2], [y, y + h, y + h, y], lw=0.7, color="black", clip_on=False, zorder=4)
    ax.text((x1 + x2) / 2, y + h * 1.25, text, ha="center", va="bottom", fontsize=FS_NOTE)


loci = pd.read_csv(LOCI, sep="\t", usecols=["set", "chrom", "start0", "end0", "length", "gc", "dinuc"])
loci = loci[loci.set.isin(SETS)]
tests = pd.read_csv(TESTS_IN, sep="\t")
delta = lambda comp, m: tests.loc[(tests.comparison == comp) & (tests.metric == m), "effect_size"].iloc[0]
fmt = lambda d: f"δ = {round(d, 2) + 0.0:.2f}"

fig, axs = plt.subplots(len(METRICS), 1, figsize=(FIG_W, FIG_H), sharex=True, constrained_layout=True)
for ax, (m, lab, logy) in zip(axs, METRICS):
    vals = [loci.loc[loci.set == s, m].dropna().to_numpy() for s in SETS]
    bpl = ax.boxplot(vals, widths=BOX_W, patch_artist=True, showfliers=False, zorder=3,
                     medianprops={"color": "black", "lw": BOX_LW}, whiskerprops={"lw": BOX_LW}, capprops={"lw": BOX_LW},
                     boxprops={"lw": BOX_LW})
    for box, s in zip(bpl["boxes"], SETS):
        box.set_facecolor(COLORS[s])
    top = max(w.get_ydata().max() for w in bpl["whiskers"])
    if logy:
        ax.set_yscale("log")
    lo, hi = ax.get_ylim()
    y1 = top * 1.12 if logy else top + 0.05 * (hi - lo)
    y2 = top * 1.75 if logy else top + 0.22 * (hi - lo)
    bracket(ax, 1, 2, y1, "identical" if m == "length" else fmt(delta("Z_matched_vs_control_independent", m)), logy)
    bracket(ax, 2, 3, y2, fmt(delta("Z_unmatched_vs_Z_matched", m)), logy)
    ax.set_ylim(lo, y2 * 1.6 if logy else y2 + 0.15 * (hi - lo))
    if logy:
        lo, hi = ax.get_ylim()
        ax.yaxis.set_major_locator(FixedLocator([t for t in LENGTH_TICKS if lo <= t <= hi]))
        ax.yaxis.set_minor_locator(NullLocator())
        ax.yaxis.set_major_formatter(FuncFormatter(lambda v, _: f"{v:g}"))
    else:
        ax.set_yticks([t for t in ax.get_yticks() if 0 <= t <= 1])
    ax.set_ylabel(lab, fontsize=FS_AXIS)
    ax.tick_params(labelsize=FS_TICK, width=TICK_W, length=TICK_LEN)
    ax.grid(True, axis="y", which="major", **GRID)
    ax.set_axisbelow(True)
axs[-1].set_xticks(range(1, len(SETS) + 1), [SET_LABELS[s] for s in SETS])
for ext in ("pdf", "png"):
    fig.savefig(PLOTS / f"{STEM}.{ext}", dpi=DPI if ext == "png" else None)
plt.close(fig)

box = []
for m, lab, _ in METRICS:
    for s_ in SETS:
        x = loci.loc[loci.set == s_, m].dropna().to_numpy()
        q1, med, q3 = np.percentile(x, [25, 50, 75])
        iqr = q3 - q1
        box.append({"Metric": lab, "Set": SET_NAMES[s_], "Loci (n)": len(x),
                    "Lower whisker": x[x >= q1 - 1.5 * iqr].min(), "Q1": q1, "Median": med, "Q3": q3,
                    "Upper whisker": x[x <= q3 + 1.5 * iqr].max()})
pd.DataFrame(box).to_csv(TESTS / f"{STEM}.tsv", sep="\t", index=False, float_format="%.6g")
tests.assign(comparison=tests.comparison.map(COMPARISONS))[[c for c in STATS_COLUMNS if c in tests.columns]] \
    .rename(columns=STATS_COLUMNS).to_csv(TESTS / f"{STEM}_statistics.tsv", sep="\t", index=False, float_format="%.6g")
n = loci.set.value_counts()
print("n: " + ", ".join(f"{s} {int(n.get(s, 0)):,}" for s in SETS))
```

#### Suppl. Fig. 2d

```python
"""Suppl. Fig. 2d - relative density of matched Z-DNA and controls within +/-WINDOW bp of SNVs per pyrimidine-collapsed
SNV type (complementary change below), 95% chromosome-bootstrap CIs, Wilcoxon BH stars, fold enrichment. Reads 13.2.
"""
from pathlib import Path

import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

BASE = Path("/path/to/pangenomes_zdna")
TABLE = BASE / "13_zdna_snv_type_window_stats.tsv"
PLOTS, TESTS = BASE / "plots", BASE / "tests"
STEM = "suppl_fig2d_snv_type_window_enrichment"
WS_COLUMNS = {
    "window_bp": "Window (+/- bp)", "stratum": "Stratum", "anchor": "Anchor", "label": "Variant category",
    "snv_type": "SNV type", "n_events": "Variants (n)",
    "Z_rel": "Z-DNA relative density", "Z_rel_lo": "Z-DNA relative density, 95% CI lower",
    "Z_rel_hi": "Z-DNA relative density, 95% CI upper", "C_rel": "Control relative density",
    "C_rel_lo": "Control relative density, 95% CI lower", "C_rel_hi": "Control relative density, 95% CI upper",
    "fold_enrichment": "Fold enrichment (Z-DNA/control)", "fold_lo": "Fold enrichment, 95% CI lower",
    "fold_hi": "Fold enrichment, 95% CI upper",
    "Z_count_window": "Z-DNA overlaps in window (n)", "n_offsets_window": "Offsets in window (n)",
    "Z_count_profile": "Z-DNA overlaps over +/-500 bp (n)", "C_count_window": "Control overlaps in window (n)",
    "C_count_profile": "Control overlaps over +/-500 bp (n)", "n_offsets_profile": "Offsets over +/-500 bp (n)",
    "wilcoxon_stat": "Wilcoxon statistic", "p_wilcoxon": "Wilcoxon P value", "q_wilcoxon": "Wilcoxon q value (BH)",
    "sig_wilcoxon": "Significance (Wilcoxon)", "p_bootstrap": "Bootstrap P value", "q_bootstrap": "Bootstrap q value (BH)",
    "sig_bootstrap": "Significance (bootstrap)", "n_chromosomes_tested": "Chromosomes tested (n)",
    "n_inside_Z": "Variants inside Z-DNA (n)", "n_inside_C": "Variants inside controls (n)",
    "Z_per_kb": "Variants per kb inside Z-DNA", "C_per_kb": "Variants per kb inside controls",
    "inside_density_ratio": "Inside density ratio (Z-DNA/control)",
    "inside_ratio_lo": "Inside density ratio, 95% CI lower", "inside_ratio_hi": "Inside density ratio, 95% CI upper",
}

WINDOW = 10
SIG_COLUMN = "sig_wilcoxon"
SHOW_FOLD = True
SHOW_COMPLEMENT = True
ORDER = ["C>A", "C>G", "C>T", "T>A", "T>C", "T>G"]
FIG_W, FIG_H = 3.0, 2.3
BAR_W = 0.38
AXIS_W, TICK_W, TICK_LEN = 1.1, 0.9, 3
ERR_LW, CAPSIZE = 0.7, 1.4
FS_TICK, FS_XTICK, FS_AXIS, FS_SIG, FS_FOLD = 8.5, 8, 9, 8, 6.5
GRID = dict(linestyle="--", linewidth=0.4, color="#B0B0B0", alpha=0.5, zorder=0)
FONT, DPI = "Nimbus Sans", 600
Z_COLOR, C_COLOR = "#B3A7E0", "#A6A6A6"
COMP = str.maketrans("ACGT", "TGCA")

plt.rcParams.update({"font.family": FONT, "pdf.fonttype": 42, "ps.fonttype": 42, "axes.linewidth": AXIS_W,
                     "axes.spines.top": False, "axes.spines.right": False})
PLOTS.mkdir(exist_ok=True)
TESTS.mkdir(exist_ok=True)

tab = pd.read_csv(TABLE, sep="\t")
t = tab[tab.window_bp == WINDOW].set_index("snv_type")
t = t.loc[[o for o in ORDER if o in t.index]].reset_index()
if t.empty:
    raise SystemExit(f"No rows for window {WINDOW} bp in {TABLE}")
n = len(t)
x = np.arange(n)
bc = pd.read_csv(
    BASE / "13_zdna_snv_type_profiles_bychrom.tsv.gz",
    sep="\t",
    usecols=["snv_type", "chrom", "offset", "Z_raw", "C_raw"]
)

bc = bc[bc.snv_type.isin(t.snv_type)].copy()
bc["win"] = bc.offset.abs() <= WINDOW

nw = bc.loc[bc.win, "offset"].nunique()
na = bc.offset.nunique()

s = bc.groupby(["snv_type", "chrom"]).apply(
    lambda g: pd.Series({
        "Z_rel": (g.loc[g.win, "Z_raw"].sum() / nw) /
                 (g.Z_raw.sum() / na),
        "C_rel": (g.loc[g.win, "C_raw"].sum() / nw) /
                 (g.C_raw.sum() / na)
    })
).reset_index()

s = s.replace([np.inf, -np.inf], np.nan).dropna()
fig, ax = plt.subplots(figsize=(FIG_W, FIG_H), constrained_layout=True)
for vals, lo, hi, dx, col in [(t.Z_rel, t.Z_rel_lo, t.Z_rel_hi, -BAR_W / 2, Z_COLOR),
                              (t.C_rel, t.C_rel_lo, t.C_rel_hi, BAR_W / 2, C_COLOR)]:
    ax.bar(x + dx, vals, BAR_W, color=col, edgecolor="black", linewidth=0.6, zorder=3,
           yerr=[np.clip(vals - lo, 0, None), np.clip(hi - vals, 0, None)], capsize=CAPSIZE,
           error_kw={"linewidth": ERR_LW, "ecolor": "black", "capthick": ERR_LW, "zorder": 4})
pos = {snv: i for i, snv in enumerate(t.snv_type)}

jit = np.random.default_rng(0).uniform(
    -BAR_W * 0.25,
    BAR_W * 0.25,
    len(s)
)

for col, dx in (
    ("Z_rel", -BAR_W / 2),
    ("C_rel", BAR_W / 2)
):
    ax.scatter(
        s.snv_type.map(pos) + dx + jit,
        s[col],
        s=4,
        color="black",
        alpha=0.5,
        linewidths=0,
        zorder=5
    )
ax.axhline(1, ls=":", color="black", lw=0.8, zorder=2)
top = np.fmax(
    np.fmax(t.Z_rel_hi, t.C_rel_hi).to_numpy(),
    s.groupby("snv_type")[["Z_rel", "C_rel"]]
     .max()
     .max(axis=1)
     .reindex(t.snv_type)
     .to_numpy()
)
pad = 0.05 * np.nanmax(top)
for i, row in t.iterrows():
    sig = row[SIG_COLUMN] if isinstance(row[SIG_COLUMN], str) else "ns"
    y = top[i] + pad
    if sig == "ns":
        ax.text(i, y, "n.s.", ha="center", va="bottom", fontsize=FS_FOLD, color="#555555", style="italic")
        continue
    ax.plot([i - BAR_W / 2, i - BAR_W / 2, i + BAR_W / 2, i + BAR_W / 2], [y, y + pad * 0.4, y + pad * 0.4, y],
            lw=0.6, color="black", zorder=4)
    ax.text(i, y + pad * 0.3, sig, ha="center", va="bottom", fontsize=FS_SIG)
    if SHOW_FOLD:
        ax.text(i, y + pad * 2.0, f"{row.fold_enrichment:.2f}×", ha="center", va="bottom", fontsize=FS_FOLD,
                color="#333333")
ax.set_ylim(0, np.nanmax(top) + pad * 5)
ax.set_xlim(-0.6, n - 0.4)
ax.set_xticks(x)
ax.set_xticklabels([f"{s}\n{s.translate(COMP)}" if SHOW_COMPLEMENT else s for s in t.snv_type], fontsize=FS_XTICK)
ax.set_ylabel(f"Relative density (±{WINDOW} bp)", fontsize=FS_AXIS)
ax.tick_params(axis="y", labelsize=FS_TICK)
ax.tick_params(width=TICK_W, length=TICK_LEN)
ax.grid(True, axis="y", which="major", **GRID)
ax.set_axisbelow(True)
for ext in ("pdf", "png"):
    fig.savefig(PLOTS / f"{STEM}.{ext}", dpi=DPI if ext == "png" else None)
plt.close(fig)

PLOT_COLS = ["window_bp", "snv_type", "n_events", "Z_count_window", "n_offsets_window", "Z_count_profile",
             "n_offsets_profile", "Z_rel", "Z_rel_lo", "Z_rel_hi", "C_count_window", "C_count_profile", "C_rel", "C_rel_lo",
             "C_rel_hi", "fold_enrichment", "fold_lo", "fold_hi", "q_wilcoxon", SIG_COLUMN]
t[PLOT_COLS].rename(columns=WS_COLUMNS).to_csv(TESTS / f"{STEM}.tsv", sep="\t", index=False, float_format="%.6g")
w = tab[tab.window_bp == WINDOW]
w[[c for c in WS_COLUMNS if c in w.columns]].rename(columns=WS_COLUMNS).to_csv(
    TESTS / f"{STEM}_statistics.tsv", sep="\t", index=False, float_format="%.6g")
print(t[["snv_type", "n_events", "fold_enrichment", SIG_COLUMN]].to_string(index=False))
s.rename(columns={
    "snv_type": "SNV type",
    "chrom": "Chromosome",
    "Z_rel": "Z-DNA relative density",
    "C_rel": "Control relative density"
}).to_csv(
    TESTS / f"{STEM}_per_chromosome.tsv",
    sep="\t",
    index=False,
    float_format="%.6g"
)
```

## Results 4 – Variant size, ZSeeker score and tract length (Fig. 4, Suppl. Fig. 3a–b, Supplementary Table 3)

### Analysis

#### 16 · Enrichment by variant size → Fig. 4a, Suppl. Fig. 3a

```python
"""16 - Z-DNA enrichment within +/-WINDOW bp of variant anchors by variant size (CHM13, HPRC v2; Fig. 4a, Suppl. Fig. 3a).

Enrichment = Z-DNA (control) fraction of the window / genome-wide fraction, averaged per size bin; pooled mean with
95% chromosome-block bootstrap CI and per-chromosome means. Size dependence: Friedman test with chromosomes as blocks,
BH over classes.
"""
import time
from pathlib import Path

import numpy as np
import pandas as pd
import pyarrow as pa
import pyarrow.compute as pc
import pyarrow.csv as pacsv
from scipy.stats import friedmanchisquare
from statsmodels.stats.multitest import multipletests

BASE = Path("/path/to/pangenomes_zdna")
MASTER = BASE / "8_CHM13_controls_MASTER.tsv"
MUT_META = BASE / "9_hprc-v2.0-mc-chm13.wave.normalized.mutations.meta.tsv.gz"
FAI = BASE / "tmp/chm13/chm13_refseq/GCF_009914755.1_T2T-CHM13v2.0_genomic_ucsc.fna.fai"
OUT, OUT_CHR, OUT_TESTS = (BASE / f"16_variant_size_enrichment{s}.tsv" for s in ("", "_bychrom", "_tests"))

TOL, DINUC_MODE = "10%", "CpG_GpC"
WINDOW = 10                    # +/- bp around the anchor (as Fig. 3a)
SIZE_BOUNDARIES = [1, 2, 3, 6, 11, 21, 51, 101, 501]   # bin i = [b_i, b_{i+1} - 1]; last bin open-ended
MIN_MUTS = 500                 # variants per size bin
MIN_CHROM_EVENTS = 20          # variants per chromosome and bin for a chromosome point / Friedman block
N_BOOTS = 2000
SEED = 42
CLASSES = ["DEL", "INS", "CDEL", "CINS", "MNV_SUB"]
CHROMS = [f"chr{i}" for i in range(1, 23)] + ["chrX", "chrY"]
STRUCT = ["snp", "mnp", "substitution", "del", "ins", "complex"]
DOM = ["snp", "mnp", "substitution", "del", "ins", "del_like", "ins_like", "balanced"]
ARROW_THREADS, BLOCK_SIZE = 16, 1 << 26


def merge_intervals(s, e):
    o = np.argsort(s, kind="stable"); s, e = s[o], e[o]
    new = np.ones(len(s), bool); new[1:] = s[1:] > np.maximum.accumulate(e)[:-1]
    return s[new].astype(np.int64), np.maximum.reduceat(e, np.flatnonzero(new)).astype(np.int64)


def covered_before(x, s, e, cum):
    """Covered bases in [0, x) for disjoint sorted intervals [s, e) with cumulative lengths cum."""
    i = np.searchsorted(s, x, side="right") - 1
    ok = i >= 0
    out = np.zeros(len(x), np.int64)
    out[ok] = cum[i[ok]] - (e[i[ok]] - s[i[ok]]) + np.clip(x[ok] - s[i[ok]], 0, e[i[ok]] - s[i[ok]])
    return out


def load_sets(clen):
    cols = ["tol_level", "dinuc_mode", "Chromosome", "Start0", "End0", "Ctrl_Start", "Ctrl_End", "has_control"]
    m = pd.read_csv(MASTER, sep="\t", usecols=cols, low_memory=False,
                    dtype={"tol_level": str, "dinuc_mode": str, "Chromosome": str, "has_control": str})
    m = m[(m.tol_level == TOL) & (m.dinuc_mode == DINUC_MODE) & (m.has_control.str.strip() == "True")
          & m.Chromosome.isin(CHROMS)].drop_duplicates(["Chromosome", "Start0", "End0"])
    sets, frac = {}, {}
    for name, sc, ec in (("Z", "Start0", "End0"), ("C", "Ctrl_Start", "Ctrl_End")):
        iv = {}
        for c, d in m.groupby("Chromosome"):
            s, e = merge_intervals(d[sc].astype(float).astype(np.int64).to_numpy(),
                                   np.minimum(d[ec].astype(float).astype(np.int64).to_numpy(), clen[c]))
            iv[c] = (s, e, np.cumsum(e - s))
        sets[name], frac[name] = iv, sum(int(v[2][-1]) for v in iv.values()) / sum(clen[c] for c in iv)
    print(f"[loci] {len(m):,} matched Z-DNA loci; genome fraction: Z-DNA {frac['Z']:.5f}, controls {frac['C']:.5f}")
    return sets, frac


def load_variants():
    pa.set_cpu_count(ARROW_THREADS)
    types = {"seqID": pa.string(), "normalized_start0": pa.int64(), "normalized_alt": pa.string(),
             "struct_class": pa.string(), "dominant_op": pa.string(), "ref_span": pa.int64(), "event_size": pa.int64()}
    reader = pacsv.open_csv(MUT_META, read_options=pacsv.ReadOptions(block_size=BLOCK_SIZE),
                            parse_options=pacsv.ParseOptions(delimiter="\t"),
                            convert_options=pacsv.ConvertOptions(include_columns=list(types), column_types=types,
                                                                 strings_can_be_null=False))
    keys = ("ch", "s0", "rs", "L", "sc", "dom", "h")
    parts, t0 = {k: [] for k in keys}, time.time()
    for b in reader:
        b = b.filter(pc.not_equal(b.column("struct_class"), "snp"))
        chi = pc.index_in(b.column("seqID"), value_set=pa.array(CHROMS))
        b, chi = b.filter(pc.is_valid(chi)), chi.filter(pc.is_valid(chi))
        if b.num_rows == 0:
            continue
        sc = pc.index_in(b.column("struct_class"), value_set=pa.array(STRUCT)).to_numpy(zero_copy_only=False)
        dom = pc.index_in(b.column("dominant_op"), value_set=pa.array(DOM)).to_numpy(zero_copy_only=False)
        for k, v in zip(keys, (chi.to_numpy(zero_copy_only=False).astype(np.int8), b.column("normalized_start0").to_numpy(),
                               b.column("ref_span").to_numpy(), b.column("event_size").to_numpy(), sc.astype(np.int8),
                               dom.astype(np.int8),
                               pd.util.hash_array(b.column("normalized_alt").to_numpy(zero_copy_only=False)))):
            parts[k].append(v)
    v = {k: np.concatenate(x) for k, x in parts.items()}
    o = np.lexsort((v["h"], v["rs"], v["s0"], v["ch"]))
    v = {k: x[o] for k, x in v.items()}
    new = np.ones(len(o), bool)
    new[1:] = ((v["ch"][1:] != v["ch"][:-1]) | (v["s0"][1:] != v["s0"][:-1]) | (v["rs"][1:] != v["rs"][:-1])
               | (v["h"][1:] != v["h"][:-1]))
    v = {k: x[new] for k, x in v.items()}
    sc, dom, L = v["sc"], v["dom"], v["L"]
    is_ = lambda x: sc == STRUCT.index(x)
    cdel = is_("complex") & (dom == DOM.index("del_like"))
    cat = np.full(len(sc), "", dtype=object)
    for name, mask in (("DEL", is_("del")), ("INS", is_("ins")), ("CDEL", cdel),
                       ("CINS", is_("complex") & (dom == DOM.index("ins_like"))),
                       ("MNV_SUB", is_("mnp") | is_("substitution"))):
        cat[mask] = name
    anchor = np.where(is_("mnp") | ((is_("substitution") | is_("del") | cdel) & (L <= 50)), v["s0"] + (v["rs"] - 1) // 2, v["s0"])
    keep = cat != ""
    print(f"[variants] {int(keep.sum()):,} distinct non-SNV alleles ({time.time() - t0:.0f}s)")
    return pd.DataFrame({"chrom": np.array(CHROMS)[v["ch"][keep]], "anchor": anchor[keep], "size": L[keep], "class": cat[keep]})


def size_bin(sizes):
    out = np.empty(len(sizes), dtype=object)
    for i, b in enumerate(SIZE_BOUNDARIES):
        last = i + 1 == len(SIZE_BOUNDARIES)
        top = np.inf if last else SIZE_BOUNDARIES[i + 1] - 1
        out[(sizes >= b) & (sizes <= top)] = f"{b}+" if last else (str(b) if top == b else f"{b}–{top}")
    return out


def main():
    t0 = time.time()
    fai = pd.read_csv(FAI, sep="\t", header=None, usecols=[0, 1], names=["chrom", "len"])
    clen = dict(zip(fai.chrom, fai.len.astype(np.int64)))
    sets, frac = load_sets(clen)
    v = load_variants()
    v = v[v.chrom.isin(list(sets["Z"]))].reset_index(drop=True)
    for name in ("Z", "C"):
        f = np.full(len(v), np.nan)
        for chrom, idx in v.groupby("chrom", sort=False).indices.items():
            a = v.anchor.to_numpy()[idx]
            lo, hi = np.maximum(a - WINDOW, 0), np.minimum(a + WINDOW + 1, clen[chrom])
            f[idx] = (covered_before(hi, *sets[name][chrom]) - covered_before(lo, *sets[name][chrom])) / (hi - lo)
        v[f"enr_{name}"] = f / frac[name]
    v["bin"] = size_bin(v["size"].to_numpy())
    order = {b: i for i, b in enumerate(size_bin(np.array(SIZE_BOUNDARIES)))}

    rng = np.random.default_rng(SEED)
    rows, chrows, tests = [], [], []
    for cls in CLASSES:
        d = v[v["class"] == cls]
        bins = sorted((b for b, n in d.bin.value_counts().items() if n >= MIN_MUTS), key=order.get)
        if not bins:
            print(f"[WARN] {cls}: no size bin with >= {MIN_MUTS} variants")
            continue
        per_chr = {}
        for b in bins:
            g = d[d.bin == b].groupby("chrom")
            n, sz, scn = g.size(), g.enr_Z.sum(), g.enr_C.sum()
            idx = rng.integers(0, len(n), size=(N_BOOTS, len(n)))
            bz = sz.to_numpy()[idx].sum(1) / n.to_numpy()[idx].sum(1)
            bc = scn.to_numpy()[idx].sum(1) / n.to_numpy()[idx].sum(1)
            db = d[d.bin == b]
            rows.append({"class": cls, "bin": b, "bin_order": order[b], "n_variants": len(db),
                         "median_size": float(db["size"].median()), "enrich_Z": db.enr_Z.mean(),
                         "enrich_Z_lo": np.percentile(bz, 2.5), "enrich_Z_hi": np.percentile(bz, 97.5),
                         "enrich_C": db.enr_C.mean(), "enrich_C_lo": np.percentile(bc, 2.5),
                         "enrich_C_hi": np.percentile(bc, 97.5)})
            keep = n >= MIN_CHROM_EVENTS
            per_chr[b] = (sz / n)[keep]
            chrows += [{"class": cls, "bin": b, "bin_order": order[b], "chrom": c, "n_variants": int(n[c]),
                        "enrich_Z": sz[c] / n[c], "enrich_C": scn[c] / n[c]} for c in n.index[keep]]
        blocks = sorted(set.intersection(*(set(s.index) for s in per_chr.values()))) if per_chr else []
        if len(bins) >= 3 and len(blocks) >= 6:
            stat, p = friedmanchisquare(*[per_chr[b].reindex(blocks).to_numpy() for b in bins])
        else:
            stat = p = np.nan
        r = pd.DataFrame([x for x in rows if x["class"] == cls])
        pk = r.loc[r.enrich_Z.idxmax()]
        tests.append({"class": cls, "test": "Friedman (size bins; chromosomes as blocks)", "n_bins": len(bins),
                      "n_chromosomes": len(blocks), "chi2": stat, "p": p, "peak_bin": pk.bin, "peak_enrich_Z": pk.enrich_Z,
                      "peak_lo": pk.enrich_Z_lo, "peak_hi": pk.enrich_Z_hi})
    tests = pd.DataFrame(tests)
    m = tests.p.notna()
    tests["q"] = np.nan
    tests.loc[m, "q"] = multipletests(tests.p[m], method="fdr_bh")[1]
    pd.DataFrame(rows).to_csv(OUT, sep="\t", index=False, float_format="%.6g")
    pd.DataFrame(chrows).to_csv(OUT_CHR, sep="\t", index=False, float_format="%.6g")
    tests.to_csv(OUT_TESTS, sep="\t", index=False, float_format="%.6g")

    print(tests[["class", "chi2", "q", "peak_bin", "peak_enrich_Z"]].to_string(index=False, float_format=lambda x: f"{x:.3g}"))
    print(f"-> {OUT.name}, {OUT_CHR.name}, {OUT_TESTS.name} [{time.time() - t0:.0f}s]")


if __name__ == "__main__":
    main()
```

#### 17 · Paired score/length models → Fig. 4b, Suppl. Fig. 3b, Supplementary Table 3

```python
"""17 - Paired Z-DNA/control variant enrichment vs ZSeeker score per bp and tract length (Fig. 4b, Suppl. Fig. 3b,
Supplementary Table 3).

Counts of distinct normalized variant anchors in [start - WINDOW, end + WINDOW) of each matched pair. Model: conditional
Poisson as binomial GLM, logit(p) ~ standardized score/bp + log2(length) + chromosome; chromosome-clustered SEs,
chromosome-block bootstrap (classes in parallel), BH over seven classes per predictor. Sensitivity: 5% tails of score/bp
trimmed within 20 length strata. Plot input: per-chromosome Z/C ratios in tertiles balanced on the other predictor.
"""
from pathlib import Path
from concurrent.futures import ProcessPoolExecutor, as_completed
import multiprocessing as mp
import time
import numpy as np
import pandas as pd
import pyarrow as pa
import pyarrow.compute as pc
import pyarrow.csv as pacsv
from scipy.stats import spearmanr, norm
from statsmodels.stats.multitest import multipletests
import statsmodels.api as sm

BASE = Path('/path/to/pangenomes_zdna')
MASTER = BASE / '8_CHM13_controls_MASTER.tsv'
MUT_META = BASE / '9_hprc-v2.0-mc-chm13.wave.normalized.mutations.meta.tsv.gz'
FAI = BASE / 'tmp/chm13/chm13_refseq/GCF_009914755.1_T2T-CHM13v2.0_genomic_ucsc.fna.fai'
PREFIX = BASE / '17_score_length'
WINDOW = 10
TOL, DINUC_MODE = '10%', 'CpG_GpC'
CHROMS = [f'chr{i}' for i in range(1, 23)] + ['chrX', 'chrY']
CLASSES = ['SNV','MNV','small_del','small_ins','large_indel','complex_del','complex_ins']
STRUCT = ['snp','mnp','substitution','del','ins','complex']
DOM = ['snp','mnp','substitution','del','ins','del_like','ins_like','balanced']
BATCH_SIZE = 1 << 26
PA_THREADS = 16
N_BOOT = 500
BOOT_WORKERS = 7
SEED = 42
N_BALANCE_BINS = 20
SUPPORT_TRIM_FRACTION = 0.05


def load_pairs(chrom_sizes):
    cols = ['tol_level','dinuc_mode','Chromosome','Start0','End0','Ctrl_Start',
            'Ctrl_End','Z-DNA Score','has_control']
    m = pd.read_csv(MASTER, sep='\t', usecols=cols, low_memory=False,
                    dtype={'tol_level':str,'dinuc_mode':str,'has_control':str})
    m = m[(m.tol_level == TOL) & (m.dinuc_mode == DINUC_MODE) &
          m.has_control.str.strip().str.lower().eq('true') &
          m.Chromosome.isin(CHROMS)].copy()
    if m.duplicated(['Chromosome','Start0','End0']).any():
        raise ValueError('Duplicate matched Z-DNA loci for chosen tolerance')
    for col in ['Start0','End0','Ctrl_Start','Ctrl_End','Z-DNA Score']:
        m[col] = pd.to_numeric(m[col], errors='raise')
    d = pd.DataFrame({'chrom':m.Chromosome.to_numpy(),
        'z_start':m.Start0.to_numpy(np.int64),
        'z_end':m.End0.to_numpy(np.int64),
        'c_start':m.Ctrl_Start.to_numpy(np.int64),
        'c_end':m.Ctrl_End.to_numpy(np.int64),
        'score':m['Z-DNA Score'].to_numpy(float)})
    d['length'] = d.z_end - d.z_start
    if (d.length<=0).any() or not ((d.c_end-d.c_start)==d.length).all():
        raise AssertionError('Nonpositive or unmatched pair length')
    d['score_per_bp'] = d.score / d.length
    for side in ('z','c'):
        d[f'{side}_lo'] = (d[f'{side}_start']-WINDOW).clip(lower=0)
        d[f'{side}_hi'] = np.minimum(d[f'{side}_end'].to_numpy()+WINDOW,
                                    d.chrom.map(chrom_sizes).to_numpy(np.int64))
        d[f'{side}_bp'] = d[f'{side}_hi']-d[f'{side}_lo']
    valid = d.z_bp.eq(d.c_bp)
    print(f'[QC] Dropped {(~valid).sum():,} pairs with unequal expanded exposures')
    d=d.loc[valid].sort_values(['chrom','z_start']).reset_index(drop=True)
    print(f'[pairs] {len(d):,} | rho(raw score,length)={spearmanr(d.score,d.length).statistic:.3f} '
          f'| rho(score/bp,length)={spearmanr(d.score_per_bp,d.length).statistic:.3f}')
    return d


def load_variants():
    """Identical class and anchor selection to 16.1, with normalized-allele deduplication."""
    pa.set_cpu_count(PA_THREADS)
    types={'seqID':pa.string(),'normalized_start0':pa.int64(),'normalized_alt':pa.string(),
           'struct_class':pa.string(),'dominant_op':pa.string(),'ref_span':pa.int64(),'event_size':pa.int64()}
    r=pacsv.open_csv(MUT_META,read_options=pacsv.ReadOptions(block_size=BATCH_SIZE),
                     parse_options=pacsv.ParseOptions(delimiter='\t'),
                     convert_options=pacsv.ConvertOptions(include_columns=list(types),column_types=types,
                                                          strings_can_be_null=False))
    keys=['ch','s0','rs','L','sc','dom','h']; parts={k:[] for k in keys}
    for b in r:
        chi=pc.index_in(b.column('seqID'),value_set=pa.array(CHROMS))
        b,chi=b.filter(pc.is_valid(chi)),chi.filter(pc.is_valid(chi))
        if not b.num_rows: continue
        sc=pc.index_in(b.column('struct_class'),value_set=pa.array(STRUCT)).to_numpy(zero_copy_only=False)
        dom=pc.index_in(b.column('dominant_op'),value_set=pa.array(DOM)).to_numpy(zero_copy_only=False)
        vals=(chi.to_numpy(zero_copy_only=False).astype(np.int8),
              b.column('normalized_start0').to_numpy(),b.column('ref_span').to_numpy(),
              b.column('event_size').to_numpy(),sc.astype(np.int8),dom.astype(np.int8),
              pd.util.hash_array(b.column('normalized_alt').to_numpy(zero_copy_only=False)))
        for k,val in zip(keys,vals): parts[k].append(val)
    v={k:np.concatenate(val) for k,val in parts.items()}
    o=np.lexsort((v['h'],v['rs'],v['s0'],v['ch']))
    v={k:a[o] for k,a in v.items()}; unique=np.ones(len(o),bool)
    unique[1:]=( (v['ch'][1:]!=v['ch'][:-1])|(v['s0'][1:]!=v['s0'][:-1])|
                (v['rs'][1:]!=v['rs'][:-1])|(v['h'][1:]!=v['h'][:-1]) )
    v={k:a[unique] for k,a in v.items()}
    sc,dom,L=v['sc'],v['dom'],v['L']
    is_ = lambda name: sc==STRUCT.index(name)
    cdel=is_('complex')&(dom==DOM.index('del_like'))
    cls=np.full(len(sc),-1,np.int8)
    groups=[('SNV',is_('snp')),('MNV',is_('mnp')),
            ('small_del',is_('del')&(L<=50)),('small_ins',is_('ins')&(L<=50)),
            ('large_indel',(is_('del')|is_('ins'))&(L>50)),
            ('complex_del',cdel),('complex_ins',is_('complex')&(dom==DOM.index('ins_like')))]
    for name,mask in groups: cls[mask]=CLASSES.index(name)
    anchor=np.where(is_('mnp')|((is_('substitution')|is_('del')|cdel)&(L<=50)),
                    v['s0']+(v['rs']-1)//2,v['s0'])
    keep=cls>=0
    print(f'[variants] {int(keep.sum()):,} distinct selected normalized alleles')
    return v['ch'][keep],anchor[keep],cls[keep]


def count_anchors(d, ch, pos, cls, side):
    out=np.zeros((len(d),len(CLASSES)),np.int64)
    for ci,c in enumerate(CHROMS):
        ix=np.flatnonzero(d.chrom.to_numpy()==c)
        if not len(ix): continue
        p=pos[ch==ci]; typ=cls[ch==ci]
        for j in range(len(CLASSES)):
            pj=np.sort(p[typ==j]);
            out[ix,j]=np.searchsorted(pj,d[f'{side}_hi'].to_numpy()[ix],side='left')-\
                      np.searchsorted(pj,d[f'{side}_lo'].to_numpy()[ix],side='left')
    return out


def design_matrix(d):
    """Fix predictor scaling using the full matched-pair dataset."""
    d['z_spb']=(d.score_per_bp-d.score_per_bp.mean())/d.score_per_bp.std(ddof=0)
    d['log2_length']=np.log2(d.length)-np.log2(d.length.median())
    chrom = pd.get_dummies(pd.Categorical(d.chrom,categories=CHROMS),
                           prefix='chr',drop_first=True,dtype=float)
    chrom.index=d.index
    return pd.concat([pd.Series(1.,index=d.index,name='intercept'),
                      d[['z_spb','log2_length']],chrom],axis=1).astype(float)


def fit_one(d, X, cat, weights=None, clustered=True):
    yz=d[f'z_{cat}'].to_numpy(np.int64)
    yc=d[f'c_{cat}'].to_numpy(np.int64)
    total=yz+yc
    good=total>0
    if weights is not None:
        w=d.chrom.map(weights).fillna(0).to_numpy(float)
        good &= w>0
    if good.sum()<100:
        return None
    matrix=X.loc[good].copy()
    matrix=matrix.loc[:, (matrix.nunique(dropna=False)>1)|(matrix.columns=='intercept')]
    if weights is not None:
        observed = set(d.loc[good, "chrom"])
        if "chr1" not in observed:
            reference = next(c for c in CHROMS if c in observed)
            matrix = matrix.drop(columns=[f"chr_{reference}"])
    y=np.column_stack([yz[good],yc[good]])
    opts={'family':sm.families.Binomial()}
    if weights is not None:
        opts['freq_weights']=w[good]
    gl=sm.GLM(y,matrix,**opts)
    if clustered:
        groups=pd.Categorical(d.chrom[good],categories=CHROMS).codes
        fitted=gl.fit(cov_type='cluster',cov_kwds={'groups':groups},maxiter=150,disp=0)
    else:
        fitted=gl.fit(maxiter=150,disp=0)
    if not fitted.converged:
        raise RuntimeError(f'{cat} model did not converge')
    return fitted,int(good.sum())


def model_table(d,X,scenario):
    rows=[]
    for cat in CLASSES:
        res=fit_one(d,X,cat)
        if res is None: raise RuntimeError(f'Insufficient informative pairs: {cat}')
        fit,n=res
        for term in ('z_spb','log2_length'):
            beta,se=float(fit.params[term]),float(fit.bse[term])
            p=2*norm.sf(abs(beta/se))
            rows.append({'scenario':scenario,'class':cat,'term':term,
                         'estimate':np.exp(beta),'ci_lo':np.exp(beta-1.96*se),
                         'ci_hi':np.exp(beta+1.96*se),'p':p,
                         'n_informative_pairs':n,'converged':True})
    tab=pd.DataFrame(rows)
    tab['q']=np.nan
    for term,g in tab.groupby('term'):
        tab.loc[g.index,'q']=multipletests(g.p,method='fdr_bh')[1]
    return tab


def balance_tertiles(x, matching_variable):
    """Three display groups *within* each of 20 rank-quantile nuisance strata."""
    strata=pd.qcut(matching_variable.rank(method='first'),N_BALANCE_BINS,labels=False)
    return x.groupby(strata).transform(
        lambda a:pd.qcut(a.rank(method='first'),3,labels=False)
    ).to_numpy(np.int8)


def chromosome_tertiles(d):
    tmp=d.copy()
    tmp['score_bin']=balance_tertiles(tmp.score_per_bp,tmp.length)
    tmp['length_bin']=balance_tertiles(tmp.length.astype(float),tmp.score_per_bp)
    rows=[]
    for predictor,column in [('score_per_bp','score_bin'),('tract_length','length_bin')]:
        for (chrom,g),sub in tmp.groupby(['chrom',column],sort=False):
            if sub.z_bp.sum()!=sub.c_bp.sum():raise AssertionError('Pair exposure changed')
            for cat in CLASSES:
                nz=int(sub[f'z_{cat}'].sum());nc=int(sub[f'c_{cat}'].sum())
                rows.append({'predictor':predictor,'class':cat,'chrom':chrom,
                             'group':['Low','Mid','High'][int(g)],'n_loci':len(sub),
                             'z_count':nz,'c_count':nc,
                             'window_bp':int(sub.z_bp.sum()),
                             'ratio_raw':(nz/nc if nc>0 else np.nan),
                             'ratio_display':(nz+0.5)/(nc+0.5)})
    return pd.DataFrame(rows)


def score_support(d):
    strata=pd.qcut(d.length.rank(method='first'),N_BALANCE_BINS,labels=False)
    rank=d.groupby(strata).score_per_bp.rank(pct=True)
    return rank.between(SUPPORT_TRIM_FRACTION,1-SUPPORT_TRIM_FRACTION).to_numpy()


def bootstrap_class(cat,d,X,nrep,seed):
    rng=np.random.default_rng(seed)
    out=[]
    for iteration in range(nrep):
        sampled=rng.integers(0,len(CHROMS),len(CHROMS))
        n=np.bincount(sampled,minlength=len(CHROMS))
        weights=dict(zip(CHROMS,n))
        try:
            res=fit_one(d,X,cat,weights=weights,clustered=False)
            if res is None:continue
            fit,_=res
            for term in ('z_spb','log2_length'):
                out.append({'class':cat,'term':term,'replicate':iteration,
                            'estimate':float(np.exp(fit.params[term]))})
        except (ValueError, np.linalg.LinAlgError, RuntimeError) as exc:
            print(
                f"[BOOTSTRAP FAILURE] {cat}, replicate {iteration}: "
                f"{type(exc).__name__}: {exc}",
                flush=True
            )
            continue
    return out


def main():
    t0=time.time()
    sizes=pd.read_csv(FAI,sep='\t',header=None,usecols=[0,1],names=['chrom','bp'])
    sizes=dict(zip(sizes.chrom,sizes.bp.astype(np.int64)))
    d=load_pairs(sizes)
    chrom,pos,cat=load_variants()
    for side in ('z','c'):
        counts=count_anchors(d,chrom,pos,cat,side)
        for k,cls in enumerate(CLASSES):d[f'{side}_{cls}']=counts[:,k]
    del chrom,pos,cat
    X=design_matrix(d)
    primary=model_table(d,X,'primary')
    mask=score_support(d)
    sensitivity=model_table(d.loc[mask],X.loc[mask],'score_support_trim')
    models=pd.concat([primary,sensitivity],ignore_index=True)
    models.to_csv(f'{PREFIX}_models.tsv',sep='\t',index=False,float_format='%.8g')
    chromosome_tertiles(d).to_csv(f'{PREFIX}_chromosome_tertiles.tsv',sep='\t',index=False)
    d.to_csv(f'{PREFIX}_pairs.tsv.gz',sep='\t',index=False,compression='gzip')
    pd.DataFrame([{'metric':'all_pairs','n':len(d)},
                  {'metric':'score_support_pairs','n':int(mask.sum())},
                  {'metric':'window_flank_bp','n':WINDOW},
                  {'metric':'bootstrap_replicates','n':N_BOOT}]).to_csv(
                     f'{PREFIX}_qc.tsv',sep='\t',index=False)
    boot=[]
    if N_BOOT>0:
        print(f'\n[BOOTSTRAP] {N_BOOT} resamples/class; {BOOT_WORKERS} workers')
        with ProcessPoolExecutor(max_workers=min(BOOT_WORKERS,len(CLASSES)),
                                 mp_context=mp.get_context('fork')) as pool:
            futures={pool.submit(bootstrap_class,cls,d,X,N_BOOT,SEED+100*i):cls
                     for i,cls in enumerate(CLASSES)}
            for fut in as_completed(futures):
                result=fut.result();boot.extend(result)
    boot=pd.DataFrame(boot,columns=['class','term','replicate','estimate'])
    boot.to_csv(f'{PREFIX}_bootstrap_replicates.tsv.gz',sep='\t',index=False)
    if len(boot):
        boot_ci=(boot.groupby(['class','term']).estimate.agg(
            n_success='count',ci_boot_lo=lambda a:a.quantile(0.025),
            ci_boot_hi=lambda a:a.quantile(0.975)).reset_index())
    else:
        boot_ci=pd.DataFrame(columns=['class','term','n_success','ci_boot_lo','ci_boot_hi'])
    boot_ci.to_csv(f'{PREFIX}_bootstrap_ci.tsv',sep='\t',index=False,float_format='%.8g')
    print(models[models.scenario=='primary'][['class','term','estimate','q']].to_string(index=False,float_format=lambda x:f'{x:.3g}'))
    print(f'-> {PREFIX.name}_* [{time.time()-t0:.0f}s]')


if __name__=='__main__':
    main()
```

#### Fig. 4a, Suppl. Fig. 3a

```python
"""Fig. 4a and Suppl. Fig. 3a - Z-DNA and control enrichment within +/-10 bp of variants by variant size: pooled mean with
95% chromosome-bootstrap CI (lines) and per-chromosome means (points); Friedman BH q per panel + legend. Reads 16.
"""
from pathlib import Path

import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from matplotlib.lines import Line2D
from matplotlib.ticker import FixedLocator, FuncFormatter, NullLocator

BASE = Path("/path/to/pangenomes_zdna")
SUMMARY, BYCHROM, TESTS_IN = (BASE / f"16_variant_size_enrichment{s}.tsv" for s in ("", "_bychrom", "_tests"))
PLOTS, TESTS = BASE / "plots", BASE / "tests"
FIGURES = {"fig4a_variant_size_enrichment": (["DEL", "INS", "CDEL", "CINS"], (2, 2), (8.2, 5.4)),
           "suppl_fig3a_variant_size_enrichment": (["MNV_SUB"], (1, 1), (4.2, 2.8))}
LEGEND_STEM = "fig4a_variant_size_enrichment_legend"
COLUMNS = {"record": "Record", "class": "Variant class", "bin": "Size bin (bp)", "chrom": "Chromosome",
           "n_variants": "Variants (n)", "median_size": "Median size (bp)", "enrich_Z": "Z-DNA enrichment",
           "enrich_Z_lo": "Z-DNA enrichment, 95% CI lower", "enrich_Z_hi": "Z-DNA enrichment, 95% CI upper",
           "enrich_C": "Control enrichment", "enrich_C_lo": "Control enrichment, 95% CI lower",
           "enrich_C_hi": "Control enrichment, 95% CI upper"}
STATS_COLUMNS = {"class": "Variant class", "test": "Test", "n_bins": "Size bins (n)", "n_chromosomes": "Chromosome blocks (n)",
                 "chi2": "Friedman chi-square", "p": "P value", "q": "q value (BH)", "peak_bin": "Peak size bin",
                 "peak_enrich_Z": "Peak Z-DNA enrichment", "peak_lo": "Peak, 95% CI lower", "peak_hi": "Peak, 95% CI upper"}
TITLES = {"DEL": "Deletions", "INS": "Insertions", "CDEL": "Complex deletions", "CINS": "Complex insertions",
          "MNV_SUB": "MNVs and substitutions"}
Y_TICKS = [0.5, 1, 2, 5, 10, 20, 50]
DODGE, JITTER = 0.17, 0.05
POINT_S, POINT_ALPHA, MS, LINE_W = 11, 0.75, 5.5, 1.4
AXIS_W, TICK_W, TICK_LEN = 1.1, 0.9, 3
FS_TICK, FS_AXIS, FS_TITLE, FS_NOTE, FS_LEGEND = 10, 11, 11, 9.5, 9.5
X_ROT = 35
GRID = dict(linestyle="--", linewidth=0.4, color="#B0B0B0", alpha=0.5, zorder=0)
FONT, DPI = "Nimbus Sans", 600
COLORS = {"Z": ("#B3A7E0", "#6E5BC4"), "C": ("#BDBDBD", "#7A7A7A")}

plt.rcParams.update({"font.family": FONT, "pdf.fonttype": 42, "ps.fonttype": 42, "axes.linewidth": AXIS_W,
                     "axes.spines.top": False, "axes.spines.right": False})
PLOTS.mkdir(exist_ok=True)
TESTS.mkdir(exist_ok=True)
summ, bych, tests = (pd.read_csv(p, sep="\t") for p in (SUMMARY, BYCHROM, TESTS_IN))
rng = np.random.default_rng(0)


def sci(p):
    if not np.isfinite(p):
        return "n.a."
    if p >= 0.001:
        return f"{p:.3f}"
    m, e = f"{p:.1e}".split("e")
    return f"{m} × 10$^{{{int(e)}}}$"


def panel(ax, cls):
    s = summ[summ["class"] == cls].sort_values("bin_order")
    x = {b: i for i, b in enumerate(s.bin)}
    for key, dx in (("C", DODGE), ("Z", -DODGE)):
        pts, line = COLORS[key]
        c = bych[bych["class"] == cls]
        ax.scatter(c.bin.map(x) + dx + rng.uniform(-JITTER, JITTER, len(c)), c[f"enrich_{key}"], s=POINT_S, color=pts,
                   alpha=POINT_ALPHA, edgecolors="black", linewidths=0.3, zorder=3)
        xi = np.arange(len(s)) + dx
        ax.plot(xi, s[f"enrich_{key}"], color=line, lw=LINE_W, zorder=4)
        ax.errorbar(xi, s[f"enrich_{key}"], yerr=[(s[f"enrich_{key}"] - s[f"enrich_{key}_lo"]).clip(lower=0),
                                                  (s[f"enrich_{key}_hi"] - s[f"enrich_{key}"]).clip(lower=0)],
                    fmt="o", ms=MS, mfc=line, mec="black", mew=0.5, ecolor="black", elinewidth=0.8, capsize=1.6, zorder=5)
    ax.axhline(1, ls=":", color="black", lw=0.8, zorder=2)
    ax.set_yscale("log")
    lo, hi = ax.get_ylim()
    ax.yaxis.set_major_locator(FixedLocator([t for t in Y_TICKS if lo <= t <= hi]))
    ax.yaxis.set_minor_locator(NullLocator())
    ax.yaxis.set_major_formatter(FuncFormatter(lambda v, _: f"{v:g}"))
    ax.set_xticks(range(len(s)), s.bin, rotation=X_ROT, ha="right", rotation_mode="anchor")
    ax.set_xlim(-0.6, len(s) - 0.4)
    t = tests.set_index("class").loc[cls]
    ax.set_title(f"{TITLES[cls]}", fontsize=FS_TITLE, loc="center")
    if np.isfinite(t.q):
        note = f"q = {sci(t.q)}"
    else:
        note = f"q = n.a. (n = {int(t.n_chromosomes)} chromosomes)"

    ax.text(
        0.99, 0.03, note,
        transform=ax.transAxes,
        ha="right", va="bottom",
        fontsize=FS_NOTE
)
    ax.tick_params(labelsize=FS_TICK, width=TICK_W, length=TICK_LEN)
    ax.grid(True, axis="y", which="major", **GRID)
    ax.set_axisbelow(True)


for stem, (classes, grid, size) in FIGURES.items():
    fig, axs = plt.subplots(*grid, figsize=size, constrained_layout=True, squeeze=False)
    for ax, cls in zip(axs.ravel(), classes):
        panel(ax, cls)
    for ax in axs[:, 0]:
        ax.set_ylabel("Enrichment (±10 bp)", fontsize=FS_AXIS)
    for ax in axs[-1, :]:
        ax.set_xlabel("Variant size (bp)", fontsize=FS_AXIS)
    for ext in ("pdf", "png"):
        fig.savefig(PLOTS / f"{stem}.{ext}", dpi=DPI if ext == "png" else None)
    plt.close(fig)
    d = pd.concat([summ[summ["class"].isin(classes)].assign(record="Pooled"),
                   bych[bych["class"].isin(classes)].assign(record="Chromosome")], ignore_index=True)
    d.assign(**{"class": d["class"].map(TITLES)})[[c for c in COLUMNS if c in d.columns]].rename(columns=COLUMNS).to_csv(
        TESTS / f"{stem}.tsv", sep="\t", index=False, float_format="%.6g")
    t = tests[tests["class"].isin(classes)]
    t.assign(**{"class": t["class"].map(TITLES)})[list(STATS_COLUMNS)].rename(columns=STATS_COLUMNS).to_csv(
        TESTS / f"{stem}_statistics.tsv", sep="\t", index=False, float_format="%.6g")

h = [Line2D([0], [0], color=COLORS[k][1], marker="o", ms=MS, mfc=COLORS[k][1], mec="black", mew=0.5, lw=LINE_W, label=l)
     for k, l in (("Z", "Z-DNA"), ("C", "Matched control"))]
h += [Line2D([0], [0], ls="none", marker="o", ms=4, color=COLORS[k][0], markeredgecolor="black",
             markeredgewidth=0.3, alpha=POINT_ALPHA, label=l)
      for k, l in (("Z", "Chromosome (Z-DNA)"), ("C", "Chromosome (control)"))]
lf, lax = plt.subplots(figsize=(2.4, 1.0))
lax.axis("off")
lax.legend(handles=h, loc="center", frameon=False, fontsize=FS_LEGEND)
for ext in ("pdf", "png"):
    lf.savefig(PLOTS / f"{LEGEND_STEM}.{ext}", dpi=DPI if ext == "png" else None, bbox_inches="tight")
plt.close(lf)
```

#### Fig. 4b, Suppl. Fig. 3b, Supplementary Table 3

```python
"""Fig. 4b and Suppl. Fig. 3b - Z-DNA/control variant-count ratios (locus +/- 10 bp, log scale) by ZSeeker score per bp
(top; tertiles balanced over length) and tract length (bottom; balanced over score per bp); per-chromosome points and
boxes; stars = BH q of the continuous paired model + legend. Also writes Supplementary_Table_3.tsv. Reads 17.
"""
from pathlib import Path

import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from matplotlib.patches import Patch
from matplotlib.ticker import FixedLocator, FuncFormatter, NullLocator

BASE = Path("/path/to/pangenomes_zdna")
PREFIX = BASE / "17_score_length"
TABLE_S3 = BASE / "Supplementary_Table_3.tsv"
MODELS, CHROM_DATA, BOOT = (Path(f"{PREFIX}_{k}.tsv") for k in ("models", "chromosome_tertiles", "bootstrap_ci"))
PLOTS, TESTS = BASE / "plots", BASE / "tests"
SCENARIO = "primary"
FIGURES = {"fig4b_score_length_ratios": (["SNV", "small_del", "small_ins", "complex_ins"], (8.2, 5.2)),
           "suppl_fig3b_score_length_ratios": (["MNV", "large_indel", "complex_del"], (6.4, 5.2))}
PREDICTOR_NAMES = {"score_per_bp": "ZSeeker score per bp", "tract_length": "Tract length"}
TERM_NAMES = {"z_spb": "Score per bp (per SD)", "log2_length": "Tract length (per doubling)"}
SCENARIO_NAMES = {"primary": "Primary", "score_support_trim": "Score-support trimmed (5% tails)"}
COLUMNS = {"predictor": "Predictor", "class": "Variant class", "chrom": "Chromosome", "group": "Tertile",
           "n_loci": "Matched pairs (n)", "z_count": "Variants at Z-DNA (n)", "c_count": "Variants at controls (n)",
           "window_bp": "Window per cohort (bp)", "ratio_raw": "Z-DNA/control ratio",
           "ratio_display": "Z-DNA/control ratio (+0.5, plotted)", "model_term": "Model term", "model_q": "Model q value (BH)",
           "model_sig": "Model significance"}
MODEL_COLUMNS = {"scenario": "Scenario", "class": "Variant class", "term": "Model term",
                 "estimate": "Ratio multiplier", "ci_lo": "95% CI lower (cluster-robust)",
                 "ci_hi": "95% CI upper (cluster-robust)", "p": "P value", "q": "q value (BH)",
                 "n_informative_pairs": "Informative pairs (n)", "n_success": "Bootstrap fits (n)",
                 "ci_boot_lo": "95% CI lower (chromosome bootstrap)", "ci_boot_hi": "95% CI upper (chromosome bootstrap)"}
ROWS = [("score_per_bp", "z_spb", "Score per bp tertile (length-balanced)", ["#F0EEDC", "#DEDAAC", "#CCCB7A"]),
        ("tract_length", "log2_length", "Length tertile (score-per-bp-balanced)", ["#D8EDE2", "#A9D4BE", "#76B593"])]
GROUPS = ["Low", "Mid", "High"]
CLASS_LABELS = {"SNV": "SNVs", "MNV": "MNVs", "small_del": "Deletions ≤50 bp", "small_ins": "Insertions ≤50 bp",
                "large_indel": "Indels >50 bp", "complex_del": "Complex deletions", "complex_ins": "Complex insertions"}
Y_TICKS = [0.25, 0.5, 1, 2, 4, 8, 16, 32]
BOX_W, BOX_LW = 0.55, 0.9
POINT_S, POINT_ALPHA, JITTER = 9, 0.7, 0.1
AXIS_W, TICK_W, TICK_LEN = 1.1, 0.9, 3
FS_TICK, FS_AXIS, FS_TITLE, FS_ROW, FS_SIG = 9.5, 10, 10, 10.5, 9.5
GRID = dict(linestyle="--", linewidth=0.4, color="#B0B0B0", alpha=0.5, zorder=0)
FONT, DPI = "Nimbus Sans", 600

plt.rcParams.update({"font.family": FONT, "pdf.fonttype": 42, "ps.fonttype": 42, "axes.linewidth": AXIS_W,
                     "axes.spines.top": False, "axes.spines.right": False})
PLOTS.mkdir(exist_ok=True)
TESTS.mkdir(exist_ok=True)
models = pd.read_csv(MODELS, sep="\t")
data = pd.read_csv(CHROM_DATA, sep="\t")
boots = pd.read_csv(BOOT, sep="\t") if BOOT.exists() and BOOT.stat().st_size > 0 else pd.DataFrame(columns=["class", "term"])
primary = models[models.scenario == SCENARIO].set_index(["class", "term"])
rng = np.random.default_rng(0)


def readable(t):
    t = t.assign(scenario=t.scenario.map(SCENARIO_NAMES), term=t.term.map(TERM_NAMES), **{"class": t["class"].map(CLASS_LABELS)})
    return t[[c for c in MODEL_COLUMNS if c in t.columns]].rename(columns=MODEL_COLUMNS)


def stars(q):
    if not np.isfinite(q):
        return "n.a."
    return "***" if q < 1e-3 else "**" if q < 1e-2 else "*" if q < 0.05 else "n.s."


def panel(ax, predictor, term, cls, palette, title):
    d = data[(data["class"] == cls) & (data.predictor == predictor)]
    vals = [np.log2(d.loc[d.group == g, "ratio_display"].to_numpy(float)) for g in GROUPS]
    if any((~np.isfinite(v)).any() for v in vals):
        raise RuntimeError(f"Non-positive display ratio: {cls} / {predictor}")
    bp = ax.boxplot(vals, positions=[1, 2, 3], widths=BOX_W, patch_artist=True, showfliers=False, zorder=2,
                    medianprops={"color": "black", "lw": BOX_LW}, whiskerprops={"lw": BOX_LW}, capprops={"lw": BOX_LW},
                    boxprops={"lw": BOX_LW})
    for box, col in zip(bp["boxes"], palette):
        box.set_facecolor(col)
    for i, v in enumerate(vals, start=1):
        ax.scatter(i + rng.uniform(-JITTER, JITTER, len(v)), v, s=POINT_S, color="black", alpha=POINT_ALPHA, lw=0, zorder=3)
    ax.axhline(0, ls=":", color="black", lw=0.8, zorder=1)
    allv = np.concatenate(vals)
    lo, hi = min(allv.min(), 0.0), max(allv.max(), 0.0)
    span = max(hi - lo, 0.5)
    ax.set_ylim(lo - 0.06 * span, hi + 0.22 * span)
    if (cls, term) not in primary.index:
        raise RuntimeError(f"Missing {SCENARIO} model for {cls} / {term}")
    s = "Model: " + stars(float(primary.loc[(cls, term), "q"]))
    ax.text(2, hi + 0.08 * span, s, ha="center", va="bottom", fontsize=FS_SIG, style="italic" if s == "n.s." else "normal")
    lo_t, hi_t = ax.get_ylim()
    ax.yaxis.set_major_locator(FixedLocator([np.log2(t) for t in Y_TICKS if lo_t <= np.log2(t) <= hi_t]))
    ax.yaxis.set_minor_locator(NullLocator())
    ax.yaxis.set_major_formatter(FuncFormatter(lambda v, _: f"{2 ** v:g}"))
    ax.set_xticks([1, 2, 3], GROUPS)
    ax.set_xlim(0.4, 3.6)
    if title:
        ax.set_title(CLASS_LABELS.get(cls, cls), fontsize=FS_TITLE)
    ax.tick_params(labelsize=FS_TICK, width=TICK_W, length=TICK_LEN)
    ax.grid(True, axis="y", which="major", **GRID)
    ax.set_axisbelow(True)
    return d.assign(model_term=term, model_q=float(primary.loc[(cls, term), "q"]), model_sig=s)


for stem, (classes, size) in FIGURES.items():
    fig = plt.figure(figsize=size, constrained_layout=True)
    subfigs = fig.subfigures(len(ROWS), 1)
    drawn = []
    for r, (sf, (predictor, term, row_title, palette)) in enumerate(zip(subfigs, ROWS)):
        axs = np.atleast_1d(sf.subplots(1, len(classes)))
        for ax, cls in zip(axs, classes):
            drawn.append(panel(ax, predictor, term, cls, palette, title=(r == 0)))
        axs[0].set_ylabel("Z-DNA/control ratio", fontsize=FS_AXIS)
        sf.supxlabel(row_title, fontsize=FS_ROW)
    for ext in ("pdf", "png"):
        fig.savefig(PLOTS / f"{stem}.{ext}", dpi=DPI if ext == "png" else None)
    plt.close(fig)
    d = pd.concat(drawn, ignore_index=True)
    d.assign(predictor=d.predictor.map(PREDICTOR_NAMES), model_term=d.model_term.map(TERM_NAMES),
             **{"class": d["class"].map(CLASS_LABELS)})[list(COLUMNS)].rename(columns=COLUMNS).to_csv(
        TESTS / f"{stem}.tsv", sep="\t", index=False, float_format="%.6g")
    terms = [t for _, t, _, _ in ROWS]
    st = models[models["class"].isin(classes) & models.term.isin(terms)].merge(boots.assign(scenario="primary"), on=["scenario", "class", "term"], how="left")
    readable(st).to_csv(TESTS / f"{stem}_statistics.tsv", sep="\t", index=False, float_format="%.6g")

readable(models.merge(boots.assign(scenario="primary"), on=["scenario", "class", "term"], how="left")).to_csv(TABLE_S3, sep="\t", index=False, float_format="%.4g")
print(f"-> {TABLE_S3.name}")

LEGEND_STEM = "fig4b_score_length_legend"
handles = [Patch(facecolor=ROWS[0][3][2], edgecolor="black", linewidth=0.5, label="Z-DNA score per bp"),
           Patch(facecolor=ROWS[1][3][2], edgecolor="black", linewidth=0.5, label="Z-DNA tract length"),
           Patch(facecolor="#E8E8E8", edgecolor="black", linewidth=0.5, label="Low"),
           Patch(facecolor="#AAAAAA", edgecolor="black", linewidth=0.5, label="Mid"),
           Patch(facecolor="#555555", edgecolor="black", linewidth=0.5, label="High"),
           Patch(facecolor="none", edgecolor="none", label="Model q (score per bp and length together):"),
           Patch(facecolor="none", edgecolor="none", label="***   q < 0.001"),
           Patch(facecolor="none", edgecolor="none", label="**     q < 0.01"),
           Patch(facecolor="none", edgecolor="none", label="*       q < 0.05"),
           Patch(facecolor="none", edgecolor="none", label="n.s.   not significant")]
lf, lax = plt.subplots(figsize=(3.2, 2.9))
lax.axis("off")
lax.legend(handles=handles, loc="center", frameon=False, fontsize=9, handlelength=1.3, handleheight=1.0,
           handletextpad=0.7, labelspacing=0.45)
for ext in ("pdf", "png"):
    lf.savefig(PLOTS / f"{LEGEND_STEM}.{ext}", dpi=DPI if ext == "png" else None, bbox_inches="tight")
plt.close(lf)
```

## Results 5 – Genomic context, allele frequency and indel mechanism (Fig. 5, Suppl. Fig. 3c–e, Supplementary Table 4)

### Analysis

#### 18.1 · Compartment membership → Suppl. Fig. 3c

```python
"""18.1 - Per-locus membership of controls, matched and unmatched Z-DNA in coarse genomic compartments (unions of
6.3 classes; Intergenic / Outside repeats = complements); a locus belongs to every compartment it overlaps by >= 1 bp
(Suppl. Fig. 3c; membership reused by 18.2, 19, 20).
"""
from pathlib import Path

import numpy as np
import pandas as pd

BASE = Path("/path/to/pangenomes_zdna")
LOCI = BASE / "8_CHM13_matching_loci.tsv.gz"
FEATDIR = BASE / "6_chm13_features"
FAI = BASE / "tmp/chm13/chm13_refseq/GCF_009914755.1_T2T-CHM13v2.0_genomic_ucsc.fna.fai"
OUT_LOCI = BASE / "18_feature_membership_loci.tsv.gz"
OUT = BASE / "18_feature_membership.tsv"

COHORTS = ["control", "Z_matched", "Z_unmatched"]
CENSAT_SAT = [f"censat_{c}" for c in ("censat", "hor", "dhor", "mon", "hsat1A", "hsat1B", "hsat2", "hsat3", "gsat", "bsat")]
GENES = ["gene_body", "rDNA_arrays", "tRNA_arrays"]                  # complement = Intergenic
REPEATS = ["rmsk_all", "telomere", "rDNA_arrays"] + CENSAT_SAT     # complement = Outside repeats
GROUPS = {
    "Protein-coding genes": ("genes", ["gene_protein_coding"]),
    "CDS": ("genes", ["pc_CDS"]),
    "Exons": ("genes", ["pc_exon"]),
    "Introns": ("genes", ["pc_intron"]),
    "lncRNA genes": ("genes", ["gene_lncRNA"]),
    "Other genes": ("genes", ["gene_miRNA", "gene_snoRNA", "gene_pseudogene", "gene_other"]),
    "Intergenic": ("genes", "NOT:GENES"),
    "Simple repeats": ("repeats", ["rmsk_Simple_repeat"]),
    "Low complexity": ("repeats", ["rmsk_Low_complexity"]),
    "Satellites": ("repeats", ["rmsk_Satellite"] + CENSAT_SAT),
    "LINEs": ("repeats", ["rmsk_LINE"]),
    "SINEs": ("repeats", ["rmsk_SINE"]),
    "LTRs": ("repeats", ["rmsk_LTR"]),
    "DNA transposons": ("repeats", ["rmsk_DNA_transposon", "rmsk_DNA_other"]),
    "rDNA": ("repeats", ["rDNA_arrays"]),
    "Outside repeats": ("repeats", "NOT:REPEATS"),
}
CHROMS = [f"chr{i}" for i in range(1, 23)] + ["chrX", "chrY"]


def merge(s, e):
    if len(s) == 0:
        return np.empty(0, np.int64), np.empty(0, np.int64)
    o = np.argsort(s, kind="stable"); s, e = np.asarray(s)[o], np.asarray(e)[o]
    new = np.ones(len(s), bool); new[1:] = s[1:] > np.maximum.accumulate(e)[:-1]
    return s[new].astype(np.int64), np.maximum.reduceat(e, np.flatnonzero(new)).astype(np.int64)


def load_union(classes):
    """Union of 6.3 classes, per chromosome; missing files are reported and skipped."""
    parts = []
    for c in classes:
        p = FEATDIR / f"{c}.bed"
        if not p.exists():
            print(f"  [missing] {p.name}")
            continue
        parts.append(pd.read_csv(p, sep="\t", header=None, usecols=[0, 1, 2], names=["chrom", "s", "e"], dtype={"chrom": str}))
    if not parts:
        return {}
    d = pd.concat(parts)
    return {ch: merge(g.s.to_numpy(np.int64), g.e.to_numpy(np.int64)) for ch, g in d.groupby("chrom") if ch in CHROMS}


def touches(s, e, iv):
    """True where [s, e) overlaps any interval of the sorted, disjoint set iv by >= 1 bp."""
    fs, fe = iv
    if len(fs) == 0:
        return np.zeros(len(s), bool)
    i = np.searchsorted(fe, s, side="right")
    ok = i < len(fs)
    out = np.zeros(len(s), bool)
    out[ok] = fs[i[ok]] < e[ok]
    return out


def main():
    fai = pd.read_csv(FAI, sep="\t", header=None, usecols=[0, 1], names=["chrom", "len"])
    clen = {c: int(l) for c, l in zip(fai.chrom, fai.len) if c in CHROMS}
    genome = sum(clen.values())
    loci = pd.read_csv(LOCI, sep="\t", usecols=["set", "chrom", "start0", "end0"])
    loci = loci[loci.set.isin(COHORTS) & loci.chrom.isin(CHROMS)].reset_index(drop=True)
    lists = {"GENES": GENES, "REPEATS": REPEATS}
    rows = []
    for name, (section, spec) in GROUPS.items():
        complement = isinstance(spec, str)
        iv = load_union(lists[spec.split(":")[1]] if complement else spec)
        hit = np.zeros(len(loci), bool)
        for ch, g in loci.groupby("chrom"):
            if ch in iv:
                hit[g.index] = touches(g.start0.to_numpy(), g.end0.to_numpy(), iv[ch])
        if complement:
            hit = ~hit
        loci[name] = hit.astype(np.int8)
        bp = sum(int((e - s).sum()) for s, e in iv.values())
        gfrac = 1 - bp / genome if complement else bp / genome
        for c in COHORTS:
            m = loci.set == c
            rows.append({"section": section, "group": name, "cohort": c, "n_loci": int(m.sum()),
                         "n_in_group": int(hit[m].sum()), "pct_loci": 100 * hit[m].mean(), "genome_pct_bp": 100 * gfrac})
    out = pd.DataFrame(rows)
    out.to_csv(OUT, sep="\t", index=False, float_format="%.6g")
    loci.to_csv(OUT_LOCI, sep="\t", index=False)

    print("n loci: " + ", ".join(f"{c} {int((loci.set == c).sum()):,}" for c in COHORTS) + f" -> {OUT.name}, {OUT_LOCI.name}")


if __name__ == "__main__":
    main()
```

#### 18.2 · Compartment rate ratios and representation → Fig. 5a, Suppl. Fig. 3d, 3e

```python
"""18.2 - Z-DNA/control variant-rate ratios within genomic compartments and variant representation (Fig. 5a, Suppl.
Fig. 3d, 3e).

Rate = variants with anchor in locus +/- FLANK bp / window bp (matched Z-DNA vs controls of the same compartment);
95% CI and two-sided p from a chromosome-block bootstrap, BH over compartment x class cells per variant set
(all; high_conf = haplotype call rate >= MIN_CALL_RATE). Representation: call rate, multiallelic and singleton fractions.
"""
import time
from pathlib import Path

import numpy as np
import pandas as pd
import pyarrow as pa
import pyarrow.compute as pc
import pyarrow.csv as pacsv
from statsmodels.stats.multitest import multipletests

BASE = Path("/path/to/pangenomes_zdna")
MEMBERSHIP = BASE / "18_feature_membership_loci.tsv.gz"
MUT_META = BASE / "9_hprc-v2.0-mc-chm13.wave.normalized.mutations.meta.tsv.gz"
FAI = BASE / "tmp/chm13/chm13_refseq/GCF_009914755.1_T2T-CHM13v2.0_genomic_ucsc.fna.fai"
OUT_RATES = BASE / "18_compartment_rates.tsv"
OUT_REP = BASE / "18_variant_representation.tsv"

FLANK = 10                     # bp added on each side of a locus (as Fig. 3a)
MIN_LOCI = 1000                # matched loci and controls required in a compartment
MIN_CELL_VARIANTS = 20         # variants (Z-DNA + control) needed to test a cell
MIN_CHROMS = 10                # chromosomes with both matched loci and controls in the compartment
MIN_CALL_RATE = 0.90
N_BOOTS = 10000
SEED = 42
COHORTS = {"Z": "Z_matched", "C": "control"}
CLASSES = ["SNV", "MNV", "small_del", "small_ins", "large_indel", "complex_del", "complex_ins"]
CHROMS = [f"chr{i}" for i in range(1, 23)] + ["chrX", "chrY"]
STRUCT = ["snp", "mnp", "substitution", "del", "ins", "complex"]
DOM = ["snp", "mnp", "substitution", "del", "ins", "del_like", "ins_like", "balanced"]
ARROW_THREADS, BLOCK_SIZE = 16, 1 << 26


def merge(s, e):
    if len(s) == 0:
        return np.empty(0, np.int64), np.empty(0, np.int64)
    o = np.argsort(s, kind="stable"); s, e = np.asarray(s)[o], np.asarray(e)[o]
    new = np.ones(len(s), bool); new[1:] = s[1:] > np.maximum.accumulate(e)[:-1]
    return s[new].astype(np.int64), np.maximum.reduceat(e, np.flatnonzero(new)).astype(np.int64)


def inside(p, iv):
    s, e = iv
    if len(s) == 0:
        return np.zeros(len(p), bool)
    i = np.searchsorted(s, p, side="right") - 1
    return (i >= 0) & (p < e[np.maximum(i, 0)])


def load_variants():
    pa.set_cpu_count(ARROW_THREADS)
    types = {"seqID": pa.string(), "normalized_start0": pa.int64(), "normalized_alt": pa.string(), "struct_class": pa.string(),
             "dominant_op": pa.string(), "ref_span": pa.int64(), "event_size": pa.int64(), "AC_computed": pa.int64(),
             "AN_computed": pa.int64(), "n_missing_full": pa.int64(), "n_missing_partial": pa.int64(), "subvariants": pa.int64()}
    reader = pacsv.open_csv(MUT_META, read_options=pacsv.ReadOptions(block_size=BLOCK_SIZE),
                            parse_options=pacsv.ParseOptions(delimiter="\t"),
                            convert_options=pacsv.ConvertOptions(include_columns=list(types), column_types=types,
                                                                 strings_can_be_null=False))
    keys = ["ch", "s0", "rs", "L", "sc", "dom", "h", "ac", "an", "mf", "mp", "sub"]
    parts, t0 = {k: [] for k in keys}, time.time()
    for b in reader:
        chi = pc.index_in(b.column("seqID"), value_set=pa.array(CHROMS))
        b, chi = b.filter(pc.is_valid(chi)), chi.filter(pc.is_valid(chi))
        if b.num_rows == 0:
            continue
        col = lambda n: b.column(n).to_numpy()
        sc = pc.index_in(b.column("struct_class"), value_set=pa.array(STRUCT)).to_numpy(zero_copy_only=False)
        dom = pc.index_in(b.column("dominant_op"), value_set=pa.array(DOM)).to_numpy(zero_copy_only=False)
        for k, v in zip(keys, (chi.to_numpy(zero_copy_only=False).astype(np.int8), col("normalized_start0"), col("ref_span"),
                               col("event_size"), sc.astype(np.int8), dom.astype(np.int8),
                               pd.util.hash_array(b.column("normalized_alt").to_numpy(zero_copy_only=False)),
                               col("AC_computed"), col("AN_computed"), col("n_missing_full"), col("n_missing_partial"),
                               col("subvariants"))):
            parts[k].append(v)
    v = {k: np.concatenate(x) for k, x in parts.items()}
    o = np.lexsort((v["h"], v["rs"], v["s0"], v["ch"]))
    v = {k: x[o] for k, x in v.items()}
    new = np.ones(len(o), bool)
    new[1:] = ((v["ch"][1:] != v["ch"][:-1]) | (v["s0"][1:] != v["s0"][:-1]) | (v["rs"][1:] != v["rs"][:-1])
               | (v["h"][1:] != v["h"][:-1]))
    v = {k: x[new] for k, x in v.items()}
    sc, dom, L = v["sc"], v["dom"], v["L"]
    is_ = lambda x: sc == STRUCT.index(x)
    cdel = is_("complex") & (dom == DOM.index("del_like"))
    cls = np.full(len(sc), -1, np.int8)
    for name, mask in (("SNV", is_("snp")), ("MNV", is_("mnp")), ("small_del", is_("del") & (L <= 50)),
                       ("small_ins", is_("ins") & (L <= 50)), ("large_indel", (is_("del") | is_("ins")) & (L > 50)),
                       ("complex_del", cdel), ("complex_ins", is_("complex") & (dom == DOM.index("ins_like")))):
        cls[mask] = CLASSES.index(name)
    called = v["an"] + 2 * v["mf"] + v["mp"]
    cr = np.where(called > 0, v["an"] / np.maximum(called, 1), np.nan)
    keep = cls >= 0
    out = pd.DataFrame({"ch": v["ch"], "pos": np.where(is_("mnp") | ((is_("del") | cdel) & (L <= 50)),
                                                     v["s0"] + (v["rs"] - 1) // 2, v["s0"]),
                        "cls": cls, "call_rate": cr, "multi": v["sub"] > 1, "singleton": v["ac"] == 1,
                        "hc": np.isfinite(cr) & (cr >= MIN_CALL_RATE)})[keep].reset_index(drop=True)
    print(f"[variants] {len(out):,} distinct normalized alleles; {int(out.hc.sum()):,} high-confidence ({time.time() - t0:.0f}s)")
    return out


def count_windows(loci, var, sel):
    """Variants (rows of var where sel) per locus window and class."""
    counts = np.zeros((len(loci), len(CLASSES)), np.int64)
    for ci, c in enumerate(CHROMS):
        li = np.flatnonzero(loci.chrom.to_numpy() == c)
        vm = sel & (var.ch.to_numpy() == ci)
        if len(li) == 0 or not vm.any():
            continue
        p, k = var.pos.to_numpy()[vm], var.cls.to_numpy()[vm]
        o = np.argsort(p); p, k = p[o], k[o]
        lo = np.searchsorted(p, loci.ws.to_numpy()[li], "left")
        n = np.searchsorted(p, loci.we.to_numpy()[li], "left") - lo
        np.add.at(counts, (np.repeat(li, n), k[np.arange(n.sum()) + np.repeat(lo - (np.cumsum(n) - n), n)]), 1)
    return counts


def main():
    t0 = time.time()
    rng = np.random.default_rng(SEED)
    fai = pd.read_csv(FAI, sep="\t", header=None, usecols=[0, 1], names=["chrom", "len"])
    clen = dict(zip(fai.chrom, fai.len.astype(np.int64)))
    loci = pd.read_csv(MEMBERSHIP, sep="\t")
    loci = loci[loci.set.isin(COHORTS.values())].reset_index(drop=True)
    groups = ["All loci"] + [c for c in loci.columns if c not in ("set", "chrom", "start0", "end0")]
    loci["All loci"] = 1
    loci["ws"] = np.maximum(loci.start0 - FLANK, 0)
    loci["we"] = np.minimum(loci.end0 + FLANK, loci.chrom.map(clen))
    loci["bp"] = loci.we - loci.ws
    var = load_variants()
    cnt = {"all": count_windows(loci, var, np.ones(len(var), bool)), "high_conf": count_windows(loci, var, var.hc.to_numpy())}

    ch = pd.Categorical(loci.chrom, categories=CHROMS).codes
    nch = len(CHROMS)
    W = np.zeros((N_BOOTS, nch))
    np.add.at(W, (np.repeat(np.arange(N_BOOTS), nch), rng.integers(0, nch, N_BOOTS * nch)), 1)
    rows = []
    for vs, counts in cnt.items():
        for g in groups:
            sel = {k: (loci.set == c).to_numpy() & (loci[g].to_numpy() == 1) for k, c in COHORTS.items()}
            n_loci = {k: int(s.sum()) for k, s in sel.items()}
            bp = {k: np.bincount(ch[s], weights=loci.bp.to_numpy()[s], minlength=nch) for k, s in sel.items()}
            for j, c in enumerate(CLASSES):
                n = {k: np.bincount(ch[s], weights=counts[s, j], minlength=nch) for k, s in sel.items()}
                r = {"variant_set": vs, "compartment": g, "class": c, "n_loci_Z": n_loci["Z"], "n_loci_C": n_loci["C"],
                     "n_Z": int(n["Z"].sum()), "n_C": int(n["C"].sum()), "bp_Z": int(bp["Z"].sum()), "bp_C": int(bp["C"].sum()),
                     "n_chrom": int(((bp["Z"] > 0) & (bp["C"] > 0)).sum())}
                r["tested"] = (min(n_loci.values()) >= MIN_LOCI and r["n_chrom"] >= MIN_CHROMS
                               and r["n_Z"] + r["n_C"] >= MIN_CELL_VARIANTS and r["n_C"] > 0)
                r["Z_per_kb"] = 1e3 * r["n_Z"] / r["bp_Z"] if r["bp_Z"] else np.nan
                r["C_per_kb"] = 1e3 * r["n_C"] / r["bp_C"] if r["bp_C"] else np.nan
                if r["tested"]:
                    r["rate_ratio"] = r["Z_per_kb"] / r["C_per_kb"]
                    with np.errstate(divide="ignore", invalid="ignore"):
                        b = (W @ n["Z"] / (W @ bp["Z"])) / (W @ n["C"] / (W @ bp["C"]))
                    lb = np.log(b[np.isfinite(b) & (b > 0)])
                    if len(lb) >= 0.5 * N_BOOTS:
                        r["ci_lo"], r["ci_hi"] = np.exp(np.percentile(lb, [2.5, 97.5]))
                        r["p"] = min(1.0, 2 * (min((lb <= 0).sum(), (lb >= 0).sum()) + 1) / (len(lb) + 1))
                    else:
                        r["ci_lo"] = r["ci_hi"] = r["p"] = np.nan
                rows.append(r)
    rates = pd.DataFrame(rows)
    rates["q"] = np.nan
    for vs, g in rates.groupby("variant_set"):
        m = g.p.notna()
        rates.loc[g.index[m], "q"] = multipletests(g.p[m], method="fdr_bh")[1]
    rates["sig"] = pd.cut(rates.q, [-1, 1e-3, 1e-2, 0.05, 2], labels=["***", "**", "*", "ns"]).astype(str)
    rates.to_csv(OUT_RATES, sep="\t", index=False, float_format="%.6g")

    rep = []
    for k, cname in COHORTS.items():
        d = loci[loci.set == cname]
        win = {c: merge(g.ws.to_numpy(), g.we.to_numpy()) for c, g in d.groupby("chrom")}
        flag = np.zeros(len(var), bool)
        for ci, c in enumerate(CHROMS):
            m = var.ch.to_numpy() == ci
            if c in win and m.any():
                flag[m] = inside(var.pos.to_numpy()[m], win[c])
        for j, c in enumerate(CLASSES):
            s = flag & (var.cls.to_numpy() == j)
            rep.append({"cohort": cname, "class": c, "n": int(s.sum()),
                        "mean_call_rate": float(np.nanmean(var.call_rate.to_numpy()[s])) if s.any() else np.nan,
                        "frac_multiallelic": float(var.multi.to_numpy()[s].mean()) if s.any() else np.nan,
                        "frac_singleton": float(var.singleton.to_numpy()[s].mean()) if s.any() else np.nan,
                        "frac_high_conf": float(var.hc.to_numpy()[s].mean()) if s.any() else np.nan})
    rep = pd.DataFrame(rep)
    rep.to_csv(OUT_REP, sep="\t", index=False, float_format="%.6g")

    a = rates[(rates.variant_set == "all") & (rates.compartment == "All loci")]
    print("All loci:\n" + a[["class", "rate_ratio", "ci_lo", "ci_hi", "q"]].to_string(index=False, float_format=lambda x: f"{x:.3g}"))
    print(f"-> {OUT_RATES.name}, {OUT_REP.name} [{time.time() - t0:.0f}s]")


if __name__ == "__main__":
    main()
```

#### 19 · SNV allele frequency → Fig. 5b

```python
"""19 - Z-DNA/control SNV rate ratios by allele frequency within compartments (Fig. 5b).

Distinct normalized SNVs with AN >= MIN_AN in singleton / <1% / 1-5% / >=5% bins; windows = matched loci and controls
+/- FLANK bp (merged). RR with 95% chromosome-block bootstrap CI; trend = slope of log2 RR across bins, two-sided
bootstrap p, BH over compartments.
"""
import time
from pathlib import Path

import numpy as np
import pandas as pd
import pyarrow as pa
import pyarrow.compute as pc
import pyarrow.csv as pacsv
from statsmodels.stats.multitest import multipletests

BASE = Path("/path/to/pangenomes_zdna")
MEMBERSHIP = BASE / "18_feature_membership_loci.tsv.gz"
MUT_META = BASE / "9_hprc-v2.0-mc-chm13.wave.normalized.mutations.meta.tsv.gz"
FAI = BASE / "tmp/chm13/chm13_refseq/GCF_009914755.1_T2T-CHM13v2.0_genomic_ucsc.fna.fai"
OUT_RATES, OUT_CHROM, OUT_TESTS = (BASE / f"19_snv_af_{k}.tsv" for k in ("rates", "bychrom", "tests"))

COMPARTMENTS = [
    "All loci",
    "Protein-coding genes",
    "CDS",
    "Exons",
    "Introns",
    "lncRNA genes",
    "Intergenic",
    "Simple repeats",
    "LINEs",
    "SINEs"
]
FLANK = 10
MIN_AN = 250                   # minimum called haplotypes (max 462)
AF_EDGES = (0.0, 0.01, 0.05, 1.0)
MIN_CELL_SNVS = 50             # compartment x bin cells with fewer SNVs in Z-DNA + controls are not estimated
N_BOOTS = 10000
SEED = 42
COHORTS = {"Z": "Z_matched", "C": "control"}
CHROMS = [f"chr{i}" for i in range(1, 23)] + ["chrX", "chrY"]
ARROW_THREADS, BLOCK_SIZE = 16, 1 << 26


def merge(s, e):
    if len(s) == 0:
        return np.empty(0, np.int64), np.empty(0, np.int64)
    o = np.argsort(s, kind="stable"); s, e = np.asarray(s)[o], np.asarray(e)[o]
    new = np.ones(len(s), bool); new[1:] = s[1:] > np.maximum.accumulate(e)[:-1]
    return s[new].astype(np.int64), np.maximum.reduceat(e, np.flatnonzero(new)).astype(np.int64)


def inside(p, iv):
    s, e = iv
    if len(s) == 0:
        return np.zeros(len(p), bool)
    i = np.searchsorted(s, p, side="right") - 1
    return (i >= 0) & (p < e[np.maximum(i, 0)])


def load_snvs():
    pa.set_cpu_count(ARROW_THREADS)
    types = {"seqID": pa.string(), "normalized_start0": pa.int64(), "normalized_ref": pa.string(),
             "normalized_alt": pa.string(), "struct_class": pa.string(), "AC_computed": pa.int64(), "AN_computed": pa.int64()}
    reader = pacsv.open_csv(MUT_META, read_options=pacsv.ReadOptions(block_size=BLOCK_SIZE),
                            parse_options=pacsv.ParseOptions(delimiter="\t"),
                            convert_options=pacsv.ConvertOptions(include_columns=list(types), column_types=types,
                                                                 strings_can_be_null=False))
    acgt = pa.array(list("ACGT"))
    parts, t0 = [], time.time()
    for b in reader:
        b = b.filter(pc.equal(b.column("struct_class"), "snp"))
        ch = pc.index_in(b.column("seqID"), value_set=pa.array(CHROMS))
        ref, alt = pc.index_in(b.column("normalized_ref"), value_set=acgt), pc.index_in(b.column("normalized_alt"), value_set=acgt)
        ok = pc.and_(pc.and_(pc.is_valid(ch), pc.is_valid(ref)), pc.is_valid(alt))
        b, ch, alt = b.filter(ok), ch.filter(ok), alt.filter(ok)
        if b.num_rows:
            parts.append(pd.DataFrame({"ch": ch.to_numpy(zero_copy_only=False).astype(np.int8),
                                       "pos": b.column("normalized_start0").to_numpy(),
                                       "alt": alt.to_numpy(zero_copy_only=False).astype(np.int8),
                                       "ac": b.column("AC_computed").to_numpy(), "an": b.column("AN_computed").to_numpy()}))
    v = pd.concat(parts, ignore_index=True).drop_duplicates(["ch", "pos", "alt"])
    v = v[(v.ac > 0) & (v.an >= MIN_AN)].reset_index(drop=True)
    v["af"] = v.ac / v.an
    k = len(AF_EDGES) - 1
    b = np.clip(np.searchsorted(AF_EDGES, v.af.to_numpy(), side="right") - 1, 0, k - 1)
    v["bin"] = np.where(v.ac.to_numpy() == 1, 0, b + 1)
    pct = lambda x: f"{100 * x:g}%"
    labels = ["Singleton"] + [f"<{pct(AF_EDGES[1])} (AC≥2)" if i == 0 else f"≥{pct(AF_EDGES[i])}" if i == k - 1
                              else f"{pct(AF_EDGES[i])}–{pct(AF_EDGES[i + 1])}" for i in range(k)]
    print(f"[snvs] {len(v):,} distinct SNV alleles with AC > 0 and AN >= {MIN_AN} ({time.time() - t0:.0f}s)")
    return v, labels


def boot_p(x):
    x = x[np.isfinite(x)]
    return min(1.0, 2 * (min((x <= 0).sum(), (x >= 0).sum()) + 1) / (len(x) + 1)) if len(x) else np.nan


@np.errstate(divide="ignore", invalid="ignore")
def main():
    t0 = time.time()
    rng = np.random.default_rng(SEED)
    fai = pd.read_csv(FAI, sep="\t", header=None, usecols=[0, 1], names=["chrom", "len"])
    clen = dict(zip(fai.chrom, fai.len.astype(np.int64)))
    loci = pd.read_csv(MEMBERSHIP, sep="\t")
    loci = loci[loci.set.isin(COHORTS.values())].reset_index(drop=True)
    loci["All loci"] = 1
    missing = [c for c in COMPARTMENTS if c not in loci.columns]
    if missing:
        raise SystemExit(f"Compartments not in {MEMBERSHIP.name}: {missing}")
    snv, labels = load_snvs()
    nb, nch = len(labels), len(CHROMS)
    ch_snv, pos = snv.ch.to_numpy(), snv.pos.to_numpy()
    W = np.zeros((N_BOOTS, nch))
    np.add.at(W, (np.repeat(np.arange(N_BOOTS), nch), rng.integers(0, nch, N_BOOTS * nch)), 1)
    rr = lambda nz, nc, bz, bc: (nz / bz) / (nc / bc)

    rates, bych, tests = [], [], []
    for comp in COMPARTMENTS:
        n = {k: np.zeros((nb, nch)) for k in COHORTS}
        bp = {k: np.zeros(nch) for k in COHORTS}
        for k, cname in COHORTS.items():
            d = loci[(loci.set == cname) & (loci[comp] == 1)]
            for ci, c in enumerate(CHROMS):
                g = d[d.chrom == c]
                if g.empty:
                    continue
                iv = merge(np.maximum(g.start0.to_numpy() - FLANK, 0), np.minimum(g.end0.to_numpy() + FLANK, clen[c]))
                bp[k][ci] = (iv[1] - iv[0]).sum()
                m = ch_snv == ci
                hit = inside(pos[m], iv)
                n[k][:, ci] = np.bincount(snv.bin.to_numpy()[m][hit], minlength=nb)
        logb = np.full((nb, N_BOOTS), np.nan)
        for g in range(nb):
            nz, nc = n["Z"][g].sum(), n["C"][g].sum()
            ok = nz + nc >= MIN_CELL_SNVS and nc > 0
            est = rr(nz, nc, bp["Z"].sum(), bp["C"].sum()) if ok else np.nan
            if ok:
                lb = np.log2(rr(W @ n["Z"][g], W @ n["C"][g], W @ bp["Z"], W @ bp["C"]))
                logb[g] = np.where(np.isfinite(lb), lb, np.nan)
            fin = logb[g][np.isfinite(logb[g])]
            ci = 2 ** np.percentile(fin, [2.5, 97.5]) if len(fin) else (np.nan, np.nan)
            rates.append({"compartment": comp, "bin": g, "bin_label": labels[g], "n_Z": int(nz), "n_C": int(nc),
                          "bp_Z": int(bp["Z"].sum()), "bp_C": int(bp["C"].sum()),
                          "Z_per_kb": 1e3 * nz / bp["Z"].sum() if bp["Z"].sum() else np.nan,
                          "C_per_kb": 1e3 * nc / bp["C"].sum() if bp["C"].sum() else np.nan,
                          "rate_ratio": est, "ci_lo": ci[0], "ci_hi": ci[1], "n_boots": N_BOOTS})
            for ci_, c in enumerate(CHROMS):
                if n["Z"][g, ci_] + n["C"][g, ci_] >= 20 and n["C"][g, ci_] > 0 and bp["Z"][ci_] > 0:
                    bych.append({"compartment": comp, "bin": g, "bin_label": labels[g], "chrom": c,
                                 "n_Z": int(n["Z"][g, ci_]), "n_C": int(n["C"][g, ci_]),
                                 "bp_Z": int(bp["Z"][ci_]), "bp_C": int(bp["C"][ci_]),
                                 "rate_ratio": rr(n["Z"][g, ci_], n["C"][g, ci_], bp["Z"][ci_], bp["C"][ci_])})
        okb = np.isfinite(logb).any(axis=1)
        if okb.sum() >= 3:
            x = np.arange(nb, dtype=float)[okb]
            xc = x - x.mean()
            obs = np.log2([r["rate_ratio"] for r in rates if r["compartment"] == comp])[okb]
            slopes = (xc[:, None] * logb[okb]).sum(0) / (xc ** 2).sum()
            tests.append({"compartment": comp, "test": "trend (slope of log2 RR across AF bins)",
                          "slope_log2": float((xc * obs).sum() / (xc ** 2).sum()),
                          "ci_lo": np.nanpercentile(slopes, 2.5), "ci_hi": np.nanpercentile(slopes, 97.5),
                          "p": boot_p(slopes), "n_bins": int(okb.sum())})
    tests = pd.DataFrame(tests)
    tests["q"] = multipletests(tests.p, method="fdr_bh")[1]
    tests["sig"] = pd.cut(tests.q, [-1, 1e-3, 1e-2, 0.05, 2], labels=["***", "**", "*", "ns"]).astype(str)
    tests = tests.assign(sided="two-sided", method="chromosome-block bootstrap", n_boots=N_BOOTS, min_an=MIN_AN,
                         correction="BH over compartments")
    pd.DataFrame(rates).to_csv(OUT_RATES, sep="\t", index=False, float_format="%.6g")
    pd.DataFrame(bych).to_csv(OUT_CHROM, sep="\t", index=False, float_format="%.6g")
    tests.to_csv(OUT_TESTS, sep="\t", index=False, float_format="%.6g")

    print(tests[["compartment", "slope_log2", "q"]].to_string(index=False, float_format=lambda x: f"{x:.3g}"))
    print(f"-> {OUT_RATES.name}, {OUT_CHROM.name}, {OUT_TESTS.name} [{time.time() - t0:.0f}s]")


if __name__ == "__main__":
    main()
```

#### 20.1 · ID83 classification (not rerun; keep `20_id83_indels.tsv.gz`)

```python
# """20.1 - ID83 classification of distinct normalized simple indels <= MAX_LEN bp (CHM13, HPRC v2) -> 20_id83_indels.tsv.gz
# (input to 20.2 and 20.3; Fig. 5c, Supplementary Table 4).
#
# COSMIC ID83 (Alexandrov et al. 2020) on left-aligned alleles: copies of the indel unit 3' of the event, else
# microhomology for deletions; validated against SigProfilerMatrixGenerator. Groups: 1-bp, repeat-unit, microhomology and
# no-repeat deletions/insertions. QC: deleted sequence matches CHM13 and indels are left-aligned.
# """
# import time
# from concurrent.futures import ProcessPoolExecutor
# from pathlib import Path
#
# import numpy as np
# import pandas as pd
# import pyarrow as pa
# import pyarrow.compute as pc
# import pyarrow.csv as pacsv
# import pysam
#
# BASE = Path("/path/to/pangenomes_zdna")
# MUT_META = BASE / "9_hprc-v2.0-mc-chm13.wave.normalized.mutations.meta.tsv.gz"
# FASTA = BASE / "tmp/chm13/chm13_refseq/GCF_009914755.1_T2T-CHM13v2.0_genomic_ucsc.fna"   # needs .fai
# OUT_INDELS = BASE / "20_id83_indels.tsv.gz"
#
# MAX_LEN = 50                   # simple indels up to this size (bp), as small_del / small_ins
# CAP = 5                        # ID83 caps (5 = 5+): copies, microhomology length and indel length
# NPROC = 24
# CHROMS = [f"chr{i}" for i in range(1, 23)] + ["chrX", "chrY"]
# ARROW_THREADS, BLOCK_SIZE = 16, 1 << 26
#
# LABELS = ([f"1:{t}:{b}:{k}" for t in ("Del", "Ins") for b in ("C", "T") for k in range(6)]
#           + [f"{L}:{t}:R:{k}" for t in ("Del", "Ins") for L in range(2, 6) for k in range(6)]
#           + [f"{L}:Del:M:{k}" for L in range(2, 6) for k in range(1, (L if L < 5 else 6))])
# assert len(LABELS) == 83 and len(set(LABELS)) == 83
# LIDX = {l: i for i, l in enumerate(LABELS)}
# GROUPS = ["del_1bp_homopolymer", "ins_1bp_homopolymer", "del_repeat_unit", "ins_repeat_unit", "del_microhomology",
#           "del_no_repeat", "ins_no_repeat"]
#
#
# def group_of(label):
#     L, t, x, k = label.split(":")
#     if L == "1":
#         return f"{t.lower()}_1bp_homopolymer"
#     if x == "M":
#         return "del_microhomology"
#     return f"{t.lower()}_repeat_unit" if int(k) > 0 else f"{t.lower()}_no_repeat"
#
#
# LABEL_GROUP = np.array([GROUPS.index(group_of(l)) for l in LABELS])
#
#
# def copies_3p(seq, pos, unit):
#     """Consecutive copies of unit starting at seq[pos], counted up to CAP."""
#     L, n = len(unit), 0
#     while n < CAP and seq[pos + n * L:pos + (n + 1) * L] == unit:
#         n += 1
#     return n
#
#
# def microhomology(seq, s0, d):
#     """Longest partial match (< len(d)) of d with the 3' flank (prefix) or the 5' flank (suffix), capped."""
#     L, best = len(d), 0
#     for k in range(1, min(L - 1, CAP) + 1):
#         if d[:k] == seq[s0 + L:s0 + L + k] or (s0 - k >= 0 and d[-k:] == seq[s0 - k:s0]):
#             best = k
#     return best
#
#
# def id83(seq, s0, kind, d):
#     """kind 'del': d = seq[s0:s0+L] is deleted; kind 'ins': d is inserted before seq[s0]."""
#     L = len(d)
#     if kind == "del":
#         n = copies_3p(seq, s0 + L, d)
#         if L == 1:
#             return f"1:Del:{'C' if d in 'CG' else 'T'}:{n}"
#         if n > 0:
#             return f"{min(L, CAP)}:Del:R:{n}"
#         mh = microhomology(seq, s0, d)
#         return f"{min(L, CAP)}:Del:M:{mh}" if mh > 0 else f"{min(L, CAP)}:Del:R:0"
#     n = copies_3p(seq, s0, d)
#     if L == 1:
#         return f"1:Ins:{'C' if d in 'CG' else 'T'}:{n}"
#     return f"{min(L, CAP)}:Ins:R:{n}"
#
#
# def run_self_tests():
#     cases = [("GGTCCCAGG", 3, "del", "C", "1:Del:C:2"),
#              ("AGGGT", 1, "del", "G", "1:Del:C:2"),
#              ("ATGCA", 1, "del", "T", "1:Del:T:0"),
#              ("ATTTG", 1, "ins", "T", "1:Ins:T:3"),
#              ("AGCT", 1, "ins", "A", "1:Ins:T:0"),
#              ("GCACACAT", 1, "del", "CA", "2:Del:R:2"),
#              ("TACGTTACGAA", 1, "del", "ACGTT", "5:Del:M:3"),
#              ("TAGCTG", 2, "del", "GC", "2:Del:R:0"),
#              ("ACACACG", 0, "ins", "AC", "2:Ins:R:3"),
#              ("GAAAAAAAAT", 1, "del", "A", "1:Del:T:5"),
#              ("G" + "CAGTC" * 4 + "A", 1, "del", "CAGTC", "5:Del:R:3"),
#              ("GA" + "CAGTCAGT" + "TT", 2, "del", "CAGTCAGT", "5:Del:R:0")]
#     for seq, s0, kind, d, want in cases:
#         got = id83(seq, s0, kind, d)
#         if got != want:
#             raise RuntimeError(f"ID83 self-test failed: {kind} {d} at {s0} in {seq}: {got} != {want}")
#     print(f"ID83 self-tests: PASS ({len(cases)} cases)")
#
#
# def classify_chrom(args):
#     chrom, s0, kind, seqs = args
#     with pysam.FastaFile(str(FASTA)) as fa:
#         ref = fa.fetch(chrom).upper()
#     lab = np.empty(len(s0), np.int16)
#     mismatch = not_left = 0
#     for i in range(len(s0)):
#         p, d = int(s0[i]), seqs[i]
#         if kind[i] == 0 and ref[p:p + len(d)] != d:
#             mismatch += 1
#         if p > 0 and ref[p - 1] == d[-1]:
#             not_left += 1
#         lab[i] = LIDX[id83(ref, p, "del" if kind[i] == 0 else "ins", d)]
#     return chrom, lab, (mismatch, not_left)
#
#
# def load_indels():
#     pa.set_cpu_count(ARROW_THREADS)
#     types = {"seqID": pa.string(), "normalized_start0": pa.int64(), "normalized_ref": pa.string(),
#              "normalized_alt": pa.string(), "struct_class": pa.string(), "event_size": pa.int64()}
#     reader = pacsv.open_csv(MUT_META, read_options=pacsv.ReadOptions(block_size=BLOCK_SIZE),
#                             parse_options=pacsv.ParseOptions(delimiter="\t"),
#                             convert_options=pacsv.ConvertOptions(include_columns=list(types), column_types=types,
#                                                                  strings_can_be_null=False))
#     parts, t0 = [], time.time()
#     for b in reader:
#         b = b.filter(pc.and_(pc.is_in(b.column("struct_class"), value_set=pa.array(["del", "ins"])),
#                              pc.less_equal(b.column("event_size"), MAX_LEN)))
#         b = b.filter(pc.is_in(b.column("seqID"), value_set=pa.array(CHROMS)))
#         if b.num_rows:
#             parts.append(b.to_pandas())
#     v = pd.concat(parts, ignore_index=True)
#     v["kind"] = (v.struct_class == "ins").astype(np.int8)
#     v["seq"] = np.where(v.kind == 0, v.normalized_ref, v.normalized_alt).astype(str)
#     v["seq"] = v.seq.str.upper()
#     n0 = len(v)
#     v = v.drop_duplicates(["seqID", "normalized_start0", "kind", "seq"]).reset_index(drop=True)
#     v = v[v.seq.str.fullmatch("[ACGT]+")].reset_index(drop=True)
#     v["length"] = v.seq.str.len()
#     print(f"[indels] {n0:,} simple indel alleles <= {MAX_LEN} bp -> {len(v):,} distinct ACGT indels ({time.time() - t0:.0f}s)")
#     return v
#
#
# @np.errstate(divide="ignore", invalid="ignore")
# def main():
#     t0 = time.time()
#     run_self_tests()
#     v = load_indels()
#
#     jobs = [(c, g.normalized_start0.to_numpy(np.int64), g.kind.to_numpy(np.int8), g.seq.tolist())
#             for c, g in v.groupby("seqID", sort=False)]
#     idx = {c: g.index.to_numpy() for c, g in v.groupby("seqID", sort=False)}
#     v["id83"] = -1
#     mism = not_left = 0
#     with ProcessPoolExecutor(min(NPROC, len(jobs))) as ex:
#         for c, lab, (mm, nl) in ex.map(classify_chrom, jobs):
#             v.loc[idx[c], "id83"] = lab
#             mism += mm
#             not_left += nl
#     print(f"[QC] deletions not matching CHM13 at their position: {mism:,}; indels not left-aligned: {not_left:,} (both must be 0)")
#     if mism or not_left:
#         raise RuntimeError("indel coordinates / representation failed QC")
#     v["label"] = np.array(LABELS, dtype=object)[v.id83.to_numpy()]
#     v["group"] = np.array(GROUPS, dtype=object)[LABEL_GROUP[v.id83.to_numpy()]]
#     v[["seqID", "normalized_start0", "kind", "length", "seq", "label", "group"]].rename(
#         columns={"seqID": "chrom", "normalized_start0": "start0"}).assign(
#         kind=lambda d: np.where(d.kind == 0, "del", "ins")).to_csv(OUT_INDELS, sep="\t", index=False)
#     print(f"-> {OUT_INDELS.name} [{time.time() - t0:.0f}s]")
#
#
# if __name__ == "__main__":
#     main()
```

#### 20.2 · ID83 group rate ratios → Supplementary Table 4

```python
"""20.2 - Z-DNA/control rate ratios of ID83 indel groups in all loci, simple repeats and outside repeats
(Fig. 5c text, Supplementary Table 4).

Reads the classified indels (20.1) and locus membership (18.1). Windows = loci +/- FLANK bp merged per cohort;
deletion anchor = midpoint, insertion anchor = insertion site. 95% CI and two-sided p from a chromosome-block bootstrap
(N_BOOTS); BH over the 7 groups x 3 compartments. The '*_1bp_homopolymer' keys hold all 1-bp indels.
"""
from pathlib import Path
import time
import numpy as np
import pandas as pd
from statsmodels.stats.multitest import multipletests

BASE = Path('/path/to/pangenomes_zdna')
INDELS = BASE / '20_id83_indels.tsv.gz'
MEMBERSHIP = BASE / '18_feature_membership_loci.tsv.gz'
FAI = BASE / 'tmp/chm13/chm13_refseq/GCF_009914755.1_T2T-CHM13v2.0_genomic_ucsc.fna.fai'
OUT_RATES = BASE / '20_id83_rates.tsv'
TABLE_S4 = BASE / 'Supplementary_Table_4.tsv'
OUT_CHROM = BASE / '20_id83_rates_bychrom.tsv'
OUT_SPECTRA = BASE / '20_id83_spectra.tsv'
COMPARTMENTS = ['All loci', 'Simple repeats', 'Outside repeats']
COHORTS = ['Z_matched', 'control']
GROUPS = ['del_1bp_homopolymer', 'ins_1bp_homopolymer',
          'del_repeat_unit', 'ins_repeat_unit', 'del_microhomology',
          'del_no_repeat', 'ins_no_repeat']
CHROMS = [f'chr{i}' for i in range(1, 23)] + ['chrX', 'chrY']
FLANK = 10
N_BOOTS = 10_000
MIN_CELL = 20
SEED = 42
GROUP_NAMES = {'del_1bp_homopolymer': '1-bp deletions', 'ins_1bp_homopolymer': '1-bp insertions',
               'del_repeat_unit': 'Repeat-unit deletions', 'ins_repeat_unit': 'Repeat-unit insertions',
               'del_microhomology': 'Microhomology deletions', 'del_no_repeat': 'No-repeat deletions',
               'ins_no_repeat': 'No-repeat insertions', 'all_indels': 'All indels <= 50 bp'}
TABLE_COLUMNS = {'compartment': 'Compartment', 'group': 'Indel group', 'n_Z': 'Indels in Z-DNA windows (n)',
                 'n_C': 'Indels in control windows (n)', 'bp_Z': 'Z-DNA windows (bp)', 'bp_C': 'Control windows (bp)',
                 'Z_per_kb': 'Indels per kb, Z-DNA', 'C_per_kb': 'Indels per kb, controls',
                 'rate_ratio': 'Rate ratio (Z-DNA/control)', 'ci_lo': '95% CI lower', 'ci_hi': '95% CI upper',
                 'p': 'P value', 'q': 'q value (BH)', 'sig': 'Significance', 'n_valid_boot': 'Valid bootstrap replicates (n)'}


def merge(s, e):
    s, e = np.asarray(s, dtype=np.int64), np.asarray(e, dtype=np.int64)
    if not len(s):
        return np.empty(0, np.int64), np.empty(0, np.int64)
    o = np.argsort(s, kind='stable')
    s, e = s[o], e[o]
    starts = np.r_[True, s[1:] > np.maximum.accumulate(e)[:-1]]
    ix = np.flatnonzero(starts)
    return s[ix], np.maximum.reduceat(e, ix)


def inside(p, iv):
    s, e = iv
    if not len(s):
        return np.zeros(len(p), dtype=bool)
    ix = np.searchsorted(s, p, side='right') - 1
    return (ix >= 0) & (p < e[np.maximum(ix, 0)])


def group_from_label(lab):
    length, typ, mode, repeats = lab.split(':')
    if length == '1':
        return f'{typ.lower()}_1bp_homopolymer'
    if mode == 'M':
        return 'del_microhomology'
    return f'{typ.lower()}_repeat_unit' if int(repeats) > 0 else f'{typ.lower()}_no_repeat'


def main():
    start = time.time()
    for p in [INDELS, MEMBERSHIP, FAI]:
        if not p.exists():
            raise FileNotFoundError(p)
    fai = pd.read_csv(FAI, sep='\t', header=None, usecols=[0, 1], names=['chrom', 'bp'])
    clen = {c: int(bp) for c, bp in zip(fai.chrom, fai.bp) if c in CHROMS}
    m = pd.read_csv(MEMBERSHIP, sep='\t', usecols=['set', 'chrom', 'start0', 'end0',
                                                   'Simple repeats', 'Outside repeats'])
    m = m[m['set'].isin(COHORTS) & m.chrom.isin(CHROMS)].copy()
    if (m['set'].value_counts().reindex(COHORTS).fillna(0) != 227540).any():
        raise AssertionError('Expected 227,540 loci in each matched cohort; inspect membership file')

    nc, ng, nch = len(COMPARTMENTS), len(GROUPS), len(CHROMS)
    windows = {}
    bp = np.zeros((nc, 2, nch), dtype=np.int64)
    for k, comp in enumerate(COMPARTMENTS):
        for h, cohort in enumerate(COHORTS):
            part = m[m['set'].eq(cohort)]
            if comp != 'All loci':
                part = part[part[comp].eq(1)]
            for ci, c in enumerate(CHROMS):
                g = part[part.chrom.eq(c)]
                if g.empty:
                    windows[k, h, ci] = merge([], [])
                    continue
                s = np.maximum(g.start0.to_numpy(np.int64) - FLANK, 0)
                e = np.minimum(g.end0.to_numpy(np.int64) + FLANK, clen[c])
                iv = merge(s, e)
                windows[k, h, ci] = iv
                bp[k, h, ci] = int(np.sum(iv[1] - iv[0]))

    v = pd.read_csv(INDELS, sep='\t', usecols=['chrom', 'start0', 'kind', 'length', 'label', 'group'],
                    dtype={'chrom': 'category', 'start0': 'int64', 'kind': 'category',
                           'length': 'int16', 'label': 'category', 'group': 'category'})
    print(f'[indels] {len(v):,} preclassified, distinct indels (no reclassification)')
    ch = pd.Categorical(v.chrom, categories=CHROMS).codes
    group = pd.Categorical(v.group, categories=GROUPS).codes
    labels = pd.Categorical(v.label)
    lab = labels.codes
    label_names = list(labels.categories)
    if (ch < 0).any() or (group < 0).any() or (lab < 0).any():
        raise ValueError('Unexpected chromosome, ID83 group or null classification')
    check_group_codes = np.array([GROUPS.index(group_from_label(s)) for s in label_names])
    if not np.array_equal(group, check_group_codes[lab]):
        raise AssertionError('Existing group labels do not agree with ID83 categories')
    pos = v.start0.to_numpy(np.int64).copy()
    dele = v.kind.eq('del').to_numpy()
    if not np.all(dele | v.kind.eq('ins').to_numpy()):
        raise ValueError('Expected only simple deletion/insertion kinds')
    pos[dele] += (v.length.to_numpy(np.int64)[dele] - 1) // 2
    del v

    cnt = np.zeros((nc, 2, nch, ng), dtype=np.int64)
    spec_count = np.zeros((2, len(label_names)), dtype=np.int64)
    for ci, c in enumerate(CHROMS):
        ix = np.flatnonzero(ch == ci)
        p, g, l = pos[ix], group[ix], lab[ix]
        for k in range(nc):
            for h in range(2):
                mask = inside(p, windows[k, h, ci])
                cnt[k, h, ci] = np.bincount(g[mask], minlength=ng)
                if k == 0:
                    spec_count[h] += np.bincount(l[mask], minlength=len(label_names))

    spec = pd.DataFrame({'id83': label_names,
                         'group': [group_from_label(x) for x in label_names],
                         'n_Z': spec_count[0], 'n_C': spec_count[1],
                         'n_genome': np.bincount(lab, minlength=len(label_names))})
    for co in ['Z', 'C', 'genome']:
        spec[f'frac_{co}'] = spec[f'n_{co}'] / spec[f'n_{co}'].sum()
    spec.to_csv(OUT_SPECTRA, sep='\t', index=False, float_format='%.9g')
    print(f'[QC] All-loci indels: Z={spec.n_Z.sum():,}, controls={spec.n_C.sum():,}, genome={spec.n_genome.sum():,}')

    rng = np.random.default_rng(SEED)
    W = np.zeros((N_BOOTS, nch), dtype=np.int16)
    rows = np.repeat(np.arange(N_BOOTS), nch)
    np.add.at(W, (rows, rng.integers(0, nch, N_BOOTS * nch)), 1)
    W = W.astype(float)
    summaries, chrom_rows = [], []
    for k, comp in enumerate(COMPARTMENTS):
        bz, bc = bp[k, 0].astype(float), bp[k, 1].astype(float)
        for j, grp in enumerate(GROUPS + ['all_indels']):
            nz = cnt[k, 0, :, j].astype(float) if j < ng else cnt[k, 0].sum(axis=1).astype(float)
            nc_ = cnt[k, 1, :, j].astype(float) if j < ng else cnt[k, 1].sum(axis=1).astype(float)
            nzt, nct = int(nz.sum()), int(nc_.sum())
            tested = (nzt + nct >= MIN_CELL) and nzt > 0 and nct > 0 and bz.sum() > 0 and bc.sum() > 0
            est = ((nzt / bz.sum()) / (nct / bc.sum())) if nct > 0 and bc.sum() > 0 and bz.sum() > 0 else np.nan
            r = {'compartment': comp, 'group': grp, 'n_Z': nzt, 'n_C': nct,
                 'bp_Z': int(bz.sum()), 'bp_C': int(bc.sum()),
                 'Z_per_kb': 1000*nzt/bz.sum() if bz.sum() > 0 else np.nan,
                 'C_per_kb': 1000*nct/bc.sum() if bc.sum() > 0 else np.nan,
                 'rate_ratio': est, 'tested': tested, 'n_boots': N_BOOTS,
                 'ci_lo': np.nan, 'ci_hi': np.nan, 'p': np.nan, 'n_valid_boot': 0}
            if tested:
                with np.errstate(divide='ignore', invalid='ignore'):
                    boot = (W @ nz / (W @ bz)) / (W @ nc_ / (W @ bc))
                    lb = np.log(boot)
                lb = lb[np.isfinite(lb)]
                r['n_valid_boot'] = len(lb)
                if len(lb) >= N_BOOTS // 2:
                    r['ci_lo'], r['ci_hi'] = np.exp(np.quantile(lb, [0.025, 0.975]))
                    r['p'] = min(1., 2 * (min((lb <= 0).sum(), (lb >= 0).sum()) + 1) / (len(lb) + 1))
                else:
                    r['tested'] = False
            summaries.append(r)
            if j < ng:
                for ci, c in enumerate(CHROMS):
                    xz, xc = int(nz[ci]), int(nc_[ci])
                    if bz[ci] > 0 and bc[ci] > 0:
                        ratio = ((xz + .5)/bz[ci])/((xc + .5)/bc[ci])
                        chrom_rows.append({'compartment': comp, 'group': grp, 'chrom': c,
                                           'n_Z': xz, 'n_C': xc, 'bp_Z': int(bz[ci]), 'bp_C': int(bc[ci]),
                                           'n_total': xz + xc, 'ratio_display': ratio,
                                           'rate_ratio': ((xz/bz[ci])/(xc/bc[ci])) if xc > 0 else np.nan})
    rates = pd.DataFrame(summaries)
    rates['q'] = np.nan
    valid = rates['group'].ne('all_indels') & rates.p.notna()
    rates.loc[valid, 'q'] = multipletests(rates.loc[valid, 'p'], method='fdr_bh')[1]
    rates['sig'] = rates.q.map(lambda q: '***' if q < .001 else '**' if q < .01 else '*' if q < .05 else 'ns' if np.isfinite(q) else '')
    rates.to_csv(OUT_RATES, sep='\t', index=False, float_format='%.9g')
    pd.DataFrame(chrom_rows).to_csv(OUT_CHROM, sep='\t', index=False, float_format='%.9g')
    rates.assign(group=rates.group.map(GROUP_NAMES))[list(TABLE_COLUMNS)].rename(columns=TABLE_COLUMNS).to_csv(
        TABLE_S4, sep='\t', index=False, float_format='%.4g')
    a = rates[rates.compartment.eq('All loci')]
    print(a[['group', 'rate_ratio', 'ci_lo', 'ci_hi', 'q']].to_string(index=False, float_format=lambda x: f'{x:.3g}'))
    print(f'-> {OUT_RATES.name}, {OUT_CHROM.name}, {OUT_SPECTRA.name}, {TABLE_S4.name} [{time.time()-start:.0f}s]')


if __name__ == '__main__':
    main()
```

#### 20.3 · ID83 positional profiles → Fig. 5c

```python
"""20.3 - Positional Z-DNA/control enrichment (+/-R bp) around anchors of ID83 indel groups (Fig. 5c).

RR(k) = [anchors with anchor+k in merged Z-DNA / Z-DNA bp] / [same for controls] on unexpanded matched loci;
pointwise 95% CI from a chromosome-block bootstrap (N_BOOTS). Groups are counted in parallel.
"""
from pathlib import Path
import os
import time
from multiprocessing import get_context
import numpy as np
import pandas as pd

BASE = Path('/path/to/pangenomes_zdna')
INDELS = BASE / '20_id83_indels.tsv.gz'
MASTER = BASE / '8_CHM13_controls_MASTER.tsv'
FAI = BASE / 'tmp/chm13/chm13_refseq/GCF_009914755.1_T2T-CHM13v2.0_genomic_ucsc.fna.fai'
OUT = BASE / '20_id83_zc_profiles.tsv.gz'
OUT_CHR = BASE / '20_id83_zc_profiles_bychrom.tsv.gz'
TOL, DINUC_MODE = '10%', 'CpG_GpC'
CHROMS = [f'chr{i}' for i in range(1, 23)] + ['chrX', 'chrY']
GROUPS = ['del_repeat_unit', 'del_microhomology', 'del_no_repeat', 'del_1bp_homopolymer',
          'ins_repeat_unit', 'ins_no_repeat', 'ins_1bp_homopolymer']
R = 500
OFFSETS = np.arange(-R, R + 1)
N_BOOTS = 2_000
SEED = 42
INTERVAL_CHUNK = 10_000
N_WORKERS = min(len(GROUPS), os.cpu_count() or 1)
SHARED = {}


def merge(s, e):
    s, e = np.asarray(s, np.int64), np.asarray(e, np.int64)
    if not len(s):
        return np.empty(0, np.int64), np.empty(0, np.int64)
    o = np.argsort(s, kind='stable')
    s, e = s[o], e[o]
    ix = np.flatnonzero(np.r_[True, s[1:] > np.maximum.accumulate(e)[:-1]])
    return s[ix], np.maximum.reduceat(e, ix)


def sweep_counts(anchors, iv_s, iv_e):
    """At each k, count anchors a with a+k in union of disjoint [s,e) loci."""
    width = 2 * R + 1
    diff = np.zeros(width + 1, dtype=np.int64)
    if not len(anchors) or not len(iv_s):
        return diff[:width]
    anchors = np.sort(np.asarray(anchors, dtype=np.int64))
    for i0 in range(0, len(iv_s), INTERVAL_CHUNK):
        s = iv_s[i0:i0 + INTERVAL_CHUNK]
        e = iv_e[i0:i0 + INTERVAL_CHUNK]
        lo = np.searchsorted(anchors, s - R, side='left')
        hi = np.searchsorted(anchors, e - 1 + R, side='right')
        n = hi - lo
        tot = int(n.sum())
        if not tot:
            continue
        interval_index = np.repeat(np.arange(len(s)), n)
        relative_start = np.repeat(lo - (np.cumsum(n) - n), n)
        a = anchors[np.arange(tot) + relative_start]
        left = np.maximum(s[interval_index] - a, -R) + R
        right_exclusive = np.minimum(e[interval_index] - 1 - a, R) + R + 1
        if (left < 0).any() or (right_exclusive > width).any() or (right_exclusive <= left).any():
            raise AssertionError('Invalid sweep-count offset bounds')
        diff += np.bincount(left, minlength=width + 1)
        diff -= np.bincount(right_exclusive, minlength=width + 1)
    ans = np.cumsum(diff)[:width]
    if (ans < 0).any():
        raise AssertionError('Negative sweep count')
    return ans


def count_group(gi):
    a, gcode, ch, iv = SHARED['a'], SHARED['gcode'], SHARED['ch'], SHARED['iv']
    out = np.zeros((2, len(CHROMS), len(OFFSETS)), np.int64)
    n = np.zeros(len(CHROMS), np.int64)
    for ci in range(len(CHROMS)):
        aa = np.sort(a[(gcode == gi) & (ch == ci)])
        n[ci] = len(aa)
        for h in range(2):
            out[h, ci] = sweep_counts(aa, *iv[h, ci])
    return gi, out, n


def main():
    t0 = time.time()
    for p in [INDELS, MASTER, FAI]:
        if not p.exists():
            raise FileNotFoundError(p)
    fai = pd.read_csv(FAI, sep='\t', header=None, usecols=[0, 1], names=['chrom', 'bp'])
    clen = {c: int(bp) for c, bp in zip(fai.chrom, fai.bp) if c in CHROMS}
    genome_bp = sum(clen.values())
    m = pd.read_csv(MASTER, sep='\t', usecols=['tol_level', 'dinuc_mode', 'Chromosome', 'Start0', 'End0',
                                               'Ctrl_Start', 'Ctrl_End', 'has_control'],
                    dtype={'tol_level':str,'dinuc_mode':str,'Chromosome':str,'has_control':str})
    m = m[(m.tol_level == TOL) & (m.dinuc_mode == DINUC_MODE) &
          m.has_control.str.strip().str.lower().eq('true') & m.Chromosome.isin(CHROMS)]
    m = m.drop_duplicates(['Chromosome', 'Start0', 'End0'])
    if len(m) != 227_540:
        raise AssertionError(f'Expected 227,540 matched pairs, received {len(m):,}')
    if not ((m.End0 - m.Start0).to_numpy() == (m.Ctrl_End - m.Ctrl_Start).to_numpy()).all():
        raise AssertionError('Matched Z-DNA/control pair lengths differ')
    iv = {}
    bp = np.zeros((2, len(CHROMS)), np.int64)
    for h, (start, end) in enumerate([('Start0','End0'), ('Ctrl_Start','Ctrl_End')]):
        for ci,c in enumerate(CHROMS):
            g = m[m.Chromosome.eq(c)]
            if g.empty:
                iv[h,ci] = merge([],[])
                continue
            s = np.maximum(g[start].to_numpy(np.int64), 0)
            e = np.minimum(g[end].to_numpy(np.int64), clen[c])
            iv[h,ci] = merge(s,e)
            bp[h,ci] = int(np.sum(iv[h,ci][1] - iv[h,ci][0]))
    print(f'[matched loci] {len(m):,} pairs | Z={bp[0].sum()/genome_bp:.6f} genome fraction | '
          f'C={bp[1].sum()/genome_bp:.6f} genome fraction')

    d = pd.read_csv(INDELS, sep='\t', usecols=['chrom','start0','kind','length','group'],
                    dtype={'chrom':'category','start0':'int64','kind':'category','length':'int16','group':'category'})
    ch = pd.Categorical(d.chrom, categories=CHROMS).codes
    gcode = pd.Categorical(d.group, categories=GROUPS).codes
    if (ch < 0).any() or (gcode < 0).any():
        raise AssertionError('Unknown chromosome or ID83 group')
    a = d.start0.to_numpy(np.int64).copy()
    dele = d.kind.eq('del').to_numpy()
    a[dele] += (d.length.to_numpy(np.int64)[dele] - 1)//2
    del d

    count = np.zeros((2, len(GROUPS), len(CHROMS), len(OFFSETS)), np.int64)
    n_anchors = np.zeros((len(GROUPS), len(CHROMS)), np.int64)
    SHARED.update(a=a, gcode=gcode, ch=ch, iv=iv)
    with get_context('fork').Pool(N_WORKERS) as pool:
        for gi, cnt, n in pool.imap_unordered(count_group, range(len(GROUPS))):
            count[:, gi], n_anchors[gi] = cnt, n
    rows_bychr = [pd.DataFrame({'group': group, 'chrom': chrom, 'offset': OFFSETS, 'n_anchors': int(n_anchors[gi, ci]),
                                'count_Z': count[0, gi, ci], 'count_C': count[1, gi, ci],
                                'bp_Z': int(bp[0, ci]), 'bp_C': int(bp[1, ci])})
                  for gi, group in enumerate(GROUPS) for ci, chrom in enumerate(CHROMS)]
    pd.concat(rows_bychr, ignore_index=True).to_csv(OUT_CHR, sep='\t', index=False, compression='gzip')

    rng = np.random.default_rng(SEED)
    nch = len(CHROMS)
    W = np.zeros((N_BOOTS,nch),np.int16)
    np.add.at(W,(np.repeat(np.arange(N_BOOTS),nch), rng.integers(0,nch,size=N_BOOTS*nch)), 1)
    W = W.astype(float)
    boot_bpZ, boot_bpC = W @ bp[0], W @ bp[1]
    out = []
    for gi,g in enumerate(GROUPS):
        nz, nc = count[0,gi].sum(axis=0), count[1,gi].sum(axis=0)
        with np.errstate(divide='ignore',invalid='ignore'):
            est = (nz / bp[0].sum()) / (nc / bp[1].sum())
            bZ, bC = W @ count[0,gi], W @ count[1,gi]
            boot = (bZ / boot_bpZ[:,None]) / (bC / boot_bpC[:,None])
        boot[~np.isfinite(boot)] = np.nan
        nvalid = np.isfinite(boot).sum(axis=0)
        if (nvalid < N_BOOTS // 2).any():
            raise RuntimeError(f'{g}: fewer than 50% valid bootstrap RRs at some positions')
        out.append(pd.DataFrame({'group':g,'offset':OFFSETS,'n_anchors':int(n_anchors[gi].sum()),
                                  'count_Z':nz, 'count_C':nc,
                                  'bp_Z':int(bp[0].sum()), 'bp_C':int(bp[1].sum()),
                                  'rate_ratio':est,
                                  'ci_lo':np.nanpercentile(boot,2.5,axis=0),
                                  'ci_hi':np.nanpercentile(boot,97.5,axis=0),
                                  'n_valid_boot':nvalid}))
    result = pd.concat(out,ignore_index=True)
    result.to_csv(OUT,sep='\t',index=False,compression='gzip',float_format='%.8g')
    r0 = result[result.offset.eq(0)]
    print(r0[['group', 'n_anchors', 'rate_ratio']].to_string(index=False, float_format=lambda x: f'{x:.3g}'))
    print(f'-> {OUT.name}, {OUT_CHR.name} [{time.time()-t0:.0f}s]')


if __name__=='__main__':
    main()
```

#### Fig. 5a, Suppl. Fig. 3d, 3e

```python
"""Fig. 5a, Suppl. Fig. 3e and Suppl. Fig. 3d - Z-DNA/control variant-rate ratio heatmaps across compartments (all
variants -> Fig. 5a; call rate >= 0.90 -> Suppl. Fig. 3e; log2 colour, BH stars, colour bar, significance legend) and
variant representation (call rate, multiallelic, singleton fractions) + legend. Reads 18.1 and 18.2.
"""
from pathlib import Path
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from matplotlib.cm import ScalarMappable
from matplotlib.colors import LinearSegmentedColormap, TwoSlopeNorm
from matplotlib.lines import Line2D

BASE = Path("/path/to/pangenomes_zdna")
RATES, REP = BASE / "18_compartment_rates.tsv", BASE / "18_variant_representation.tsv"
MEMBERSHIP_TABLE = BASE / "18_feature_membership.tsv"
PLOTS, TESTS = BASE / "plots", BASE / "tests"
HEATMAPS = {"all": "fig5a_compartment_rate_ratios", "high_conf": "suppl_fig3e_compartment_rate_ratios_high_call_rate"}
REP_STEM = "suppl_fig3d_variant_representation"
CLASSES = ["SNV", "MNV", "small_del", "small_ins", "large_indel", "complex_del", "complex_ins"]
CLASS_LABELS = {"SNV": "SNV", "MNV": "MNV", "small_del": "Del ≤50 bp", "small_ins": "Ins ≤50 bp",
                "large_indel": "Indel >50 bp", "complex_del": "Complex del.", "complex_ins": "Complex ins."}
EXCLUDE = {"All repeats", "Telomeres"}
RATE_COLUMNS = {"compartment": "Compartment", "class": "Variant class", "n_loci_Z": "Matched Z-DNA loci (n)",
                "n_loci_C": "Controls (n)", "n_Z": "Variants in Z-DNA windows (n)", "n_C": "Variants in control windows (n)",
                "bp_Z": "Z-DNA windows (bp)", "bp_C": "Control windows (bp)", "Z_per_kb": "Variants per kb, Z-DNA",
                "C_per_kb": "Variants per kb, controls", "rate_ratio": "Rate ratio (Z-DNA/control)"}
STATS_COLUMNS = {"compartment": "Compartment", "class": "Variant class", "rate_ratio": "Rate ratio (Z-DNA/control)",
                 "ci_lo": "95% CI lower", "ci_hi": "95% CI upper", "p": "Bootstrap P value", "q": "q value (BH)",
                 "sig": "Significance", "n_chrom": "Chromosomes (n)"}
REP_COLUMNS = {"cohort": "Set", "class": "Variant class", "n": "Variants (n)", "mean_call_rate": "Mean haplotype call rate",
               "frac_multiallelic": "Fraction at multiallelic sites", "frac_singleton": "Fraction singletons",
               "frac_high_conf": "Fraction with call rate >= 0.90"}
SET_NAMES = {"Z_matched": "Matched Z-DNA", "control": "Matched control"}
LOG2_LIM = 3.0
HEAT_SIZE = (8.2, 3.3)
REP_SIZE = (9, 2.2)
GAP_LW = 5
FS_TICK, FS_CELL, FS_AXIS, FS_LEGEND = 10, 8, 11, 10
AXIS_W, TICK_W, TICK_LEN = 1.1, 0.9, 3
GRID = dict(linestyle="--", linewidth=0.4, color="#B0B0B0", alpha=0.5, zorder=0)
FONT, DPI = "Nimbus Sans", 600
CMAP = LinearSegmentedColormap.from_list("div", ["#3C5A78", "#F4F4F4", "#B5654A"])
Z_COLOR, C_COLOR = "#84508C", "#A6A6A6"

plt.rcParams.update({"font.family": FONT, "pdf.fonttype": 42, "ps.fonttype": 42, "axes.linewidth": AXIS_W,
                     "axes.spines.top": False, "axes.spines.right": False})
PLOTS.mkdir(exist_ok=True)
TESTS.mkdir(exist_ok=True)
rates = pd.read_csv(RATES, sep="\t")
section = pd.read_csv(MEMBERSHIP_TABLE, sep="\t").drop_duplicates("group").set_index("group").section
nrm = TwoSlopeNorm(vmin=-LOG2_LIM, vcenter=0, vmax=LOG2_LIM)


def save(fig, stem, tight=False):
    for ext in ("pdf", "png"):
        fig.savefig(PLOTS / f"{stem}.{ext}", dpi=DPI if ext == "png" else None, bbox_inches="tight" if tight else None)
    plt.close(fig)


for vs, stem in HEATMAPS.items():
    r = rates[
        (rates.variant_set == vs)
        & rates.tested
        & rates.p.notna()
        & ~rates.compartment.isin(EXCLUDE)
    ]
    cols = [c for c in r.compartment.drop_duplicates()]
    cols = ["All loci"] + [c for c in cols if c != "All loci"]
    rows = [c for c in CLASSES if c in set(r["class"])]
    V = r.pivot(index="class", columns="compartment", values="rate_ratio").reindex(index=rows, columns=cols)
    S = r.pivot(index="class", columns="compartment", values="sig").reindex(index=rows, columns=cols)
    L = np.log2(V.astype(float))
    fig, ax = plt.subplots(figsize=HEAT_SIZE, constrained_layout=True)
    ax.imshow(np.ma.masked_invalid(np.clip(L.to_numpy(), -LOG2_LIM, LOG2_LIM)), cmap=CMAP.with_extremes(bad="#E6E6E6"),
              norm=nrm, aspect="auto", interpolation="nearest")
    for i in range(len(rows)):
        for j in range(len(cols)):
            v = V.iloc[i, j]
            if pd.isna(v):
                continue
            rr, gg, bb, _ = CMAP(nrm(np.clip(L.iloc[i, j], -LOG2_LIM, LOG2_LIM)))
            s = S.iloc[i, j]
            ax.text(j, i, f"{v:.2f}" + (f"\n{s}" if s in ("*", "**", "***") else ""), ha="center", va="center",
                    fontsize=FS_CELL, linespacing=0.9, color="black" if 0.299 * rr + 0.587 * gg + 0.114 * bb > 0.5 else "white")
    ax.set_xticks(range(len(cols)),
              [str(c).replace(" ", "\n") if len(str(c).split()) == 2 else c for c in cols],
              rotation=45, ha="right", rotation_mode="anchor", fontsize=FS_TICK)
    ax.set_yticks(range(len(rows)), [CLASS_LABELS[c] for c in rows], fontsize=FS_TICK)
    ax.set_xticks(np.arange(-0.5, len(cols)), minor=True)
    ax.set_yticks(np.arange(-0.5, len(rows)), minor=True)
    ax.grid(which="minor", color="white", lw=1.0)
    sec = ["all"] + [section.get(c, "") for c in cols[1:]]
    for j in range(1, len(cols)):
        if sec[j] != sec[j - 1]:
            ax.axvline(j - 0.5, color="white", lw=GAP_LW)
    ax.tick_params(which="both", length=0)
    for s_ in ax.spines.values():
        s_.set_visible(False)
    save(fig, stem)
    out = r[r.compartment.isin(cols) & r["class"].isin(rows)]
    out = out.assign(**{"class": out["class"].map(CLASS_LABELS)})
    out[list(RATE_COLUMNS)].rename(columns=RATE_COLUMNS).to_csv(TESTS / f"{stem}.tsv", sep="\t", index=False, float_format="%.6g")
    out[list(STATS_COLUMNS)].rename(columns=STATS_COLUMNS).to_csv(TESTS / f"{stem}_statistics.tsv", sep="\t", index=False,
                                                                  float_format="%.6g")
    print(f"{stem}: {len(rows)} classes x {len(cols)} compartments")

cf = plt.figure(figsize=(1.0, 2.0))
cax = cf.add_axes([0.1, 0.05, 0.18, 0.9])
cb = cf.colorbar(ScalarMappable(norm=nrm, cmap=CMAP), cax=cax)
cb.set_label("log$_2$ rate ratio (Z-DNA / control)", fontsize=FS_AXIS)
cb.ax.tick_params(labelsize=FS_TICK)
cb.outline.set_linewidth(0.5)
save(cf, f"{HEATMAPS['all']}_colorbar", tight=True)

rep = pd.read_csv(REP, sep="\t")
rows = [c for c in CLASSES if c in set(rep["class"])]
y = np.arange(len(rows))[::-1]
fig, axs = plt.subplots(1, 3, figsize=REP_SIZE, sharey=True, constrained_layout=True)
for ax, col, lab in zip(axs, ("mean_call_rate", "frac_multiallelic", "frac_singleton"),
                        ("Mean haplotype call rate", "Fraction multiallelic", "Fraction singletons")):
    for cohort, colr, dy in (("Z_matched", Z_COLOR, 0.12), ("control", C_COLOR, -0.12)):
        d = rep[rep.cohort == cohort].set_index("class").reindex(rows)
        ax.scatter(d[col], y + dy, s=42, color=colr, edgecolor="black", linewidth=0.7, zorder=3)
    ax.set_xlabel(lab, fontsize=FS_AXIS)
    ax.tick_params(labelsize=FS_TICK, width=TICK_W, length=TICK_LEN)
    ax.grid(True, axis="x", which="major", **GRID)
    ax.set_axisbelow(True)
axs[0].set_yticks(y, [CLASS_LABELS[c] for c in rows])
save(fig, REP_STEM)
lf, lax = plt.subplots(figsize=(2.2, 0.6))
lax.axis("off")
lax.legend(handles=[Line2D([0], [0], ls="none", marker="o", ms=6.6, mfc=Z_COLOR, mec="black", mew=0.75, label="Z-DNA"),
                    Line2D([0], [0], ls="none", marker="o", ms=6.6, mfc=C_COLOR, mec="black", mew=0.75, label="Matched control")],
           loc="center", frameon=False, fontsize=FS_LEGEND)
save(lf, f"{REP_STEM}_legend", tight=True)
rep.assign(cohort=rep.cohort.map(SET_NAMES), **{"class": rep["class"].map(CLASS_LABELS)})[list(REP_COLUMNS)] \
    .rename(columns=REP_COLUMNS).to_csv(TESTS / f"{REP_STEM}.tsv", sep="\t", index=False, float_format="%.6g")


sf, ax = plt.subplots(figsize=(2.2, 1.2))
ax.axis("off")

ax.text(
    0.05, 0.95,
    "Significance (BH-adjusted)\n"
    "***   q < 0.001\n"
    "**     q < 0.01\n"
    "*       q < 0.05",
    transform=ax.transAxes,
    ha="left", va="top",
    fontsize=FS_LEGEND,
    linespacing=1.5
)

save(sf, "fig5a_compartment_significance_legend", tight=True)
```

#### Fig. 5b

```python
"""Fig. 5b - Z-DNA/control SNV rate ratio by allele frequency in all loci and nine compartments (2 x 5 panels):
pooled ratio with 95% chromosome-bootstrap CI and per-chromosome ratios; trend q per panel. Reads 19.
"""
from pathlib import Path

import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from matplotlib.ticker import FixedLocator, FuncFormatter, NullLocator

BASE = Path("/path/to/pangenomes_zdna")
RATES, BYCHROM, TESTS_IN = (BASE / f"19_snv_af_{k}.tsv" for k in ("rates", "bychrom", "tests"))
PLOTS, TESTS = BASE / "plots", BASE / "tests"
STEM = "fig5b_snv_allele_frequency"
COLUMNS = {"record": "Record", "compartment": "Compartment", "bin_label": "Allele-frequency bin", "chrom": "Chromosome",
           "n_Z": "SNVs in Z-DNA windows (n)", "n_C": "SNVs in control windows (n)", "bp_Z": "Z-DNA windows (bp)",
           "bp_C": "Control windows (bp)", "Z_per_kb": "SNVs per kb, Z-DNA", "C_per_kb": "SNVs per kb, controls",
           "rate_ratio": "Rate ratio (Z-DNA/control)", "ci_lo": "95% CI lower", "ci_hi": "95% CI upper"}
STATS_COLUMNS = {"compartment": "Compartment", "test": "Test", "slope_log2": "Slope of log2 rate ratio per bin",
                 "ci_lo": "Slope, 95% CI lower", "ci_hi": "Slope, 95% CI upper", "p": "Bootstrap P value",
                 "q": "q value (BH)", "sig": "Significance", "n_bins": "Bins (n)", "n_boots": "Bootstrap replicates (n)",
                 "min_an": "Minimum allele number"}
N_ROWS, N_COLS = 2, 5
FIG_W, FIG_H = 11.8, 5.4
Y_TICKS = [0.5, 0.75, 1, 1.5, 2, 3, 4]
Y_PAD_LOW, Y_PAD_HIGH = 0.92, 1.45
JITTER = 0.13
POINT_S, POINT_ALPHA, POINT_EDGE_LW = 18, 0.75, 0.4
MS, MARKER_EDGE_LW, ERR_LW, CAPSIZE = 7.5, 0.8, 1.3, 3
AXIS_W, TICK_W, TICK_LEN = 1.1, 0.9, 3
FS_TICK, FS_XTICK, FS_AXIS, FS_TITLE, FS_NOTE = 11.5, 11, 12.5, 12.5, 11
X_ROT = 45
GRID = dict(linestyle="--", linewidth=0.4, color="#B0B0B0", alpha=0.5, zorder=0)
FONT, DPI = "Nimbus Sans", 600
Z_COLOR, POINT_COLOR = "#B3A7E0", "#9A9A9A"
SHORT = {"Singleton": "Singleton", "<1% (AC≥2)": "<1%", "1%–5%": "1–5%", "≥5%": "≥5%"}

plt.rcParams.update({"font.family": FONT, "pdf.fonttype": 42, "ps.fonttype": 42, "axes.linewidth": AXIS_W,
                     "axes.spines.top": False, "axes.spines.right": False})
PLOTS.mkdir(exist_ok=True)
TESTS.mkdir(exist_ok=True)
rates, bych, tests = (pd.read_csv(p, sep="\t") for p in (RATES, BYCHROM, TESTS_IN))
comps = list(dict.fromkeys(rates.compartment))
if len(comps) > N_ROWS * N_COLS:
    raise SystemExit(f"{len(comps)} compartments do not fit a {N_ROWS} x {N_COLS} grid")
bins = rates.drop_duplicates("bin").sort_values("bin")
rng = np.random.default_rng(0)


def sci(p):
    if not np.isfinite(p):
        return "n.a."
    if p >= 0.001:
        return f"{p:.3f}"
    m, e = f"{p:.1e}".split("e")
    return f"{m} × 10$^{{{int(e)}}}$"


pts = bych.loc[bych.rate_ratio > 0, "rate_ratio"]
y_lo = min(pts.min(), rates.ci_lo.min(), 1) * Y_PAD_LOW
y_hi = max(pts.max(), rates.ci_hi.max(), 1) * Y_PAD_HIGH

fig, axs = plt.subplots(N_ROWS, N_COLS, figsize=(FIG_W, FIG_H), sharex=True, sharey=True, constrained_layout=True)
axes = axs.ravel()
for k, (ax, comp) in enumerate(zip(axes, comps)):
    c = bych[(bych.compartment == comp) & (bych.rate_ratio > 0)]
    ax.scatter(c.bin + rng.uniform(-JITTER, JITTER, len(c)), c.rate_ratio, s=POINT_S, color=POINT_COLOR, alpha=POINT_ALPHA,
               edgecolor="black", linewidth=POINT_EDGE_LW, zorder=2)
    d = rates[rates.compartment == comp].sort_values("bin")
    ok = d.rate_ratio.notna()
    ax.errorbar(d.bin[ok], d.rate_ratio[ok], yerr=[(d.rate_ratio - d.ci_lo)[ok].clip(lower=0),
                                                   (d.ci_hi - d.rate_ratio)[ok].clip(lower=0)],
                fmt="o", ms=MS, mfc=Z_COLOR, mec="black", mew=MARKER_EDGE_LW, ecolor="black", elinewidth=ERR_LW,
                capsize=CAPSIZE, capthick=ERR_LW, zorder=4)
    ax.axhline(1, ls=":", color="black", lw=0.9, zorder=1)
    q = tests.set_index("compartment").q.get(comp, np.nan)
    ax.set_title(comp, fontsize=FS_TITLE, pad=4)
    ax.text(0.5, 0.985, f"q = {sci(q)}", transform=ax.transAxes, ha="center", va="top", fontsize=FS_NOTE)
    ax.tick_params(labelsize=FS_TICK, width=TICK_W, length=TICK_LEN)
    ax.grid(True, axis="y", which="major", **GRID)
    ax.set_axisbelow(True)
    if k // N_COLS == N_ROWS - 1 or k + N_COLS >= len(comps):
        ax.set_xticks(bins.bin, [SHORT.get(b, b) for b in bins.bin_label], rotation=X_ROT, ha="right",
                      rotation_mode="anchor", fontsize=FS_XTICK)
        ax.tick_params(axis="x", labelbottom=True)
    else:
        ax.tick_params(axis="x", labelbottom=False)
    if k % N_COLS == 0:
        ax.set_ylabel("SNV rate ratio\n(Z-DNA / control)", fontsize=FS_AXIS)
for ax in axes[len(comps):]:
    ax.set_visible(False)
ax0 = axes[0]
ax0.set_xlim(bins.bin.min() - 0.5, bins.bin.max() + 0.5)
ax0.set_yscale("log")
ax0.set_ylim(y_lo, y_hi)
ax0.yaxis.set_major_locator(FixedLocator([v for v in Y_TICKS if y_lo <= v <= y_hi]))
ax0.yaxis.set_minor_locator(NullLocator())
ax0.yaxis.set_major_formatter(FuncFormatter(lambda v, _: f"{v:g}"))
fig.supxlabel("Allele frequency", fontsize=FS_AXIS)
for ext in ("pdf", "png"):
    fig.savefig(PLOTS / f"{STEM}.{ext}", dpi=DPI if ext == "png" else None)
plt.close(fig)
d = pd.concat([rates.assign(record="Pooled"), bych.assign(record="Chromosome")], ignore_index=True)
d[[c for c in COLUMNS if c in d.columns]].rename(columns=COLUMNS).to_csv(TESTS / f"{STEM}.tsv", sep="\t", index=False,
                                                                        float_format="%.6g")
tests[[c for c in STATS_COLUMNS if c in tests.columns]].rename(columns=STATS_COLUMNS).to_csv(
    TESTS / f"{STEM}_statistics.tsv", sep="\t", index=False, float_format="%.6g")
```

#### Fig. 5c

```python
"""Fig. 5c - Z-DNA/control positional enrichment (+/-XLIM bp, log scale, 95% chromosome-bootstrap bands) around
ID83-derived deletion and insertion groups + legend. Reads 20.3.
"""
from pathlib import Path
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from matplotlib.lines import Line2D
from matplotlib.ticker import FixedLocator, FuncFormatter, NullLocator

BASE = Path('/path/to/pangenomes_zdna')
PROFILES = BASE / '20_id83_zc_profiles.tsv.gz'
PLOTS, TESTS = BASE / 'plots', BASE / 'tests'
STEM = 'fig5c_id83_metaprofiles'
GROUP_NAMES = {'del_1bp_homopolymer': '1-bp deletions', 'ins_1bp_homopolymer': '1-bp insertions',
               'del_repeat_unit': 'Repeat-unit deletions', 'ins_repeat_unit': 'Repeat-unit insertions',
               'del_microhomology': 'Microhomology deletions', 'del_no_repeat': 'No-repeat deletions',
               'ins_no_repeat': 'No-repeat insertions'}
COLUMNS = {'group': 'Indel group', 'offset': 'Offset (bp)', 'n_anchors': 'Indel anchors (n)',
           'count_Z': 'Anchors overlapping Z-DNA at offset (n)', 'count_C': 'Anchors overlapping controls at offset (n)',
           'bp_Z': 'Z-DNA (bp)', 'bp_C': 'Controls (bp)', 'rate_ratio': 'Positional enrichment (Z-DNA/control)',
           'ci_lo': '95% CI lower', 'ci_hi': '95% CI upper', 'n_valid_boot': 'Valid bootstrap replicates (n)'}
DPI, FONT = 600, 'Nimbus Sans'
XLIM = 200
PROFILE_TICKS = [0.25, 0.5, 1, 2, 5, 10, 20, 50]
PANELS = [
    ('Deletions', ['del_repeat_unit', 'del_microhomology', 'del_no_repeat', 'del_1bp_homopolymer']),
    ('Insertions', ['ins_repeat_unit', 'ins_no_repeat', 'ins_1bp_homopolymer']),
]
COLORS = {
    'del_repeat_unit': '#6E5BC4', 'ins_repeat_unit': '#6E5BC4',
    'del_microhomology': '#B5654A',
    'del_no_repeat': '#7D7D7D', 'ins_no_repeat': '#7D7D7D',
    'del_1bp_homopolymer': '#3C5A78', 'ins_1bp_homopolymer': '#3C5A78',
}
GRID = dict(linestyle='--', linewidth=0.4, color='#B0B0B0', alpha=0.5)
plt.rcParams.update({
    'font.family': FONT, 'pdf.fonttype': 42, 'ps.fonttype': 42,
    'axes.linewidth': 1.1, 'axes.spines.top': False, 'axes.spines.right': False,
})


def save(fig, stem, tight=False):
    for ext in ('pdf', 'png'):
        fig.savefig(PLOTS / f'{stem}.{ext}', dpi=DPI if ext == 'png' else None,
                    bbox_inches='tight' if tight else None)
    plt.close(fig)


def main():
    if not PROFILES.exists():
        raise FileNotFoundError(PROFILES)
    PLOTS.mkdir(parents=True, exist_ok=True)
    TESTS.mkdir(parents=True, exist_ok=True)
    prof = pd.read_csv(PROFILES, sep='\t')
    required = {'group', 'offset', 'rate_ratio', 'ci_lo', 'ci_hi', 'n_anchors'}
    if missing := required - set(prof):
        raise ValueError(f'Missing profile columns: {sorted(missing)}')
    prof = prof.loc[prof.offset.abs() <= XLIM].copy()

    fig, axs = plt.subplots(1, 2, figsize=(8.2, 3.0), sharey=True, constrained_layout=True)
    for ax, (title, groups) in zip(axs, PANELS):
        for group in groups:
            d = prof.loc[prof.group.eq(group)].sort_values('offset')
            if d.empty:
                raise RuntimeError(f'Missing metaprofile: {group}')
            if len(d) != 2 * XLIM + 1:
                raise RuntimeError(f'Incomplete offset coverage for {group}: {len(d)} positions')
            x = d.offset.to_numpy(float)
            mid, lo, hi = [d[c].to_numpy(float) for c in ('rate_ratio', 'ci_lo', 'ci_hi')]
            if not (np.isfinite(mid).all() and np.isfinite(lo).all() and np.isfinite(hi).all()):
                raise RuntimeError(f'Nonfinite profile values in {group}')
            col = COLORS[group]
            ax.fill_between(x, lo, hi, color=col, alpha=0.20, lw=0, zorder=2)
            ax.plot(x, mid, color=col, lw=1.65, zorder=3)
        ax.axvline(0, ls='--', color='black', lw=0.8, zorder=1)
        ax.axhline(1, ls=':', color='black', lw=0.8, zorder=1)
        ax.set_xlim(-XLIM, XLIM)
        ax.set_yscale('log')
        ax.set_title(title, fontsize=10.5, loc='left')
        ax.set_xlabel('Distance from indel anchor (bp)', fontsize=10.5)
        ax.tick_params(labelsize=9.5, width=0.9, length=3)
        ax.grid(True, axis='both', which='major', **GRID)
        ax.set_axisbelow(True)
    val = prof[['rate_ratio', 'ci_lo', 'ci_hi']].to_numpy(float)
    val = val[np.isfinite(val) & (val > 0)]
    ylo = max(0.1, float(val.min()) * 0.8)
    yhi = float(val.max()) * 1.12
    axs[0].set_ylim(ylo, yhi)
    axs[0].yaxis.set_major_locator(FixedLocator([v for v in PROFILE_TICKS if ylo <= v <= yhi]))
    axs[0].yaxis.set_minor_locator(NullLocator())
    axs[0].yaxis.set_major_formatter(FuncFormatter(lambda v, _: f'{v:g}'))
    axs[0].set_ylabel('Z-DNA/control positional enrichment', fontsize=10.5)
    save(fig, STEM)

    prof.assign(group=prof.group.map(GROUP_NAMES))[list(COLUMNS)].rename(columns=COLUMNS).to_csv(
        TESTS / f'{STEM}.tsv', sep='\t', index=False, float_format='%.8g')
    lf, la = plt.subplots(figsize=(2.7, 1.6))
    la.axis('off')
    mechanisms = [('Repeat unit', '#6E5BC4'), ('Microhomology (deletions)', '#B5654A'),
                  ('No repeat', '#7D7D7D'), ('1-bp indel', '#3C5A78')]
    la.legend(handles=[Line2D([0], [0], color=color, lw=2, label=label)
                       for label, color in mechanisms],
              loc='center', frameon=False, fontsize=9)
    save(lf, STEM + '_legend', tight=True)


if __name__ == '__main__':
    main()
```

#### Suppl. Fig. 3c

```python
"""Suppl. Fig. 3c - % of controls, matched and unmatched Z-DNA loci overlapping each compartment (bars) and % of genome
bp in the compartment (line); genes and repeats separated + legend. Reads 18.1.
"""
from pathlib import Path

import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from matplotlib.lines import Line2D
from matplotlib.patches import Patch
import matplotlib.patheffects as pe

BASE = Path("/path/to/pangenomes_zdna")
TABLE = BASE / "18_feature_membership.tsv"
PLOTS, TESTS = BASE / "plots", BASE / "tests"
STEM = "suppl_fig3c_feature_membership"
COLUMNS = {"section": "Section", "group": "Compartment", "cohort": "Set", "n_loci": "Loci in set (n)",
           "n_in_group": "Loci overlapping compartment (n)", "pct_loci": "Loci overlapping compartment (%)",
           "genome_pct_bp": "Genome bp in compartment (%)"}
COHORTS = ["control", "Z_matched", "Z_unmatched"]
LABELS = {"control": "Controls", "Z_matched": "Matched Z-DNA", "Z_unmatched": "Unmatched Z-DNA"}
COLORS = {"control": "#A6A6A6", "Z_matched": "#B3A7E0", "Z_unmatched": "#513AB5"}
FIG_W, FIG_H = 9, 3.4
BAR_W, SECTION_GAP = 0.26, 0.8
AXIS_W, TICK_W, TICK_LEN = 1.1, 0.9, 3
FS_TICK, FS_XTICK, FS_AXIS, FS_SECTION, FS_LEGEND = 10, 9.5, 11, 10.5, 9.5
X_ROT = 45
GRID = dict(linestyle="--", linewidth=0.4, color="#B0B0B0", alpha=0.5, zorder=0)
FONT, DPI = "Nimbus Sans", 600

plt.rcParams.update({"font.family": FONT, "pdf.fonttype": 42, "ps.fonttype": 42, "axes.linewidth": AXIS_W,
                     "axes.spines.top": False, "axes.spines.right": False})
PLOTS.mkdir(exist_ok=True)
TESTS.mkdir(exist_ok=True)
t = pd.read_csv(TABLE, sep="\t")
groups = t.drop_duplicates("group")[["section", "group", "genome_pct_bp"]].reset_index(drop=True)
x = np.arange(len(groups), dtype=float) + np.where(groups.section == groups.section.iloc[0], 0, SECTION_GAP)

fig, ax = plt.subplots(figsize=(FIG_W, FIG_H), constrained_layout=True)
for k, c in enumerate(COHORTS):
    v = t[t.cohort == c].set_index("group").pct_loci.reindex(groups.group)
    ax.bar(x + (k - 1) * BAR_W, v, BAR_W, color=COLORS[c], edgecolor="black", linewidth=0.4, zorder=3)

ax.hlines(
    groups.genome_pct_bp,
    x - 1.5 * BAR_W,
    x + 1.5 * BAR_W,
    color="#B6BF65",
    lw=2.2,
    path_effects=[pe.Stroke(linewidth=3.4, foreground="black"), pe.Normal()],
    zorder=4
)

ax.set_xticks(
    x,
    [str(g).replace(" ", "\n") for g in groups.group],
    rotation=X_ROT,
    ha="right",
    rotation_mode="anchor",
    fontsize=FS_XTICK
)
ax.set_xlim(x.min() - 0.6, x.max() + 0.6)
ax.set_ylim(0, 100)
ax.set_ylabel("Loci overlapping the compartment (%)", fontsize=FS_AXIS)

for sec, label in (("genes", "Genes"), ("repeats", "Repeats")):
    xs = x[groups.section == sec]
    if len(xs):
        ax.text(xs.mean(), 103, label, ha="center", va="bottom", fontsize=FS_SECTION)

ax.tick_params(axis="y", labelsize=FS_TICK)
ax.tick_params(width=TICK_W, length=TICK_LEN)
ax.grid(True, axis="y", which="major", **GRID)
ax.set_axisbelow(True)

for ext in ("pdf", "png"):
    fig.savefig(PLOTS / f"{STEM}.{ext}", dpi=DPI if ext == "png" else None)
plt.close(fig)

h = [Patch(facecolor=COLORS[c], edgecolor="black", linewidth=0.4, label=LABELS[c]) for c in COHORTS]
h.append(
    Line2D(
        [0], [0],
        color="#B6BF65",
        lw=2.2,
        path_effects=[pe.Stroke(linewidth=3.4, foreground="black"), pe.Normal()],
        label="Genome (% of bp)"
    )
)

lf, lax = plt.subplots(figsize=(2.2, 1.0))
lax.axis("off")
lax.legend(handles=h, loc="center", frameon=False, fontsize=FS_LEGEND)

for ext in ("pdf", "png"):
    lf.savefig(PLOTS / f"{STEM}_legend.{ext}", dpi=DPI if ext == "png" else None, bbox_inches="tight")
plt.close(lf)

t.assign(cohort=t.cohort.map(LABELS), section=t.section.str.capitalize())[list(COLUMNS)].rename(columns=COLUMNS).to_csv(
    TESTS / f"{STEM}.tsv", sep="\t", index=False, float_format="%.6g")
```

## Supplementary Data 1

```python
"""Supplementary Data 1 - every tests/ TSV as one sheet of Supplementary_Data_1.xlsx (main figures, then supplementary;
sheet = panel, '<panel> statistics' for test results)."""
import importlib.util
import re
from pathlib import Path
import tempfile
import pandas as pd

BASE = Path("/path/to/pangenomes_zdna")
TESTS = BASE / "tests"
OUT = BASE / "Supplementary_Data_1.xlsx"
MAX_ROWS = 1_048_575
MAX_NAME = 31


def parse(stem):
    m = re.match(r"(suppl_)?fig(\d+)([a-z])_(.+)$", stem)
    if not m:
        raise ValueError(f"Not a panel file name: {stem}")
    return bool(m.group(1)), int(m.group(2)), m.group(3), m.group(4)


def sheet_names(stems):
    by_panel = {}
    for st in stems:
        by_panel.setdefault(parse(st)[:3], []).append(st)
    names = {}
    for (sup, fig, pan), group in by_panel.items():
        label = f"{'Suppl. ' if sup else ''}Fig. {fig}{pan}"
        data = [st for st in group if not st.endswith("_statistics")]
        tokens = [parse(st)[3].split("_") for st in data]
        k = 0
        while len(data) > 1 and all(len(t) > k for t in tokens) and len({t[k] for t in tokens}) == 1:
            k += 1
        for st in group:
            if st.endswith("_statistics"):
                names[st] = f"{label} statistics"
                continue
            words = parse(st)[3].split("_")[k:] if len(data) > 1 else []
            while len(" ".join([label] + words)) > MAX_NAME:
                words = words[:-1]
            names[st] = " ".join([label] + words)
    seen = {}
    for st in list(names):
        seen[names[st]] = seen.get(names[st], 0) + 1
        if seen[names[st]] > 1:
            names[st] = f"{names[st][:MAX_NAME - 4]} ({seen[names[st]]})"
    return names


def main():
    files = sorted(TESTS.glob("*.tsv"), key=lambda p: (parse(p.stem)[:3], p.stem.endswith("_statistics"), p.stem))
    names = sheet_names([p.stem for p in files])
    skipped = []
    with tempfile.TemporaryDirectory(dir=BASE) as tmp:
        if importlib.util.find_spec("xlsxwriter"):
            writer = pd.ExcelWriter(OUT, engine="xlsxwriter",
                                    engine_kwargs={"options": {"tmpdir": tmp}})
        else:
            writer = pd.ExcelWriter(OUT, engine="openpyxl")
        with writer:
            for p in files:
                d = pd.read_csv(p, sep="\t", low_memory=False)
                if len(d) > MAX_ROWS:
                    skipped.append(f"{p.name} ({len(d):,} rows)")
                    continue
                d.to_excel(writer, sheet_name=names[p.stem], index=False)
    print(f"{len(files) - len(skipped)} sheets -> {OUT.name}" + (f"; too long for Excel: {', '.join(skipped)}" if skipped else ""))


if __name__ == "__main__":
    main()
```
